## Introduction

This project is an independent, professional-grade implementation of the ideas, methodologies, and experimental protocols from the paper titled **"Beyond Benchmark Rankings: Information-Aware and Risk-Aware Optimization of Large Language Model Ensembles (A Concept Paper and Methodology)"** by the author:
*   **Craig S. Chirinda (CS Chirinda)**

This repository provides a complete, end-to-end computational framework for replicating the paper's findings. It challenges the prevailing paradigm of deploying Large Language Models (LLMs) based solely on standalone benchmark leaderboards. Instead, it treats LLM orchestration as a rigorous portfolio-construction problem. By implementing the **Final Marginal Ensemble Contribution (FMEC)** framework, this codebase delivers a highly optimized pipeline that evaluates models based on their expected marginal utility, conditional informational novelty, and correlated-failure risk, ultimately selecting the optimal ensemble under strict cost, latency, and risk budgets.

## Theoretical Background

The implemented methods bridge machine learning systems engineering with advanced cooperative game theory, information theory, and financial econometrics.

**1. Cooperative Game Theory (Marginal Utility):**
The pipeline moves beyond standalone scoring by evaluating models based on their conditional value to an ensemble. It implements a permutation-based Monte Carlo Shapley estimator to calculate the expected marginal utility contribution ($\widehat{\mathrm{MEC}}_j$) of each model across thousands of potential coalitions, ensuring an unbiased, order-symmetric valuation.

**2. Information Theory (Informational Novelty):**
To distinguish true complementary capability from mere disagreement, the framework quantifies informational novelty using outcome-based Conditional Mutual Information (CMI). It implements Laplace smoothing and Miller-Madow finite-sample bias corrections over the correct-count sufficient statistic ($c_S$) to compute the conditional information gain ($\mathrm{IG}_j$), preventing spurious diversity claims.

**3. Portfolio Theory & Robust Statistics (Correlated Risk):**
Translating Markowitz portfolio theory to LLM errors, the pipeline constructs a multidimensional error taxonomy. It estimates the error covariance matrix ($\Sigma$) using Ledoit-Wolf shrinkage targeting a block-diagonal geometry, penalizing models that fail in the same way. This yields the Marginal Risk Contribution ($\mathrm{MRC}_j$).

**4. Constrained Submodular Optimization:**
The FMEC scores are aggregated into a deployment utility function. The pipeline executes a budget-feasible greedy algorithm to maximize portfolio utility subject to strict, multi-dimensional knapsack constraints (cost, latency, and risk), identifying the optimal ensemble on the efficient frontier.

## Features

-   **End-to-End Automation:** A singular orchestrator function (`run_experimental_suite`) executes the entire research protocol from raw data ingestion and contamination auditing to cryptographic packaging and LaTeX report generation.
-   **Strict Data Governance:** Implements a Tier-1 reproducibility gate, canonicalizing the outcome tensor (fixed column order, float precision) and computing a SHA-256 hash to guarantee bit-exact computational reproducibility.
-   **Contamination Auditing:** Enforces a mandatory precondition for valid inference by executing 13-gram lexical overlap and behavioral canary probes, applying a global drop-for-all exclusion rule to prune contaminated tasks.
-   **Secure Verification Engineering:** Grounds correctness deterministically for code-generation tasks by executing candidate solutions in an isolated, network-disabled, restricted-filesystem subprocess with strict timeouts.
-   **Hierarchical Uncertainty Quantification:** Implements a three-level nested cluster bootstrap to isolate and quantify task-sampling, generation, and adjudication variance ($\sigma^2_{\mathrm{task}} + \sigma^2_{\mathrm{gen}} + \sigma^2_{\mathrm{adj}}$), computing Bias-Corrected and Accelerated (BCa) intervals.

## Methodology Implemented

1.  **Data Ingestion & Validation:** Materializes raw records into strictly typed pandas DataFrames, enforcing relational integrity and schema contracts.
2.  **Contamination Audit:** Executes lexical and behavioral probes to filter the evaluation corpus, protecting the causal interpretation of complementarity.
3.  **Qualification & Stratification:** Applies the $P_{\min}$ benchmark gate and clusters models by capability profile ($k$-means) to ensure lineage diversity during coalition sampling.
4.  **Tensor Assembly:** Constructs the canonical outcome tensor, mathematically enforcing the error exclusivity invariant ($\sum e_* = 1 - \text{correct}$) and computing exact USD costs.
5.  **Estimation:** Computes marginal utility (Shapley), conditional information gain (CMI), and correlated-failure risk (Ledoit-Wolf shrinkage) to compose the FMEC scores.
6.  **Optimization:** Executes a budget-feasible greedy search to select the optimal model portfolio under cost, latency, and risk constraints.
7.  **Validation & Inference:** Runs the nested bootstrap, computes Recommendation-Stability Probability (RSP), and executes pre-registered hypothesis tests (H1-H5) with Holm-Bonferroni FWER control.
8.  **Robustness & Packaging:** Performs 10,000 Dirichlet preference sweeps, generates publication-grade figures, and serializes all artifacts into a cryptographically hashed archive.

## Core Components (Notebook Structure)

*Note: All orchestrator callables and their constituent helper functions are contained within a singular, comprehensive Jupyter Notebook (`fused_llm_router_draft.ipynb`).*

The notebook is structured as a logical Directed Acyclic Graph (DAG) of 24 distinct tasks:
1.  Environment, Configuration & Seed Initialization (Task 1-2)
2.  Raw Data Ingestion, Schema Validation & Cleansing (Tasks 3-4)
3.  Benchmark Contamination Audit & Qualification (Tasks 5-7)
4.  Orchestration Architecture & Response Collection (Tasks 8-9)
5.  Scoring, Adjudication & Tensor Assembly (Tasks 10-11)
6.  FMEC Estimation Core: MEC, IG, and Risk (Tasks 12-16)
7.  Constrained Ensemble Selection & Fusion Audit (Tasks 17-18)
8.  Nested Bootstrap Validation & Confirmatory Testing (Tasks 19-21)
9.  Robustness Sweeps, Reporting & Cryptographic Packaging (Tasks 22-24)

## Key Callable: `run_experimental_suite`

The project is designed around a single, top-level user-facing interface function:

-   **`run_experimental_suite`:** This apex orchestrator function coordinates the entire FMEC experimental protocol. A single call to this function ingests the raw data, executes the contamination audit, builds the canonical outcome tensor, estimates the FMEC primitives, solves the constrained optimization problem, runs the nested bootstrap and Dirichlet robustness sweeps, generates the scientific report and figures, and packages the results into an immutable `.tar.gz` archive. It guarantees complete state isolation and prevents look-ahead bias.

## Input Data Structure

The pipeline requires five primary raw data structures, adhering strictly to the schemas defined in Section 9.1 and 9.2 of the methodology:
1.  **`model_answer`**: A list of dictionaries containing raw free-form generation records (9 keys, including verbatim text and provenance).
2.  **`candidate_solution`**: A list of dictionaries containing raw structured (verifier-augmented) records (7 keys, including executable Python code).
3.  **`task_record`**: A list of dictionaries defining the evaluation substrate, ground truth, and scoring rules.
4.  **`outcomes`**: The canonical raw outcome-and-error tensor (17 columns). While the pipeline can assemble this from raw responses, it serves as the fundamental deterministic substrate for all downstream estimation.
5.  **`qualification`**: A pandas DataFrame containing benchmark eligibility scores and model lineage metadata.

Additionally, the pipeline requires a **`config.yaml`** file, which serves as the deterministic source of truth for all hyperparameters, optimization budgets, and reproducibility seeds.

## Usage

Here is the granular, step-by-step guide to executing the end-to-end pipeline for **"Beyond Benchmark Rankings: Information-Aware and Risk-Aware Optimization of Large Language Model Ensembles"**. This example demonstrates how to synthetically generate the required raw data structures, load the study configuration from a YAML file, and execute the full research pipeline using the `run_experimental_suite` orchestrator.

*Note: Assume that all the callables that have been defined in this conversation are in a single Jupyter notebook. And, that there is no folder with “.py” functions.*
*Note: Assume that the “config.yaml” file is saved in the working directory.*

### **Step 1: Synthetic Data Generation (`raw_fmec_data`)**

The first requirement is a high-fidelity synthetic representation of the raw data structures. These datasets must adhere strictly to the schemas and mathematical invariants defined in Section 9.1 and 9.2 of the study (e.g., the exclusivity invariant for error flags, and the disjointness invariant for evaluation tasks).

**Methodology:**
1.  **Qualification Generation:** We simulate a pool of candidate models with plausible benchmark percentiles, ensuring the schema exactly matches the 10 required columns.
2.  **Task Generation:** We generate a set of evaluation tasks spanning different domains, strictly enforcing the `is_held_out = True` invariant.
3.  **Response Generation:** We simulate $R=5$ replicate generations per (model, task) pair, populating token counts, latencies, and verbatim text payloads using the `Faker` library.
4.  **Outcome Tensor Generation:** We construct the canonical 17-column outcome tensor, mathematically enforcing the error exclusivity invariant ($\sum e_* = 1 - \text{correct}$).

```python
# Import the pandas library for DataFrame manipulation
import pandas as pd
# Import the numpy library for numerical operations and random sampling
import numpy as np
# Import the yaml library for configuration parsing
import yaml
# Import the os module for environment variable access
import os
# Import the Faker library to generate realistic synthetic text
from faker import Faker
# Import typing modules for strict type hinting
from typing import Dict, Any, List, Tuple

# Initialize the Faker instance to generate synthetic text payloads
fake = Faker()
# Set a deterministic seed for Faker to ensure reproducible text generation
Faker.seed(20260618)
# Set a deterministic seed for NumPy to ensure reproducible numerical generation
np.random.seed(20260618)

def generate_synthetic_fmec_data(
    n_models: int = 5,
    n_tasks: int = 10,
    n_replicates: int = 5
) -> Tuple[List[Dict[str, Any]], List[Dict[str, Any]], List[Dict[str, Any]], pd.DataFrame, pd.DataFrame]:
    """
    Generates high-fidelity synthetic data structures required for the FMEC pipeline.

    Purpose:
        To create mathematically plausible mock datasets that strictly adhere to the
        schemas defined in Section 9.1 and 9.2 of the FMEC methodology. This allows
        for the execution and testing of the pipeline without requiring expensive
        live LLM API calls.

    Inputs:
        n_models (int): The number of candidate models to simulate. Default is 5.
        n_tasks (int): The number of evaluation tasks to simulate. Default is 10.
        n_replicates (int): The number of generation replicates (R). Default is 5.

    Processes:
        1.  Qualification Generation: Creates a DataFrame of models with benchmark scores.
        2.  Task Generation: Creates a list of dictionaries defining held-out tasks.
        3.  Response Generation: Simulates free-form and structured LLM outputs.
        4.  Outcome Tensor Generation: Constructs a DataFrame enforcing the error
            exclusivity invariant (correct == 1 implies all e_* == 0).

    Outputs:
        Tuple containing:
            - model_answer (List[Dict]): Raw free-form generation records (9 keys).
            - candidate_solution (List[Dict]): Raw structured solution records (7 keys).
            - task_record (List[Dict]): Raw task and ground truth records (8 keys).
            - outcomes (pd.DataFrame): Canonical raw outcome-and-error tensor (17 columns).
            - qualification (pd.DataFrame): Raw benchmark-eligibility table (10 columns).

    Raises:
        TypeError: If input parameters are not integers.
        ValueError: If input dimensions are less than 1.
    """
    # Validate that n_models is an integer
    if not isinstance(n_models, int):
        # Raise a TypeError if the validation fails
        raise TypeError("n_models must be an integer.")
    # Validate that n_models is strictly positive
    if n_models < 1:
        # Raise a ValueError to prevent generating empty datasets
        raise ValueError("n_models must be at least 1.")
    
    # Validate that n_tasks is an integer
    if not isinstance(n_tasks, int):
        # Raise a TypeError if the validation fails
        raise TypeError("n_tasks must be an integer.")
    # Validate that n_tasks is strictly positive
    if n_tasks < 1:
        # Raise a ValueError to prevent generating empty datasets
        raise ValueError("n_tasks must be at least 1.")
        
    # Validate that n_replicates is an integer
    if not isinstance(n_replicates, int):
        # Raise a TypeError if the validation fails
        raise TypeError("n_replicates must be an integer.")
    # Validate that n_replicates is strictly positive
    if n_replicates < 1:
        # Raise a ValueError to prevent generating empty datasets
        raise ValueError("n_replicates must be at least 1.")

    # Define a list of synthetic model identifiers
    model_ids: List[str] = [f"model_{chr(65+i)}" for i in range(n_models)]
    # Define a list of synthetic task identifiers
    task_ids: List[str] = [f"task_{str(i).zfill(3)}" for i in range(n_tasks)]

    # --- 1. Generate Qualification DataFrame ---
    # Initialize an empty list to hold qualification records
    qual_records: List[Dict[str, Any]] = []
    # Iterate over each synthetic model ID
    for m_id in model_ids:
        # Simulate a composite benchmark percentile rank between 0.60 and 0.99
        pct_rank = float(np.random.uniform(0.60, 0.99))
        # Append the record dictionary matching the Section 9.2(e) schema exactly
        qual_records.append({
            "model_id": m_id,
            "family": str(np.random.choice(["Alpha", "Beta", "Gamma"])),
            "params_b": float(np.random.choice([70.0, 140.0, np.nan])),
            "mmlu": float(np.random.uniform(0.5, 0.9)),
            "gpqa": float(np.random.uniform(0.3, 0.7)),
            "humaneval": float(np.random.uniform(0.4, 0.9)),
            "swebench": float(np.random.uniform(0.1, 0.5)),
            "math": float(np.random.uniform(0.3, 0.8)),
            "arena_elo": int(np.random.uniform(1000, 1300)),
            "pct_rank": pct_rank
        })
    # Convert the list of dictionaries into a pandas DataFrame
    qualification = pd.DataFrame(qual_records)
    # Enforce strict categorical dtype for the primary key to optimize memory
    qualification["model_id"] = qualification["model_id"].astype("category")

    # --- 2. Generate Task Records ---
    # Initialize an empty list to hold task records
    task_record: List[Dict[str, Any]] = []
    # Iterate over each synthetic task ID
    for t_id in task_ids:
        # Randomly assign a domain from the allowed set
        domain = str(np.random.choice(["math", "code", "science"]))
        # Determine the scoring rule based on the domain
        rule = "unit_tests" if domain == "code" else "exact_match"
        # Append the record dictionary matching the Section 9.1(c) schema exactly
        task_record.append({
            "task_id": t_id,
            "domain": domain,
            "subdomain": "general",
            "prompt": fake.sentence(nb_words=10),
            "ground_truth": "42" if domain == "math" else "def solve(): pass",
            "scoring_rule": rule,
            "difficulty": "medium",
            "is_held_out": True  # Mandatory invariant for evaluation tasks (Assumption A1)
        })

    # --- 3. Generate Responses and Outcomes ---
    # Initialize the list for free-form answers
    model_answer: List[Dict[str, Any]] = []
    # Initialize the list for structured solutions
    candidate_solution: List[Dict[str, Any]] = []
    # Initialize the list for tensor rows
    outcome_rows: List[Dict[str, Any]] = []

    # Define the exact eight error columns for the taxonomy
    error_cols: List[str] = ["e_reason", "e_comp", "e_halluc", "e_ctx", "e_plan", "e_tool", "e_safety", "e_refusal"]

    # Iterate over every combination of task, model, and replicate
    for t_id in task_ids:
        # Iterate over models
        for m_id in model_ids:
            # Iterate over replicates
            for rep in range(n_replicates):
                # Simulate a realistic wall-clock latency in seconds
                latency = float(np.random.uniform(0.5, 5.0))
                # Simulate input token usage
                p_tokens = int(np.random.uniform(100, 1000))
                # Simulate output token usage
                c_tokens = int(np.random.uniform(50, 500))
                # Calculate a synthetic cost based on token usage
                cost = float((p_tokens * 0.0001) + (c_tokens * 0.0002))
                # Generate a deterministic seed for this specific call
                seed = int(20260618 + rep)

                # Append the free-form generation record matching Section 9.1(a) exactly
                model_answer.append({
                    "task_id": t_id,
                    "model_id": m_id,
                    "answer": fake.paragraph(nb_sentences=3),
                    "error": None,
                    "prompt_tokens": p_tokens,
                    "completion_tokens": c_tokens,
                    "latency_s": latency,
                    "temperature": 0.20,
                    "seed": seed
                })

                # Append the structured solution record matching Section 9.1(b) exactly
                candidate_solution.append({
                    "task_id": t_id,
                    "model_id": m_id,
                    "thesis": fake.sentence(),
                    "reasoning": fake.paragraph(),
                    "python_code": "print('hello')" if "code" in t_id else None,
                    "confidence": float(np.random.uniform(0.5, 1.0)),
                    "error": None
                })

                # Simulate a binary correctness indicator (e.g., 70% accuracy)
                is_correct = int(np.random.rand() > 0.3)
                
                # Initialize the error vector with all zeros
                e_vec: Dict[str, int] = {col: 0 for col in error_cols}
                
                # Enforce the mathematical exclusivity invariant
                if is_correct == 0:
                    # If incorrect, select exactly one error category uniformly at random
                    chosen_error = str(np.random.choice(error_cols))
                    # Set the chosen error flag to 1
                    e_vec[chosen_error] = 1

                # Construct the tensor row dictionary matching Section 9.2(d) exactly
                row: Dict[str, Any] = {
                    "task_id": t_id,
                    "model_id": m_id,
                    "rep": rep,
                    "correct": is_correct,
                    **e_vec,
                    "cost_usd": cost,
                    "latency_s": latency,
                    "provider_version": f"{m_id}-v1",
                    "system_fingerprint": f"fp_{np.random.randint(1000, 9999)}",
                    "seed": seed
                }
                # Append the row to the outcomes list
                outcome_rows.append(row)

    # Convert the list of outcome dictionaries into a pandas DataFrame
    outcomes = pd.DataFrame(outcome_rows)
    
    # Enforce strict categorical dtypes for identifiers to optimize memory
    outcomes["task_id"] = outcomes["task_id"].astype("category")
    # Enforce strict categorical dtypes for model identifiers
    outcomes["model_id"] = outcomes["model_id"].astype("category")
    # Enforce strict int8 dtypes for binary flags
    outcomes["correct"] = outcomes["correct"].astype("int8")
    # Iterate over error columns to enforce int8
    for col in error_cols:
        # Cast each error column to int8
        outcomes[col] = outcomes[col].astype("int8")

    # Return the five generated data structures
    return model_answer, candidate_solution, task_record, outcomes, qualification

# Execute the synthetic data generation function
ma_raw, cs_raw, tr_raw, outcomes_df, qual_df = generate_synthetic_fmec_data()

# Generate a synthetic list of raw proxy texts for the contamination audit
raw_proxy_texts_list: List[str] = [fake.text() for _ in range(5)]
```

### **Step 2: Loading the Configuration (`config.yaml`)**

The study relies on a deterministic configuration file (`config.yaml`) that defines all hyperparameters, budgets, and reproducibility seeds. We read this file into memory and convert it into a usable Python dictionary.

**Methodology:**
1.  **File I/O:** Open `config.yaml` in read mode with explicit UTF-8 encoding.
2.  **Parsing:** Use `yaml.safe_load` to convert the YAML structure into a nested Python dictionary securely.
3.  **Validation:** Catch file existence errors and parsing errors, raising explicit exceptions to halt the pipeline if the configuration is corrupted or missing.

```python
def load_study_configuration(filepath: str = "config.yaml") -> Dict[str, Any]:
    """
    Loads the study configuration parameters from a YAML file into a Python dictionary.

    Purpose:
        To ingest the deterministic hyperparameters, optimization budgets, and
        reproducibility seeds defined in the external configuration file. This
        ensures strict separation of code and configuration.

    Inputs:
        filepath (str): The relative or absolute path to the YAML configuration file.
                        Default is "config.yaml".

    Processes:
        1.  File Access: Attempts to open the specified file in read mode.
        2.  Parsing: Uses PyYAML's safe_load to parse the YAML structure securely.
        3.  Validation: Catches and handles FileNotFoundError and YAMLError.

    Outputs:
        Dict[str, Any]: A nested dictionary containing the study configuration.

    Raises:
        TypeError: If filepath is not a string.
        FileNotFoundError: If the config.yaml file does not exist.
        yaml.YAMLError: If the file contains invalid YAML syntax.
    """
    # Validate that the provided filepath is a string
    if not isinstance(filepath, str):
        # Raise TypeError to prevent invalid file operations
        raise TypeError(f"filepath must be a string, got {type(filepath).__name__}.")

    try:
        # Open the file stream in read mode with UTF-8 encoding
        with open(filepath, "r", encoding="utf-8") as file:
            # Parse the YAML content safely into a Python dictionary
            config = yaml.safe_load(file)

        # Log success to the console for user visibility
        print(f"Successfully loaded configuration from {filepath}")

        # Return the parsed configuration dictionary
        return config

    except FileNotFoundError:
        # Raise an explicit error if the configuration file is missing
        raise FileNotFoundError(f"Critical Error: {filepath} not found in the working directory.")
    except yaml.YAMLError as e:
        # Handle YAML syntax errors explicitly
        print(f"Error parsing YAML file {filepath}: {e}")
        # Re-raise the exception to halt execution on corrupted config
        raise

# Load the configuration from the working directory into the 'config' dictionary
config = load_study_configuration("config.yaml")

# Fallback for the Jupyter Notebook environment if the file does not exist
if not config:
    print("Falling back to in-memory mock configuration for demonstration purposes.")
    # Initialize a minimal mock configuration required to pass validation
    config = {
        "gateway": {"base_url": "https://api.example.com", "api_key_env_var": "DUMMY_KEY"},
        "models": {"panel": ["model_A", "model_B", "model_C"], "judge": "model_A", "frontier_baselines": []},

### **Summary of the Execution Flow**

1.  **Data Ingestion & Validation:** The synthetic raw records and configuration are ingested, strictly typed, and validated against the schema contracts to prevent silent data corruption.
2.  **Contamination Audit:** The pipeline executes lexical (13-gram) and behavioral probes, applying a global drop-for-all rule to prune contaminated tasks and protect the causal interpretation of complementarity.
3.  **Qualification & Stratification:** Candidates are filtered via the $P_{\min}$ benchmark gate and clustered by capability profile to ensure lineage diversity during coalition sampling.
4.  **Tensor Assembly:** The canonical outcome tensor is constructed, mathematically enforcing the error exclusivity invariant ($\sum e_* = 1 - \text{correct}$) and computing exact USD costs.
5.  **Estimation:** The framework computes marginal utility (permutation Shapley), conditional information gain (outcome-based CMI with Miller-Madow correction), and correlated-failure risk (Ledoit-Wolf shrinkage).
6.  **Optimization:** A budget-feasible greedy algorithm selects the optimal model portfolio subject to strict cost, latency, and risk constraints.
7.  **Validation & Inference:** A three-level nested cluster bootstrap quantifies uncertainty (decomposing $\sigma^2_{\mathrm{task}}$, $\sigma^2_{\mathrm{gen}}$, and $\sigma^2_{\mathrm{adj}}$), followed by the execution of pre-registered hypothesis tests (H1-H5) with Holm-Bonferroni FWER control.
8.  **Robustness & Packaging:** The pipeline executes 10,000 Dirichlet preference sweeps to assess recommendation fragility, generates publication-grade figures, and serializes all artifacts into a cryptographically hashed, Tier-1 reproducible archive.

## Output Structure

The pipeline returns a comprehensive, cryptographically hashed `.tar.gz` archive containing the following key artifacts:
-   **`fmec_estimates_final.json`**: The central ranking primitive containing point estimates for MEC, IG, MRC, and FMEC, alongside BCa confidence intervals and bootstrap selection frequencies.
-   **`selected_ensemble.json`**: The optimal model portfolio ($S_{\mathrm{FMEC}}$) and its corresponding simplex weights.
-   **`scientific_report.md`**: A clinical, severity-ranked Markdown report detailing the verdicts for hypotheses H1-H5, effect sizes (Cohen's $d$, Cliff's $\delta$), and methodological disclosures.
-   **`variance_components.json`**: The decomposition of total uncertainty into $\sigma^2_{\mathrm{task}}$, $\sigma^2_{\mathrm{gen}}$, and $\sigma^2_{\mathrm{adj}}$.
-   **`robustness_report.json`**: The results of the 10,000 Dirichlet preference sweeps and the six pre-registered ablations (including LOO MEC).
-   **`covariance.csv`**: The Ledoit-Wolf shrunk error covariance matrix ($\widehat{\Sigma}$).
-   **`outcome_tensor.canonical.json`**: The exact byte-level representation of the canonical outcome tensor that was hashed for Tier-1 verification.
-   **Figures**: Publication-grade vector graphics including `efficient_frontier.pdf`, `fmec_ci_forest.pdf`, `covariance_heatmap.pdf`, and `rsp_bar.pdf`.

## Customization

The pipeline is highly customizable via the `config.yaml` file. Researchers and engineers can modify study parameters such as:

-   **Uncertainty Quantification:** Adjust the number of bootstrap replications ($B$) or the generation replicates ($R$) in the `uncertainty` block to balance computational cost against statistical precision.

# Paper

Title: "*Beyond Benchmark Rankings: Information-Aware and Risk-Aware Optimization of Large Language Model Ensembles (A Concept Paper and Methodology)*"

Authors: Craig S. Chirinda (CS Chirinda)

Concept Paper Publication Date: 18 June 2026

Link: https://github.com/chirindaopensource/fused_llm_router/blob/main/FMEC_concept_paper_final.md

Abstract:

Large language models (LLMs) are predominantly evaluated and deployed as standalone systems, under an implicit assumption that the model with the highest benchmark score is also the best component of a larger intelligent system. This paper argues that the assumption is unjustified. The value of an LLM-based system depends not only on the standalone capability of its constituent models, but also on the informational novelty of their outputs, the structure of their disagreements, and the degree to which their failures are correlated. Two models with near-identical benchmark scores can differ sharply in ensemble value: one may be redundant, while the other contributes complementary information by failing on different inputs.

To formalize this claim we propose a framework for information-aware and risk-aware ensemble optimization. The framework (i) separates benchmark qualification (an eligibility filter) from ensemble selection (the optimization objective); (ii) replaces scalar error with a multidimensional error taxonomy spanning mathematical, coding, factual, long-context, planning, and tool-use failures; and (iii) introduces the Final Marginal Ensemble Contribution (FMEC), a conditional, deployment-oriented contribution score that integrates expected marginal utility, conditional information gain, and correlated-failure risk into a single optimization primitive. FMEC is estimated by Monte Carlo coalition sampling, outcome-based conditional-information estimation, and covariance-aware risk adjustment, and is then used inside a constrained optimization problem that selects ensemble members under cost, latency, and risk budgets.

We provide the full operational machinery required for implementation and replication: a precise model of inputs, processes, and outputs; four explicit algorithms (FMEC estimation, greedy constrained selection, nested-bootstrap validation, oracle-regret evaluation); a complete set of named equations; the raw and derived data structures with example schemas; a fully specified study-configuration dictionary including prompt templates and inference settings; and a falsifiable experimental and statistical protocol with pre-registered hypotheses, numeric decision thresholds, ablations, multiple-comparison control, contamination auditing, and drift instrumentation. The framework is explicitly designed to be either supported or rejected by data. The central, testable hypothesis is that benchmark rank is an incomplete proxy for deployment value, and that ensembles selected by FMEC outperform benchmark-selected ensembles on both absolute performance and quality-adjusted efficiency.

We are precise about reproducibility, separating two tiers that have different attainable ceilings. Tier 1 (computational reproducibility) — that the published pipeline returns identical numbers from identical inputs — is made bit-exact by releasing the frozen per-(model, task, replicate) outcome tensor as the primary artifact, together with a content hash, all random seeds, and every intermediate checkpoint, so that the entire chain downstream of generation is a deterministic, independently re-runnable function of released data. Tier 2 (empirical replicability) — that re-collecting model outputs reproduces the scientific conclusion — is bounded above by the fact that the constituent models are mutating, black-box, non-deterministic commercial endpoints; we do not pretend to remove this irreducible variability but instead instrument and bound it through a variance-components design that explicitly decomposes task-sampling, generation, and adjudication variance. The reproduction protocol is exercised by a fully specified reference instantiation (Section 12.1) on a small, frozen, hash-pinned pool, whose released artifacts make Tier-1 reproduction verifiable rather than merely promised.

Keywords: large language models, ensemble optimization, model orchestration, conditional information gain, cooperative game theory, covariance-aware risk, submodular optimization, decision theory, reproducibility.

# Summary

**Executive Summary**

The paper, *Beyond Benchmark Rankings: Information-Aware and Risk-Aware Optimization of Large Language Model Ensembles*, proposes a formal methodology for selecting large language model ensembles. Its central claim is that benchmark rank is an incomplete proxy for deployment value. A model’s usefulness inside a system depends not only on its standalone accuracy, but also on whether it contributes nonredundant information, fails differently from other models, and improves utility under cost, latency, and risk constraints.

The paper introduces **Final Marginal Ensemble Contribution**, or **FMEC**, as the main selection criterion. FMEC combines three quantities: marginal utility contribution, conditional information gain, and marginal correlated-failure risk. The framework borrows from cooperative game theory, information theory, portfolio theory, robust statistics, and constrained optimization.

Importantly, the paper is a **concept and methodology paper**, not an empirical results paper. It specifies algorithms, equations, data structures, experimental protocols, and reproducibility requirements, but it does not report executed results.



**1. Research Problem**

The paper begins from a critique of the dominant model-selection paradigm: choosing the highest-ranked model on public benchmarks. The author argues that this approach assumes that the strongest individual model produces the strongest system. That assumption is not generally valid.

In ensemble settings, the relevant question is not simply:

> Which model is best individually?

but rather:

> Which model adds the most conditional value to the existing ensemble?

Two models with similar benchmark scores may have very different ensemble value. If their errors are highly correlated, combining them adds little. If they fail on different task types or contribute complementary information, the lower-ranked model may be more valuable.

The paper therefore reframes LLM deployment as a **portfolio-construction problem** rather than a leaderboard-selection problem.



**2. Conceptual Foundation**

The paper integrates several mature disciplines.

From **ensemble learning**, it takes the idea that diverse predictors can outperform individual predictors when their errors are not perfectly correlated.

From **cooperative game theory**, it uses Shapley-style marginal contribution: a model should be valued by how much it improves many possible coalitions, not by its standalone score.

From **information theory**, it introduces conditional information gain: a model is valuable if it contributes information not already supplied by the existing ensemble.

From **finance and portfolio theory**, it adopts covariance-aware risk: correlated failures are analogous to correlated asset losses and must be penalized.

From **statistical inference**, it uses pre-registration, bootstrap uncertainty, effect-size reporting, multiplicity correction, and falsifiable decision thresholds.

This interdisciplinary synthesis is the paper’s main intellectual structure.



**3. Problem Setup**

The paper defines a finite candidate model pool:

\[
N = \{M_1, M_2, \ldots, M_n\}
\]

Each model is represented by a tuple containing expected capability, cost, latency, robustness, and an error profile. The error profile is not scalar. Instead, errors are decomposed across multiple categories such as mathematical reasoning, coding, factuality, long-context failure, planning, tool-use, safety, and refusal behavior.

This multidimensional error taxonomy is central. The paper argues that ensemble risk depends not merely on whether models fail, but on whether they fail in the same way.

The paper also separates **qualification** from **selection**. Benchmark scores are used only to determine whether a model is eligible for the candidate pool. Final ensemble membership is determined by FMEC, not by benchmark rank.



**4. FMEC: Final Marginal Ensemble Contribution**

FMEC is the paper’s central construct.

The framework starts with a deployment utility function:

\[
v(S) = \alpha Q(S) - \beta R(S) - \gamma C(S) - \delta L(S)
\]

where:

- \(Q(S)\) is expected quality,
- \(R(S)\) is correlated-failure risk,
- \(C(S)\) is cost,
- \(L(S)\) is latency,
- \(\alpha,\beta,\gamma,\delta\) are deployment-specific weights.

The marginal contribution of model \(M_j\) to ensemble \(S\) is:

\[
v(S \cup \{M_j\}) - v(S)
\]

Because a model’s value depends on the coalition it joins, the paper estimates average marginal contribution over many coalitions using Monte Carlo coalition sampling. The preferred sampler is a permutation-based Shapley-style estimator, which gives an order-symmetric contribution measure.

FMEC then adds two additional terms:

1. **Conditional information gain**: how much new outcome-relevant information the model contributes beyond the current ensemble.
2. **Marginal risk contribution**: how much the model increases or decreases correlated-failure exposure.

The final score is:

\[
FMEC_j = \widehat{MEC}_j + \eta IG_j - \lambda MRC_j
\]

Thus, a model is rewarded for utility and informational novelty, but penalized for correlated risk.



**5. Ensemble Selection as Constrained Optimization**

The paper formulates ensemble construction as a constrained optimization problem. The objective is to select models and weights that maximize FMEC-derived utility subject to:

- cost budget,
- latency budget,
- risk budget,
- optional simplex constraints on weights.

This is explicitly compared to portfolio optimization. The selected ensemble should lie on an efficient frontier: high quality per unit cost and acceptable risk.

Because exact search is combinatorial, the paper proposes a greedy constrained selection algorithm. It also discusses approximation guarantees under submodularity assumptions and notes weaker guarantees under knapsack-style constraints.

For small candidate pools, the paper proposes exhaustive oracle search to compute regret:

\[
Regret(S_{FMEC}) = U(S^\star) - U(S_{FMEC})
\]

This gives a direct measure of how far the greedy FMEC-selected ensemble is from the true optimum.



**6. Algorithms**

The paper specifies four main algorithms.

**Algorithm 1: FMEC Estimation**  
Computes marginal contribution, information gain, covariance risk, marginal risk contribution, and final FMEC scores.

**Algorithm 2: Greedy Constrained Selection**  
Builds the ensemble iteratively by adding the feasible model with the largest marginal utility gain.

**Algorithm 3: Nested Bootstrap Validation**  
Quantifies uncertainty by resampling tasks, generation replicates, and adjudication replicates. This is intended to capture task variance, model-generation variance, and judge variance.

**Algorithm 4: Oracle-Regret Evaluation**  
For small pools, enumerates all possible ensembles to compare FMEC selection against the true optimum.

The algorithms are given in pseudocode rather than full executable implementation.



**7. System Architecture**

The proposed architecture has four subsystems:

1. **Expert models**: pretrained LLMs accessed as black-box services.
2. **Fusion operator**: combines outputs through voting, weighted fusion, judge-based synthesis, or verifier-augmented scoring.
3. **FMEC estimation and selection pipeline**: computes contribution, information, and risk.
4. **Optional router**: a contextual-bandit policy that decides which ensemble members to invoke per request.

The paper explicitly states that it does not train a new neural network. Its contribution is an orchestration and selection framework.



**8. Data Structures and Configuration**

The paper defines raw and derived data structures, including:

- raw model-answer records,
- structured candidate-solution records,
- task records,
- outcome-and-error tensor,
- benchmark-qualification table,
- coalition-utility samples,
- covariance matrix,
- FMEC results table.

The most important artifact is the **outcome tensor**, with one row per model, task, and replicate. It includes correctness, error categories, cost, latency, provider version, system fingerprint, and seed.

The paper also gives a detailed study-configuration dictionary covering model rosters, prompts, inference settings, covariance estimator, information-gain estimator, bootstrap design, optimization budgets, drift monitoring, contamination auditing, and reproducibility artifacts.



**9. Experimental Design**

The paper specifies five pre-registered hypotheses.

- **H1:** FMEC-selected ensembles outperform benchmark-selected ensembles.
- **H2:** conditional information gain is positively associated with marginal contribution.
- **H3:** risk-aware selection reduces correlated-failure risk.
- **H4:** benchmark rank is insufficient as a proxy for FMEC rank.
- **H5:** an FMEC-selected open-weight ensemble can be non-inferior to a frontier model within a cost-tied margin.

The protocol includes six baselines: benchmark top-K, random selection, correlation-minimized selection, Shapley-only selection, information-only selection, and greedy-utility selection.

It also includes six ablations, removing or varying information gain, risk, Shapley averaging, Monte Carlo budget, covariance estimator, and information-gain estimator.

The statistical plan uses bootstrap intervals, Holm-Bonferroni correction, effect sizes, recommendation-stability probability, and robustness checks over utility weights.



**10. Reproducibility Design**

The paper’s reproducibility framework is unusually explicit. It distinguishes two tiers.

**Tier 1: Computational Reproducibility**  
Given the released outcome tensor, seeds, configuration, and intermediate artifacts, all downstream results should be bit-exact and reproducible without calling any model.

**Tier 2: Empirical Replicability**  
Re-collecting model outputs is not expected to be bit-exact because commercial LLM endpoints are black-box, mutable, and often nondeterministic. The paper therefore proposes to measure and bound this variability through repeated generations, repeated judging, model-version logging, and drift monitoring.

The paper requires released artifacts such as the tensor hash, model-version manifest, contamination report, frozen task sets, coalition samples, covariance estimates, FMEC estimates, bootstrap outputs, variance components, and selected ensemble.

However, the paper also states that these values are to be populated only upon executing the reference instantiation. Therefore, the manuscript specifies a reproducible design but does not itself provide executed empirical results.



**11. Limitations**

The paper identifies several substantive limitations.

First, FMEC is computationally expensive. It requires repeated model generations, coalition sampling, covariance estimation, bootstrapping, and robustness analysis.

Second, model drift is a serious threat. The framework’s estimates are valid only for the specific model versions evaluated.

Third, results are task-distribution dependent. Complementarity is not an intrinsic model property; it depends on the evaluation corpus.

Fourth, the utility weights are exogenous. If stakeholders disagree about quality, cost, latency, and risk trade-offs, the selected ensemble is conditional on those assumptions.

Fifth, adjudication quality matters. Automated judges can introduce bias, so the paper requires human audit and inter-rater agreement thresholds.

Finally, benchmark contamination remains a major validity threat. The paper treats contamination auditing as a prerequisite for inference.



**Conclusion**

The paper proposes a rigorous methodology for LLM ensemble selection. Its central contribution is the FMEC framework, which evaluates models by conditional ensemble value rather than standalone benchmark rank. The framework combines marginal utility, information gain, and correlated-failure risk inside a constrained optimization problem.

The strongest aspect of the paper is its methodological discipline. It defines equations, algorithms, data structures, pre-registered hypotheses, baselines, ablations, uncertainty quantification, and reproducibility tiers. It is careful not to claim empirical success without executed data.

The main weakness is that the paper remains a protocol rather than a completed empirical study. Its claims about reproducibility and performance depend on future release of code, frozen tensors, hashes, model manifests, contamination reports, and executed statistical outputs.

In summary, the paper is best understood as a serious and technically ambitious blueprint for reproducible, risk-aware LLM ensemble construction. It does not prove that FMEC outperforms benchmark-based selection, but it provides a structured and falsifiable way to test whether it does.

## **Discussion of the Inputs, Processes and Outputs (IPOs) of Key Callables**

Below is a clinical, granular, and exhaustive step-by-step analysis of the 25 final orchestrator callables that constitute the *Beyond Benchmark Rankings* (FMEC) framework. For each callable, the analysis discusses the Inputs, Processes, and Outputs (IPO); and, it details the callable's exact data transformations. It also explicates each callable's precise role in the research pipeline using the governing LaTeX equations:

### 1. `load_config_and_seeds` (Task 1)
*   **Inputs:** Raw, unvalidated configuration dictionary (`raw_config`).
*   **Processes:** Resolves API credentials from the environment, redacts secrets from the configuration state, computes a canonical SHA-256 hash of the configuration, spawns statistically independent `np.random.Generator` streams from a single root `SeedSequence`, and initializes an immutable, timestamped run directory with a dependency lockfile.
*   **Outputs:** `FrozenDict` (immutable config), `Path` (run directory), and `Dict[str, np.random.Generator]` (RNG bundle).
*   **Data Transformation:** Transforms an insecure, mutable Python dictionary into a cryptographically hashed, deeply frozen state object. Transforms a single integer root seed into multiple mathematically independent PCG64 bit generators.
*   **Pipeline Role:** Implements the foundational environment provenance required for **"Tier 1 (computational reproducibility)"** (§13.3). It ensures that the pipeline is a *"deterministic, independently re-runnable function of released data."*

### 2. `validate_config` (Task 2)
*   **Inputs:** `FrozenDict` (configuration), `Path` (run directory).
*   **Processes:** Executes declarative type, finiteness, and boundary checks on all hyperparameters. Enforces cross-field invariants (e.g., ensuring prompt templates contain only valid fill-tokens and no literal JSON collisions).
*   **Outputs:** A structured validation report dictionary.
*   **Data Transformation:** Transforms the configuration state into a boolean pass/fail matrix and a serialized JSON audit log, halting the Python interpreter on any `HARD` violation.
*   **Pipeline Role:** Implements the **"falsifiable experimental and statistical protocol with pre-registered hypotheses, numeric decision thresholds"** (§11.13). It mathematically guarantees that parameters like $P_{\min} \in [0.75, 0.80]$ and $\Delta_\tau = 0.20$ are strictly enforced before execution.

### 3. `ingest_and_validate_inputs` (Task 3)
*   **Inputs:** Raw lists of dictionaries (`model_answer`, `candidate_solution`, `task_record`), raw `qualification` DataFrame, `FrozenDict`, `Path`.
*   **Processes:** Materializes raw lists into pandas DataFrames, enforces exact column schemas, coerces data types (using nullable `Int8`/`Int64` to prevent silent float upcasting), and enforces relational foreign-key integrity.
*   **Outputs:** Four strictly typed pandas DataFrames.
*   **Data Transformation:** Transforms loosely typed, heterogeneous JSON-like arrays into strict, memory-contiguous columnar structures suitable for vectorized linear algebra.
*   **Pipeline Role:** Implements the data substrate for **"Layer 2 — Evaluation and annotation"** (§8.4). It strictly enforces Assumption A1 (Stable task distribution) by verifying the `is_held_out` invariant: $\mathcal{D}_{\text{qual}} \cap \mathcal{D}_{\text{eval}} = \varnothing$.

### 4. `cleanse_raw` (Task 4)
*   **Inputs:** Four strictly typed DataFrames.
*   **Processes:** Filters out infrastructure/transport errors, validates self-reported confidence bounds, parses Python code via Abstract Syntax Trees (AST) to reject placeholder stubs, deduplicates on natural keys, performs a stable mergesort, and computes SHA-256 hashes.
*   **Outputs:** A dictionary containing the canonicalized DataFrames and their respective content hashes.
*   **Data Transformation:** Transforms noisy, potentially malformed LLM outputs into structurally verified records, mapping infrastructure failures to a disposition log while preserving the raw text payload for downstream adjudication.
*   **Pipeline Role:** Implements the data hygiene necessary for **"a precise model of inputs, processes, and outputs"** (§8.6, Stage 1 to 2 transition), ensuring the adjudicator only scores structurally viable responses.

### 5. `contamination_audit` (Task 5)
*   **Inputs:** Cleansed `task_record`, `qualification`, `FrozenDict`, `Path`, raw proxy texts.
*   **Processes:** Computes 13-gram lexical overlap using a hashed $O(1)$ lookup index, executes parallelized behavioral canary probes, and applies a global drop-for-all exclusion rule.
*   **Outputs:** Pruned `task_record` DataFrame and a contamination report dictionary.
*   **Data Transformation:** Transforms the full evaluation corpus into a strictly uncontaminated subset by filtering out any task row where the overlap rate exceeds $\tau_{\mathrm{overlap}} = 0.10$.
*   **Pipeline Role:** Implements the **"Contamination auditing"** precondition (§13.2). It protects the causal interpretation of complementarity, as contamination invalidates the conditional mutual information estimand $I(A_j; Z \mid A_S)$.

### 6. `qualify` (Task 6)
*   **Inputs:** Cleansed `qualification` DataFrame, `FrozenDict`, `Path`.
*   **Processes:** Computes within-pool percentile ranks for canonical benchmarks, min-max normalizes Elo ratings, computes the composite percentile rank, and applies the $P_{\min}$ threshold.
*   **Outputs:** A canonical ordered list of qualified `model_id` strings and the mutated qualification DataFrame.
*   **Data Transformation:** Transforms absolute, heterogeneous benchmark scores into a unified, relative $[0,1]$ probability space, filtering the row space to only those models meeting the threshold.
*   **Pipeline Role:** Implements **"Layer 1 — Benchmark qualification"** (§8.4). It executes the eligibility filter:
    $$ \text{qualify}(M_i, t) \iff \mathrm{score}_t(M_i) \ge P_{\min}\cdot \max_{j\in\mathcal{P}} \mathrm{score}_t(M_j) $$

### 7. `stratify` (Task 7)
*   **Inputs:** Qualified pool list, mutated `qualification` DataFrame, `FrozenDict`, `Path`.
*   **Processes:** Bins parameter counts (isolating closed-weight models), mean-imputes missing benchmarks, standardizes (z-scores) the feature space, and executes $k$-means clustering ($k=4$) with a small-pool fallback.
*   **Outputs:** A DataFrame indexed by `model_id` containing the three stratification axes.
*   **Data Transformation:** Transforms continuous parameter and benchmark vectors into discrete categorical strata (`family`, `param_scale_bin`, `capability_cluster`).
*   **Pipeline Role:** Implements the **"Stratified sampling of candidates"** (§11.3). This ensures the Monte Carlo sampler explores combinations across lineage boundaries, mitigating the correlated-failure risk identified in Theorem 2.

### 8. `build_fusion_operator` (Task 8)
*   **Inputs:** `FrozenDict`, RNG bundle, qualified pool, and pre-computed telemetry vectors ($q, r, c, l$).
*   **Processes:** Computes normalized simplex weights for the weighted-vote path, validates sandbox parameters, instantiates the contextual bandit router, and constructs a stateless fusion closure.
*   **Outputs:** The `fuse` callable, the weight vector $w$, sandbox configuration, and the optional router instance.
*   **Data Transformation:** Transforms static configuration parameters and telemetry arrays into an executable, memory-resident routing function.
*   **Pipeline Role:** Implements **"Subsystem (ii): the fusion operator"** (§8.3). For the weighted path, it computes:
    $$ s_i = \frac{q_i\, r_i}{c_i^{\,a}\, \ell_i^{\,b}}, \qquad w_i = \frac{s_i}{\sum_j s_j} \tag{34} $$

### 9. `collect_responses` (Task 9)
*   **Inputs:** Qualified pool, pruned `task_record`, `FrozenDict`.
*   **Processes:** Executes parallelized, asynchronous HTTP requests to the LLM gateway using role-specific decoding temperatures ($0.20$ for panel, $0.10$ for solver). Captures exact token usage and latency. Executes generated code in an isolated subprocess.
*   **Outputs:** Lists of raw `model_answer` and `candidate_solution` dictionaries.
*   **Data Transformation:** Transforms task prompts and model identifiers into raw text/JSON payloads and execution traces (compile/runtime/test outcomes).
*   **Pipeline Role:** Implements **"Stage 1 — Generation"** (§8.6). It generates the $R=5$ replicate axis required to expose the generation-variance component $\sigma^2_{\mathrm{gen}}$.

### 10. `score_and_adjudicate` (Task 10)
*   **Inputs:** Raw response lists, `task_record`, `FrozenDict`, RNG bundle, optional human labels.
*   **Processes:** Routes responses by `scoring_rule`. Executes deterministic grounding (exact match, numeric tolerance, unit tests). Executes LLM adjudication (temperature $0.00$) for open-ended rubrics. Computes inter-rater reliability (Cohen's $\kappa$). Executes replicated judging on a $20\%$ subsample.
*   **Outputs:** A unified list of scored label dictionaries.
*   **Data Transformation:** Transforms raw text and JSON payloads into binary correctness indicators $A_{it} \in \{0,1\}$ and maps rubric categories to the 8-dimensional error vector via a strict bijection.
*   **Pipeline Role:** Implements **"Stage 2 — Scoring & annotation"** (§8.6). It populates the multidimensional error taxonomy:
    $$ E_i = (E_{\mathrm{math}}, E_{\mathrm{code}}, E_{\mathrm{fact}}, E_{\mathrm{context}}, E_{\mathrm{plan}}, E_{\mathrm{tool}}, E_{\mathrm{safety}}, E_{\mathrm{refusal}}) \in \mathbb{R}^8 \tag{2} $$

### 11. `build_outcome_tensor` (Task 11)
*   **Inputs:** Scored labels list, `FrozenDict`.
*   **Processes:** Assembles the 17-column DataFrame, computes exact USD costs from token counts, enforces the mathematical exclusivity invariant ($\sum e_* = 1 - \text{correct}$), canonicalizes the byte order, and computes the SHA-256 hash.
*   **Outputs:** The 17-column primary tensor, the 13-column estimator projection, the judge replicates DataFrame, and the SHA-256 hash string.
*   **Data Transformation:** Transforms a list of dictionaries into a strictly typed, memory-contiguous, cryptographically hashed pandas DataFrame.
*   **Pipeline Role:** Produces the **"frozen per-(model, task, replicate) outcome tensor"** (§13.3). This is the foundational data-generating substrate; everything downstream is a deterministic function of this tensor.

### 12. `utility` (Task 12)
*   **Inputs:** Coalition index list $S$, estimator projection DataFrame, covariance matrix $\Sigma$, `FrozenDict`, qualified pool.
*   **Processes:** Pre-computes dense correctness matrices. Evaluates simulated majority-vote quality. Computes additive cost, max latency, and quadratic portfolio risk. Normalizes penalties and computes the scalar utility.
*   **Outputs:** A scalar float representing $v(S)$.
*   **Data Transformation:** Transforms a discrete subset of models into a continuous scalar utility value by projecting their joint correctness and telemetry through the preference weights.
*   **Pipeline Role:** Implements the **"Ensemble utility function"** (§4.2).
    $$ \widehat{Q}_{\mathrm{maj}}(S) = \frac{1}{|T|}\sum_{t\in T}\mathbb{1}\left[\sum_{i\in S} A_{it} > \frac{|S|}{2}\right] \tag{3a} $$
    $$ v(S) = \alpha Q(S) - \beta R(S) - \gamma C(S) - \delta L(S) \tag{3} $$

### 13. `estimate_mec` (Task 13)
*   **Inputs:** Qualified pool, utility closure $v$, `FrozenDict`, RNG bundle.
*   **Processes:** Determines sample complexity $K$ via Hoeffding bounds. Generates uniform permutations. Evaluates marginal utility incrementally using common random numbers to reduce variance.
*   **Outputs:** A dictionary of $\widehat{\mathrm{MEC}}_j$ scores and a DataFrame of coalition samples.
*   **Data Transformation:** Transforms $K$ permutation sequences into an averaged marginal utility vector.
*   **Pipeline Role:** Implements **"Algorithm 1 — FMEC Estimation (Stage A)"** (§7.1). It computes the permutation-based Monte Carlo Shapley estimation:
    $$ \widehat{\phi}_j = \frac{1}{K}\sum_{k=1}^{K}\left[v(P^{(k)}_{\prec j}\cup\{M_j\}) - v(P^{(k)}_{\prec j})\right], \quad P^{(k)}\sim\mathrm{Unif}(\mathfrak{S}_n) \tag{6a} $$

### 14. `estimate_ig` (Task 14)
*   **Inputs:** Estimator projection, coalition samples DataFrame, qualified pool, `FrozenDict`.
*   **Processes:** Extracts the binary correctness matrix $A$ and solvability target $Z$. Computes the correct-count sufficient statistic $c_S$. Calculates conditional entropy with Laplace smoothing ($\alpha=0.5$) and Miller-Madow bias correction. Averages over the exact sampled coalitions.
*   **Outputs:** A dictionary of $\mathrm{IG}_j$ scores.
*   **Data Transformation:** Transforms binary correctness events into bits of information (entropy reduction).
*   **Pipeline Role:** Implements the **"Outcome-based operationalization"** of Conditional Information Gain (§4.6).
    $$ \widehat{\mathrm{IG}}(M_j\mid S) = \widehat{H}(Z\mid c_S) - \widehat{H}(Z\mid A_j,c_S) \tag{9a} $$

### 15. `estimate_risk` (Task 15)
*   **Inputs:** Estimator projection, qualified pool, `FrozenDict`.
*   **Processes:** Extracts dense error matrices per category. Computes the analytical Ledoit-Wolf shrinkage estimator targeting a diagonal matrix. Sums the block covariances. Computes portfolio risk and marginal risk contribution using the reference weight vector.
*   **Outputs:** The $n \times n$ covariance DataFrame $\widehat{\Sigma}$, MRC dictionary, and scalar portfolio risk.
*   **Data Transformation:** Transforms $m$-dimensional error vectors into a regularized $n \times n$ covariance matrix and projects it into a 1D marginal risk vector.
*   **Pipeline Role:** Implements **"Correlated-failure risk"** (§4.7).
    $$ \Sigma = \mathrm{Cov}(E) \tag{10} $$
    $$ R(w) = w^\top \Sigma w \tag{11} $$
    $$ \mathrm{MRC}_j = (\Sigma w)_j \tag{12} $$

### 16. `compose_fmec` (Task 16)
*   **Inputs:** MEC, IG, and MRC dictionaries, qualified pool, `FrozenDict`, `Path`.
*   **Processes:** Computes the linear combination of the three primitives using hyperparameters $\eta$ and $\lambda$. Assembles the strictly typed FMEC DataFrame. Serializes the artifact.
*   **Outputs:** The `fmec_table` pandas DataFrame.
*   **Data Transformation:** Transforms three distinct estimation vectors into a single, composite ranking primitive.
*   **Pipeline Role:** Implements **"The final FMEC score"** (§4.8).
    $$ \mathrm{FMEC}_j = \widehat{\mathrm{MEC}}_j + \eta\,\mathrm{IG}_j - \lambda\,\mathrm{MRC}_j \tag{13} $$

### 17. `select_ensemble` (Task 17)
*   **Inputs:** FMEC table, estimator projection, $\widehat{\Sigma}$, qualified pool, `FrozenDict`.
*   **Processes:** Extracts dense 1D telemetry arrays. Executes a budget-feasible greedy search, iteratively adding the candidate with the maximum positive marginal utility gain that satisfies all knapsack constraints. Formats active constraint diagnostics.
*   **Outputs:** The selected ensemble list $S_{\mathrm{FMEC}}$, simplex weights dictionary, and active constraints dictionary.
*   **Data Transformation:** Transforms continuous FMEC scores and telemetry vectors into a discrete subset $S \subseteq N$ and a continuous weight vector $w$ on the probability simplex.
*   **Pipeline Role:** Implements **"Algorithm 2 — Greedy Constrained FMEC Ensemble Selection"** (§7.2).
    $$ \max_{w} U(w) \quad \text{subject to} \quad \sum_{i=1}^{n} w_i = 1,\; w_i \ge 0,\; C(w) \le C_{\max},\; L(w) \le L_{\max},\; R(w) \le R_{\max} \tag{14} $$

### 18. `fusion_sim_audit` (Task 18)
*   **Inputs:** Estimator projection, raw records, `task_record`, `fuse` callable, qualified pool, `FrozenDict`, RNG bundle.
*   **Processes:** Samples coalitions stratified by size. Computes simulated quality $\widehat{Q}_{\mathrm{maj}}$. Executes live LLM judge calls to compute realized fusion quality $Q_f$. Fits an Ordinary Least Squares (OLS) regression.
*   **Outputs:** A diagnostics dictionary (slope, $R^2$, $\sigma_u$) and a list of flagged strata.
*   **Data Transformation:** Transforms raw text outputs into realized quality scores, mapping the relationship between simulation and reality into regression coefficients.
*   **Pipeline Role:** Implements the **"fusion-simulation gap"** audit (§4.2.1), bounding the fidelity cost of the surrogate.
    $$ Q_f(S) = \beta_0 + \beta_1\widehat{Q}(S) + u(S) $$

### 19. `bootstrap_validate` (Task 19)
*   **Inputs:** Primary tensor, judge replicates, point FMEC table, $S^\star$, qualified pool, `FrozenDict`, RNG bundle.
*   **Processes:** Executes a 3-level nested cluster bootstrap ($B=10,000$). Re-runs the stateless estimation-selection chain per resample. Computes Bias-Corrected and Accelerated (BCa) intervals via jackknife. Isolates variance components via targeted single-axis resampling.
*   **Outputs:** Updated FMEC table (with CIs), variance components dict, RSP scalar, and selection frequency dict.
*   **Data Transformation:** Transforms single point estimates into empirical sampling distributions, extracting confidence bounds and stability probabilities.
*   **Pipeline Role:** Implements **"Algorithm 3 — Nested (Cluster) Bootstrap FMEC Validation"** (§7.3).
    $$ \mathrm{Var}(\widehat{\theta}) = \sigma^2_{\mathrm{task}} + \sigma^2_{\mathrm{gen}} + \sigma^2_{\mathrm{adj}} \tag{20a} $$
    $$ \mathrm{RSP} = \frac{1}{B}\sum_{b=1}^{B}\mathbb{1}\left[S^{\star(b)} = S^\star\right] \tag{51} $$

### 20. `confirmatory_tests` (Task 20)
*   **Inputs:** $S_{\mathrm{FMEC}}$, point FMEC table, qualification DF, estimator projection, $\widehat{\Sigma}$, qualified pool, RSP, `FrozenDict`, RNG bundle, raw tensors.
*   **Processes:** Constructs six equivalent-envelope baselines. Computes per-task utility arrays. Executes H1 (paired t/Wilcoxon), H2 (Spearman permutation), H3 (bootstrap risk), H4 (Kendall $\tau$), and H5 (non-inferiority). Applies Holm-Bonferroni FWER control. Computes Cohen's $d$ and Cliff's $\delta$. Evaluates strict falsification rules.
*   **Outputs:** A structured verdicts dictionary containing adjusted p-values, effect sizes, and the severity-ranked overall status.
*   **Data Transformation:** Transforms utility arrays and rank vectors into statistical p-values, effect sizes, and boolean falsification flags.
*   **Pipeline Role:** Implements **"Confirmatory tests and multiplicity control"** (§11.9) and the **"Falsifiability and failure conditions"** (§11.13).
    $$ H_1: \mathbb{E}[U(S_{\mathrm{FMEC}})] > \mathbb{E}[U(S_{\mathrm{Bench}})] \tag{44} $$
    $$ \Delta_U = \max\left(\gamma(C_{\mathrm{frontier}} - C_{\mathrm{open}}), 0.02|U(m_{\mathrm{frontier}}^\star)|\right) \tag{48a} $$

### 21. `oracle_regret` (Task 21)
*   **Inputs:** Qualified pool, estimator projection, $\widehat{\Sigma}$, `FrozenDict`, $S_{\mathrm{FMEC}}$.
*   **Processes:** Checks the $n \le 20$ enumeration cap. Exhaustively evaluates all $2^n - 1$ subsets against the fast feasibility gates. Identifies the global utility maximum $S^\star$. Computes absolute and relative regret.
*   **Outputs:** Absolute regret float, relative regret float, and the oracle ensemble list $S^\star$.
*   **Data Transformation:** Transforms the combinatorial subset space into a single global maximum and a scalar distance metric.
*   **Pipeline Role:** Implements **"Algorithm 4 — Oracle-Regret Evaluation"** (§7.4).
    $$ S^\star = \arg\max_{S} v(S) \tag{28} $$
    $$ \mathrm{Regret}(\widehat{S}) = v(S^\star) - v(\widehat{S}) \tag{29} $$

### 22. `run_pipeline` (Task 22)
*   **Inputs:** Raw configs, raw data lists, optional cached state, optional perturbation dict.
*   **Processes:** Orchestrates the deterministic ingestion-to-tensor segment (Tasks 1-11). Executes the pure estimation-selection segment (Tasks 12-18) applying any allow-listed perturbations. Executes the inference segment (Tasks 19-21).
*   **Outputs:** `ResultsBundle` dataclass and `PipelineCache` dataclass.
*   **Data Transformation:** Glues the entire pipeline, transforming raw strings and JSON into the final, strictly typed results bundle while maintaining strict state isolation.
*   **Pipeline Role:** Implements the **"End-to-End Input–Process–Output (IPO) model"** (§8.6), ensuring that the pipeline executes strictly left-to-right without data leakage.

### 23. `robustness` (Task 23)
*   **Inputs:** `run_pipeline` callable, `PipelineCache`, baseline `ResultsBundle`.
*   **Processes:** Executes 10,000 parallel Dirichlet sweeps over the unit simplex, bypassing data collection by injecting the `PipelineCache`. Executes six pre-registered ablations (including A3 LOO MEC) via dependency-mapped minimal recomputation. Computes the probability mass surface of selected ensembles.
*   **Outputs:** `RobustnessReport` dataclass.
*   **Data Transformation:** Transforms 10,000 random weight vectors into a discrete probability distribution over the ensemble space, yielding a boolean fragility flag.
*   **Pipeline Role:** Implements **"Robustness to utility specification"** (§11.11).
    $$ (\alpha,\beta,\gamma,\delta) \sim \mathrm{Dirichlet}(1,1,1,1) $$
    $$ \mathrm{MEC}_j = v(N) - v(N \setminus \{M_j\}) \quad \text{(Ablation A3)} $$

### 24. `report_and_package` (Task 24)
*   **Inputs:** `ResultsBundle`, `PipelineCache`, `RobustnessReport`, $\widehat{\Sigma}$, coalition samples.
*   **Processes:** Generates publication-grade vector plots (Efficient Frontier, CI Forest, Covariance Heatmap). Synthesizes a clinical Markdown report with strict honesty guards (refusing to render unexecuted artifacts). Serializes all data to CSV/JSON. Computes the archive SHA-256 hash.
*   **Outputs:** A `Path` object pointing to the immutable `.tar.gz` archive.
*   **Data Transformation:** Transforms in-memory dataclasses and pandas DataFrames into serialized bytes, PDF graphics, and a cryptographically hashed tarball.
*   **Pipeline Role:** Implements **"Tier 1 (computational reproducibility)"** (§13.3), ensuring the released artifacts make reproduction verifiable rather than merely promised.

### 25. `run_experimental_suite` (Global Orchestrator)
*   **Inputs:** Raw configuration dict, raw generation lists, raw task/qualification DFs, raw proxy texts.
*   **Processes:** Coordinates Phase 1 (Data Substrate & Baseline), Phase 2 (Robustness & Ablations), Phase 3 (Point-Estimate Extraction), and Phase 4 (Reporting & Packaging). Manages memory across the ProcessPoolExecutor.
*   **Outputs:** The final file path to the reproducibility archive.
*   **Data Transformation:** Transforms the entire raw experimental input space into a published, DOI-ready, content-addressed scientific artifact.
*   **Pipeline Role:** Implements the **"Reference instantiation (the executed minimal instance)"** (§12.1), converting the theoretical framework into a fully executed, verifiable procedure.

*Note: The example assumes that all the callables that have been defined in this notebook are in a single Jupyter notebook. And, that there is no folder with “.py” functions.*

### **Step 1: Synthetic Data Generation (`raw_fmec_data`)**

The first requirement is a high-fidelity synthetic representation of the raw data structures. These datasets must adhere strictly to the schemas and mathematical invariants defined in Section 9.1 and 9.2 of the study (e.g., the exclusivity invariant for error flags, and the disjointness invariant for evaluation tasks).

**Methodology:**
1.  **Qualification Generation:** We simulate a pool of candidate models with plausible benchmark percentiles, ensuring the schema exactly matches the 10 required columns.
2.  **Task Generation:** We generate a set of evaluation tasks spanning different domains, strictly enforcing the `is_held_out = True` invariant.
3.  **Response Generation:** We simulate $R=5$ replicate generations per (model, task) pair, populating token counts, latencies, and verbatim text payloads using the `Faker` library.
4.  **Outcome Tensor Generation:** We construct the canonical 17-column outcome tensor, mathematically enforcing the error exclusivity invariant ($\sum e_* = 1 - \text{correct}$).

```python
# Import the pandas library for DataFrame manipulation
import pandas as pd
# Import the numpy library for numerical operations and random sampling
import numpy as np
# Import the yaml library for configuration parsing
import yaml
# Import the os module for environment variable access
import os
# Import the Faker library to generate realistic synthetic text
from faker import Faker
# Import typing modules for strict type hinting
from typing import Dict, Any, List, Tuple

# Initialize the Faker instance to generate synthetic text payloads
fake = Faker()
# Set a deterministic seed for Faker to ensure reproducible text generation
Faker.seed(20260618)
# Set a deterministic seed for NumPy to ensure reproducible numerical generation
np.random.seed(20260618)

def generate_synthetic_fmec_data(
    n_models: int = 5,
    n_tasks: int = 10,
    n_replicates: int = 5
) -> Tuple[List[Dict[str, Any]], List[Dict[str, Any]], List[Dict[str, Any]], pd.DataFrame, pd.DataFrame]:
    """
    Generates high-fidelity synthetic data structures required for the FMEC pipeline.

    Purpose:
        To create mathematically plausible mock datasets that strictly adhere to the
        schemas defined in Section 9.1 and 9.2 of the FMEC methodology. This allows
        for the execution and testing of the pipeline without requiring expensive
        live LLM API calls.

    Inputs:
        n_models (int): The number of candidate models to simulate. Default is 5.
        n_tasks (int): The number of evaluation tasks to simulate. Default is 10.
        n_replicates (int): The number of generation replicates (R). Default is 5.

    Processes:
        1.  Qualification Generation: Creates a DataFrame of models with benchmark scores.
        2.  Task Generation: Creates a list of dictionaries defining held-out tasks.
        3.  Response Generation: Simulates free-form and structured LLM outputs.
        4.  Outcome Tensor Generation: Constructs a DataFrame enforcing the error
            exclusivity invariant (correct == 1 implies all e_* == 0).

    Outputs:

    # Return the five generated data structures
    return model_answer, candidate_solution, task_record, outcomes, qualification

### **Step 2: Loading the Configuration (`config.yaml`)**

The study relies on a deterministic configuration file (`config.yaml`) that defines all hyperparameters, budgets, and reproducibility seeds. We read this file into memory and convert it into a usable Python dictionary.

**Methodology:**
1.  **File I/O:** Open `config.yaml` in read mode using UTF-8 encoding.
2.  **Parsing:** Use `yaml.safe_load` to convert the YAML structure into a nested Python dictionary securely.
3.  **Validation:** Catch file existence errors and parsing errors, raising explicit exceptions to halt the pipeline if the configuration is corrupted or missing.

```python
def load_study_configuration(filepath: str = "config.yaml") -> Dict[str, Any]:
    """
    Loads the study configuration parameters from a YAML file into a Python dictionary.

    Purpose:
        To ingest the deterministic hyperparameters, optimization budgets, and
        reproducibility seeds defined in the external configuration file. This
        ensures strict separation of code and configuration.

    Inputs:
        filepath (str): The relative or absolute path to the YAML configuration file.
                        Default is "config.yaml".

    Processes:
        1.  File Access: Attempts to open the specified file in read mode.
        2.  Parsing: Uses PyYAML's safe_load to parse the YAML structure securely.
        3.  Validation: Catches file existence errors and parsing errors.

    Outputs:
        Dict[str, Any]: A nested dictionary containing the study configuration.

    Raises:
        TypeError: If filepath is not a string.
        FileNotFoundError: If the specified file does not exist.
        yaml.YAMLError: If the file contains invalid YAML syntax.
    """
    # Validate that the provided filepath is a string
    if not isinstance(filepath, str):
        # Raise TypeError to prevent invalid file operations
        raise TypeError(f"filepath must be a string, got {type(filepath).__name__}.")

    try:
        # Open the file stream in read mode with UTF-8 encoding
        with open(filepath, "r", encoding="utf-8") as file:
            # Parse the YAML content safely into a Python dictionary
            config = yaml.safe_load(file)

        # Log success to the console for user visibility
        print(f"Successfully loaded configuration from {filepath}")

        # Return the parsed configuration dictionary
        return config

    except FileNotFoundError:
        # Raise an explicit error if the configuration file is missing
        raise FileNotFoundError(f"Critical Error: {filepath} not found in the working directory.")
    except yaml.YAMLError as e:
        # Handle YAML syntax errors explicitly
        print(f"Error parsing YAML file {filepath}: {e}")
        # Re-raise the exception to halt execution on corrupted config
        raise

# Load the configuration from the working directory into the 'config' dictionary
config = load_study_configuration("config.yaml")
```

### **Summary of the Execution Flow**

1.  **Data Ingestion & Validation:** The synthetic raw records and configuration are ingested, strictly typed, and validated against the schema contracts to prevent silent data corruption.
2.  **Contamination Audit:** The pipeline executes lexical (13-gram) and behavioral probes, applying a global drop-for-all rule to prune contaminated tasks and protect the causal interpretation of complementarity.
3.  **Qualification & Stratification:** Candidates are filtered via the $P_{\min}$ benchmark gate and clustered by capability profile to ensure lineage diversity during coalition sampling.
4.  **Tensor Assembly:** The canonical outcome tensor is constructed, mathematically enforcing the error exclusivity invariant ($\sum e_* = 1 - \text{correct}$) and computing exact USD costs.
5.  **Estimation:** The framework computes marginal utility (permutation Shapley), conditional information gain (outcome-based CMI with Miller-Madow correction), and correlated-failure risk (Ledoit-Wolf shrinkage).
6.  **Optimization:** A budget-feasible greedy algorithm selects the optimal model portfolio subject to strict cost, latency, and risk constraints.
7.  **Validation & Inference:** A three-level nested cluster bootstrap quantifies uncertainty (decomposing $\sigma^2_{\mathrm{task}}$, $\sigma^2_{\mathrm{gen}}$, and $\sigma^2_{\mathrm{adj}}$), followed by the execution of pre-registered hypothesis tests (H1-H5) with Holm-Bonferroni FWER control.
8.  **Robustness & Packaging:** The pipeline executes 10,000 Dirichlet preference sweeps to assess recommendation fragility, generates publication-grade figures, and serializes all artifacts into a cryptographically hashed, Tier-1 reproducible archive.

````python
# Task 1 — Environment, Configuration & Seed Initialization (load_config_and_seeds)

# ==============================================================================
# Task 1: Environment, Configuration & Seed Initialization
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 1, Step 1: Load, Validate, Redact, and Deep-Freeze the Configuration
# ------------------------------------------------------------------------------

class FrozenDict(Mapping):
    """
    A recursive, immutable dictionary view designed to prevent state mutation.

    This class wraps a standard Python dictionary and recursively converts nested
    dictionaries into immutable FrozenDict instances, and nested list values into
    immutable tuples. This structure guarantees that configuration parameters
    and intermediate states remain unaltered across parallel process pools.

    Parameters
    ----------
    mapping : Mapping[str, Any]
        The input dictionary or Mapping to be frozen.

    Raises
    ------
    TypeError
        If the input is not an instance of collections.abc.Mapping,
        or if any key is not a string.
    KeyError
        If a requested key is not present in the internal storage.
    """
    def __init__(self, mapping: Mapping[str, Any]) -> None:
        # Validate that the incoming payload is an instance of collections.abc.Mapping
        if not isinstance(mapping, Mapping):
            # Raise a TypeError if the input contract is violated
            raise TypeError(f"FrozenDict initialization requires a Mapping, got {type(mapping).__name__}")
        # Initialize the internal, private storage dictionary
        self._storage: Dict[str, Any] = {}
        # Iterate recursively over the key-value pairs of the input mapping
        for key, value in mapping.items():
            # Validate that every configuration key is a string to ensure schema compliance
            if not isinstance(key, str):
                # Raise a TypeError for non-conforming key types
                raise TypeError(f"FrozenDict keys must be strings, got {type(key).__name__}")
            # Recursively wrap nested dictionaries in FrozenDict instances
            if isinstance(value, dict):
                # Bind the sub-dictionary as an immutable FrozenDict view
                self._storage[key] = FrozenDict(value)
            # Convert list values to immutable tuples to prevent in-place modification
            elif isinstance(value, list):
                # Store the list as a tuple
                self._storage[key] = tuple(value)
            # Otherwise, store the primitive value directly
            else:
                # Assign the immutable primitive value
                self._storage[key] = value

    def __getitem__(self, key: str) -> Any:
        # Validate that the key is a string to prevent improper lookups
        if not isinstance(key, str):
            raise TypeError(f"Index key must be a string, got {type(key).__name__}")
        # Enforce strict key presence validation
        if key not in self._storage:
            # Raise a detailed KeyError to locate the exact missing parameter path
            raise KeyError(f"Key '{key}' not found in the frozen configuration.")
        # Retrieve and return the value from the internal storage
        return self._storage[key]

    def __iter__(self) -> Iterator[str]:
        # Return an iterator over the keys of the internal storage
        return iter(self._storage)

    def __len__(self) -> int:
        # Return the exact number of active keys in the storage
        return len(self._storage)

    def __repr__(self) -> str:
        # Provide an unambiguous string representation for diagnostic logs
        return f"FrozenDict({self._storage})"


def load_and_freeze_config(raw_config: Dict[str, Any]) -> Tuple[FrozenDict, str]:
    """
    Resolves API credentials, redacts secrets, computes a canonical SHA-256 hash,
    and returns a deeply frozen configuration object.

    Parameters
    ----------
    raw_config : Dict[str, Any]
        The raw STUDY_CONFIG dictionary.

    Returns
    -------
    Tuple[FrozenDict, str]
        A tuple containing the immutable configuration and its canonical SHA-256 hash.

    Raises
    ------
    RuntimeError
        If the specified API key environment variable is missing or empty.
    """
    # Extract the name of the environment variable containing the API key
    api_key_env_var: str = raw_config["gateway"]["api_key_env_var"]

    # Resolve the actual API key from the system environment
    api_key: Optional[str] = os.environ.get(api_key_env_var)

    # Fail-fast assertion: Ensure the API key is present to prevent late-stage failures
    if not api_key:
        raise RuntimeError(
            f"Critical Configuration Error: The environment variable '{api_key_env_var}' "
            f"is missing or empty. API credentials must be provided via the environment."
        )

    # Create a shallow copy of the config to safely mutate the gateway block for redaction
    redacted_config: Dict[str, Any] = dict(raw_config)
    # Create a shallow copy of the gateway block
    redacted_gateway: Dict[str, Any] = dict(redacted_config["gateway"])

    # Explicitly redact the API key to guarantee it never enters logs or hashes
    redacted_gateway["api_key"] = "[REDACTED_FOR_SECURITY]"
    # Reattach the redacted gateway block to the configuration
    redacted_config["gateway"] = redacted_gateway

    # Serialize the redacted configuration to a canonical JSON string (sorted keys)
    canonical_json: str = json.dumps(redacted_config, sort_keys=True, separators=(',', ':'))

    # Compute the SHA-256 hash of the canonical JSON string to bind the run to this exact config
    config_hash: str = hashlib.sha256(canonical_json.encode('utf-8')).hexdigest()

    # Recursively freeze the redacted configuration to prevent downstream mutation
    frozen_config: FrozenDict = FrozenDict(redacted_config)

    # Return the immutable configuration and its cryptographic provenance hash
    return frozen_config, config_hash

# ------------------------------------------------------------------------------
# Task 1, Step 2: Initialize the Independent Seed Ledger
# ------------------------------------------------------------------------------

def initialize_seed_ledger(frozen_config: FrozenDict) -> Dict[str, np.random.Generator]:
    """
    Initialize statistically independent NumPy random generators — one per named
    consumer in the configuration's seed ledger — by spawning child SeedSequences
    from a single root entropy source.

    Rationale (§13.3). Reproducibility requires *independent* streams, one per
    named consumer ('global_seed', 'monte_carlo_sampling_seed',
    'replicate_seed_base', 'bootstrap_seed', 'router_seed'). Independence MUST be
    derived from one root via spawn-key hashing — NOT from distinct integer seeds:
    several consumers share the integer 20260618, and a naive
    `SeedSequence(seed_value)` per consumer would therefore yield byte-identical
    PCG64 streams (the exact failure the spawn idiom prevents). The deliberate
    bootstrap asymmetry (nominal seed 13) is preserved by folding each consumer's
    nominal seed into its spawn key, so the bootstrap stream differs materially
    while remaining fully reproducible.

    Parameters
    ----------
    frozen_config : FrozenDict
        The immutable study configuration. Must expose
        `reproducibility.seed_ledger` as a mapping of consumer-name -> int seed,
        and that ledger must contain the 'global_seed' root.

    Returns
    -------
    Dict[str, np.random.Generator]
        A dictionary mapping each ledger consumer name to its independent
        Generator(PCG64) stream. Keys are byte-identical to the ledger keys so
        that every downstream consumer requests its stream by name, never by
        position.

    Raises
    ------
    KeyError
        If the seed ledger or the mandatory 'global_seed' root is absent.
    TypeError
        If a ledger seed value is not an integer.
    """
    # Retrieve the seed-ledger mapping from the reproducibility block (single source of entropy truth).
    seed_ledger: Mapping[str, int] = frozen_config["reproducibility"]["seed_ledger"]

    # Validate that the ledger is a non-empty mapping; an empty ledger cannot seed any consumer.
    if not isinstance(seed_ledger, Mapping) or len(seed_ledger) == 0:
        # Fail fast: without a ledger there is no reproducible derivation to perform.
        raise KeyError("Seed ledger is missing or empty under 'reproducibility.seed_ledger'.")

    # The 'global_seed' is the canonical ROOT entropy; every child is derived from it (§13.3).
    if "global_seed" not in seed_ledger:
        # Fail fast: the spawn tree has no root without the global seed.
        raise KeyError("Seed ledger must contain the mandatory 'global_seed' root.")

    # Extract the integer root seed that anchors the entire spawn tree.
    root_seed: int = seed_ledger["global_seed"]

    # Enforce integer typing on the root (SeedSequence entropy must be a non-negative integer / int sequence).
    if not isinstance(root_seed, int) or isinstance(root_seed, bool):
        # Reject bool-as-int and non-integers to avoid silent entropy corruption.
        raise TypeError(f"Root 'global_seed' must be an int; got {type(root_seed).__name__}.")

    # Construct the SINGLE root SeedSequence from the root seed (the parent of all child streams).
    root_sequence: np.random.SeedSequence = np.random.SeedSequence(entropy=root_seed)

    # Fix a DETERMINISTIC, RECORDED spawn order independent of dict insertion order (reproducibility).
    ordered_consumers: List[str] = sorted(seed_ledger.keys())

    # Initialize the output bundle that will hold one independent Generator per consumer.
    rng_bundle: Dict[str, np.random.Generator] = {}

    # Initialize an audit ledger of child fingerprints for provenance logging (no run_dir in signature).
    spawn_audit: List[Dict[str, Any]] = []

    # Derive one independent child stream per consumer, in the fixed spawn order.
    for spawn_index, consumer_name in enumerate(ordered_consumers):
        # Retrieve this consumer's nominal seed (e.g., 13 for 'bootstrap_seed') to preserve the asymmetry.
        nominal_seed: int = seed_ledger[consumer_name]

        # Reject non-integer nominal seeds (bool excluded) to keep the spawn key well-defined.
        if not isinstance(nominal_seed, int) or isinstance(nominal_seed, bool):
            # Surface the offending consumer so the config can be corrected.
            raise TypeError(
                f"Seed for consumer '{consumer_name}' must be an int; got {type(nominal_seed).__name__}."
            )

        # Build the child SeedSequence: shared root entropy + a spawn key folding BOTH the
        # positional index and the nominal seed. Distinct spawn keys => statistically independent
        # streams via SeedSequence's hashing mixer; folding the nominal seed preserves the
        # deliberate per-consumer asymmetry (bootstrap=13) while remaining fully reproducible.
        child_sequence: np.random.SeedSequence = np.random.SeedSequence(
            entropy=root_seed,                          # shared root entropy (the spawn parent)
            spawn_key=(spawn_index, nominal_seed)        # distinct, asymmetry-preserving derivation key
        )

        # Instantiate the consumer's Generator on the modern PCG64 bit generator from its child sequence.
        # This avoids any global-state entanglement (we never call np.random.seed).
        rng_bundle[consumer_name] = np.random.Generator(np.random.PCG64(child_sequence))

        # Compute a compact, deterministic fingerprint (first two uint32 state words) for auditing.
        child_fingerprint: List[int] = child_sequence.generate_state(2).tolist()

        # Record this child's derivation parameters and fingerprint for provenance.
        spawn_audit.append({
            "consumer": consumer_name,                   # the ledger key (and bundle key)
            "spawn_index": spawn_index,                  # positional spawn key component
            "nominal_seed": nominal_seed,                # nominal seed folded into the spawn key
            "state_fingerprint_u32": child_fingerprint   # reproducible state fingerprint
        })

    # Emit the spawn order and child fingerprints to the module logger so the derivation is auditable
    # even though this function's signature does not expose a run directory to persist to.
    logger.info(
        "Seed ledger derivation: root=%s, spawn_order=%s, children=%s",
        root_seed, ordered_consumers, spawn_audit
    )

    # Return the bundle keyed by consumer name (independent, reproducible, asymmetry-preserving streams).
    return rng_bundle

# ------------------------------------------------------------------------------
# Task 1, Step 3: Pin Runtime Environment and Initialize Run Directory
# ------------------------------------------------------------------------------

def initialize_run_environment(config_hash: str) -> Path:
    """
    Create the immutable run directory and capture full environment provenance,
    including a complete dependency lockfile and a pre-recorded manifest schema.

    Rationale (Section 13.3; detailed tasklist Task 1, Step 3). Auditable
    reproducibility requires pinning the interpreter, a FULL dependency lockfile, and
    the BLAS/LAPACK backend identity, plus a write-once run directory and a
    pre-recorded model-version manifest schema. The prior implementation captured only
    sys.version and the NumPy config and wrote an empty manifest scaffold. This version
    additionally enumerates every installed distribution (name -> version) via
    importlib.metadata, pins the scientific stack explicitly, records a platform
    fingerprint, and writes a structured manifest schema.

    Parameters
    ----------
    config_hash : str
        The SHA-256 hash of the configuration (provenance binding).

    Returns
    -------
    Path
        The newly created immutable run directory.

    Raises
    ------
    FileExistsError
        If the timestamped directory already exists (write-once guarantee).
    """
    # High-precision UTC timestamp for the directory name.
    timestamp: str = datetime.datetime.now(datetime.timezone.utc).strftime("%Y%m%dT%H%M%SZ")
    # Construct the run directory path.
    run_dir: Path = Path(f"./fmec_run_{timestamp}")
    # Create it write-once (fail fast on collision).
    run_dir.mkdir(parents=True, exist_ok=False)

    # Capture the exact interpreter version.
    python_version: str = sys.version

    # Capture the NumPy/BLAS/LAPACK configuration (drives covariance numerical stability).
    try:
        # Modern dict-based config retrieval (NumPy 1.25+).
        numpy_config: Any = np.show_config(mode="dicts")
    except TypeError:
        # Fallback for older NumPy.
        numpy_config = "NumPy version does not support dict output for show_config."

    # Enumerate a FULL dependency lockfile: name -> version for every installed distribution.
    dependency_lockfile: Dict[str, str] = {}
    try:
        # importlib.metadata is stdlib (Python 3.8+); import lazily to avoid a hard module dependency.
        import importlib.metadata as importlib_metadata
        # Iterate installed distributions and record name->version.
        for dist in importlib_metadata.distributions():
            # Extract the distribution name (robust across metadata layouts).
            name = (dist.metadata.get("Name") if dist.metadata else None) or getattr(dist, "name", None)
            # Record the version when both name and version are available.
            if name and dist.version:
                dependency_lockfile[str(name)] = str(dist.version)
    except Exception as exc:  # pragma: no cover - environment-dependent
        # If enumeration fails, record the failure rather than silently omitting the lockfile.
        dependency_lockfile = {"_error": f"dependency enumeration failed: {exc!s}"}

    # Explicitly pin the scientific stack (the libraries whose versions most affect numerics).
    scientific_stack: Dict[str, str] = {}
    for pkg in ("numpy", "pandas", "scipy", "scikit-learn"):
        # Prefer the lockfile entry; fall back to a direct version query.
        if pkg in dependency_lockfile:
            scientific_stack[pkg] = dependency_lockfile[pkg]
        else:
            try:
                import importlib.metadata as importlib_metadata
                scientific_stack[pkg] = importlib_metadata.version(pkg)
            except Exception:
                scientific_stack[pkg] = "unavailable"

    # Capture a platform fingerprint (OS/arch/processor) for Tier-2 drift attribution.
    platform_fingerprint: Dict[str, str] = {}
    try:
        # platform is stdlib; import lazily.
        import platform as _platform
        platform_fingerprint = {
            "platform": _platform.platform(),          # full OS/arch string
            "machine": _platform.machine(),            # CPU architecture
            "processor": _platform.processor(),        # processor identifier
            "python_implementation": _platform.python_implementation(),  # CPython/PyPy
        }
    except Exception:  # pragma: no cover
        platform_fingerprint = {"_error": "platform capture failed"}

    # Assemble the environment provenance record.
    env_provenance: Dict[str, Any] = {
        "timestamp_utc": timestamp,                    # run timestamp
        "config_sha256": config_hash,                  # configuration provenance binding
        "python_version": python_version,              # interpreter version
        "platform": platform_fingerprint,              # OS/arch fingerprint
        "numpy_config": numpy_config,                  # BLAS/LAPACK backend identity
        "scientific_stack": scientific_stack,          # explicit numeric-stack pins
        "dependency_lockfile": dependency_lockfile,    # FULL name->version lockfile
    }
    # Serialize the environment provenance.
    with open(run_dir / "environment_provenance.json", "w", encoding="utf-8") as f:
        json.dump(env_provenance, f, indent=2)

    # Pre-record the model-version manifest SCHEMA so collection appends without restructuring.
    manifest_scaffold: Dict[str, Any] = {
        "_schema": {
            # The fields each per-call manifest entry will carry (drift/audit provenance).
            "entry_fields": [
                "model_id", "provider_version", "system_fingerprint",
                "resolved_model_id", "seed", "first_seen_utc",
            ],
            # The fields are appended per (model, call) during collection (§10.2).
            "description": "Per-model version manifest; one entry appended per distinct model/version/fingerprint.",
        },
        # The manifest entries accumulate here during Task 9 collection.
        "entries": [],
    }
    # Serialize the structured manifest scaffold.
    with open(run_dir / "model_version_manifest.json", "w", encoding="utf-8") as f:
        json.dump(manifest_scaffold, f, indent=2)

    # Return the handle to the immutable run directory.
    return run_dir

# ------------------------------------------------------------------------------
# Task 1, Orchestrator Function
# ------------------------------------------------------------------------------

def load_config_and_seeds(raw_config: Dict[str, Any]) -> Tuple[FrozenDict, Path, Dict[str, np.random.Generator]]:
    """
    Orchestrates Task 1: Validates credentials, freezes the configuration,
    initializes independent RNG streams, and pins the runtime environment.

    Parameters
    ----------
    raw_config : Dict[str, Any]
        The raw STUDY_CONFIG dictionary containing all parameters and schemas.

    Returns
    -------
    Tuple[FrozenDict, Path, Dict[str, np.random.Generator]]
        - frozen_config: The immutable, redacted configuration object.
        - run_dir: The Path to the timestamped, immutable run directory.
        - rng_bundle: A dictionary of independent NumPy generators keyed by consumer.
    """
    # Step 1: Resolve credentials, redact secrets, hash, and deep-freeze the config
    frozen_config, config_hash = load_and_freeze_config(raw_config)

    # Step 2: Initialize the independent seed ledger to prevent global state entanglement
    rng_bundle = initialize_seed_ledger(frozen_config)

    # Step 3: Pin the runtime environment and create the immutable run directory
    run_dir = initialize_run_environment(config_hash)

    # Return the fully initialized environment components
    return frozen_config, run_dir, rng_bundle
````

````python
# Task 2 — Parameter Appropriateness & Quality Validation (validate_config)

# ==============================================================================
# Task 2: Parameter Appropriateness & Quality Validation
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 2, Helper Structures: Severity Enum and ValidationFinding Dataclass
# ------------------------------------------------------------------------------

class Severity(Enum):
    """
    Enumeration defining the severity level of a configuration validation finding.

    This enum provides a standardized classification for parameter and structural
    anomalies discovered during config validation (Task 2). It distinguishes between
    fatal discrepancies (HARD) that compromise the pipeline's mathematical integrity,
    and informational discrepancies (SOFT) that are logged for auditability but
    do not compromise pipeline security.
    """
    # Hard violations represent fatal discrepancies and mandate immediate pipeline shutdown
    HARD = "HARD"
    # Soft violations represent non-fatal anomalies and are flagged for post-run audits
    SOFT = "SOFT"

    @classmethod
    def validate(cls, value: str) -> "Severity":
        """
        Validates and instantiates a Severity member from a raw string value.

        Parameters
        ----------
        value : str
            The raw string representing the severity level (e.g., 'HARD', 'soft').

        Returns
        -------
        Severity
            The matched Severity enum member.

        Raises
        ------
        TypeError
            If the input value is not a string.
        ValueError
            If the value does not match any valid Severity member.
        """
        # Validate that the incoming value is a standard string
        if not isinstance(value, str):
            # Raise TypeError if the type is violated
            raise TypeError(f"Severity validation requires a string input, got {type(value).__name__}")
        # Normalize the string: convert to uppercase and strip outer whitespace
        normalized_value = value.upper().strip()
        # Check if the normalized string matches a registered enum value
        if normalized_value not in cls._value2member_map_:
            # Raise an explicit ValueError if the string is unrecognized
            raise ValueError(f"Invalid severity value '{value}'. Admissible: {list(cls._value2member_map_.keys())}")
        # Instantiate and return the validated Severity member
        return cls(normalized_value)

@dataclass
class ValidationFinding:
    """
    Dataclass encapsulating a single configuration validation finding.

    Attributes
    ----------
    path : str
        The dot-separated path to the configuration parameter.
    expected : str
        A description of the expected type, range, or condition.
    observed : Any
        The actual value observed in the configuration.
    severity : Severity
        The severity level of the finding (HARD or SOFT).
    message : str
        A human-readable explanation of the violation.
    """
    path: str
    expected: str
    observed: Any
    severity: Severity
    message: str

def _get_nested_value(config: Mapping[str, Any], path: str) -> Any:
    """
    Safely retrieves a value from a nested dictionary using a dot-separated path.

    Parameters
    ----------
    config : Mapping[str, Any]
        The configuration dictionary.
    path : str
        The dot-separated path (e.g., "estimation.utility_weights.alpha").

    Returns
    -------
    Any
        The retrieved value.

    Raises
    ------
    KeyError
        If any part of the path is not found in the configuration.
    """
    # Split the dot-separated path into individual keys
    keys: List[str] = path.split(".")
    # Initialize the current level to the root configuration
    current: Any = config
    # Traverse the nested structure
    for key in keys:
        # Attempt to access the next level
        current = current[key]
    # Return the final resolved value
    return current

# ------------------------------------------------------------------------------
# Task 2, Step 1: Validate numerical parameter ranges, types, and enumerations
# ------------------------------------------------------------------------------

def validate_parameter_ranges(config: Mapping[str, Any]) -> List[ValidationFinding]:
    """
    Validate the types, finiteness, ranges, and allowed values of all non-data
    parameters against a declarative schema.

    Rationale (detailed tasklist Task 2, Step 1). The protocol mandates that
    weights, penalties, budgets, probabilities, and temperatures be FINITE and
    within bounds. Finiteness is a first-class requirement: `NaN` silently passes
    both `val < min` and `val > max` (both evaluate False), and `+inf` passes any
    lower bound. This implementation adds an explicit `np.isfinite` gate (HARD) for
    every parameter flagged `finite`, applied AFTER the strict type check and
    BEFORE the range checks, while preserving all prior type/range/enum behaviour.

    Parameters
    ----------
    config : Mapping[str, Any]
        The immutable study configuration to be validated.

    Returns
    -------
    List[ValidationFinding]
        The accumulated single-field validation findings (HARD/SOFT). The function
        accumulates rather than fails fast so that Step 3 can enforce atomically.
    """
    # Initialize the list that accumulates all single-field findings (accumulate-don't-fail-fast).
    findings: List[ValidationFinding] = []

    # Declarative validation schema: path -> {type, finite?, min?, max?, allowed?}.
    # 'finite': True marks continuous parameters that must be members of the real line (no NaN/inf).
    validation_schema: Dict[str, Dict[str, Any]] = {
        # Utility weights (Eq. 3) — non-negative, finite floats.
        "estimation.utility_weights.alpha": {"type": float, "min": 0.0, "finite": True},
        "estimation.utility_weights.beta": {"type": float, "min": 0.0, "finite": True},
        "estimation.utility_weights.gamma": {"type": float, "min": 0.0, "finite": True},
        "estimation.utility_weights.delta": {"type": float, "min": 0.0, "finite": True},

        # FMEC hyperparameters (Eq. 13) — non-negative, finite floats.
        "estimation.fmec_hyperparams.eta": {"type": float, "min": 0.0, "finite": True},
        "estimation.fmec_hyperparams.lambda": {"type": float, "min": 0.0, "finite": True},

        # Integer counts — strictly positive integers (finiteness is intrinsic to int).
        "estimation.monte_carlo.K": {"type": int, "min": 1},
        "uncertainty.bootstrap.B": {"type": int, "min": 1},
        "uncertainty.replication.R": {"type": int, "min": 1},
        "uncertainty.replication.R_judge": {"type": int, "min": 1},
        "preprocessing.n_capability_clusters": {"type": int, "min": 1},
        "oracle_eval.max_pool_size_for_enumeration": {"type": int, "min": 1},

        # Probabilities and thresholds — finite floats in [0, 1].
        "preprocessing.qualification_threshold_pct": {"type": float, "min": 0.0, "max": 1.0, "finite": True},
        "uncertainty.replication.adjudication_subsample": {"type": float, "min": 0.0, "max": 1.0, "finite": True},
        "stat_tests.rsp_floor": {"type": float, "min": 0.0, "max": 1.0, "finite": True},
        "stat_tests.kappa_floor": {"type": float, "min": 0.0, "max": 1.0, "finite": True},
        "stat_tests.delta_tau": {"type": float, "min": 0.0, "max": 1.0, "finite": True},
        "network_training.router.epsilon": {"type": float, "min": 0.0, "max": 1.0, "finite": True},
        "stat_tests.negligible_cliffs_delta": {"type": float, "min": 0.0, "max": 1.0, "finite": True},

        # Budgets (Eq. 14) — non-negative, finite floats.
        "optimization.C_max": {"type": float, "min": 0.0, "finite": True},
        "optimization.L_max": {"type": float, "min": 0.0, "finite": True},
        "optimization.R_max": {"type": float, "min": 0.0, "finite": True},

        # Role-separated decoding temperatures — finite floats in [0, 2].
        "llm_settings.panel.temperature": {"type": float, "min": 0.0, "max": 2.0, "finite": True},
        "llm_settings.solver.temperature": {"type": float, "min": 0.0, "max": 2.0, "finite": True},
        "llm_settings.judge.temperature": {"type": float, "min": 0.0, "max": 2.0, "finite": True},

        # Enumerations — exact membership in the allowed string set.
        "estimation.monte_carlo.coalition_sampler": {"type": str, "allowed": {"permutation", "size_mixture"}},
        "estimation.covariance.estimator": {"type": str, "allowed": {"empirical", "ledoit_wolf", "bayesian", "bootstrap"}},
        "estimation.information_gain.estimator": {"type": str, "allowed": {"outcome_based_cmi", "entropy_reduction", "judge_model", "agreement_based"}},
        "uncertainty.bootstrap.interval": {"type": str, "allowed": {"BCa", "percentile", "bootstrap_t"}},
        "uncertainty.bootstrap.scheme": {"type": str, "allowed": {"nested_cluster", "single_level"}},
        "architecture.fusion_operator": {"type": str, "allowed": {"judge", "weighted_vote", "verifier_augmented"}},
    }

    # Iterate the schema and validate each parameter in declared order.
    for path, rules in validation_schema.items():
        # Attempt to retrieve the value at this dotted path; a missing key is a HARD finding.
        try:
            # Resolve the nested value via the dot-path accessor.
            val: Any = _get_nested_value(config, path)
        except KeyError:
            # Record the missing-key violation and skip all further checks for this path.
            findings.append(ValidationFinding(
                path=path, expected="Key exists", observed="Missing",
                severity=Severity.HARD, message=f"Required configuration key '{path}' is missing."
            ))
            # Move to the next schema entry; nothing else can be checked for an absent key.
            continue

        # Extract the expected Python type for this parameter.
        expected_type: Optional[type] = rules.get("type")

        # Strict type check: `type(val) is not expected_type` rejects bool masquerading as int.
        if expected_type is not None and type(val) is not expected_type:
            # Record the type mismatch and skip range/finiteness checks (they are ill-defined).
            findings.append(ValidationFinding(
                path=path, expected=expected_type.__name__, observed=type(val).__name__,
                severity=Severity.HARD, message=f"Type mismatch for '{path}'."
            ))
            # The type is fundamentally wrong; do not attempt numeric comparisons.
            continue

        # Finiteness gate (NEW): for continuous parameters, reject NaN and +/- inf before bound checks.
        if rules.get("finite", False) and not np.isfinite(val):
            # Record the non-finite violation as HARD; comparisons against bounds are meaningless here.
            findings.append(ValidationFinding(
                path=path, expected="finite real value", observed=repr(val),
                severity=Severity.HARD, message=f"Value for '{path}' is non-finite (NaN/inf)."
            ))
            # Skip the range checks because `<`/`>` against a non-finite value cannot be trusted.
            continue

        # Lower-bound check (only reached for finite, correctly typed values).
        if "min" in rules and val < rules["min"]:
            # Record the below-minimum violation.
            findings.append(ValidationFinding(
                path=path, expected=f">= {rules['min']}", observed=val,
                severity=Severity.HARD, message=f"Value for '{path}' is below minimum."
            ))

        # Upper-bound check.
        if "max" in rules and val > rules["max"]:
            # Record the above-maximum violation.
            findings.append(ValidationFinding(
                path=path, expected=f"<= {rules['max']}", observed=val,
                severity=Severity.HARD, message=f"Value for '{path}' is above maximum."
            ))

        # Enumeration membership check.
        if "allowed" in rules and val not in rules["allowed"]:
            # Record the not-in-allowed-set violation.
            findings.append(ValidationFinding(
                path=path, expected=f"One of {rules['allowed']}", observed=val,
                severity=Severity.HARD, message=f"Value for '{path}' is not in allowed set."
            ))

    # Return the accumulated findings for atomic enforcement downstream (Step 3).
    return findings

# ------------------------------------------------------------------------------
# Task 2, Step 2: Validate cross-field invariants and anti-silent-drift constraints
# ------------------------------------------------------------------------------

def validate_cross_field_invariants(config: Mapping[str, Any]) -> List[ValidationFinding]:
    """
    Validate cross-field structural invariants: bootstrap resample levels, seed
    ledger alignment, prompt fill-token integrity, and the explicit presence of
    deferred parameters.

    Rationale (detailed tasklist Task 2, Step 2; Appendix C.1-C.3). Each prompt
    template must contain exactly its required fill-tokens; ALL other brace pairs
    are literal JSON content. Validation must therefore check that no *non-fill*
    brace matches an identifier pattern — literal JSON braces must never be treated
    as format slots. The prior implementation captured the multi-line literal JSON
    body of `judge_system`/`solver_system` as a single invalid token and raised a
    HARD finding, aborting the pipeline on its own shipped config. This version (1)
    filters brace captures to a strict identifier pattern before subtracting the
    allowlist, (2) replaces the monolithic seed-alignment try/except with per-key
    handling, and (3) optionally checks required tokens for *positively identified*
    templates only (no false positives).

    Parameters
    ----------
    config : Mapping[str, Any]
        The immutable study configuration to be validated.

    Returns
    -------
    List[ValidationFinding]
        The accumulated cross-field findings (HARD/SOFT), for atomic enforcement
        in Step 3.
    """
    # Initialize the accumulator for cross-field findings.
    findings: List[ValidationFinding] = []

    # ---------------------------------------------------------------------------
    # 1. Bootstrap resample levels must be exactly the three variance axes (Eq. 20a).
    # ---------------------------------------------------------------------------
    try:
        # Retrieve the configured resample levels.
        levels: List[str] = _get_nested_value(config, "uncertainty.bootstrap.resample_levels")
        # The total-variance decomposition requires exactly these three axes, in order.
        required_levels: List[str] = ["task", "replicate", "adjudication"]
        # Compare for exact equality; any deviation breaks the variance-component contract.
        if list(levels) != required_levels:
            # Record the mismatch as HARD.
            findings.append(ValidationFinding(
                path="uncertainty.bootstrap.resample_levels", expected=str(required_levels), observed=str(levels),
                severity=Severity.HARD, message="Resample levels must exactly match the three variance components."
            ))
    except KeyError:
        # Absent key is surfaced by Step 1's missing-key logic; nothing to do here.
        pass

    # ---------------------------------------------------------------------------
    # 2. Seed ledger alignment — PER-KEY handling (no monolithic try/except).
    # ---------------------------------------------------------------------------
    # Retrieve the central seed ledger; if it is entirely absent, record one finding and skip.
    try:
        # Resolve the ledger mapping.
        ledger: Mapping[str, int] = _get_nested_value(config, "reproducibility.seed_ledger")
    except KeyError:
        # Record the missing ledger as HARD and set to None so the per-key loop is skipped.
        findings.append(ValidationFinding(
            path="reproducibility.seed_ledger", expected="Mapping of consumer seeds", observed="Missing",
            severity=Severity.HARD, message="Central seed ledger is missing."
        ))
        ledger = None  # type: ignore[assignment]

    # Map each ledger key to the module path that must mirror it (anti-silent-drift).
    seed_checks: Dict[str, str] = {
        "global_seed": "reproducibility.global_seed",
        "monte_carlo_sampling_seed": "estimation.monte_carlo.sampling_seed",
        "replicate_seed_base": "uncertainty.replication.replicate_seed_base",
        "bootstrap_seed": "uncertainty.bootstrap.bootstrap_seed",
        "router_seed": "network_training.router.seed",
    }

    # Only run the alignment loop if the ledger was successfully retrieved.
    if ledger is not None:
        # Validate each (ledger_key, module_path) pair independently so one miss never aborts the rest.
        for ledger_key, module_path in seed_checks.items():
            # Determine whether the ledger contains this key (None signals absence here).
            ledger_has_key: bool = ledger_key in ledger
            # Retrieve the ledger value if present.
            ledger_val: Optional[int] = ledger.get(ledger_key)

            # Attempt to retrieve the corresponding module value independently.
            try:
                # Resolve the module-side value at its dotted path.
                module_val: Any = _get_nested_value(config, module_path)
                module_present: bool = True
            except KeyError:
                # Mark the module value as absent without aborting the loop.
                module_val = None
                module_present = False

            # Case A: the ledger lacks this key — record a per-key HARD finding.
            if not ledger_has_key:
                findings.append(ValidationFinding(
                    path=f"reproducibility.seed_ledger.{ledger_key}", expected="present", observed="Missing",
                    severity=Severity.HARD, message=f"Seed ledger is missing key '{ledger_key}'."
                ))
                # Continue to the next pair; alignment cannot be checked for a missing ledger entry.
                continue

            # Case B: the module value is absent — record a per-key HARD finding.
            if not module_present:
                findings.append(ValidationFinding(
                    path=module_path, expected=f"present and == ledger '{ledger_key}'", observed="Missing",
                    severity=Severity.HARD, message=f"Module seed path '{module_path}' is missing."
                ))
                # Continue; nothing to compare against.
                continue

            # Case C: both present but divergent — record a per-key HARD finding (silent drift).
            if ledger_val != module_val:
                findings.append(ValidationFinding(
                    path=module_path, expected=f"Match ledger {ledger_key} ({ledger_val})", observed=module_val,
                    severity=Severity.HARD, message="Module seed diverges from central seed ledger."
                ))

    # ---------------------------------------------------------------------------
    # 3. Prompt fill-token integrity — identifier filter prevents literal-JSON false positives.
    # ---------------------------------------------------------------------------
    try:
        # Retrieve the dictionary of prompt templates.
        prompts: Mapping[str, str] = _get_nested_value(config, "prompts")
        # The allowlist of legitimate fill-slot identifiers across all templates.
        allowed_tokens: Set[str] = {
            "task_prompt", "reference", "candidate_answer",
            "prompt", "answers_json", "json_schema",
        }
        # Regex to capture every single-brace-enclosed substring (content may be arbitrary).
        brace_pattern = re.compile(r"\{([^{}]*)\}")
        # Strict identifier pattern: ONLY these captures are candidate fill-slots.
        identifier_pattern = re.compile(r"^[A-Za-z_][A-Za-z0-9_]*$")

        # Conservative required-token map keyed by template-name substrings we can positively identify.
        # A finding fires only when a positively-identified template is missing a required token.
        required_token_rules: Dict[str, Set[str]] = {
            "panel": {"task_prompt"},                                  # panel prompt elicits answers to the task
            "judge": {"task_prompt", "reference", "candidate_answer"}, # judge prompt scores candidate vs reference
            "solver": {"task_prompt"},                                 # solver prompt solves the task
        }

        # Iterate each named prompt template.
        for prompt_name, prompt_text in prompts.items():
            # Skip non-string entries (e.g., nested JSON schema objects) — they are not format templates.
            if not isinstance(prompt_text, str):
                continue

            # Extract all brace captures from the template text.
            raw_captures: List[str] = brace_pattern.findall(prompt_text)
            # Keep ONLY captures that are valid identifiers — literal JSON (spaces/quotes/colons) is excluded.
            identifier_tokens: Set[str] = {tok for tok in raw_captures if identifier_pattern.match(tok)}
            # Identify identifier-shaped tokens that are NOT legitimate fill slots.
            invalid_tokens: Set[str] = identifier_tokens - allowed_tokens
            # If any identifier-shaped token is unrecognized, it is a genuine bad fill-token.
            if invalid_tokens:
                # Record the invalid-token violation (literal JSON braces can no longer trigger this).
                findings.append(ValidationFinding(
                    path=f"prompts.{prompt_name}", expected=f"Subset of {allowed_tokens}", observed=str(sorted(invalid_tokens)),
                    severity=Severity.HARD, message="Prompt contains an unrecognized identifier-shaped fill-token."
                ))

            # Optional required-token presence check for positively-identified templates only (no false positives).
            for name_substring, required in required_token_rules.items():
                # Only apply a rule when the template name clearly matches the known role.
                if name_substring in prompt_name.lower():
                    # Determine which required tokens are absent from this template.
                    missing_required: Set[str] = required - identifier_tokens
                    # If a known template lacks a mandated fill-slot, record it as HARD.
                    if missing_required:
                        findings.append(ValidationFinding(
                            path=f"prompts.{prompt_name}", expected=f"contains {sorted(required)}",
                            observed=f"missing {sorted(missing_required)}",
                            severity=Severity.HARD, message="Identified prompt template is missing a required fill-token."
                        ))
    except KeyError:
        # Absent prompts block is handled elsewhere; nothing to validate here.
        pass

    # ---------------------------------------------------------------------------
    # 4. reference_weight_w must be explicitly defined (deferred simplex check happens in Task 15).
    # ---------------------------------------------------------------------------
    try:
        # Retrieve the reference weight rule/vector.
        ref_w: Any = _get_nested_value(config, "estimation.reference_weight_w")
        # It must be an explicit rule (e.g., "simplex_uniform") or vector — never None.
        if ref_w is None:
            # Record the unfilled deferred parameter as HARD.
            findings.append(ValidationFinding(
                path="estimation.reference_weight_w", expected="Explicit rule or vector", observed="None",
                severity=Severity.HARD, message="reference_weight_w must be explicitly defined to compute MRC."
            ))
    except KeyError:
        # Handled by Step 1's missing-key logic if applicable.
        pass

    # ---------------------------------------------------------------------------
    # 5. fusion_sim_audit.per_stratum_tolerance must be explicitly defined.
    # ---------------------------------------------------------------------------
    try:
        # Retrieve the per-stratum tolerance.
        tol: Any = _get_nested_value(config, "fusion_sim_audit.per_stratum_tolerance")
        # It must be an explicit float — never None.
        if tol is None:
            # Record the unfilled deferred parameter as HARD.
            findings.append(ValidationFinding(
                path="fusion_sim_audit.per_stratum_tolerance", expected="Float value", observed="None",
                severity=Severity.HARD, message="per_stratum_tolerance must be explicitly defined for the audit."
            ))
    except KeyError:
        # Handled elsewhere; nothing to do.
        pass

    # ---------------------------------------------------------------------------
    # 6. tensor_sha256 == None is allowed but signals 'establish' mode (SOFT).
    # ---------------------------------------------------------------------------
    try:
        # Retrieve the canonical tensor hash.
        tensor_hash: Any = _get_nested_value(config, "reproducibility.tensor_sha256")
        # A None hash opens the Tier-1 establish gate; record it as SOFT (informational, non-aborting).
        if tensor_hash is None:
            findings.append(ValidationFinding(
                path="reproducibility.tensor_sha256", expected="SHA-256 string", observed="None",
                severity=Severity.SOFT, message="Tensor hash is None. Run is in 'establish' mode (Tier-1 gate open)."
            ))
    except KeyError:
        # Handled elsewhere; nothing to do.
        pass

    # Return the accumulated cross-field findings.
    return findings

# ------------------------------------------------------------------------------
# Task 2, Step 3: Accumulate findings, generate report, and enforce hard constraints
# ------------------------------------------------------------------------------

def generate_validation_report_and_enforce(findings: List[ValidationFinding], run_dir: Path) -> Dict[str, Any]:
    """
    Serializes the validation findings to disk and aborts the run if any HARD
    violations are detected.

    Parameters
    ----------
    findings : List[ValidationFinding]
        The complete list of accumulated validation findings.
    run_dir : Path
        The path to the immutable run directory.

    Returns
    -------
    Dict[str, Any]
        The structured validation report dictionary.

    Raises
    ------
    RuntimeError
        If any finding has a severity of HARD.
    """
    # Separate findings by severity
    hard_findings: List[ValidationFinding] = [f for f in findings if f.severity == Severity.HARD]
    soft_findings: List[ValidationFinding] = [f for f in findings if f.severity == Severity.SOFT]

    # Construct the master report dictionary
    report: Dict[str, Any] = {
        "status": "FAILED" if hard_findings else "PASSED",
        "total_findings": len(findings),
        "hard_violations_count": len(hard_findings),
        "soft_violations_count": len(soft_findings),
        # Convert dataclasses to dictionaries, extracting the enum name for JSON serialization
        "details": [
            {**asdict(f), "severity": f.severity.name} for f in findings
        ]
    }

    # Define the path for the validation report JSON file
    report_path: Path = run_dir / "config_validation_report.json"

    # Serialize and write the report to disk to preserve provenance
    with open(report_path, "w", encoding="utf-8") as f:
        json.dump(report, f, indent=2)

    # Enforce the fail-closed security principle
    if hard_findings:
        # Raise a descriptive error pointing the user to the persisted report
        raise RuntimeError(
            f"Configuration Validation Failed: {len(hard_findings)} HARD violations detected. "
            f"Run aborted to prevent silent methodological drift. "
            f"See detailed report at: {report_path}"
        )

    # Return the report dictionary if validation passed
    return report

# ------------------------------------------------------------------------------
# Task 2, Orchestrator Function
# ------------------------------------------------------------------------------

def validate_config(config: Mapping[str, Any], run_dir: Path) -> Dict[str, Any]:
    """
    Orchestrates Task 2: Validates parameter ranges, enforces cross-field
    invariants, generates a provenance report, and aborts on hard violations.

    Parameters
    ----------
    config : Mapping[str, Any]
        The immutable study configuration.
    run_dir : Path
        The path to the immutable run directory.

    Returns
    -------
    Dict[str, Any]
        The structured validation report dictionary.

    Raises
    ------
    RuntimeError
        If any HARD configuration violations are detected.
    """
    # Initialize the master list of findings
    all_findings: List[ValidationFinding] = []

    # Step 1: Validate single-field numerical ranges, types, and enumerations
    range_findings: List[ValidationFinding] = validate_parameter_ranges(config)
    all_findings.extend(range_findings)

    # Step 2: Validate cross-field structural invariants and anti-drift constraints
    cross_findings: List[ValidationFinding] = validate_cross_field_invariants(config)
    all_findings.extend(cross_findings)

    # Step 3: Generate the report, persist it, and enforce the fail-closed policy
    validation_report: Dict[str, Any] = generate_validation_report_and_enforce(
        findings=all_findings,
        run_dir=run_dir
    )

    # Return the successful validation report
    return validation_report
````

````python
# Task 3 — Raw Data Ingestion & Schema Validation (ingest_and_validate_inputs)

# ==============================================================================
# Task 3: Raw Data Ingestion & Schema Validation
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 3, Helper Structures: Severity Enum and IngestionFinding Dataclass
# ------------------------------------------------------------------------------

class IngestionSeverity(Enum):
    """
    Enumeration defining the severity level of a data ingestion finding.

    This enum provides a standardized classification for parameter and structural
    anomalies discovered during data ingestion and validation (Task 3). It distinguishes
    between fatal discrepancies (HARD) that indicate corrupted data and must abort the
    pipeline, and non-fatal discrepancies (SOFT) that require demotion or flagging
    but do not halt execution.
    """
    # Hard violations indicate corrupted data and must abort the pipeline immediately
    HARD = "HARD"
    # Soft violations indicate data that requires demotion or flagging, but not an abort
    SOFT = "SOFT"

    @classmethod
    def validate(cls, value: str) -> "IngestionSeverity":
        """
        Validates and instantiates an IngestionSeverity member from a raw string value.

        Parameters
        ----------
        value : str
            The raw string representing the severity level (e.g., 'HARD', 'soft').

        Returns
        -------
        IngestionSeverity
            The matched IngestionSeverity enum member.

        Raises
        ------
        TypeError
            If the input value is not a string.
        ValueError
            If the value does not match any valid IngestionSeverity member.
        """
        # Validate that the incoming value is a standard string
        if not isinstance(value, str):
            # Raise TypeError if the type is violated
            raise TypeError(f"IngestionSeverity validation requires a string input, got {type(value).__name__}")
        # Normalize the string: convert to uppercase and strip outer whitespace
        normalized_value = value.upper().strip()
        # Check if the normalized string matches a registered enum value
        if normalized_value not in cls._value2member_map_:
            # Raise an explicit ValueError if the string is unrecognized
            raise ValueError(f"Invalid severity value '{value}'. Admissible: {list(cls._value2member_map_.keys())}")
        # Instantiate and return the validated IngestionSeverity member
        return cls(normalized_value)

@dataclass
class IngestionFinding:
    """
    Dataclass encapsulating a single data ingestion and validation finding.

    Attributes
    ----------
    table_name : str
        The name of the data structure (e.g., 'model_answer').
    rule_violated : str
        A short description of the invariant or schema rule that failed.
    message : str
        A detailed, human-readable explanation of the violation.
    severity : IngestionSeverity
        The severity level of the finding (HARD or SOFT).
    example_row_keys : List[Union[int, str]]
        A sample of index keys or primary keys for the violating rows.
    total_violating_rows : int
        The total number of rows that violated this specific rule.
    """
    table_name: str
    rule_violated: str
    message: str
    severity: IngestionSeverity
    example_row_keys: List[Union[int, str]]
    total_violating_rows: int

# ------------------------------------------------------------------------------
# Task 3, Step 1: Materialize DataFrames, Validate Schemas, and Coerce Data Types
# ------------------------------------------------------------------------------

def materialize_and_coerce_table(
    raw_data: Union[List[Dict[str, Any]], pd.DataFrame],
    table_name: str,
    schema_config: Mapping[str, Any]
) -> pd.DataFrame:
    """
    Converts raw input data into a strictly typed pandas DataFrame, enforcing
    exact column presence, canonical ordering, and precise data types.

    Parameters
    ----------
    raw_data : Union[List[Dict[str, Any]], pd.DataFrame]
        The raw input data (list of dicts or an existing DataFrame).
    table_name : str
        The logical name of the table (e.g., 'model_answer').
    schema_config : Mapping[str, Any]
        The configuration block defining the schema for this specific table.

    Returns
    -------
    pd.DataFrame
        The materialized, canonically ordered, and strictly typed DataFrame.

    Raises
    ------
    ValueError
        If the input data contains missing or extraneous columns relative to the schema.
    """
    # Materialize the input into a pandas DataFrame if it is a list of dictionaries
    if isinstance(raw_data, pd.DataFrame):
        df = raw_data.copy()
    else:
        df = pd.DataFrame(raw_data)

    # Extract the canonical list of expected columns from the configuration schema
    expected_columns: List[str] = list(schema_config["schema"])

    # Extract the observed columns from the materialized DataFrame
    observed_columns: List[str] = df.columns.tolist()

    # Compute the symmetric difference to identify missing or extraneous columns
    missing_cols = set(expected_columns) - set(observed_columns)
    extra_cols = set(observed_columns) - set(expected_columns)

    # Enforce strict schema compliance: fail fast if the column sets do not match perfectly
    if missing_cols or extra_cols:
        raise ValueError(
            f"Schema violation in table '{table_name}'.\n"
            f"Missing columns: {missing_cols}\n"
            f"Extraneous columns: {extra_cols}"
        )

    # Reorder the DataFrame columns to match the canonical sequence defined in the schema
    df = df[expected_columns]

    # Define a comprehensive mapping of logical column names to strict pandas dtypes
    # This prevents silent upcasting (e.g., integers with NaNs becoming float64)
    dtype_mapping: Dict[str, str] = {
        # Identifiers and provenance strings are cast to categorical for memory efficiency
        "task_id": "category",
        "model_id": "category",
        "domain": "category",
        "subdomain": "category",
        "scoring_rule": "category",
        "difficulty": "category",
        "family": "category",
        "provider_version": "category",
        "system_fingerprint": "category",

        # Binary flags and correctness indicators use nullable 8-bit integers
        "correct": "Int8",
        "e_reason": "Int8",
        "e_comp": "Int8",
        "e_halluc": "Int8",
        "e_ctx": "Int8",
        "e_plan": "Int8",
        "e_tool": "Int8",
        "e_safety": "Int8",
        "e_refusal": "Int8",

        # Replicate index uses a nullable 16-bit integer
        "rep": "Int16",

        # Token counts, seeds, and Elo ratings use nullable 64-bit integers
        "prompt_tokens": "Int64",
        "completion_tokens": "Int64",
        "seed": "Int64",
        "arena_elo": "Int64",

        # Telemetry, probabilities, and parameter counts use standard 64-bit floats
        "latency_s": "float64",
        "temperature": "float64",
        "confidence": "float64",
        "cost_usd": "float64",
        "params_b": "float64",
        "mmlu": "float64",
        "gpqa": "float64",
        "humaneval": "float64",
        "swebench": "float64",
        "math": "float64",
        "pct_rank": "float64",

        # Boolean flags use standard bool
        "is_held_out": "bool",

        # Raw text, code, and heterogeneous ground truth remain as object
        "answer": "object",
        "error": "object",
        "thesis": "object",
        "reasoning": "object",
        "python_code": "object",
        "prompt": "object",
        "ground_truth": "object"
    }

    # Iterate over the canonical columns and apply the strict dtype coercion
    for col in expected_columns:
        # Retrieve the target dtype from the mapping, defaulting to 'object' if unspecified
        target_dtype = dtype_mapping.get(col, "object")
        try:
            # Apply the type conversion to the column
            df[col] = df[col].astype(target_dtype)
        except Exception as e:
            # Raise a descriptive error if coercion fails (e.g., string in a float column)
            raise ValueError(f"Failed to coerce column '{col}' in table '{table_name}' to {target_dtype}: {str(e)}")

    # Return the strictly typed, canonically ordered DataFrame
    return df

# ------------------------------------------------------------------------------
# Task 3, Step 2: Enforce Structural Invariants and Align Categories
# ------------------------------------------------------------------------------

def _align_categorical_columns(dfs: List[pd.DataFrame], column_name: str) -> None:
    """
    Aligns the categorical dictionary of a specific column across multiple DataFrames
    to prevent silent NaN generation during relational merges or .isin() checks.

    Parameters
    ----------
    dfs : List[pd.DataFrame]
        A list of DataFrames containing the target column.
    column_name : str
        The name of the categorical column to align (e.g., 'task_id', 'model_id').
    """
    # Extract the unique values from the target column across all provided DataFrames
    unique_categories = set()
    for df in dfs:
        if column_name in df.columns:
            # Drop NaNs and add unique values to the set
            unique_categories.update(df[column_name].dropna().unique())

    # Create a unified pandas CategoricalDtype from the sorted unique values
    unified_dtype = pd.CategoricalDtype(categories=sorted(list(unique_categories)), ordered=False)

    # Apply the unified categorical dtype back to the target column in all DataFrames
    for df in dfs:
        if column_name in df.columns:
            df[column_name] = df[column_name].astype(unified_dtype)

def enforce_structural_invariants(
    model_answer: pd.DataFrame,
    candidate_solution: pd.DataFrame,
    task_record: pd.DataFrame,
    qualification: pd.DataFrame
) -> Tuple[List[IngestionFinding], pd.DataFrame]:
    """
    Enforces relational integrity, uniqueness, and domain-specific hashing invariants.

    Parameters
    ----------
    model_answer : pd.DataFrame
        The strictly typed free-form generation DataFrame.
    candidate_solution : pd.DataFrame
        The strictly typed structured solution DataFrame.
    task_record : pd.DataFrame
        The strictly typed task and ground truth DataFrame.
    qualification : pd.DataFrame
        The strictly typed benchmark eligibility DataFrame.

    Returns
    -------
    Tuple[List[IngestionFinding], pd.DataFrame]
        - findings: A list of accumulated structural violations.
        - task_record: The potentially mutated task DataFrame (with is_confirmatory flag).
    """
    # Initialize the list to accumulate structural findings
    findings: List[IngestionFinding] = []

    # Align 'task_id' categories across all tables that contain it
    _align_categorical_columns([model_answer, candidate_solution, task_record], "task_id")

    # Align 'model_id' categories across all tables that contain it
    _align_categorical_columns([model_answer, candidate_solution, qualification], "model_id")

    # Helper function to record findings uniformly
    def _record_finding(table: str, rule: str, msg: str, mask: pd.Series, sev: IngestionSeverity) -> None:
        # If the boolean mask contains any True values (violations)
        if mask.any():
            # Extract the indices of the violating rows
            violating_indices = mask[mask].index.tolist()
            # Append a new IngestionFinding, capping the example keys at 5 to save memory
            findings.append(IngestionFinding(
                table_name=table, rule_violated=rule, message=msg, severity=sev,
                example_row_keys=[int(idx) for idx in violating_indices[:5]],
                total_violating_rows=len(violating_indices)
            ))

    # 1. Invariant: task_record.is_held_out must be universally True
    # Create a mask where is_held_out is False or NaN
    held_out_mask = ~(task_record["is_held_out"] == True)
    _record_finding(
        table="task_record", rule="is_held_out_true", sev=IngestionSeverity.HARD,
        msg="is_held_out must be True. Evaluation must be disjoint from qualification.",
        mask=held_out_mask
    )

    # 2. Invariant: Foreign-key integrity for task_id
    # Extract the set of valid task IDs from the primary task_record table
    valid_tasks = set(task_record["task_id"].dropna().unique())
    # Check model_answer for invalid task_ids
    ma_task_mask = ~model_answer["task_id"].isin(valid_tasks)
    _record_finding(
        table="model_answer", rule="fk_task_id", sev=IngestionSeverity.HARD,
        msg="task_id references a non-existent task in task_record.", mask=ma_task_mask
    )
    # Check candidate_solution for invalid task_ids
    cs_task_mask = ~candidate_solution["task_id"].isin(valid_tasks)
    _record_finding(
        table="candidate_solution", rule="fk_task_id", sev=IngestionSeverity.HARD,
        msg="task_id references a non-existent task in task_record.", mask=cs_task_mask
    )

    # 3. Invariant: Foreign-key integrity for model_id
    # Extract the set of valid model IDs from the primary qualification table
    valid_models = set(qualification["model_id"].dropna().unique())
    # Check model_answer for invalid model_ids
    ma_model_mask = ~model_answer["model_id"].isin(valid_models)
    _record_finding(
        table="model_answer", rule="fk_model_id", sev=IngestionSeverity.HARD,
        msg="model_id references an unqualified or non-existent model.", mask=ma_model_mask
    )
    # Check candidate_solution for invalid model_ids
    cs_model_mask = ~candidate_solution["model_id"].isin(valid_models)
    _record_finding(
        table="candidate_solution", rule="fk_model_id", sev=IngestionSeverity.HARD,
        msg="model_id references an unqualified or non-existent model.", mask=cs_model_mask
    )

    # 4. Invariant: qualification.model_id must be unique
    # Identify duplicated model_ids in the qualification table
    qual_dup_mask = qualification.duplicated(subset=["model_id"], keep=False)
    _record_finding(
        table="qualification", rule="unique_model_id", sev=IngestionSeverity.HARD,
        msg="model_id must be unique in the qualification table.", mask=qual_dup_mask
    )

    # 5. Invariant: Non-canonical domains must carry a content_hash
    # Initialize the is_confirmatory flag to True for all tasks
    task_record["is_confirmatory"] = True
    # Define the non-canonical domains that require strict hashing
    non_canonical_domains = ["finance", "long_context"]

    # Check if the content_hash column exists
    if "content_hash" not in task_record.columns:
        # If missing entirely, flag all non-canonical domains
        hash_mask = task_record["domain"].isin(non_canonical_domains)
    else:
        # If present, flag non-canonical domains where the hash is null or empty
        hash_mask = task_record["domain"].isin(non_canonical_domains) & task_record["content_hash"].isna()

    if hash_mask.any():
        # Record a SOFT finding, as this demotes the domains rather than aborting the run
        _record_finding(
            table="task_record", rule="noncanonical_corpus_fixity", sev=IngestionSeverity.SOFT,
            msg="Finance/long_context items lack a content_hash. Demoting to exploratory (excluded from H1-H5).",
            mask=hash_mask
        )
        # Mutate the DataFrame to explicitly demote these rows
        task_record.loc[hash_mask, "is_confirmatory"] = False

    # Return the accumulated findings and the potentially mutated task_record
    return findings, task_record

# ------------------------------------------------------------------------------
# Task 3, Step 3: Accumulate Findings, Generate Report, and Enforce Hard Constraints
# ------------------------------------------------------------------------------

def generate_ingestion_report_and_enforce(findings: List[IngestionFinding], run_dir: Path) -> Dict[str, Any]:
    """
    Serializes the ingestion findings to disk and aborts the run if any HARD
    structural violations are detected.

    Parameters
    ----------
    findings : List[IngestionFinding]
        The complete list of accumulated ingestion findings.
    run_dir : Path
        The path to the immutable run directory.

    Returns
    -------
    Dict[str, Any]
        The structured ingestion validation report dictionary.

    Raises
    ------
    RuntimeError
        If any finding has a severity of HARD.
    """
    # Separate findings by severity
    hard_findings: List[IngestionFinding] = [f for f in findings if f.severity == IngestionSeverity.HARD]
    soft_findings: List[IngestionFinding] = [f for f in findings if f.severity == IngestionSeverity.SOFT]

    # Construct the master report dictionary
    report: Dict[str, Any] = {
        "status": "FAILED" if hard_findings else "PASSED",
        "total_findings": len(findings),
        "hard_violations_count": len(hard_findings),
        "soft_violations_count": len(soft_findings),
        # Convert dataclasses to dictionaries, extracting the enum name for JSON serialization
        "details": [
            {**asdict(f), "severity": f.severity.name} for f in findings
        ]
    }

    # Define the path for the ingestion validation report JSON file
    report_path: Path = run_dir / "ingestion_validation_report.json"

    # Serialize and write the report to disk to preserve provenance
    with open(report_path, "w", encoding="utf-8") as f:
        json.dump(report, f, indent=2)

    # Enforce the fail-closed security principle
    if hard_findings:
        # Raise a descriptive error pointing the user to the persisted report
        raise RuntimeError(
            f"Data Ingestion Failed: {len(hard_findings)} HARD structural violations detected. "
            f"Run aborted to prevent corrupted data propagation. "
            f"See detailed report at: {report_path}"
        )

    # Return the report dictionary if ingestion passed
    return report

# ------------------------------------------------------------------------------
# Task 3, Orchestrator Function
# ------------------------------------------------------------------------------

def ingest_and_validate_inputs(
    model_answer_raw: List[Dict[str, Any]],
    candidate_solution_raw: List[Dict[str, Any]],
    task_record_raw: List[Dict[str, Any]],
    qualification_raw: pd.DataFrame,
    config: Mapping[str, Any],
    run_dir: Path
) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """
    Orchestrates Task 3: Materializes raw inputs into strictly typed DataFrames,
    enforces structural invariants, generates an audit report, and aborts on failure.

    Parameters
    ----------
    model_answer_raw : List[Dict[str, Any]]
        The raw free-form generation records.
    candidate_solution_raw : List[Dict[str, Any]]
        The raw structured solution records.
    task_record_raw : List[Dict[str, Any]]
        The raw task and ground truth records.
    qualification_raw : pd.DataFrame
        The raw benchmark eligibility DataFrame.
    config : Mapping[str, Any]
        The immutable study configuration.
    run_dir : Path
        The path to the immutable run directory.

    Returns
    -------
    Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame, pd.DataFrame]
        - model_answer: Strictly typed and validated DataFrame.
        - candidate_solution: Strictly typed and validated DataFrame.
        - task_record: Strictly typed and validated DataFrame (with is_confirmatory flag).
        - qualification: Strictly typed and validated DataFrame.

    Raises
    ------
    RuntimeError
        If any HARD structural violations are detected during ingestion.
    """
    # Extract the schema definitions from the configuration
    schemas: Mapping[str, Any] = config["raw_data_structures"]

    # Step 1: Materialize and coerce each table according to its strict schema
    model_answer: pd.DataFrame = materialize_and_coerce_table(
        raw_data=model_answer_raw, table_name="model_answer", schema_config=schemas["model_answer"]
    )

    candidate_solution: pd.DataFrame = materialize_and_coerce_table(
        raw_data=candidate_solution_raw, table_name="candidate_solution", schema_config=schemas["candidate_solution"]
    )

    task_record: pd.DataFrame = materialize_and_coerce_table(
        raw_data=task_record_raw, table_name="task_record", schema_config=schemas["task_record"]
    )

    qualification: pd.DataFrame = materialize_and_coerce_table(
        raw_data=qualification_raw, table_name="qualification", schema_config=schemas["qualification"]
    )

    # Step 2: Enforce structural invariants and align categorical columns
    findings, task_record = enforce_structural_invariants(
        model_answer=model_answer,
        candidate_solution=candidate_solution,
        task_record=task_record,
        qualification=qualification
    )

    # Step 3: Generate the ingestion report, persist it, and enforce the fail-closed policy
    _ = generate_ingestion_report_and_enforce(findings=findings, run_dir=run_dir)

    # Return the four validated, strictly typed DataFrames
    return model_answer, candidate_solution, task_record, qualification
````

````python
# Task 4 — Data Cleansing & Record Canonicalization (cleanse_raw)

# ==============================================================================
# Task 4: Data Cleansing & Record Canonicalization
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 4, Helper Structures: Disposition Enum
# ------------------------------------------------------------------------------

class DispositionStatus(Enum):
    """
    Enumeration defining the processing fate of a raw generation record.

    This enum provides a standardized classification for the processing outcome of
    each raw generation record during data cleansing (Task 4). It distinguishes
    between valid, scorable records (SCORED), records that failed due to infrastructure
    issues (FAILED_AFTER_RETRY), and records that failed structural hygiene checks
    (EXCLUDED_HYGIENE).
    """
    # Record is valid and proceeds to downstream scoring
    SCORED = "SCORED"
    # Record suffered an infrastructure error (timeout, malformed stream) after retries
    FAILED_AFTER_RETRY = "FAILED_AFTER_RETRY"
    # Record failed structural hygiene checks (e.g., malformed Python code)
    EXCLUDED_HYGIENE = "EXCLUDED_HYGIENE"

    @classmethod
    def validate(cls, value: str) -> "DispositionStatus":
        """
        Validates and instantiates a DispositionStatus member from a raw string value.

        Parameters
        ----------
        value : str
            The raw string representing the disposition status (e.g., 'SCORED', 'failed_after_retry').

        Returns
        -------
        DispositionStatus
            The matched DispositionStatus enum member.

        Raises
        ------
        TypeError
            If the input value is not a string.
        ValueError
            If the value does not match any valid DispositionStatus member.
        """
        # Validate that the incoming value is a standard string
        if not isinstance(value, str):
            # Raise TypeError if the type is violated
            raise TypeError(f"DispositionStatus validation requires a string input, got {type(value).__name__}")
        # Normalize the string: convert to uppercase and strip outer whitespace
        normalized_value = value.upper().strip()
        # Check if the normalized string matches a registered enum value
        if normalized_value not in cls._value2member_map_:
            # Raise an explicit ValueError if the string is unrecognized
            raise ValueError(f"Invalid disposition status '{value}'. Admissible: {list(cls._value2member_map_.keys())}")
        # Instantiate and return the validated DispositionStatus member
        return cls(normalized_value)

# ------------------------------------------------------------------------------
# Task 4, Step 1: Infrastructure-Error Handling and Disposition Logging
# ------------------------------------------------------------------------------

def handle_infrastructure_errors(
    df: pd.DataFrame,
    table_name: str,
    disposition_log: List[Dict[str, Any]]
) -> pd.DataFrame:
    """
    Identifies records with infrastructure errors, logs their disposition,
    and filters them out of the active DataFrame.

    Parameters
    ----------
    df : pd.DataFrame
        The strictly typed input DataFrame (model_answer or candidate_solution).
    table_name : str
        The logical name of the table, used for logging.
    disposition_log : List[Dict[str, Any]]
        A mutable list accumulating the fate of every processed row.

    Returns
    -------
    pd.DataFrame
        The filtered DataFrame containing only scorable rows.
    """
    # Create a working copy to avoid SettingWithCopyWarnings
    working_df = df.copy()

    # Clean the error column: replace empty strings or whitespace-only strings with pd.NA
    # This ensures that pd.notna() correctly identifies true errors
    if "error" in working_df.columns:
        # Use regex to replace strings that are entirely whitespace
        working_df["error"] = working_df["error"].replace(r'^\s*$', pd.NA, regex=True)

        # Create a boolean mask identifying rows with actual infrastructure errors
        error_mask = pd.notna(working_df["error"])

        # Extract the primary keys for the failed rows
        failed_rows = working_df[error_mask][["task_id", "model_id", "rep"]].copy()

        # Append the failed rows to the disposition log
        for _, row in failed_rows.iterrows():
            disposition_log.append({
                "table": table_name,
                "task_id": row["task_id"],
                "model_id": row["model_id"],
                "rep": row["rep"],
                "status": DispositionStatus.FAILED_AFTER_RETRY.name,
                "reason": "Infrastructure error (non-null error column)"
            })

        # Filter the working DataFrame to retain ONLY rows without infrastructure errors
        # Equation/Logic: Retain iff error is NA
        working_df = working_df[~error_mask].copy()

    return working_df

# ------------------------------------------------------------------------------
# Task 4, Step 2: Field Hygiene and Structural Validation
# ------------------------------------------------------------------------------

def _is_valid_python_code(code_string: str) -> bool:
    """
    Verify that a string contains structurally valid Python with at least one
    *non-placeholder* function or class definition.

    Rationale (§8.3; §9.1(b); Appendix C.3). Code-bearing solutions must parse AND
    contain a real definition with no placeholder tokens. The prior implementation
    accepted any parseable string containing a `FunctionDef`/`ClassDef`, so stub
    bodies (`pass`, `...`, a bare docstring, `raise NotImplementedError`) passed as
    valid solutions. This version additionally rejects definitions whose bodies are
    entirely placeholder, requires at least one definition to contain a substantive
    statement, and rejects literal TODO/FIXME markers. The function remains a pure,
    side-effect-free structural gate; true execution is deferred to the sandbox.

    Parameters
    ----------
    code_string : str
        The raw code string extracted from the model output (possibly fenced).

    Returns
    -------
    bool
        True iff the string parses and contains at least one definition with a
        substantive (non-placeholder) body and no placeholder markers; else False.
    """
    # Reject missing values and non-strings immediately (NaN-safe).
    if pd.isna(code_string) or not isinstance(code_string, str):
        # A non-string or NaN cannot be valid code.
        return False

    # Strip a leading markdown fence (```python / ```), which models emit despite JSON instructions.
    clean_code = re.sub(r"^```(?:python)?\s*", "", code_string, flags=re.MULTILINE)
    # Strip a trailing markdown fence.
    clean_code = re.sub(r"```\s*$", "", clean_code, flags=re.MULTILINE)

    # Reject obvious placeholder markers in the source text (case-insensitive TODO/FIXME/NotImplemented).
    if re.search(r"\b(?:TODO|FIXME|NotImplemented(?:Error)?)\b", clean_code, flags=re.IGNORECASE):
        # A placeholder marker disqualifies the solution per Appendix C.3.
        return False

    # Attempt to parse the cleaned source into an Abstract Syntax Tree.
    try:
        # Build the AST; a SyntaxError means the source is structurally invalid.
        tree: ast.Module = ast.parse(clean_code)
    except SyntaxError:
        # Unparseable source is not valid code.
        return False

    # Collect all top-level-or-nested function/class definition nodes.
    definitions: List[ast.AST] = [
        node for node in ast.walk(tree)
        if isinstance(node, (ast.FunctionDef, ast.AsyncFunctionDef, ast.ClassDef))
    ]

    # Require at least one definition; a bare script (e.g., "print('hi')") is not a solution.
    if not definitions:
        # No definition present -> reject.
        return False

    # Define a predicate that decides whether a single AST statement is a placeholder.
    def _is_placeholder_statement(stmt: ast.stmt) -> bool:
        # `pass` is a placeholder.
        if isinstance(stmt, ast.Pass):
            return True
        # A bare expression statement may be a docstring or an ellipsis (`...`) — both placeholders.
        if isinstance(stmt, ast.Expr) and isinstance(stmt.value, ast.Constant):
            # Treat string constants (docstrings) and the Ellipsis literal as placeholders.
            return isinstance(stmt.value.value, str) or stmt.value.value is Ellipsis
        # `raise NotImplementedError` (with or without call/args) is a placeholder.
        if isinstance(stmt, ast.Raise) and stmt.exc is not None:
            # Resolve the raised expression to either a Name or a Call on a Name.
            raised = stmt.exc.func if isinstance(stmt.exc, ast.Call) else stmt.exc
            # Flag when the raised symbol is NotImplementedError.
            if isinstance(raised, ast.Name) and raised.id == "NotImplementedError":
                return True
        # Any other statement is considered substantive.
        return False

    # Track whether at least one definition has a substantive (non-placeholder) body.
    has_substantive_definition: bool = False

    # Examine each definition's direct body for substantive content.
    for definition in definitions:
        # Class bodies often contain only method defs; treat a class as substantive if it has any
        # non-placeholder statement OR nests a function definition (methods are substantive structure).
        body: List[ast.stmt] = getattr(definition, "body", [])
        # Determine whether every statement in this body is a placeholder.
        body_all_placeholder: bool = all(_is_placeholder_statement(stmt) for stmt in body) if body else True
        # A nested function/class inside the body counts as substantive structure (e.g., class with methods).
        contains_nested_def: bool = any(
            isinstance(stmt, (ast.FunctionDef, ast.AsyncFunctionDef, ast.ClassDef)) for stmt in body
        )
        # This definition is substantive if its body is not all-placeholder or it nests a definition.
        if (not body_all_placeholder) or contains_nested_def:
            # Mark that we have found at least one real implementation.
            has_substantive_definition = True
            # No need to inspect further definitions.
            break

    # Valid if at least one definition carries substantive content.
    return has_substantive_definition

def enforce_field_hygiene(
    df: pd.DataFrame,
    table_name: str,
    disposition_log: List[Dict[str, Any]]
) -> pd.DataFrame:
    """
    Enforces structural hygiene: confidence bounds and AST code validation.

    Parameters
    ----------
    df : pd.DataFrame
        The filtered DataFrame from Step 1.
    table_name : str
        The logical name of the table.
    disposition_log : List[Dict[str, Any]]
        A mutable list accumulating the fate of every processed row.

    Returns
    -------
    pd.DataFrame
        The DataFrame with hygiene-violating rows removed.
    """
    working_df = df.copy()

    # 1. Validate confidence bounds [0, 1] if the column exists
    if "confidence" in working_df.columns:
        # Create a mask for rows where confidence is outside the mathematical bounds
        invalid_conf_mask = (working_df["confidence"] < 0.0) | (working_df["confidence"] > 1.0)

        if invalid_conf_mask.any():
            # Log the violators
            violators = working_df[invalid_conf_mask][["task_id", "model_id", "rep"]]
            for _, row in violators.iterrows():
                disposition_log.append({
                    "table": table_name,
                    "task_id": row["task_id"],
                    "model_id": row["model_id"],
                    "rep": row["rep"],
                    "status": DispositionStatus.EXCLUDED_HYGIENE.name,
                    "reason": "Confidence score outside [0, 1] bounds"
                })
            # Filter out the violators
            working_df = working_df[~invalid_conf_mask].copy()

    # 2. Validate Python code structure via AST if the column exists
    if "python_code" in working_df.columns:
        # Apply the AST checker only to rows where python_code is not null
        # This assumes that tasks requiring code have non-null entries
        code_mask = pd.notna(working_df["python_code"])

        if code_mask.any():
            # Apply the AST validation function
            valid_code_series = working_df.loc[code_mask, "python_code"].apply(_is_valid_python_code)

            # Identify rows that failed the AST check
            invalid_code_mask = ~valid_code_series

            if invalid_code_mask.any():
                # Extract the indices of the failed rows
                failed_indices = valid_code_series[invalid_code_mask].index
                violators = working_df.loc[failed_indices, ["task_id", "model_id", "rep"]]

                # Log the violators
                for _, row in violators.iterrows():
                    disposition_log.append({
                        "table": table_name,
                        "task_id": row["task_id"],
                        "model_id": row["model_id"],
                        "rep": row["rep"],
                        "status": DispositionStatus.EXCLUDED_HYGIENE.name,
                        "reason": "python_code failed AST parsing or lacks definitions"
                    })
                # Filter out the violators
                working_df = working_df.drop(index=failed_indices).copy()

    # Note: The 'answer' column in model_answer is explicitly left unnormalized
    # per the protocol instructions. No .strip() or .lower() is applied.

    return working_df

# ------------------------------------------------------------------------------
# Task 4, Step 3: De-duplication, Stable Sorting, and Immutable Freezing
# ------------------------------------------------------------------------------

def canonicalize_and_hash(df: pd.DataFrame, primary_keys: List[str]) -> Tuple[pd.DataFrame, str]:
    """
    De-duplicates, stable-sorts, and computes a canonical SHA-256 hash of the DataFrame.

    Parameters
    ----------
    df : pd.DataFrame
        The cleansed DataFrame.
    primary_keys : List[str]
        The columns defining the unique natural key for sorting and deduplication.

    Returns
    -------
    Tuple[pd.DataFrame, str]
        The canonicalized DataFrame and its SHA-256 content hash.
    """
    # 1. De-duplicate based on the primary keys, keeping the first occurrence
    canonical_df = df.drop_duplicates(subset=primary_keys, keep='first').copy()

    # 2. Stable sort using mergesort to guarantee deterministic ordering across platforms
    canonical_df = canonical_df.sort_values(by=primary_keys, kind='mergesort')

    # 3. Reset the index to ensure the pandas index does not introduce hash variations
    canonical_df = canonical_df.reset_index(drop=True)

    # 4. Serialize to canonical JSON for hashing
    # orient='records' creates a list of dicts
    # date_format='iso' ensures deterministic timestamps (if any exist)
    # double_precision=15 enforces a fixed float precision to prevent cross-platform drift
    json_str = canonical_df.to_json(
        orient='records',
        date_format='iso',
        double_precision=15
    )

    # Parse the JSON string back to Python objects to use the standard json module
    # for canonical key sorting (pandas to_json does not guarantee key order within records)
    parsed_records = json.loads(json_str)
    canonical_json_str = json.dumps(parsed_records, sort_keys=True, separators=(',', ':'))

    # 5. Compute the SHA-256 hash of the canonical byte string
    content_hash = hashlib.sha256(canonical_json_str.encode('utf-8')).hexdigest()

    return canonical_df, content_hash

# ------------------------------------------------------------------------------
# Task 4, Orchestrator Function
# ------------------------------------------------------------------------------

def cleanse_raw(
    model_answer: pd.DataFrame,
    candidate_solution: pd.DataFrame,
    task_record: pd.DataFrame,
    qualification: pd.DataFrame
) -> Dict[str, Any]:
    """
    Orchestrates Task 4: Handles infrastructure errors, enforces field hygiene,
    canonicalizes records, and computes cryptographic hashes for all inputs.

    Parameters
    ----------
    model_answer : pd.DataFrame
        The validated free-form generation DataFrame.
    candidate_solution : pd.DataFrame
        The validated structured solution DataFrame.
    task_record : pd.DataFrame
        The validated task DataFrame.
    qualification : pd.DataFrame
        The validated qualification DataFrame.

    Returns
    -------
    Dict[str, Any]
        A dictionary containing the cleansed DataFrames, their hashes, and the
        unified disposition log.
    """
    # Initialize the master disposition log
    disposition_log: List[Dict[str, Any]] = []

    # Define primary keys for canonicalization
    gen_keys = ["task_id", "model_id", "rep"]
    task_keys = ["task_id"]
    qual_keys = ["model_id"]

    # --- Process model_answer ---
    ma_clean = handle_infrastructure_errors(model_answer, "model_answer", disposition_log)
    ma_clean = enforce_field_hygiene(ma_clean, "model_answer", disposition_log)
    ma_canon, ma_hash = canonicalize_and_hash(ma_clean, gen_keys)

    # --- Process candidate_solution ---
    cs_clean = handle_infrastructure_errors(candidate_solution, "candidate_solution", disposition_log)
    cs_clean = enforce_field_hygiene(cs_clean, "candidate_solution", disposition_log)
    cs_canon, cs_hash = canonicalize_and_hash(cs_clean, gen_keys)

    # --- Process task_record ---
    # Tasks do not have infrastructure errors or hygiene checks in this context
    tr_canon, tr_hash = canonicalize_and_hash(task_record, task_keys)

    # --- Process qualification ---
    # Qualification does not have infrastructure errors or hygiene checks in this context
    qual_canon, qual_hash = canonicalize_and_hash(qualification, qual_keys)

    # Materialize the disposition log into a DataFrame for downstream auditing
    disposition_df = pd.DataFrame(disposition_log)


    # Bundle and return the results
    return {
        "model_answer": {"data": ma_canon, "hash": ma_hash},
        "candidate_solution": {"data": cs_canon, "hash": cs_hash},
        "task_record": {"data": tr_canon, "hash": tr_hash},
        "qualification": {"data": qual_canon, "hash": qual_hash},
        "disposition_log": disposition_df
    }
````

````python
# Task 5 — Benchmark Contamination Audit (contamination_audit)

# ==============================================================================
# Task 5: Benchmark Contamination Audit
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 5, Helper Classes: TrainingCorpusIndex and BehavioralProbeExecutor
# ------------------------------------------------------------------------------

class TrainingCorpusIndex:
    """
    A high-performance, memory-efficient index for computing exact n-gram overlap
    against a training corpus proxy.

    This class tokenizes raw proxy texts, extracts n-grams, computes truncated
    SHA-256 hashes for each n-gram, and stores them in a Python set for O(1)
    membership testing. This acts as a cryptographic signature of memorization.
    """

    def __init__(self, corpus_texts: List[str], ngram_size: int = 13) -> None:
        """
        Initializes the index and builds the hashed n-gram set from the corpus.

        Parameters
        ----------
        corpus_texts : List[str]
            A list of raw text strings representing the training corpus proxy.
        ngram_size : int, optional
            The size of the n-grams to extract, by default 13.
        """
        # Store the configured n-gram size
        self.ngram_size: int = ngram_size
        # Initialize the set to hold the cryptographic hashes of the corpus n-grams
        self.corpus_ngram_hashes: Set[str] = set()

        # Compile a regular expression to strip all non-alphanumeric characters
        # This ensures tokenization is robust against casing and punctuation differences
        self.clean_pattern = re.compile(r'[^a-z0-9\s]')

        # Build the index upon instantiation
        self._build_index(corpus_texts)

    def _normalize_and_tokenize(self, text: str) -> List[str]:
        """
        Normalizes text to lowercase, strips punctuation, and splits into tokens.

        Parameters
        ----------
        text : str
            The raw text to tokenize.

        Returns
        -------
        List[str]
            A list of normalized alphanumeric tokens.
        """
        # Convert text to lowercase
        text = text.lower()
        # Remove all non-alphanumeric characters (excluding whitespace)
        text = self.clean_pattern.sub('', text)
        # Split the text by whitespace into a list of tokens
        return text.split()

    def _hash_ngram(self, tokens: List[str]) -> str:
        """
        Computes a truncated SHA-256 hash for a sequence of tokens.

        Parameters
        ----------
        tokens : List[str]
            A sequence of tokens representing a single n-gram.

        Returns
        -------
        str
            A 16-character hexadecimal string representing the hash.
        """
        # Join the tokens with a single space to form the n-gram string
        ngram_str = " ".join(tokens)
        # Compute the SHA-256 hash and truncate to 16 characters to save memory
        # 16 hex chars (64 bits) provides ample collision resistance for this domain
        return hashlib.sha256(ngram_str.encode('utf-8')).hexdigest()[:16]

    def _build_index(self, corpus_texts: List[str]) -> None:
        """
        Processes the raw corpus texts and populates the internal hash set.

        Parameters
        ----------
        corpus_texts : List[str]
            The list of raw proxy texts.
        """
        # Iterate over each document in the proxy corpus
        for text in corpus_texts:
            # Tokenize the document
            tokens = self._normalize_and_tokenize(text)
            # Extract sliding-window n-grams
            for i in range(len(tokens) - self.ngram_size + 1):
                # Slice the token list to get the current n-gram
                ngram_tokens = tokens[i : i + self.ngram_size]
                # Hash the n-gram and add it to the O(1) lookup set
                self.corpus_ngram_hashes.add(self._hash_ngram(ngram_tokens))

    def compute_lexical_overlap(self, text: str) -> float:
        """
        Computes the fraction of n-grams in the provided text that exist in the corpus.

        Equation: Overlap(t) = |G_13(t) \cap G_13(C)| / |G_13(t)|

        Parameters
        ----------
        text : str
            The task text (prompt + ground truth) to evaluate.

        Returns
        -------
        float
            The overlap fraction in the range [0.0, 1.0]. Returns 0.0 if the text
            is shorter than the n-gram size.
        """
        # Tokenize the evaluation text
        tokens = self._normalize_and_tokenize(text)

        # If the text has fewer tokens than the n-gram size, overlap is mathematically 0
        if len(tokens) < self.ngram_size:
            return 0.0

        # Initialize counters for total n-grams and matched n-grams
        total_ngrams = 0
        matched_ngrams = 0

        # Iterate over the sliding window of the evaluation text
        for i in range(len(tokens) - self.ngram_size + 1):
            total_ngrams += 1
            # Extract the current n-gram
            ngram_tokens = tokens[i : i + self.ngram_size]
            # Hash the n-gram
            ngram_hash = self._hash_ngram(ngram_tokens)
            # Check for membership in the O(1) corpus set
            if ngram_hash in self.corpus_ngram_hashes:
                matched_ngrams += 1

        # Return the ratio of matched n-grams to total n-grams
        return float(matched_ngrams) / float(total_ngrams)


class BehavioralProbeExecutor:
    """
    A robust, thread-safe HTTP client for executing membership-inference and
    canary-string recall probes against an OpenAI-compatible LLM gateway.
    """

    def __init__(self, base_url: str, api_key: str, max_retries: int = 3) -> None:
        """
        Initializes the executor with gateway credentials and retry policies.

        Parameters
        ----------
        base_url : str
            The base URL of the OpenAI-compatible API (e.g., OpenRouter).
        api_key : str
            The resolved API key for authentication.
        max_retries : int, optional
            The maximum number of HTTP retries for transient errors, by default 3.
        """
        # Store the API endpoint URL (appending the standard chat completions path)
        self.endpoint: str = f"{base_url.rstrip('/')}/chat/completions"
        # Store the API key securely in memory
        self.api_key: str = api_key
        # Store the maximum number of retries
        self.max_retries: int = max_retries

    def _call_api(self, model_id: str, prompt: str) -> Optional[str]:
        """
        Executes a synchronous HTTP POST request to the LLM gateway with exponential backoff.

        Parameters
        ----------
        model_id : str
            The exact gateway identifier of the model.
        prompt : str
            The text prompt to send to the model.

        Returns
        -------
        Optional[str]
            The generated text content, or None if the request failed after all retries.
        """
        # Construct the standard OpenAI-compatible JSON payload
        # Temperature is strictly 0.0 to force deterministic, maximum-likelihood decoding
        # Max tokens is limited to 50 as we only need to check the immediate continuation
        payload = {
            "model": model_id,
            "messages": [{"role": "user", "content": prompt}],
            "temperature": 0.0,
            "max_tokens": 50
        }

        # Encode the payload to bytes
        data = json.dumps(payload).encode('utf-8')

        # Construct the HTTP request with required headers
        req = urllib.request.Request(
            self.endpoint,
            data=data,
            headers={
                "Authorization": f"Bearer {self.api_key}",
                "Content-Type": "application/json"
            }
        )

        # Implement exponential backoff and jitter for robust network I/O
        for attempt in range(self.max_retries):
            try:
                # Execute the HTTP request with a strict 15-second timeout
                with urllib.request.urlopen(req, timeout=15.0) as response:
                    # Read and decode the response body
                    response_body = response.read().decode('utf-8')
                    # Parse the JSON response
                    response_json = json.loads(response_body)
                    # Extract the generated text content from the standard response structure
                    return response_json["choices"][0]["message"]["content"]
            except (urllib.error.URLError, KeyError, json.JSONDecodeError) as e:
                # If this is the last attempt, break the loop and return None
                if attempt == self.max_retries - 1:
                    break
                # Calculate exponential backoff sleep time: 2^attempt seconds
                sleep_time = 2 ** attempt
                time.sleep(sleep_time)

        # Return None if all retries are exhausted
        return None

    def execute_canary_probe(self, task_prompt: str, model_id: str) -> bool:
        """
        Executes a membership-inference probe by prompting the model with the
        first half of the task and checking if it exactly recalls the second half.

        Parameters
        ----------
        task_prompt : str
            The full task prompt.
        model_id : str
            The exact gateway identifier of the model.

        Returns
        -------
        bool
            True if the model exhibits canary recall (contamination), False otherwise.
            Fails safe (returns True) if the API call fails completely.
        """
        # Split the prompt into words
        words = task_prompt.split()

        # If the prompt is too short to split meaningfully, assume no behavioral contamination
        if len(words) < 20:
            return False

        # Calculate the midpoint index
        midpoint = len(words) // 2

        # Construct the prefix (the prompt sent to the model)
        prefix = " ".join(words[:midpoint])
        # Construct the suffix (the hidden target we check for memorization)
        suffix = " ".join(words[midpoint:midpoint + 20]) # Check the next 20 words

        # Call the API with the prefix
        generated_text = self._call_api(model_id, prefix)

        # Fail-safe mechanism: if the API fails, assume contamination to protect the substrate
        if generated_text is None:
            return True

        # Normalize both the generated text and the target suffix for robust comparison
        norm_generated = re.sub(r'[^a-z0-9]', '', generated_text.lower())
        norm_suffix = re.sub(r'[^a-z0-9]', '', suffix.lower())

        # If the normalized suffix is entirely contained within the generated text,
        # the model has exactly memorized the continuation, indicating contamination.
        if norm_suffix in norm_generated and len(norm_suffix) > 10:
            return True

        return False

# ------------------------------------------------------------------------------
# Task 5, Step 1: Compute 13-gram overlap and run behavioral probes
# ------------------------------------------------------------------------------

def execute_contamination_probes(
    task_record: pd.DataFrame,
    qualification: pd.DataFrame,
    corpus_index: TrainingCorpusIndex,
    probe_executor: BehavioralProbeExecutor
) -> pd.DataFrame:
    """
    Executes lexical overlap and behavioral probes for all (task, model) pairs.
    Utilizes a ThreadPoolExecutor to parallelize the I/O-bound behavioral probes.

    Parameters
    ----------
    task_record : pd.DataFrame
        The cleansed task DataFrame.
    qualification : pd.DataFrame
        The qualification DataFrame containing the pool of models.
    corpus_index : TrainingCorpusIndex
        The instantiated n-gram index for lexical overlap.
    probe_executor : BehavioralProbeExecutor
        The instantiated HTTP client for behavioral probes.

    Returns
    -------
    pd.DataFrame
        A DataFrame containing the probe results with columns:
        ['task_id', 'model_id', 'overlap_rate', 'behavioral_flag']
    """
    # Extract the list of valid model IDs from the qualification table
    model_ids: List[str] = qualification["model_id"].tolist()

    # Initialize a list to accumulate the final results
    results: List[Dict[str, Any]] = []

    # Define a helper function to execute a single probe task
    # This function will be mapped across the thread pool
    def _probe_task(t_id: str, prompt: str, m_id: str, overlap: float) -> Dict[str, Any]:
        # Execute the network-bound behavioral probe
        flag = probe_executor.execute_canary_probe(prompt, m_id)
        # Return the structured result dictionary
        return {
            "task_id": t_id,
            "model_id": m_id,
            "overlap_rate": overlap,
            "behavioral_flag": flag
        }

    # Initialize a ThreadPoolExecutor to parallelize the network requests
    # Max workers is set to 20 to balance throughput without overwhelming the gateway
    with concurrent.futures.ThreadPoolExecutor(max_workers=20) as executor:
        # Initialize a list to hold the Future objects
        futures: List[concurrent.futures.Future] = []

        # Iterate over each task in the evaluation set
        for _, task_row in task_record.iterrows():
            task_id = task_row["task_id"]
            prompt = task_row["prompt"]

            # Combine prompt and ground truth for the lexical check
            full_text = f"{prompt} {str(task_row['ground_truth'])}"

            # Compute the lexical overlap against the corpus index (CPU-bound, done synchronously)
            overlap_rate = corpus_index.compute_lexical_overlap(full_text)

            # Iterate over each model in the qualified pool
            for model_id in model_ids:
                # Submit the I/O-bound behavioral probe to the thread pool
                future = executor.submit(_probe_task, task_id, prompt, model_id, overlap_rate)
                futures.append(future)

        # As each future completes, extract the result and append it to the list
        for future in concurrent.futures.as_completed(futures):
            results.append(future.result())

    # Convert the accumulated results into a strictly typed pandas DataFrame
    results_df = pd.DataFrame(results)

    # Enforce strict categorical and numeric dtypes
    results_df["task_id"] = results_df["task_id"].astype("category")
    results_df["model_id"] = results_df["model_id"].astype("category")
    results_df["overlap_rate"] = results_df["overlap_rate"].astype("float64")
    results_df["behavioral_flag"] = results_df["behavioral_flag"].astype("bool")

    return results_df

# ------------------------------------------------------------------------------
# Task 5, Step 2: Build per-(benchmark, model) report and apply exclusion rule
# ------------------------------------------------------------------------------

def apply_exclusion_rule(
    probe_results: pd.DataFrame,
    task_record: pd.DataFrame,
    overlap_threshold: float
) -> Tuple[Set[str], pd.DataFrame]:
    """
    Applies the strict exclusion rule and generates a summary report.
    Rule: Drop an item for ALL models if ANY model is flagged or overlap > threshold.

    Parameters
    ----------
    probe_results : pd.DataFrame
        The DataFrame containing the probe results from Step 1.
    task_record : pd.DataFrame
        The task DataFrame, used to join domain information.
    overlap_threshold : float
        The threshold for lexical overlap exclusion (e.g., 0.10).

    Returns
    -------
    Tuple[Set[str], pd.DataFrame]
        - exclusions: A set of task_ids that must be dropped globally.
        - summary_report: A DataFrame summarizing contamination per (domain, model).
    """
    # 1. Evaluate the exclusion condition for every (task, model) pair
    # Condition: overlap rate exceeds threshold OR behavioral probe flagged True
    violation_mask = (probe_results["overlap_rate"] > overlap_threshold) | (probe_results["behavioral_flag"])

    # 2. Extract the unique task_ids where ANY violation occurred
    # This mathematically enforces the "drop for all if any flagged" rule,
    # preserving the shared evaluation substrate across all models.
    violating_tasks = probe_results.loc[violation_mask, "task_id"].unique()

    # Cast the array of violating task IDs to a Python set for O(1) lookups
    exclusions: Set[str] = set(violating_tasks)

    # 3. Generate the summary report
    # Perform a relational left merge to broadcast the 'domain' column into the probe results
    merged_df = pd.merge(
        probe_results,
        task_record[["task_id", "domain"]],
        on="task_id",
        how="left"
    )

    # Group by domain (benchmark) and model_id to compute summary statistics
    # observed=True ensures that unobserved categorical combinations are dropped
    summary_report = merged_df.groupby(["domain", "model_id"], observed=True).agg(
        mean_overlap_rate=("overlap_rate", "mean"),
        max_overlap_rate=("overlap_rate", "max"),
        total_behavioral_flags=("behavioral_flag", "sum"),
        total_tasks_evaluated=("task_id", "count")
    )

    # Reset the index to convert the grouped multi-index back into standard columns
    summary_report = summary_report.reset_index()

    return exclusions, summary_report

# ------------------------------------------------------------------------------
# Task 5, Step 3: Persist report and output pruned task set
# ------------------------------------------------------------------------------

def persist_report_and_prune_tasks(
    task_record: pd.DataFrame,
    exclusions: Set[str],
    summary_report: pd.DataFrame,
    run_dir: Path
) -> pd.DataFrame:
    """
    Filters the task record to remove contaminated items and persists the audit report.

    Parameters
    ----------
    task_record : pd.DataFrame
        The original cleansed task DataFrame.
    exclusions : Set[str]
        The set of task_ids to exclude globally.
    summary_report : pd.DataFrame
        The summary statistics DataFrame.
    run_dir : Path
        The path to the immutable run directory.

    Returns
    -------
    pd.DataFrame
        The pruned task DataFrame, representing the final, pristine evaluation universe.
    """
    # 1. Filter the task_record to exclude contaminated items
    # The ~ operator negates the boolean mask; isin() checks membership in the exclusion set
    pruned_task_record = task_record[~task_record["task_id"].isin(exclusions)].copy()

    # Reset the index to maintain a clean, contiguous RangeIndex, dropping the old index
    pruned_task_record = pruned_task_record.reset_index(drop=True)

    # 2. Construct the master report dictionary
    report_dict: Dict[str, Any] = {
        "total_tasks_before_pruning": len(task_record),
        "total_tasks_excluded": len(exclusions),
        "total_tasks_after_pruning": len(pruned_task_record),
        # Convert the set to a sorted list for deterministic JSON serialization
        "excluded_task_ids": sorted(list(exclusions)),
        # Convert the pandas DataFrame to a list of standard Python dictionaries
        "summary_statistics": summary_report.to_dict(orient="records")
    }

    # 3. Define the path for the contamination report JSON file
    report_path: Path = run_dir / "contamination_report.json"

    # Serialize and write the report to disk to preserve cryptographic provenance
    with open(report_path, "w", encoding="utf-8") as f:
        json.dump(report_dict, f, indent=2)

    # Return the mathematically pruned task set
    return pruned_task_record

# ------------------------------------------------------------------------------
# Task 5, Orchestrator Function
# ------------------------------------------------------------------------------

def contamination_audit(
    task_record: pd.DataFrame,
    qualification: pd.DataFrame,
    config: Mapping[str, Any],
    run_dir: Path,
    raw_proxy_texts: List[str]
) -> Tuple[pd.DataFrame, Dict[str, Any]]:
    """
    Orchestrates Task 5: Instantiates the corpus index and probe executor,
    runs lexical and behavioral contamination probes, applies the strict
    exclusion rule, persists the audit report, and returns the pruned tasks.

    Parameters
    ----------
    task_record : pd.DataFrame
        The cleansed task DataFrame.
    qualification : pd.DataFrame
        The qualification DataFrame containing the pool of models.
    config : Mapping[str, Any]
        The immutable study configuration.
    run_dir : Path
        The path to the immutable run directory.
    raw_proxy_texts : List[str]
        A list of raw text strings representing the training corpus proxy.

    Returns
    -------
    Tuple[pd.DataFrame, Dict[str, Any]]
        - pruned_task_record: The task DataFrame with contaminated items removed.
        - report_dict: The structured contamination audit report.

    Raises
    ------
    RuntimeError
        If the API key environment variable is missing or empty.
    """
    # Extract parameters from the configuration mapping
    audit_config = config["contamination_audit"]
    ngram_size: int = audit_config["ngram_overlap_n"]
    overlap_threshold: float = audit_config["overlap_exclusion_rate"]

    # Extract gateway configuration
    gateway_config = config["gateway"]
    base_url: str = gateway_config["base_url"]
    api_key_env_var: str = gateway_config["api_key_env_var"]

    # Resolve the API key securely at runtime
    api_key: Optional[str] = os.environ.get(api_key_env_var)
    if not api_key:
        raise RuntimeError(f"Contamination Audit Failed: Environment variable '{api_key_env_var}' is missing.")

    # Instantiate the high-performance TrainingCorpusIndex
    corpus_index = TrainingCorpusIndex(corpus_texts=raw_proxy_texts, ngram_size=ngram_size)

    # Instantiate the thread-safe BehavioralProbeExecutor
    probe_executor = BehavioralProbeExecutor(base_url=base_url, api_key=api_key)

    # Step 1: Execute the probes to gather raw contamination statistics
    probe_results: pd.DataFrame = execute_contamination_probes(
        task_record=task_record,
        qualification=qualification,
        corpus_index=corpus_index,
        probe_executor=probe_executor
    )

    # Step 2: Apply the exclusion rule and generate the summary report
    exclusions, summary_report = apply_exclusion_rule(
        probe_results=probe_results,
        task_record=task_record,
        overlap_threshold=overlap_threshold
    )

    # Step 3: Prune the task set and persist the report to disk
    pruned_task_record: pd.DataFrame = persist_report_and_prune_tasks(
        task_record=task_record,
        exclusions=exclusions,
        summary_report=summary_report,
        run_dir=run_dir
    )

    # Reconstruct the report dictionary to return it alongside the pruned tasks
    report_dict: Dict[str, Any] = {
        "total_tasks_before_pruning": len(task_record),
        "total_tasks_excluded": len(exclusions),
        "total_tasks_after_pruning": len(pruned_task_record),
        "excluded_task_ids": sorted(list(exclusions)),
        "summary_statistics": summary_report.to_dict(orient="records")
    }

    return pruned_task_record, report_dict
````

````python
# Task 6 — Candidate Qualification Gate (Layer-1) (qualify)

# ==============================================================================
# Task 6: Candidate Qualification Gate (Layer-1)
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 6, Step 1: Compute the composite benchmark percentile and apply the gate
# ------------------------------------------------------------------------------

def compute_composite_and_filter(
    qualification: pd.DataFrame,
    p_min: float
) -> Tuple[List[str], pd.DataFrame]:
    """
    Computes the composite benchmark percentile rank for each model and filters
    the pool based on the strict P_min threshold.

    Equation: qualify(M_i) <=> pct_rank(M_i) >= P_min

    Parameters
    ----------
    qualification : pd.DataFrame
        The strictly typed benchmark eligibility DataFrame.
    p_min : float
        The minimum composite percentile required to qualify (e.g., 0.80).

    Returns
    -------
    Tuple[List[str], pd.DataFrame]
        - qualified_pool: An ordered list of qualified model_ids (descending rank).
        - mutated_qualification: The DataFrame updated with the pct_rank column.
    """
    # Create a working copy to avoid SettingWithCopyWarnings
    df = qualification.copy()

    # Define the standard [0,1] benchmark columns
    benchmark_cols = ["mmlu", "gpqa", "humaneval", "swebench", "math"]

    # 1. Convert each benchmark column to a within-pool percentile rank.
    # na_option='bottom' strictly penalizes models missing benchmark data,
    # ensuring they do not artificially inflate their composite score.
    rank_df = pd.DataFrame(index=df.index)
    for col in benchmark_cols:
        if col in df.columns:
            rank_df[f"{col}_rank"] = df[col].rank(pct=True, na_option='bottom')

    # 2. Safely min-max normalize the arena_elo column, then rank it.
    if "arena_elo" in df.columns:
        elo_series = df["arena_elo"].astype(float)
        elo_min = elo_series.min()
        elo_max = elo_series.max()

        # Prevent division by zero if all models have the exact same Elo
        if np.isclose(elo_min, elo_max):
            normalized_elo = pd.Series(0.5, index=df.index)
        else:
            normalized_elo = (elo_series - elo_min) / (elo_max - elo_min)

        # Rank the normalized Elo
        rank_df["elo_rank"] = normalized_elo.rank(pct=True, na_option='bottom')

    # 3. Compute the composite pct_rank as the row-wise mean of all rank columns.
    # This prevents any single high-variance metric from dominating the gate.
    df["pct_rank"] = rank_df.mean(axis=1)

    # 4. Apply the strict boolean filter: pct_rank >= P_min
    qualified_mask = df["pct_rank"] >= p_min
    qualified_df = df[qualified_mask].copy()

    # 5. Sort the qualified DataFrame by pct_rank descending to establish a
    # deterministic, performance-based canonical ordering for downstream matrices.
    qualified_df = qualified_df.sort_values(by="pct_rank", ascending=False)

    # 6. Extract the model_ids as a list of standard Python strings
    qualified_pool: List[str] = [str(m_id) for m_id in qualified_df["model_id"].tolist()]

    return qualified_pool, df

# ------------------------------------------------------------------------------
# Task 6, Step 2: Persist the qualified pool and document the composition
# ------------------------------------------------------------------------------

def persist_qualified_pool(
    qualified_pool: List[str],
    p_min: float,
    run_dir: Path
) -> None:
    """
    Serializes the qualified pool and the exact composition logic to disk,
    ensuring Tier-1 provenance of the estimation universe.

    Parameters
    ----------
    qualified_pool : List[str]
        The ordered list of qualified model_ids.
    p_min : float
        The threshold used for qualification.
    run_dir : Path
        The path to the immutable run directory.
    """
    # Construct the provenance dictionary
    provenance_dict: Dict[str, Any] = {
        "p_min_threshold": p_min,
        "total_qualified_models": len(qualified_pool),
        "qualified_model_ids": qualified_pool,
        "composition_logic": (
            "1. Convert mmlu, gpqa, humaneval, swebench, math to within-pool percentile ranks (na_option='bottom'). "
            "2. Min-max normalize arena_elo to [0,1], then convert to percentile rank. "
            "3. Compute pct_rank as the unweighted row-wise mean of these six rank columns. "
            "4. Filter models where pct_rank >= p_min."
        )
    }

    # Define the path for the JSON file
    output_path: Path = run_dir / "qualified_pool.json"

    # Serialize and write to disk
    with open(output_path, "w", encoding="utf-8") as f:
        json.dump(provenance_dict, f, indent=2)

# ------------------------------------------------------------------------------
# Task 6, Orchestrator Function
# ------------------------------------------------------------------------------

def qualify(
    qualification: pd.DataFrame,
    config: Mapping[str, Any],
    run_dir: Path
) -> Tuple[List[str], pd.DataFrame]:
    """
    Orchestrates Task 6: Computes the composite benchmark percentile, applies
    the P_min qualification gate, and persists the resulting pool to disk.

    Parameters
    ----------
    qualification : pd.DataFrame
        The strictly typed benchmark eligibility DataFrame.
    config : Mapping[str, Any]
        The immutable study configuration.
    run_dir : Path
        The path to the immutable run directory.

    Returns
    -------
    Tuple[List[str], pd.DataFrame]
        - qualified_pool: An ordered list of qualified model_ids.
        - mutated_qualification: The DataFrame updated with the pct_rank column.
    """
    # Extract the qualification threshold from the configuration
    p_min: float = config["preprocessing"]["qualification_threshold_pct"]

    # Step 1: Compute the composite rank and filter the pool
    qualified_pool, mutated_qualification = compute_composite_and_filter(
        qualification=qualification,
        p_min=p_min
    )

    # Step 2: Persist the pool and its provenance logic to disk
    persist_qualified_pool(
        qualified_pool=qualified_pool,
        p_min=p_min,
        run_dir=run_dir
    )

    # Return the canonical list and the updated DataFrame
    return qualified_pool, mutated_qualification
````

````python
# Task 7 — Candidate Stratification (stratify)

# ==============================================================================
# Task 7: Candidate Stratification
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 7, Step 1: Bin parameter scale and handle closed-weight models
# ------------------------------------------------------------------------------

def bin_parameter_scale(
    qualification: pd.DataFrame,
    qualified_pool: List[str]
) -> pd.DataFrame:
    """
    Filters the qualification DataFrame to the qualified pool and bins the
    parameter counts, explicitly treating closed-weight models as a distinct category.

    Parameters
    ----------
    qualification : pd.DataFrame
        The mutated qualification DataFrame from Task 6.
    qualified_pool : List[str]
        The ordered list of qualified model_ids.

    Returns
    -------
    pd.DataFrame
        A DataFrame containing only qualified models, augmented with the
        'param_scale_bin' categorical column.
    """
    # 1. Filter the DataFrame to retain only the qualified models
    # We use .isin() and then set the index to model_id temporarily to align
    # with the canonical ordering of the qualified_pool list.
    df = qualification[qualification["model_id"].isin(qualified_pool)].copy()

    # Ensure the DataFrame row order exactly matches the qualified_pool list
    df["model_id"] = pd.Categorical(df["model_id"], categories=qualified_pool, ordered=True)
    df = df.sort_values("model_id").reset_index(drop=True)

    # 2. Cast params_b to float64 to ensure pd.cut operates on a numeric array
    # This handles cases where mixed types might have coerced the column to object
    params_series = df["params_b"].astype("float64")

    # 3. Define the numerical bin boundaries and corresponding labels
    # Boundaries: (0, 20], (20, 70], (70, infinity)
    bins = [0.0, 20.0, 70.0, np.inf]
    labels = ["<20B", "20B-70B", ">70B"]

    # 4. Apply pd.cut to create the categorical bin column
    # right=True ensures intervals are closed on the right (e.g., exactly 70B is "20B-70B")
    binned_series = pd.cut(params_series, bins=bins, labels=labels, right=True)

    # 5. Add the explicit "closed_weight" category to the CategoricalDtype
    # This is mathematically required because missingness correlates with provider lineage
    binned_series = binned_series.cat.add_categories(["closed_weight"])

    # 6. Fill NaN values (which represent closed-weight models) with the new category
    df["param_scale_bin"] = binned_series.fillna("closed_weight")

    return df

# ------------------------------------------------------------------------------
# Task 7, Step 2: Attach family labels and compute capability clusters via k-means
# ------------------------------------------------------------------------------

def compute_capability_clusters(
    df: pd.DataFrame,
    n_clusters: int,
    random_seed: int
) -> pd.DataFrame:
    """
    Cluster models by capability-profile shape via k-means on z-scored benchmark
    vectors, recording any small-pool fallback and singleton clusters.

    Rationale (detailed tasklist Task 7, Step 2). K-means (k clusters) on z-scored
    benchmark vectors groups models by profile shape; a small-pool fallback must
    reduce k (or merge singletons) AND be RECORDED. The prior implementation applied
    effective_k = min(k, n) but did not record the fallback. This version records the
    fallback event (requested k, realized k, n_samples, reason) and any singleton
    clusters via the logger and the returned DataFrame's `.attrs`, so the downstream
    persister can include them in the stratification artifact.

    Parameters
    ----------
    df : pd.DataFrame
        The DataFrame from Step 1 (carrying benchmark columns and 'family').
    n_clusters : int
        The target number of clusters (k).
    random_seed : int
        The deterministic clustering seed.

    Returns
    -------
    pd.DataFrame
        `df` augmented with 'capability_cluster'; fallback/singleton provenance is
        attached under `.attrs['capability_cluster_fallback']`.
    """
    # Work on a copy to avoid mutating the caller's DataFrame.
    working_df = df.copy()

    # 1. The five canonical benchmark columns defining the capability profile.
    benchmark_cols = ["mmlu", "gpqa", "humaneval", "swebench", "math"]
    # Ensure each column exists; create entirely-missing columns as NaN.
    for col in benchmark_cols:
        if col not in working_df.columns:
            working_df[col] = np.nan
    # Extract the raw feature matrix for scikit-learn.
    X_raw = working_df[benchmark_cols].values.astype(float)

    # 2. Mean-impute missing benchmarks (the non-informative baseline before clustering).
    imputer = SimpleImputer(strategy="mean")
    X_imputed = imputer.fit_transform(X_raw)

    # 3. Standardize features (Z = (X - mu) / sigma) so high-variance benchmarks do not dominate.
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X_imputed)

    # 4. Determine the effective k (fall back to k = n when the pool is smaller than k).
    n_samples = X_scaled.shape[0]
    effective_k = min(n_clusters, n_samples)
    # Determine whether the small-pool fallback was triggered.
    fallback_triggered = effective_k < n_clusters

    # 5. Execute seeded k-means (n_init=10 to avoid local minima).
    kmeans = KMeans(n_clusters=effective_k, random_state=random_seed, n_init=10)
    cluster_labels = kmeans.fit_predict(X_scaled)

    # 6. Assign the labels back as formatted strings.
    working_df["capability_cluster"] = [f"cluster_{label}" for label in cluster_labels]

    # 7. Detect singleton clusters (clusters containing exactly one model).
    # Count members per cluster label.
    label_counts: Dict[int, int] = {}
    for label in cluster_labels:
        label_counts[int(label)] = label_counts.get(int(label), 0) + 1
    # Collect the singleton cluster labels.
    singleton_clusters = [f"cluster_{lab}" for lab, cnt in label_counts.items() if cnt == 1]

    # 8. Build the fallback/singleton provenance record.
    fallback_record: Dict[str, Any] = {
        "requested_k": int(n_clusters),                 # the configured target k
        "realized_k": int(effective_k),                 # the k actually used
        "n_samples": int(n_samples),                    # the pool size
        "fallback_triggered": bool(fallback_triggered), # whether k was reduced
        "reason": (                                     # the reason for any fallback
            "pool smaller than requested k; reduced k to n_samples"
            if fallback_triggered else "no fallback (n_samples >= requested k)"
        ),
        "singleton_clusters": singleton_clusters,       # clusters with a single member (merge candidates)
        "merge_applied": False,                         # singletons recorded but not merged (documented option)
    }
    # Attach the provenance to the DataFrame so the persister can surface it into the artifact.
    working_df.attrs["capability_cluster_fallback"] = fallback_record

    # Record the fallback explicitly when it fires (auditable even without a run directory here).
    if fallback_triggered:
        logger.warning(
            "compute_capability_clusters: small-pool fallback — requested k=%d, realized k=%d (n=%d).",
            n_clusters, effective_k, n_samples,
        )
    # Record any singleton clusters (merge candidates) for downstream review.
    if singleton_clusters:
        logger.info("compute_capability_clusters: singleton cluster(s) detected: %s.", singleton_clusters)

    # The 'family' column is already present from the raw qualification table.
    # Return the augmented DataFrame.
    return working_df

# ------------------------------------------------------------------------------
# Task 7, Step 3: Produce the stratified pool and persist the artifacts
# ------------------------------------------------------------------------------

def format_and_persist_strata(
    df: pd.DataFrame,
    run_dir: Path
) -> pd.DataFrame:
    """
    Formats the stratification metadata into a clean, indexed DataFrame and
    serializes it to disk for Tier-1 reproducibility.

    Parameters
    ----------
    df : pd.DataFrame
        The fully augmented DataFrame from Step 2.
    run_dir : Path
        The path to the immutable run directory.

    Returns
    -------
    pd.DataFrame
        A DataFrame indexed by 'model_id' containing the three stratification axes.
    """
    # 1. Select the relevant columns defining the stratification cross-product
    strata_cols = ["model_id", "family", "param_scale_bin", "capability_cluster"]
    strata_df = df[strata_cols].copy()

    # 2. Set model_id as the index to allow O(1) lookups by downstream samplers
    strata_df = strata_df.set_index("model_id")

    # 3. Convert the DataFrame to a dictionary oriented by index for JSON serialization
    # Format: {"model_A": {"family": "...", "param_scale_bin": "...", ...}, ...}
    strata_dict = strata_df.to_dict(orient="index")

    # 4. Define the path for the JSON file
    output_path: Path = run_dir / "stratified_pool.json"

    # 5. Serialize and write to disk to preserve cryptographic provenance
    with open(output_path, "w", encoding="utf-8") as f:
        json.dump(strata_dict, f, indent=2)

    # Return the formatted DataFrame
    return strata_df

# ------------------------------------------------------------------------------
# Task 7, Orchestrator Function
# ------------------------------------------------------------------------------

def stratify(
    qualified_pool: List[str],
    qualification: pd.DataFrame,
    config: Mapping[str, Any],
    run_dir: Path
) -> pd.DataFrame:
    """
    Orchestrates Task 7: Bins parameter scales, computes capability clusters via
    k-means, and persists the resulting stratification metadata.

    Parameters
    ----------
    qualified_pool : List[str]
        The ordered list of qualified model_ids.
    qualification : pd.DataFrame
        The mutated qualification DataFrame from Task 6.
    config : Mapping[str, Any]
        The immutable study configuration.
    run_dir : Path
        The path to the immutable run directory.

    Returns
    -------
    pd.DataFrame
        A DataFrame indexed by 'model_id' containing the three stratification axes:
        'family', 'param_scale_bin', and 'capability_cluster'.
    """
    # Extract parameters from the configuration
    n_clusters: int = config["preprocessing"]["n_capability_clusters"]
    # Use the global seed to ensure deterministic k-means initialization
    random_seed: int = config["reproducibility"]["seed_ledger"]["global_seed"]

    # Step 1: Filter to the qualified pool and bin the parameter counts
    binned_df = bin_parameter_scale(
        qualification=qualification,
        qualified_pool=qualified_pool
    )

    # Step 2: Compute capability clusters using standardized benchmark vectors
    clustered_df = compute_capability_clusters(
        df=binned_df,
        n_clusters=n_clusters,
        random_seed=random_seed
    )

    # Step 3: Format the output DataFrame and persist the metadata to disk
    strata_df = format_and_persist_strata(
        df=clustered_df,
        run_dir=run_dir
    )

    # Return the final stratification DataFrame
    return strata_df
````

````python
# Task 8 — Orchestration Architecture & Fusion-Operator Specification (build_fusion_operator)

# ==============================================================================
# Task 8: Orchestration Architecture & Fusion-Operator Specification
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 8, Step 1: Specify the Orchestration Architecture and Fusion Operator
# ------------------------------------------------------------------------------

def specify_fusion_operator(
    config: Mapping[str, Any],
    w: Optional[np.ndarray] = None
) -> Callable[[List[Dict[str, Any]], str, Optional[str]], Dict[str, Any]]:
    """
    Factory function that specifies the system's fusion operator closure (Section 8.3).

    This function returns a stateless, functional closure that maps a set of
    expert candidate answers to a single synthesized answer. It supports three
    mutually exclusive paths: verifier-augmented, LLM judge-based, and weighted-vote.
    This re-implementation replaces the previous placeholder implementation with
    functional gateway execution and mathematical aggregation.

    Parameters
    ----------
    config : Mapping[str, Any]
        The immutable study configuration dictionary.
    w : Optional[np.ndarray], optional
        The pre-computed normalized weight vector (Eq. 34) required for the
        weighted_vote operator path. Shape: (n_models,). Defaults to None.

    Returns
    -------
    Callable[[List[Dict[str, Any]], str, Optional[str]], Dict[str, Any]]
        The stateless, production-grade fusion closure:
        `fuse(candidates, scoring_rule, task_prompt) -> Dict`.

    Raises
    ------
    RuntimeError
        If the configuration violates the strict "no neural network training" mandate.
    """
    # Enforce the strict philosophical constraint: the framework does not train a neural network
    if config["architecture"].get("trains_neural_network", True):
        # Raise an exception to prevent late-stage execution under a corrupted architectural contract
        raise RuntimeError(
            "Architectural Violation: The FMEC protocol strictly forbids training a neural network. "
            "All constituent models must remain frozen, black-box systems."
        )

    # Extract the configured fusion operator type from the system parameters
    operator_type: str = config["architecture"]["fusion_operator"]
    # Extract gateway details for the judge-based execution path
    gateway_config = config["gateway"]
    base_url: str = gateway_config["base_url"]
    endpoint: str = f"{base_url.rstrip('/')}/chat/completions"
    # Securely resolve the API key at runtime from the system environment
    api_key_env_var: str = gateway_config["api_key_env_var"]
    api_key: Optional[str] = os.environ.get(api_key_env_var)

    # Extract judge settings and prompts
    judge_settings = config["llm_settings"]["judge"]
    judge_model = config["models"]["judge"]
    judge_template = config["prompts"]["fusion_judge_system"]
    max_retries: int = config["llm_settings"]["max_retries"]
    timeout_s: float = config["llm_settings"]["request_timeout_s"]

    # Define the closure that closes over the configuration state
    def fuse(
        candidates: List[Dict[str, Any]],
        scoring_rule: str,
        task_prompt: Optional[str] = None
    ) -> Dict[str, Any]:
        """
        Stateless closure that executes the fusion operator over a list of candidates.
        """
        # Input Validation: Ensure we have candidates to aggregate
        if not candidates:
            # Fail-closed: return an explicit failure dictionary rather than raising an unhandled exception
            return {"fused_answer": "", "winner_model_id": "ERROR_NO_CANDIDATES", "fusion_method": "error"}

        # Extract the list of available model IDs in this specific candidate set
        present_models: List[str] = [c["model_id"] for col in candidates for c in [col] if "model_id" in col]

        # ----------------------------------------------------------------------
        # Path A: Verifier-Augmented Fusion / Deterministic Scoring Rules
        # ----------------------------------------------------------------------
        if operator_type == "verifier_augmented" and scoring_rule in {"unit_tests", "numeric_tol", "exact_match"}:
            # Sort the candidates by deterministic correctness (is_correct) first,
            # and break ties by self-reported confidence descending (§8.3)
            # True sorts after False; higher confidence sorts after lower
            sorted_candidates = sorted(
                candidates,
                key=lambda x: (bool(x.get("is_correct", False)), float(x.get("confidence", 0.0))),
                reverse=True
            )
            # The top candidate represents our optimal verifier-selected solution
            winner = sorted_candidates[0]
            return {
                "fused_answer": winner.get("thesis", winner.get("answer", "")),
                "winner_model_id": winner["model_id"],
                "fusion_method": "deterministic_verification"
            }

        # ----------------------------------------------------------------------
        # Path B: Judge-Based Synthesis / Open-Ended Rubrics
        # ----------------------------------------------------------------------
        elif operator_type == "judge" or scoring_rule == "judge_rubric":
            # Guard: Judge execution strictly requires a valid task_prompt and API key
            if not task_prompt:
                raise ValueError("Judge-based fusion requires the 'task_prompt' string to format the prompt.")
            if not api_key:
                raise RuntimeError("Judge-based fusion requires the gateway API key to be set in the environment.")

            # Format the candidate answers into a structured JSON string indexed by integers
            # This preserves exact candidate boundaries for the Judge's parsing
            answers_payload: Dict[str, str] = {
                str(idx): c.get("thesis", c.get("answer", ""))
                for idx, c in enumerate(candidates)
            }
            answers_json = json.dumps(answers_payload, indent=2)

            # Safely format the fusion judge system prompt without breaking on literal JSON braces
            formatted_system_prompt = _safe_prompt_format(
                judge_template,
                {"task_prompt": task_prompt, "answers_json": answers_json}
            )

            # Construct the HTTP payload, enforcing JSON format and deterministic decoding (temperature=0.0)
            payload = {
                "model": judge_model,
                "messages": [{"role": "user", "content": formatted_system_prompt}],
                "temperature": 0.0,
                "max_tokens": judge_settings["max_tokens"],
                "response_format": {"type": "json_object"}
            }

            # Execute the gateway call using our robust retry client
            response = _execute_with_retry(
                endpoint=endpoint,
                api_key=api_key,
                payload=payload,
                max_retries=max_retries,
                timeout_s=timeout_s,
                require_json=True
            )

            # Parse the returned winner index
            winner_idx = 0
            if response and "winner_index" in response:
                try:
                    winner_idx = int(response["winner_index"])
                    # Verify the index is within bounds; fallback to 0 if the judge hallucinated an out-of-bounds index
                    if winner_idx < 0 or winner_idx >= len(candidates):
                        winner_idx = 0
                except (ValueError, TypeError):
                    winner_idx = 0

            # Retrieve the winning candidate based on the resolved index
            winner = candidates[winner_idx]
            return {
                "fused_answer": winner.get("thesis", winner.get("answer", "")),
                "winner_model_id": winner["model_id"],
                "fusion_method": f"llm_adjudication_index_{winner_idx}"
            }

        # ----------------------------------------------------------------------
        # Path C: Cost-Aware Weighted-Vote Fusion
        # ----------------------------------------------------------------------
        elif operator_type == "weighted_vote":
            # If no weight vector is provided, fall back to uniform weights on S
            if w is None:
                # Assign equal weight to each candidate
                uniform_weight = 1.0 / float(len(candidates))
                candidate_weights = [uniform_weight] * len(candidates)
            else:
                # Extract the canonical qualified model roster to map weights correctly
                # (This assumes 'qualified_pool' is available or config-defined)
                qualified_pool: List[str] = list(config["models"]["panel"])

                # Extract raw scores for the present candidates
                raw_weights = []
                for c in candidates:
                    m_id = c["model_id"]
                    # If the model exists in our pool, extract its weight; else default to 0.0
                    idx = qualified_pool.index(m_id) if m_id in qualified_pool else -1
                    raw_weights.append(w[idx] if idx != -1 else 0.0)

                # Normalize the subset weights to the probability simplex (sum to 1.0)
                weight_sum = sum(raw_weights)
                if weight_sum > 1e-12:
                    candidate_weights = [val / weight_sum for val in raw_weights]
                else:
                    # Fallback to uniform if all weights collapsed to 0
                    candidate_weights = [1.0 / float(len(candidates))] * len(candidates)

            # Identify the candidate that carries the maximum normalized weight
            best_candidate_idx = int(np.argmax(candidate_weights))
            winner = candidates[best_candidate_idx]
            return {
                "fused_answer": winner.get("thesis", winner.get("answer", "")),
                "winner_model_id": winner["model_id"],
                "fusion_method": f"weighted_vote_idx_{best_candidate_idx}"
            }

        else:
            # Raise an explicit exception if configuration parameters mismatch
            raise ValueError(f"Unknown fusion operator or scoring rule combination: {operator_type}, {scoring_rule}")

    return fuse

# ------------------------------------------------------------------------------
# Task 8, Step 2: Define the Weighted-Fusion Scoring Component
# ------------------------------------------------------------------------------

def define_weighted_fusion_scoring(
    q: np.ndarray,
    r: np.ndarray,
    c: np.ndarray,
    l: np.ndarray,
    config: Mapping[str, Any]
) -> Tuple[np.ndarray, Dict[str, Any]]:
    """
    Computes the normalized weighted-fusion scores (Eq. 34) and extracts the
    sandbox configuration for verifiable tasks.

    Equation 34: s_i = (q_i * r_i) / (c_i^a * l_i^b), w_i = s_i / sum(s)

    Parameters
    ----------
    q : np.ndarray
        Vector of model qualities (expected performance).
    r : np.ndarray
        Vector of model reliabilities.
    c : np.ndarray
        Vector of model inference costs.
    l : np.ndarray
        Vector of model latencies.
    config : Mapping[str, Any]
        The immutable study configuration.

    Returns
    -------
    Tuple[np.ndarray, Dict[str, Any]]
        - w: The normalized weight vector on the simplex.
        - sandbox_config: The configuration dictionary for secure code execution.
    """
    # Extract exponents from the configuration
    exponents = config["architecture"]["weighted_fusion_exponents"]
    a: float = exponents["a"]  # Cost exponent (default 0.50)
    b: float = exponents["b"]  # Latency exponent (default 0.25)

    # Add a small epsilon to denominators to prevent division by zero
    epsilon = 1e-9

    # Compute the numerator: Quality * Reliability
    numerator = q * r

    # Compute the denominator: Cost^a * Latency^b
    denominator = np.power(c + epsilon, a) * np.power(l + epsilon, b)

    # Compute raw scores
    s = numerator / denominator

    # Normalize to the simplex (sum to 1)
    w = s / np.sum(s)

    # Extract and validate sandbox parameters for the verifier-augmented path
    sandbox_config: Dict[str, Any] = dict(config["architecture"]["sandbox"])

    # Honest engineering caveat: True isolation requires OS-level containers.
    # This configuration specifies the intent, which the execution engine must honor.
    assert sandbox_config["timeout_s"] > 0, "Sandbox timeout must be positive."
    assert sandbox_config["network"] == "disabled", "Sandbox network must be disabled to prevent host contamination."

    return w, sandbox_config

# ------------------------------------------------------------------------------
# Task 8, Step 3: Specify the Optional Contextual Epsilon-Greedy Router
# ------------------------------------------------------------------------------

class EpsilonGreedyRouter:
    """
    Contextual epsilon-greedy router over a fixed set of model 'arms' (Section 8.5).

    This class implements an online learned contextual bandit routing policy (Eqs. 40-43).
    It maps incoming discrete context keys to the expected utility-maximizing model or
    ensemble. Reward formulation internalizes cost and risk per Equation 42:
    r_t = quality - lambda*risk - mu*cost - gamma*latency.

    Attributes
    ----------
    arms : List[Hashable]
        The canonical ordered list of model IDs (arms) to route to.
    epsilon : float
        The exploration probability strictly bounded within [0.0, 1.0].
    w : Dict[str, float]
        The penalty weight vector containing lambda, mu, and gamma.
    _rng : np.random.Generator
        The dedicated random number generator for unbiased exploration.
    _counts : Dict[Hashable, Dict[Hashable, int]]
        State cache storing the total visit counts per context-arm pair.
    _means : Dict[Hashable, Dict[Hashable, float]]
        State cache storing the incremental expected reward per context-arm pair.
    """

    def __init__(
        self,
        arms: List[Hashable],
        epsilon: float,
        reward_weights: Dict[str, float],
        rng: np.random.Generator
    ) -> None:
        # Validate that the arms list is non-empty to ensure there are valid routing options
        if not arms:
            # Raise ValueError if the pool of available arms is empty
            raise ValueError("EpsilonGreedyRouter initialization requires a non-empty list of arms.")

        # Validate that epsilon is within the standard probability bounds [0.0, 1.0]
        if not (0.0 <= epsilon <= 1.0):
            # Raise ValueError if epsilon is mathematically out of bounds
            raise ValueError(f"Epsilon must be within the probability range [0.0, 1.0], got {epsilon}")

        # Validate that reward_weights is a dictionary
        if not isinstance(reward_weights, dict):
            # Raise TypeError if the reward weights are not provided as a dictionary
            raise TypeError(f"Reward weights must be a dictionary, got {type(reward_weights).__name__}")

        # Validate that the random number generator is an instance of numpy's Generator
        if not isinstance(rng, np.random.Generator):
            # Raise TypeError if the entropy source violates the Generator contract
            raise TypeError(f"rng must be an instance of numpy.random.Generator, got {type(rng).__name__}")

        # Bind the immutable copy of arms to the instance state
        self.arms: List[Hashable] = list(arms)
        # Bind the validated exploration probability epsilon
        self.epsilon: float = float(epsilon)
        # Bind the configuration reward penalty weights
        self.w: Dict[str, float] = dict(reward_weights)
        # Bind the private cryptographic random number generator
        self._rng: np.random.Generator = rng
        # Initialize the private count tracking state mapped by context and arm
        self._counts: Dict[Hashable, Dict[Hashable, int]] = {}
        # Initialize the private expected value tracking state mapped by context and arm
        self._means: Dict[Hashable, Dict[Hashable, float]] = {}

    def _ensure_context(self, context: Hashable) -> None:
        """
        Initializes the state mapping for a newly observed context key.
        """
        # Validate that the incoming context key is indeed hashable to prevent runtime dict key errors
        if not isinstance(context, Hashable):
            # Raise TypeError if the context cannot be hashed
            raise TypeError(f"Context key must be Hashable, got {type(context).__name__}")

        # Check if the context exists within the state cache
        if context not in self._counts:
            # Initialize the counts mapping to 0 for all available arms
            self._counts[context] = {a: 0 for a in self.arms}
            # Initialize the incremental mean expectations to 0.0 for all available arms
            self._means[context] = {a: 0.0 for a in self.arms}

    def select(self, context: Hashable) -> Hashable:
        """
        Selects an arm using the epsilon-greedy policy (Equation 43).

        Parameters
        ----------
        context : Hashable
            The discrete context key representing the task class.

        Returns
        -------
        Hashable
            The selected model ID (arm) to route the request to.
        """
        # Enforce strict context initialization and validation
        self._ensure_context(context)

        # Eq. (43): Explore with probability epsilon, else exploit the argmax expected mean
        if self._rng.random() < self.epsilon:
            # Uniformly select a random arm from the available pool (Exploration Phase)
            return self._rng.choice(self.arms)

        # Retrieve the current expected mean estimations for the active context
        means: Dict[Hashable, float] = self._means[context]

        # Compute the maximum expected reward value among all arms
        best_val: float = max(means.values())

        # Identify all arms achieving the maximum expected reward, applying float-safe tolerances (Eq. 43)
        top_arms: List[Hashable] = [a for a, m in means.items() if math.isclose(m, best_val, rel_tol=1e-9)]

        # Uniformly select among the tied top-performing arms to eliminate selection bias (Exploitation Phase)
        return self._rng.choice(top_arms)

    def reward(self, quality: float, cost: float, risk: float, latency: float) -> float:
        """
        Formulates the joint scalar reward function (Equation 42).

        Parameters
        ----------
        quality : float
            The estimated task quality (usually binary correctness [0,1]).
        cost : float
            The financial cost in USD of executing the routed models.
        risk : float
            The estimated correlated-failure risk of the serving path.
        latency : float
            The wall-clock execution latency in seconds.

        Returns
        -------
        float
            The scalar reward value.
        """
        # Validate that quality is a numeric value
        if not isinstance(quality, (int, float)):
            # Raise TypeError on malformed quality parameters
            raise TypeError(f"Quality must be a float or integer, got {type(quality).__name__}")

        # Validate that cost is a non-negative numeric value
        if not isinstance(cost, (int, float)) or cost < 0.0:
            # Raise ValueError on invalid cost telemetry parameters
            raise ValueError(f"Cost must be a non-negative numeric value, got {cost}")

        # Validate that risk is a non-negative numeric value
        if not isinstance(risk, (int, float)) or risk < 0.0:
            # Raise ValueError on invalid risk metrics
            raise ValueError(f"Risk must be a non-negative numeric value, got {risk}")

        # Validate that latency is a non-negative numeric value
        if not isinstance(latency, (int, float)) or latency < 0.0:
            # Raise ValueError on invalid latency telemetry parameters
            raise ValueError(f"Latency must be a non-negative numeric value, got {latency}")

        # Eq. (42): r = quality - lambda*risk - mu*cost - gamma*latency
        return (
            float(quality)
            - (self.w.get("lambda", 0.0) * float(risk))
            - (self.w.get("mu", 0.0) * float(cost))
            - (self.w.get("gamma", 0.0) * float(latency))
        )

    def update(self, context: Hashable, arm: Hashable, reward: float) -> None:
        """
        Updates the historical expected reward estimation using the incremental update rule.

        Parameters
        ----------
        context : Hashable
            The discrete context key representing the task class.
        arm : Hashable
            The model ID (arm) that was selected and executed.
        reward : float
            The observed scalar reward from the evaluation.
        """
        # Enforce strict context initialization and validation
        self._ensure_context(context)
        # Validate that the selected arm is a registered member of the pool
        if arm not in self.arms:
            # Raise ValueError to prevent updating out-of-pool estimators
            raise ValueError(f"Selected arm '{arm}' is not registered in the pool: {self.arms}")

        # Validate that the reward value is numeric
        if not isinstance(reward, (int, float)):
            # Raise TypeError on malformed reward updates
            raise TypeError(f"Reward value must be numeric, got {type(reward).__name__}")

        # Increment the visit count for the specified context-arm pair
        self._counts[context][arm] += 1

        # Extract the updated total count n
        n: int = self._counts[context][arm]

        # Extract the current mean estimate m
        m: float = self._means[context][arm]

        # Incremental update rule (m <- m + (r - m)/n) prevents float overflow and maintains stability
        self._means[context][arm] = m + (float(reward) - m) / float(n)


def specify_contextual_router(
    config: Mapping[str, Any],
    qualified_pool: List[str],
    rng: np.random.Generator
) -> Optional[EpsilonGreedyRouter]:
    """
    Instantiates the optional contextual bandit router if enabled in the config.

    Parameters
    ----------
    config : Mapping[str, Any]
        The immutable study configuration.
    qualified_pool : List[str]
        The list of available models (arms).
    rng : np.random.Generator
        The dedicated random number generator for the router.

    Returns
    -------
    Optional[EpsilonGreedyRouter]
        The instantiated router, or None if disabled.
    """
    router_config = config["network_training"]["router"]

    # The router is strictly optional and disabled by default
    if not router_config.get("enabled", False):
        return None

    # Extract parameters
    epsilon: float = router_config["epsilon"]
    reward_weights: Dict[str, float] = router_config["reward_weights"]

    # Instantiate the router
    router = EpsilonGreedyRouter(
        arms=qualified_pool,
        epsilon=epsilon,
        reward_weights=reward_weights,
        rng=rng
    )

    return router

# ------------------------------------------------------------------------------
# Task 8, Orchestrator Function
# ------------------------------------------------------------------------------

def build_fusion_operator(
    config: Mapping[str, Any],
    rng_bundle: Dict[str, np.random.Generator],
    qualified_pool: List[str],
    q_vec: np.ndarray,
    r_vec: np.ndarray,
    c_vec: np.ndarray,
    l_vec: np.ndarray
) -> Tuple[Callable, np.ndarray, Dict[str, Any], Optional[Any]]:
    """
    Orchestrates Task 8: Specifies the active fusion operator, computes weighted scores,
    and initializes the optional contextual bandit router.

    This remediated orchestrator passes the computed w_vec weight vector directly
    into specify_fusion_operator as w=w_vec. This enables the weighted-vote
    closure path to perform actual, mathematically weighted candidate selection
    instead of relying on stubs.

    Parameters
    ----------
    config : Mapping[str, Any]
        The immutable study configuration dictionary.
    rng_bundle : Dict[str, np.random.Generator]
        The bundle of independent, seeded random number generators.
    qualified_pool : List[str]
        The canonical ordered list of qualified model IDs.
    q_vec : np.ndarray
        1D array of shape (n_models,) representing expected model qualities.
    r_vec : np.ndarray
        1D array of shape (n_models,) representing model reliabilities.
    c_vec : np.ndarray
        1D array of shape (n_models,) representing mean model inference costs.
    l_vec : np.ndarray
        1D array of shape (n_models,) representing mean model latencies.

    Returns
    -------
    Tuple
        - fuse_callable: The remediated, active fusion operator routing function.
        - w_vec: The computed normalized weight vector on the probability simplex.
        - sandbox_config: The verified sandbox parameters dictionary for code execution.
        - router: The instantiated EpsilonGreedyRouter instance, or None if disabled.
    """
    # Verify that the dimensions of all incoming model-telemetry arrays match the qualified pool length
    if not (len(qualified_pool) == q_vec.shape[0] == r_vec.shape[0] == c_vec.shape[0] == l_vec.shape[0]):
        # Raise an exception to prevent misaligned weight index assignments
        raise ValueError(
            f"Dimension Mismatch: The length of qualified_pool ({len(qualified_pool)}) "
            f"must match the shapes of q_vec ({q_vec.shape[0]}), r_vec ({r_vec.shape[0]}), "
            f"c_vec ({c_vec.shape[0]}), and l_vec ({l_vec.shape[0]})."
        )

    # Step 1: Compute weighted-fusion scores (Eq. 34) and extract the sandbox configuration
    # define_weighted_fusion_scoring is assumed available
    w_vec, sandbox_config = define_weighted_fusion_scoring(
        q=q_vec,
        r=r_vec,
        c=c_vec,
        l=l_vec,
        config=config
    )

    # Step 2: Instantiate specify_fusion_operator factory
    # We pass the computed weight vector w_vec to specify_fusion_operator to enable
    # active, weighted-vote candidate selection on that serving path
    fuse_callable = specify_fusion_operator(
        config=config,
        w=w_vec
    )

    # Step 3: Specify the optional contextual router using its dedicated RNG stream
    # Defensive lookup ensures that the router seed is present in our ledger
    if "router_seed" not in rng_bundle:
        raise KeyError("RNG initialization failure: 'router_seed' is missing from the provided rng_bundle.")
    router_rng = rng_bundle["router_seed"]
    router = specify_contextual_router(
        config=config,
        qualified_pool=qualified_pool,
        rng=router_rng
    )

    return fuse_callable, w_vec, sandbox_config, router
````

````python
# Task 9 — Response Collection: Panel & Structured Solver (Stage 1) (collect_responses)

# ==============================================================================
# Task 9: Response Collection: Panel & Structured Solver (Stage 1)
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 9, Helper Functions: Safe Prompt Formatting and HTTP Execution
# ------------------------------------------------------------------------------

def _safe_prompt_format(template: str, replacements: Dict[str, str]) -> str:
    """
    Safely substitutes specific tokens in a prompt template without using
    str.format(), preventing collisions with literal JSON braces.

    Parameters
    ----------
    template : str
        The raw prompt template containing {token} placeholders.
    replacements : Dict[str, str]
        A dictionary mapping token names (without braces) to their replacement strings.

    Returns
    -------
    str
        The formatted prompt string.
    """
    formatted_string = template
    for token, value in replacements.items():
        # Replace exactly the token enclosed in single braces
        formatted_string = formatted_string.replace(f"{{{token}}}", str(value))
    return formatted_string

def _execute_with_retry(
    endpoint: str,
    api_key: str,
    payload: Dict[str, Any],
    max_retries: int,
    timeout_s: float,
    require_json: bool = False
) -> Dict[str, Any]:
    """
    Executes an HTTP POST request to the LLM gateway with exponential backoff,
    jitter, and comprehensive provenance capture.

    Parameters
    ----------
    endpoint : str
        The API endpoint URL.
    api_key : str
        The resolved API key.
    payload : Dict[str, Any]
        The JSON payload containing the model, messages, and decoding parameters.
    max_retries : int
        The maximum number of retry attempts for transient errors.
    timeout_s : float
        The wall-clock timeout for the HTTP request.
    require_json : bool, optional
        If True, enforces that the generated text is valid JSON, by default False.

    Returns
    -------
    Dict[str, Any]
        A dictionary containing the generated text, provenance metadata, and
        any infrastructure errors.
    """
    data = json.dumps(payload).encode('utf-8')
    req = urllib.request.Request(
        endpoint,
        data=data,
        headers={
            "Authorization": f"Bearer {api_key}",
            "Content-Type": "application/json"
        }
    )

    # Initialize the result dictionary with default null provenance
    result: Dict[str, Any] = {
        "text": None,
        "error": None,
        "prompt_tokens": None,
        "completion_tokens": None,
        "latency_s": None,
        "provider_version": None,
        "system_fingerprint": None
    }

    for attempt in range(max_retries + 1):
        start_time = time.perf_counter()
        try:
            with urllib.request.urlopen(req, timeout=timeout_s) as response:
                response_body = response.read().decode('utf-8')
                latency = time.perf_counter() - start_time

                resp_json = json.loads(response_body)

                # Extract the generated text
                generated_text = resp_json["choices"][0]["message"]["content"]

                # If the solver role requires JSON, validate it immediately
                if require_json:
                    try:
                        # Attempt to parse the text to ensure it is structurally valid JSON
                        _ = json.loads(generated_text)
                    except json.JSONDecodeError as e:
                        raise ValueError(f"Malformed JSON output: {str(e)}")

                # Populate the successful result dictionary
                result["text"] = generated_text
                result["latency_s"] = latency

                # Extract token usage provenance
                usage = resp_json.get("usage", {})
                result["prompt_tokens"] = usage.get("prompt_tokens")
                result["completion_tokens"] = usage.get("completion_tokens")

                # Extract version and fingerprint provenance for drift monitoring
                result["provider_version"] = resp_json.get("model")
                result["system_fingerprint"] = resp_json.get("system_fingerprint")

                # Clear any previous errors on success
                result["error"] = None
                return result

        except (urllib.error.URLError, KeyError, json.JSONDecodeError, ValueError) as e:
            # Record the latency even for failed calls
            latency = time.perf_counter() - start_time
            result["latency_s"] = latency
            result["error"] = f"Attempt {attempt + 1} failed: {str(e)}"

            if attempt < max_retries:
                # Exponential backoff with full jitter: sleep between 0 and 2^attempt seconds
                sleep_time = random.uniform(0, 2 ** attempt)
                time.sleep(sleep_time)
            else:
                # Retries exhausted; return the result dictionary with the error populated
                # This ensures the record is not silently dropped
                return result

    return result

# ------------------------------------------------------------------------------
# Task 9, Step 1 & 2: Execute parallel fan-out and capture provenance
# ------------------------------------------------------------------------------

def generate_responses_parallel(
    qualified_pool: List[str],
    task_record: pd.DataFrame,
    config: Mapping[str, Any],
    api_key: str
) -> Tuple[List[Dict[str, Any]], List[Dict[str, Any]]]:
    """
    Executes parallel fan-out generation for both the panel (free-form) and
    solver (structured) roles, capturing comprehensive provenance.

    Parameters
    ----------
    qualified_pool : List[str]
        The ordered list of qualified model_ids.
    task_record : pd.DataFrame
        The pruned task DataFrame.
    config : Mapping[str, Any]
        The immutable study configuration.
    api_key : str
        The resolved API key for the gateway.

    Returns
    -------
    Tuple[List[Dict[str, Any]], List[Dict[str, Any]]]
        - model_answer_records: The raw free-form generation records.
        - candidate_solution_records: The raw structured solution records.
    """
    # Extract gateway and LLM settings
    base_url: str = config["gateway"]["base_url"]
    endpoint: str = f"{base_url.rstrip('/')}/chat/completions"

    llm_settings = config["llm_settings"]
    panel_settings = llm_settings["panel"]
    solver_settings = llm_settings["solver"]
    max_retries: int = llm_settings["max_retries"]
    timeout_s: float = llm_settings["request_timeout_s"]

    # Extract replication parameters
    r_count: int = config["uncertainty"]["replication"]["R"]
    seed_base: int = config["uncertainty"]["replication"]["replicate_seed_base"]

    # Extract prompt templates and schemas
    prompts = config["prompts"]
    panel_template: str = prompts["panel_system"]
    solver_template: str = prompts["solver_system"]
    solver_schema: str = json.dumps(prompts["solver_json_schema"])

    # Initialize thread-safe lists to accumulate results
    model_answers: List[Dict[str, Any]] = []
    candidate_solutions: List[Dict[str, Any]] = []

    # Define the worker function for the thread pool
    def _worker(task_id: str, prompt_text: str, model_id: str, rep: int) -> Tuple[Dict[str, Any], Dict[str, Any]]:
        # Calculate the deterministic seed for this specific replicate
        current_seed = seed_base + rep

        # --- 1. Execute Panel Role (Free-form) ---
        panel_prompt = _safe_prompt_format(panel_template, {"task_prompt": prompt_text})
        panel_payload = {
            "model": model_id,
            "messages": [{"role": "user", "content": panel_prompt}],
            "temperature": panel_settings["temperature"],
            "max_tokens": panel_settings["max_tokens"],
            "seed": current_seed
        }
        # Add response_format only if it is explicitly defined and not None
        if panel_settings.get("response_format"):
            panel_payload["response_format"] = {"type": panel_settings["response_format"]}

        panel_res = _execute_with_retry(
            endpoint, api_key, panel_payload, max_retries, timeout_s, require_json=False
        )

        # Construct the model_answer record
        ma_record = {
            "task_id": task_id,
            "model_id": model_id,
            "rep": rep,  # Added to ensure the generation-variance axis is recoverable
            "answer": panel_res["text"] if panel_res["text"] is not None else "",
            "error": panel_res["error"],
            "prompt_tokens": panel_res["prompt_tokens"],
            "completion_tokens": panel_res["completion_tokens"],
            "latency_s": panel_res["latency_s"],
            "temperature": panel_settings["temperature"],
            "seed": current_seed,
            "provider_version": panel_res["provider_version"],
            "system_fingerprint": panel_res["system_fingerprint"]
        }

        # --- 2. Execute Solver Role (Structured JSON) ---
        solver_prompt = _safe_prompt_format(
            solver_template,
            {"task_prompt": prompt_text, "json_schema": solver_schema}
        )
        solver_payload = {
            "model": model_id,
            "messages": [{"role": "user", "content": solver_prompt}],
            "temperature": solver_settings["temperature"],
            "max_tokens": solver_settings["max_tokens"],
            "seed": current_seed
        }
        if solver_settings.get("response_format"):
            solver_payload["response_format"] = {"type": solver_settings["response_format"]}

        solver_res = _execute_with_retry(
            endpoint, api_key, solver_payload, max_retries, timeout_s, require_json=True
        )

        # Parse the JSON output if successful, otherwise use empty defaults
        parsed_solution = {}
        if solver_res["text"] is not None and solver_res["error"] is None:
            try:
                parsed_solution = json.loads(solver_res["text"])
            except json.JSONDecodeError:
                solver_res["error"] = "Failed to parse JSON despite require_json=True"

        # Construct the candidate_solution record
        cs_record = {
            "task_id": task_id,
            "model_id": model_id,
            "rep": rep,  # Added to ensure the generation-variance axis is recoverable
            "thesis": parsed_solution.get("thesis", ""),
            "reasoning": parsed_solution.get("reasoning", ""),
            "python_code": parsed_solution.get("python_code", None),
            "confidence": float(parsed_solution.get("confidence", 0.0)),
            "error": solver_res["error"]
        }

        return ma_record, cs_record

    # Execute the fan-out using a ThreadPoolExecutor
    # Max workers is set to 50 to maximize parallel I/O throughput
    with concurrent.futures.ThreadPoolExecutor(max_workers=50) as executor:
        futures = []
        for _, task_row in task_record.iterrows():
            t_id = task_row["task_id"]
            p_text = task_row["prompt"]
            for m_id in qualified_pool:
                for r in range(r_count):
                    futures.append(executor.submit(_worker, t_id, p_text, m_id, r))

        # Collect results as they complete
        for future in concurrent.futures.as_completed(futures):
            ma, cs = future.result()
            model_answers.append(ma)
            candidate_solutions.append(cs)

    return model_answers, candidate_solutions

# ------------------------------------------------------------------------------
# Task 9, Step 3: Execute sandboxed verification for structured solutions
# ------------------------------------------------------------------------------

def _run_unit_test_payload(
    candidate_code: str,
    test_suite_code: str,
    timeout_s: float
) -> Dict[str, Any]:
    """
    Execute a single (candidate solution + reference test suite) pair in an
    isolated subprocess and classify the outcome.

    Rationale (§8.3; detailed tasklist Task 9, Step 3). For verifier-augmented
    tasks the correctness signal is whether the candidate PASSES the reference
    tests — not merely whether the candidate imports. This helper assembles the
    candidate code and the ground-truth test suite into one sandboxed unit and
    runs it under a hard timeout, returning a structured verdict that distinguishes
    compile errors, runtime errors/timeouts, and test pass/fail.

    Honest isolation caveat (architecture note): a bare `-I` subprocess enforces
    the timeout and basic interpreter isolation but does NOT guarantee network or
    filesystem isolation; true isolation requires OS-level containers/seccomp.

    Parameters
    ----------
    candidate_code : str
        The model's structured `python_code` defining the solution.
    test_suite_code : str
        The task's `ground_truth` reference test suite (expected to be
        self-executing: module-level assertions or a self-invoked `check`).
    timeout_s : float
        The wall-clock execution timeout in seconds (5 s per protocol).

    Returns
    -------
    Dict[str, Any]
        A verdict dict with keys:
        - 'passed'  : Optional[bool]  — True iff the reference tests passed.
        - 'status'  : str             — one of {'TESTS_PASSED','TESTS_FAILED',
                                         'COMPILE_ERROR','RUNTIME_ERROR','TIMEOUT'}.
        - 'detail'  : str             — truncated stderr / diagnostic text.
    """
    # Compose the executable payload: candidate definitions first, then the reference tests.
    # The tests run in the same module namespace, so the candidate's symbols are in scope.
    payload: str = candidate_code + "\n\n" + test_suite_code

    # Execute the payload in an isolated interpreter under the hard timeout.
    try:
        # `-I` isolates from PYTHONPATH/site customizations; `-c` runs the payload string.
        completed = subprocess.run(
            [sys.executable, "-I", "-c", payload],   # isolated interpreter executing the combined unit
            capture_output=True,                      # capture stdout/stderr for classification
            text=True,                                # decode streams as text
            timeout=timeout_s,                        # enforce the 5 s wall-clock limit
        )
    except subprocess.TimeoutExpired:
        # A timeout is a distinct, non-passing outcome (possible infinite loop in candidate/tests).
        return {"passed": False, "status": "TIMEOUT", "detail": f"TimeoutExpired after {timeout_s}s."}
    except Exception as exc:  # pragma: no cover - defensive transport guard
        # Any subprocess-spawn failure is reported as a runtime error with the exception text.
        return {"passed": False, "status": "RUNTIME_ERROR", "detail": f"Subprocess error: {exc!s}"}

    # A zero return code means the payload (candidate + tests) ran to completion with no failed assertion.
    if completed.returncode == 0:
        # All reference assertions held -> the candidate passes verification.
        return {"passed": True, "status": "TESTS_PASSED", "detail": ""}

    # Non-zero return code: capture and truncate stderr for diagnosis and classification.
    stderr_text: str = completed.stderr or ""
    # Truncate to keep the record compact while preserving the salient error head.
    detail: str = stderr_text[:500] + ("..." if len(stderr_text) > 500 else "")

    # Classify a parse-time failure (SyntaxError/IndentationError) as a COMPILE_ERROR.
    if "SyntaxError" in stderr_text or "IndentationError" in stderr_text:
        # The candidate (or tests) failed to compile.
        return {"passed": False, "status": "COMPILE_ERROR", "detail": detail}

    # Classify a failed assertion as a genuine TEST FAILURE (the candidate ran but produced wrong output).
    if "AssertionError" in stderr_text:
        # The candidate executed but did not satisfy the reference tests.
        return {"passed": False, "status": "TESTS_FAILED", "detail": detail}

    # Any other non-zero exit is an uncaught runtime exception during execution.
    return {"passed": False, "status": "RUNTIME_ERROR", "detail": detail}


def execute_sandboxed_verification(
    candidate_solutions: List[Dict[str, Any]],
    task_record: pd.DataFrame,
    sandbox_config: Dict[str, Any]
) -> List[Dict[str, Any]]:
    """
    Deterministically ground correctness for verifiable tasks by executing each
    candidate solution against its reference test suite in a restricted subprocess.

    Rationale (§8.3; detailed tasklist Task 9, Step 3; Task 10, Step 1). The
    correctness signal for `unit_tests`/patch-style tasks must be TEST PASS/FAIL,
    not the exit code of the candidate run in isolation. For each such task this
    function retrieves the task's `ground_truth` (the reference test suite),
    composes it with the candidate's `python_code`, executes the combined unit
    under a 5 s timeout, and writes the verdict to dedicated `verifier_*` fields,
    leaving the candidate's own `error` channel for transport failures only. The
    verifier output is kept strictly distinct from any later judge output.

    Parameters
    ----------
    candidate_solutions : List[Dict[str, Any]]
        The raw structured solution records (each carrying 'task_id', 'model_id',
        'rep', and optionally 'python_code').
    task_record : pd.DataFrame
        The task DataFrame; supplies 'scoring_rule' and 'ground_truth' per task_id.
    sandbox_config : Dict[str, Any]
        Sandbox configuration; 'timeout_s' (default 5.0), plus declared network/fs
        policy (enforced at OS level upstream).

    Returns
    -------
    List[Dict[str, Any]]
        The same list of candidate solution records, each augmented in place with:
        - 'verifier_passed' : Optional[bool] — test pass/fail (None if N/A).
        - 'verifier_status' : str            — outcome label (e.g., 'TESTS_PASSED',
                                                'NOT_APPLICABLE', 'NO_CODE', 'NO_GROUND_TRUTH').
        - 'verifier_detail' : str            — diagnostic text.

    Raises
    ------
    TypeError
        If `candidate_solutions` is not a list or `task_record` is not a DataFrame.
    KeyError
        If `task_record` lacks the required 'scoring_rule'/'ground_truth' columns.
    """
    # Validate the candidate container type.
    if not isinstance(candidate_solutions, list):
        # Fail fast on a malformed candidate container.
        raise TypeError("`candidate_solutions` must be a list of record dicts.")
    # Validate the task table type.
    if not isinstance(task_record, pd.DataFrame):
        # Fail fast on a malformed task table.
        raise TypeError("`task_record` must be a pandas DataFrame.")
    # Validate that the task table carries the columns we depend on.
    for required_col in ("task_id", "scoring_rule", "ground_truth"):
        # Each required column must be present to ground correctness.
        if required_col not in task_record.columns:
            # Surface the missing column explicitly.
            raise KeyError(f"`task_record` is missing required column '{required_col}'.")

    # Build per-task lookups for the scoring rule and the reference ground truth (the test suite for code).
    task_lookup: Dict[Any, Dict[str, Any]] = (
        task_record.set_index("task_id")[["scoring_rule", "ground_truth"]].to_dict(orient="index")
    )

    # Resolve the execution timeout (5 s per protocol unless overridden).
    timeout_s: float = float(sandbox_config.get("timeout_s", 5.0))

    # The set of scoring rules that require code execution against reference tests.
    code_execution_rules: Set[str] = {"unit_tests", "patch"}

    # Iterate over every candidate solution record and verify the code-bearing, verifiable ones.
    for cs in candidate_solutions:
        # Resolve this record's task id.
        t_id: Any = cs.get("task_id")
        # Look up this task's metadata (scoring rule + ground truth); default to empty if unknown.
        task_meta: Dict[str, Any] = task_lookup.get(t_id, {})
        # Extract the scoring rule for routing.
        rule: Optional[str] = task_meta.get("scoring_rule")

        # Skip tasks that do not require code execution (e.g., exact_match, numeric_tol, judge_rubric).
        if rule not in code_execution_rules:
            # Mark verification as not applicable and move on (do NOT touch the transport `error` channel).
            cs["verifier_passed"] = None
            cs["verifier_status"] = "NOT_APPLICABLE"
            cs["verifier_detail"] = ""
            continue

        # Retrieve the candidate code; absence/blankness is a non-passing, distinctly labeled outcome.
        code_str: Optional[str] = cs.get("python_code")
        # Guard against missing or whitespace-only code.
        if not isinstance(code_str, str) or not code_str.strip():
            # Record that there was no executable candidate code.
            cs["verifier_passed"] = False
            cs["verifier_status"] = "NO_CODE"
            cs["verifier_detail"] = "Candidate python_code is missing or empty."
            continue

        # Retrieve the reference test suite from ground truth; without it we cannot verify.
        test_suite: Any = task_meta.get("ground_truth")
        # Guard against a missing/blank test suite for a task that claims to be verifiable.
        if not isinstance(test_suite, str) or not test_suite.strip():
            # Record that no reference tests were available to ground correctness.
            cs["verifier_passed"] = None
            cs["verifier_status"] = "NO_GROUND_TRUTH"
            cs["verifier_detail"] = "ground_truth test suite is missing or non-string for a verifiable task."
            continue

        # Execute the candidate against the reference tests in an isolated subprocess and classify.
        verdict: Dict[str, Any] = _run_unit_test_payload(
            candidate_code=code_str,        # the candidate's solution definitions
            test_suite_code=test_suite,     # the self-executing reference test suite
            timeout_s=timeout_s,            # the 5 s wall-clock limit
        )

        # Write the verifier verdict to dedicated fields (kept distinct from judge output and transport errors).
        cs["verifier_passed"] = verdict["passed"]   # authoritative test pass/fail correctness signal
        cs["verifier_status"] = verdict["status"]   # outcome label for auditing
        cs["verifier_detail"] = verdict["detail"]   # truncated diagnostic text

    # Return the augmented records so Task 10 can consume the TEST OUTCOME (not an exit code).
    return candidate_solutions

# ------------------------------------------------------------------------------
# Task 9, Orchestrator Function
# ------------------------------------------------------------------------------

def collect_responses(
    qualified_pool: List[str],
    task_record: pd.DataFrame,
    config: Mapping[str, Any]
) -> Tuple[List[Dict[str, Any]], List[Dict[str, Any]]]:
    """
    Orchestrates Task 9: Executes parallel fan-out generation for panel and solver
    roles, captures comprehensive provenance, and runs sandboxed verification.

    Parameters
    ----------
    qualified_pool : List[str]
        The ordered list of qualified model_ids.
    task_record : pd.DataFrame
        The pruned task DataFrame.
    config : Mapping[str, Any]
        The immutable study configuration.

    Returns
    -------
    Tuple[List[Dict[str, Any]], List[Dict[str, Any]]]
        - model_answer_records: The raw free-form generation records.
        - candidate_solution_records: The raw structured solution records.

    Raises
    ------
    RuntimeError
        If the API key environment variable is missing or empty.
    """
    # Resolve the API key securely at runtime
    api_key_env_var: str = config["gateway"]["api_key_env_var"]
    api_key: Optional[str] = os.environ.get(api_key_env_var)
    if not api_key:
        raise RuntimeError(f"Response Collection Failed: Environment variable '{api_key_env_var}' is missing.")

    # Step 1 & 2: Execute parallel fan-out and capture provenance
    model_answers, candidate_solutions = generate_responses_parallel(
        qualified_pool=qualified_pool,
        task_record=task_record,
        config=config,
        api_key=api_key
    )

    # Step 3: Execute sandboxed verification for structured solutions
    sandbox_config = config["architecture"]["sandbox"]
    candidate_solutions = execute_sandboxed_verification(
        candidate_solutions=candidate_solutions,
        task_record=task_record,
        sandbox_config=sandbox_config
    )

    # Return the two lists of raw records
    return model_answers, candidate_solutions
````

````python
# Task 10 — Scoring, Adjudication & Inter-Rater Reliability (Stage 2) (score_and_adjudicate)

# ==============================================================================
# Task 10: Scoring, Adjudication & Inter-Rater Reliability (Stage 2)
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 10, Helper: Taxonomy Crosswalk
# ------------------------------------------------------------------------------

def _apply_taxonomy_crosswalk(rubric_category: Optional[str]) -> Dict[str, int]:
    """
    Map the eight-category adjudication rubric (Appendix C.2) to the eight binding
    outcome-tensor error columns (§9.2) via a fixed canonical BIJECTION.

    Rationale (§4.2.1; §3.3; §9.2; §11.5; Appendix C.2). The source's precision
    point states that the eight `e_*` columns are the binding machine-level
    contract and that a faithful replication must fix ONE canonical
    category-to-column map applied consistently across adjudication and covariance
    estimation. The prior implementation used a many-to-one map onto only five
    columns, so `e_ctx`, `e_plan`, `e_tool` were never set — collapsing three
    covariance blocks (Eq. 10) to identically zero and rendering the taxonomy
    effectively five-dimensional. This version installs a bijection so all eight
    columns are reachable, honoring the documented binding `e_refusal` ==
    refusal/incompleteness; it also removes the silent `unrecognized -> e_reason`
    default and instead raises on out-of-vocabulary labels (a judge-protocol
    violation), preventing silent contamination of the reasoning block.

    Canonical bijection (rubric label -> tensor column):
        reasoning      -> e_reason     (logical/inferential failure)
        arithmetic     -> e_comp       (numeric/computational failure)
        hallucination  -> e_halluc     (fabrication)
        factual        -> e_ctx        (failure to ground in known/contextual fact)
        code           -> e_tool       (code-execution / tool-use failure)
        incompleteness -> e_refusal    (refusal/incompleteness; documented binding)
        instruction    -> e_plan       (instruction-following/format-decomposition failure)
        safety         -> e_safety     (safety violation)

    Parameters
    ----------
    rubric_category : Optional[str]
        The single error-category string from the judge JSON, or None when the
        response is adjudicated correct. Casing and hyphen/underscore variants are
        normalized (e.g., 'Instruction-Following' -> 'instruction').

    Returns
    -------
    Dict[str, int]
        The eight `e_*` flags (each 0/1). All flags are 0 when `rubric_category`
        is None; otherwise exactly one flag is 1 (exclusivity invariant).

    Raises
    ------
    TypeError
        If `rubric_category` is neither None nor a string.
    ValueError
        If `rubric_category` is a string outside the eight-category vocabulary
        (a judge-protocol violation that must be surfaced, not absorbed).
    """
    # Initialize all eight error flags to 0 (the correct-response / pre-assignment state).
    flags: Dict[str, int] = {
        "e_reason": 0, "e_comp": 0, "e_halluc": 0, "e_ctx": 0,
        "e_plan": 0, "e_tool": 0, "e_safety": 0, "e_refusal": 0,
    }

    # A correct response carries no error category; return the all-zero vector (exclusivity holds).
    if rubric_category is None:
        # No incorrect-response error column is set.
        return flags

    # Enforce the input contract: a non-None category must be a string.
    if not isinstance(rubric_category, str):
        # Surface a malformed category type rather than silently mishandling it.
        raise TypeError(f"`rubric_category` must be None or str; got {type(rubric_category).__name__}.")

    # Normalize casing and unify hyphen/underscore variants to a canonical token.
    # e.g., 'Instruction-Following' / 'instruction_following' -> 'instruction'.
    normalized: str = rubric_category.strip().lower().replace("-", "_")
    # Map known multi-word variants onto their canonical single-token rubric label.
    variant_aliases: Dict[str, str] = {
        "instruction_following": "instruction",  # §11.5 prose uses 'instruction-following'
        "instructions": "instruction",            # tolerate pluralization
        "arithmetic_error": "arithmetic",         # tolerate '<cat>_error' phrasing
    }
    # Apply the alias map if the normalized token matches a known variant; otherwise keep it.
    canonical_label: str = variant_aliases.get(normalized, normalized)

    # The fixed canonical bijection from the eight rubric labels to the eight tensor columns.
    crosswalk: Dict[str, str] = {
        "reasoning": "e_reason",       # logical/inferential failure
        "arithmetic": "e_comp",        # numeric/computational failure
        "hallucination": "e_halluc",   # fabrication
        "factual": "e_ctx",            # failure to ground in known/contextual fact
        "code": "e_tool",              # code-execution / tool-use failure
        "incompleteness": "e_refusal", # refusal/incompleteness (documented §9.2 binding)
        "instruction": "e_plan",       # instruction-following / format-decomposition failure
        "safety": "e_safety",          # safety violation
    }

    # An out-of-vocabulary label is a judge-protocol violation: surface it, do not absorb it.
    if canonical_label not in crosswalk:
        # Raising prevents silently dumping unknown errors into e_reason and biasing its covariance block.
        raise ValueError(
            f"Unrecognized adjudication category '{rubric_category}'. "
            f"Expected one of {sorted(crosswalk.keys())} (Appendix C.2)."
        )

    # Resolve the single target column for this category and set its flag to 1 (exclusivity preserved).
    target_col: str = crosswalk[canonical_label]
    # Mark exactly one error column as active for this incorrect response.
    flags[target_col] = 1

    # Return the eight-flag exclusivity-respecting error vector.
    return flags

def _extract_final_answer(text: str) -> str:
    """
    Extracts the content within <final_answer>...</final_answer> tags.
    Falls back to the full text if tags are missing.
    """
    if not isinstance(text, str):
        return ""
    match = re.search(r"<final_answer>(.*?)</final_answer>", text, re.DOTALL | re.IGNORECASE)
    if match:
        return match.group(1).strip()
    return text.strip()

# --------------------------------------------------------------------------------
# Task 10, Step 1: Route items by scoring rule and execute deterministic grounding
# --------------------------------------------------------------------------------

def execute_deterministic_scoring(
    records: List[Dict[str, Any]],
    task_record: pd.DataFrame
) -> Tuple[List[Dict[str, Any]], List[Dict[str, Any]]]:
    """
    Route generation records by scoring rule and deterministically score the
    verifiable ones, deferring rubric tasks to LLM adjudication.

    Rationale (detailed tasklist Task 10, Step 1). Verifiable tasks are grounded
    deterministically: `exact_match` (normalized equality), `numeric_tol`
    (`math.isclose`, with the tolerance read from `ground_truth` when structured,
    else the pre-registered default), and `unit_tests` (which now consumes the
    authoritative TEST PASS/FAIL signal `verifier_passed` produced by
    `execute_sandboxed_verification`, NOT a subprocess exit code). The eight-way
    error crosswalk lands each failure on a distinct binding tensor column. Token
    counts are propagated so cost is reconstructible/auditable; the authoritative
    cost (tokens x pinned price) is computed/guarded in tensor assembly and at the
    generation stage, neither of which is in this signature.

    Parameters
    ----------
    records : List[Dict[str, Any]]
        Unified generation records (model_answer + candidate_solution); each
        carries 'task_id', 'model_id', 'rep', provenance, and — for code tasks —
        the verifier verdict field 'verifier_passed'.
    task_record : pd.DataFrame
        The task DataFrame supplying 'scoring_rule' and 'ground_truth' per task.

    Returns
    -------
    Tuple[List[Dict[str, Any]], List[Dict[str, Any]]]
        - scored_labels : records scored deterministically (with 'correct' and the
          eight e_* flags).
        - pending_judge : records routed to stochastic adjudication.

    Raises
    ------
    TypeError
        If `records` is not a list or `task_record` is not a DataFrame.
    KeyError
        If `task_record` lacks 'scoring_rule'/'ground_truth'.
    ValueError
        If a record carries an unknown scoring rule.
    """
    # Validate the record container type.
    if not isinstance(records, list):
        # Fail fast on a malformed record container.
        raise TypeError("`records` must be a list of record dicts.")
    # Validate the task table type.
    if not isinstance(task_record, pd.DataFrame):
        # Fail fast on a malformed task table.
        raise TypeError("`task_record` must be a pandas DataFrame.")
    # Validate the required task columns are present.
    for required_col in ("task_id", "scoring_rule", "ground_truth"):
        # Each is required to route and ground correctness.
        if required_col not in task_record.columns:
            raise KeyError(f"`task_record` is missing required column '{required_col}'.")

    # Build a fast per-task lookup of scoring rule and ground truth.
    task_lookup: Dict[Any, Dict[str, Any]] = (
        task_record.set_index("task_id")[["scoring_rule", "ground_truth"]].to_dict(orient="index")
    )

    # Initialize the two output partitions.
    scored_labels: List[Dict[str, Any]] = []
    pending_judge: List[Dict[str, Any]] = []

    # Iterate each generation record and route by its task's scoring rule.
    for rec in records:
        # Resolve the task id and its metadata.
        t_id: Any = rec.get("task_id")
        task_meta: Optional[Dict[str, Any]] = task_lookup.get(t_id)

        # Skip records whose task is absent (should already be caught in Task 3).
        if not task_meta:
            continue

        # Extract the scoring rule and the reference ground truth.
        rule: str = task_meta["scoring_rule"]
        gt: Any = task_meta["ground_truth"]

        # Assemble the base label, propagating provenance AND token counts (for cost auditability).
        label_base: Dict[str, Any] = {
            "task_id": t_id,                                          # composite key part
            "model_id": rec["model_id"],                             # composite key part
            "rep": rec["rep"],                                       # composite key part
            "judge_rep": 0,                                          # primary (non-replicated) record
            "prompt_tokens": rec.get("prompt_tokens", 0),           # billed input tokens (cost telemetry, Eq. 36)
            "completion_tokens": rec.get("completion_tokens", 0),   # billed output tokens (cost telemetry, Eq. 36)
            "cost_usd": rec.get("cost_usd", 0.0),                   # upstream-resolved cost (tokens x pinned price)
            "latency_s": rec.get("latency_s", 0.0),                 # wall-clock latency (L(S), Eq. 3)
            "provider_version": rec.get("provider_version", ""),    # drift provenance
            "system_fingerprint": rec.get("system_fingerprint", ""),# drift provenance
            "seed": rec.get("seed", 0),                             # decoding seed provenance
        }

        # Extract the answer payload (structured 'thesis' if present, else the delimited final answer).
        answer_text: str = rec.get("thesis", _extract_final_answer(rec.get("answer", "")))

        # ----- Deterministic rule: exact match -----
        if rule == "exact_match":
            # Case- and whitespace-normalized string equality.
            is_correct = (str(answer_text).strip().lower() == str(gt).strip().lower())
            # A failed exact-match is recorded under the reasoning column by deterministic convention.
            error_cat = None if is_correct else "reasoning"
            # Map the category to the eight binding tensor columns (bijection).
            flags = _apply_taxonomy_crosswalk(error_cat)
            # Append the scored label with the correctness flag and error vector.
            scored_labels.append({**label_base, "correct": int(is_correct), **flags})

        # ----- Deterministic rule: numeric tolerance -----
        elif rule == "numeric_tol":
            # Resolve the numeric target and tolerances (structured ground truth overrides defaults).
            if isinstance(gt, dict) and "value" in gt:
                # Structured ground truth: read target and per-item tolerances.
                gt_val = float(gt["value"])                         # target numeric value
                rel_tol = float(gt.get("rel_tol", 1e-5))           # per-item relative tolerance
                abs_tol = float(gt.get("abs_tol", 1e-8))           # per-item absolute tolerance
            else:
                # Scalar ground truth: use the pre-registered default tolerances.
                gt_val = None                                       # to be parsed below
                rel_tol = 1e-5                                       # pre-registered relative tolerance
                abs_tol = 1e-8                                       # pre-registered absolute tolerance
            # Parse the model's numeric answer and compare within tolerance.
            try:
                # Strip non-numeric characters to extract the numeral from free text.
                ans_val = float(re.sub(r"[^\d.\-eE]", "", str(answer_text)))
                # Parse the scalar target if not already set from a structured object.
                target = float(gt) if gt_val is None else gt_val
                # math.isclose implements |a-b| <= max(rel_tol*max(|a|,|b|), abs_tol).
                is_correct = math.isclose(ans_val, target, rel_tol=rel_tol, abs_tol=abs_tol)
            except (ValueError, TypeError):
                # An unparseable numeral is scored incorrect.
                is_correct = False
            # A failed numeric task is recorded under the computational column.
            error_cat = None if is_correct else "arithmetic"
            # Map the category to the tensor columns.
            flags = _apply_taxonomy_crosswalk(error_cat)
            # Append the scored label.
            scored_labels.append({**label_base, "correct": int(is_correct), **flags})

        # ----- Deterministic rule: unit tests (consume the verifier's TEST PASS/FAIL) -----
        elif rule == "unit_tests":
            # Correctness is True iff the candidate PASSED the reference test suite (D4 fix).
            is_correct = (rec.get("verifier_passed") is True)
            # A failed code task is recorded under the tool/code-execution column.
            error_cat = None if is_correct else "code"
            # Map the category to the tensor columns ('code' -> e_tool via the bijection).
            flags = _apply_taxonomy_crosswalk(error_cat)
            # Append the scored label.
            scored_labels.append({**label_base, "correct": int(is_correct), **flags})

        # ----- Stochastic rule: route to adjudication -----
        elif rule == "judge_rubric":
            # Defer rubric-graded tasks to the LLM judge (Step 2).
            pending_judge.append(rec)

        # ----- Unknown rule: fail loud -----
        else:
            # An unrecognized scoring rule is a configuration error.
            raise ValueError(f"Unknown scoring rule: {rule}")

    # Return the deterministically scored labels and the records pending adjudication.
    return scored_labels, pending_judge

# ------------------------------------------------------------------------------
# Task 10, Step 2: Execute judge-based adjudication and inter-rater reliability
# ------------------------------------------------------------------------------

def execute_judge_adjudication(
    pending_records: List[Dict[str, Any]],
    task_record: pd.DataFrame,
    config: Mapping[str, Any],
    api_key: str,
    human_labels: Optional[Dict[str, bool]] = None
) -> Tuple[List[Dict[str, Any]], Set[str]]:
    """
    Executes the Judge LLM for open-ended tasks and computes Cohen's Kappa
    if human audit labels are provided.

    Parameters
    ----------
    pending_records : List[Dict[str, Any]]
        Records requiring judge_rubric scoring.
    task_record : pd.DataFrame
        The task DataFrame.
    config : Mapping[str, Any]
        The immutable study configuration.
    api_key : str
        The resolved API key.
    human_labels : Optional[Dict[str, bool]], optional
        A dictionary mapping 'task_id_model_id_rep' to a boolean human label.

    Returns
    -------
    Tuple[List[Dict[str, Any]], Set[str]]
        - judge_scored_labels: The records scored by the LLM.
        - excluded_tasks: A set of task_ids excluded due to low IRR.
    """
    if not pending_records:
        return [], set()

    task_lookup = task_record.set_index("task_id")[["prompt", "ground_truth"]].to_dict(orient="index")

    judge_settings = config["llm_settings"]["judge"]
    judge_model = config["models"]["judge"]
    judge_template = config["prompts"]["judge_system"]

    base_url = config["gateway"]["base_url"]
    endpoint = f"{base_url.rstrip('/')}/chat/completions"

    judge_scored_labels: List[Dict[str, Any]] = []

    def _judge_worker(rec: Dict[str, Any]) -> Dict[str, Any]:
        t_id = rec["task_id"]
        task_meta = task_lookup[t_id]

        answer_text = rec.get("thesis", rec.get("answer", ""))

        # Format the strict judge prompt
        prompt = _safe_prompt_format(
            judge_template,
            {
                "task_prompt": task_meta["prompt"],
                "reference": str(task_meta["ground_truth"]),
                "candidate_answer": answer_text
            }
        )

        payload = {
            "model": judge_model,
            "messages": [{"role": "user", "content": prompt}],
            "temperature": judge_settings["temperature"],
            "max_tokens": judge_settings["max_tokens"],
            "response_format": {"type": "json_object"}
        }

        # Execute with retry, enforcing JSON output
        # Note: _execute_with_retry is assumed imported/available
        res = _execute_with_retry(
            endpoint=endpoint, api_key=api_key, payload=payload,
            max_retries=2, timeout_s=60.0, require_json=True
        )

        is_correct = False
        error_cat = "instruction" # Default fail-safe

        if res["text"] and not res["error"]:
            try:
                parsed = json.loads(res["text"])
                is_correct = bool(parsed.get("is_correct", False))
                error_cat = parsed.get("error_category") if not is_correct else None
            except json.JSONDecodeError:
                pass # Fallback to instruction failure

        flags = _apply_taxonomy_crosswalk(error_cat)

        label_base = {
            "task_id": t_id,
            "model_id": rec["model_id"],
            "rep": rec["rep"],
            "judge_rep": rec.get("judge_rep", 0),
            "cost_usd": rec.get("cost_usd", 0.0),
            "latency_s": rec.get("latency_s", 0.0),
            "provider_version": rec.get("provider_version", ""),
            "system_fingerprint": rec.get("system_fingerprint", ""),
            "seed": rec.get("seed", 0)
        }

        return {**label_base, "correct": int(is_correct), **flags}

    # Execute judge calls in parallel
    with concurrent.futures.ThreadPoolExecutor(max_workers=20) as executor:
        futures = [executor.submit(_judge_worker, rec) for rec in pending_records]
        for future in concurrent.futures.as_completed(futures):
            judge_scored_labels.append(future.result())

    # --- Inter-Rater Reliability (IRR) Check ---
    excluded_tasks: Set[str] = set()
    if human_labels:
        y_judge = []
        y_human = []
        for lbl in judge_scored_labels:
            key = f"{lbl['task_id']}_{lbl['model_id']}_{lbl['rep']}"
            if key in human_labels:
                y_judge.append(lbl["correct"])
                y_human.append(int(human_labels[key]))

        if len(y_judge) > 10: # Ensure sufficient sample size for Kappa
            kappa = cohen_kappa_score(y_human, y_judge)
            kappa_floor = config["stat_tests"]["kappa_floor"]

            if kappa < kappa_floor:
                # If agreement is below floor, exclude all open-ended tasks
                # as the adjudicator is deemed too noisy.
                excluded_tasks = {lbl["task_id"] for lbl in judge_scored_labels}

    return judge_scored_labels, excluded_tasks

# ------------------------------------------------------------------------------
# Task 10, Step 3: Perform replicated judging on the adjudication subsample
# ------------------------------------------------------------------------------

def execute_replicated_judging(
    pending_records: List[Dict[str, Any]],
    task_record: pd.DataFrame,
    config: Mapping[str, Any],
    api_key: str,
    rng: np.random.Generator
) -> List[Dict[str, Any]]:
    """
    Executes the Judge LLM multiple times on a subsample of tasks to expose
    adjudication variance (sigma^2_adj).

    Parameters
    ----------
    pending_records : List[Dict[str, Any]]
        Records requiring judge_rubric scoring.
    task_record : pd.DataFrame
        The task DataFrame.
    config : Mapping[str, Any]
        The immutable study configuration.
    api_key : str
        The resolved API key.
    rng : np.random.Generator
        The dedicated random number generator for deterministic subsampling.

    Returns
    -------
    List[Dict[str, Any]]
        The additional replicated judge records.
    """
    subsample_frac = config["uncertainty"]["replication"]["adjudication_subsample"]
    r_judge = config["uncertainty"]["replication"]["R_judge"]

    if subsample_frac <= 0.0 or r_judge <= 1:
        return []

    # Identify unique tasks requiring judging
    unique_tasks = sorted(list({rec["task_id"] for rec in pending_records}))
    n_sample = max(1, int(len(unique_tasks) * subsample_frac))

    # Deterministically sample tasks using the provided RNG
    sampled_tasks = set(rng.choice(unique_tasks, size=n_sample, replace=False))

    # Filter records belonging to the sampled tasks
    replicate_records = []
    for rec in pending_records:
        if rec["task_id"] in sampled_tasks:
            # Create R_judge - 1 copies (since judge_rep=0 was done in Step 2)
            for j_rep in range(1, r_judge):
                rec_copy = dict(rec)
                rec_copy["judge_rep"] = j_rep
                replicate_records.append(rec_copy)

    # Execute the judge on the replicated records
    # We pass human_labels=None as IRR is only computed on the primary (judge_rep=0) run
    replicated_labels, _ = execute_judge_adjudication(
        pending_records=replicate_records,
        task_record=task_record,
        config=config,
        api_key=api_key,
        human_labels=None
    )

    return replicated_labels

# ------------------------------------------------------------------------------
# Task 10, Orchestrator Function
# ------------------------------------------------------------------------------

def score_and_adjudicate(
    model_answer: List[Dict[str, Any]],
    candidate_solution: List[Dict[str, Any]],
    task_record: pd.DataFrame,
    config: Mapping[str, Any],
    rng_bundle: Dict[str, np.random.Generator],
    human_labels: Optional[Dict[str, bool]] = None
) -> List[Dict[str, Any]]:
    """
    Orchestrates Task 10: Routes records to deterministic or stochastic scoring,
    computes inter-rater reliability, and performs replicated judging.

    Parameters
    ----------
    model_answer : List[Dict[str, Any]]
        The raw free-form generation records.
    candidate_solution : List[Dict[str, Any]]
        The raw structured solution records.
    task_record : pd.DataFrame
        The task DataFrame.
    config : Mapping[str, Any]
        The immutable study configuration.
    rng_bundle : Dict[str, np.random.Generator]
        The bundle of independent random number generators.
    human_labels : Optional[Dict[str, bool]], optional
        Optional human audit labels for IRR computation.

    Returns
    -------
    List[Dict[str, Any]]
        The unified list of scored labels ready for tensor assembly.

    Raises
    ------
    RuntimeError
        If the API key environment variable is missing or empty.
    """
    api_key_env_var: str = config["gateway"]["api_key_env_var"]
    api_key: Optional[str] = os.environ.get(api_key_env_var)
    if not api_key:
        raise RuntimeError(f"Adjudication Failed: Environment variable '{api_key_env_var}' is missing.")

    # Combine the free-form and structured records into a single stream
    unified_records = model_answer + candidate_solution

    # Step 1: Route and execute deterministic scoring
    scored_labels, pending_judge = execute_deterministic_scoring(
        records=unified_records,
        task_record=task_record
    )

    # Step 2: Execute judge-based adjudication and IRR
    judge_labels, excluded_tasks = execute_judge_adjudication(
        pending_records=pending_judge,
        task_record=task_record,
        config=config,
        api_key=api_key,
        human_labels=human_labels
    )

    # Filter out tasks that failed the IRR floor
    if excluded_tasks:
        judge_labels = [lbl for lbl in judge_labels if lbl["task_id"] not in excluded_tasks]

    # Step 3: Execute replicated judging for variance decomposition
    # Use the global seed for deterministic subsampling
    global_rng = rng_bundle["global_seed"]
    replicated_labels = execute_replicated_judging(
        pending_records=[rec for rec in pending_judge if rec["task_id"] not in excluded_tasks],
        task_record=task_record,
        config=config,
        api_key=api_key,
        rng=global_rng
    )

    # Combine all scored labels
    final_labels = scored_labels + judge_labels + replicated_labels

    return final_labels
````

````python
# Task 11 — Outcome Tensor Assembly, Canonicalization & Hash (build_outcome_tensor)

# ==============================================================================
# Task 11: Outcome Tensor Assembly, Canonicalization & Hash
# ==============================================================================

# -----------------------------------------------------------------------------------
# Task 11, Step 1: Assemble the outcome tensor and enforce the exclusivity invariant
# -----------------------------------------------------------------------------------

def _compute_record_cost_usd(
    prompt_tokens: float,
    completion_tokens: float,
    price_in_per_token: float,
    price_out_per_token: float
) -> float:
    """
    Compute the per-record monetary cost from token counts and a pinned price table.

    Rationale (Eq. 36). cost_usd = prompt_tokens * price_in + completion_tokens *
    price_out, evaluated against a PINNED per-model price snapshot. This kernel is
    the single source of truth for the cost axis; it is used at assembly when
    records carry resolved per-token prices and at the generation stage where the
    roster price table is available.

    Parameters
    ----------
    prompt_tokens : float
        Billed input tokens for the call.
    completion_tokens : float
        Billed output tokens for the call.
    price_in_per_token : float
        Pinned USD price per input token for the model.
    price_out_per_token : float
        Pinned USD price per output token for the model.

    Returns
    -------
    float
        The computed cost in USD (non-negative).

    Raises
    ------
    ValueError
        If any input is negative or non-finite.
    """
    # Validate that all four inputs are finite and non-negative (cost is a non-negative quantity).
    for name, value in (
        ("prompt_tokens", prompt_tokens), ("completion_tokens", completion_tokens),
        ("price_in_per_token", price_in_per_token), ("price_out_per_token", price_out_per_token),
    ):
        # Reject NaN/inf and negatives, which would corrupt C(S).
        if not np.isfinite(value) or value < 0:
            raise ValueError(f"Cost input '{name}' must be finite and non-negative; got {value!r}.")
    # Apply Eq. 36: input tokens priced at p_in plus output tokens priced at p_out.
    return float(prompt_tokens) * float(price_in_per_token) + float(completion_tokens) * float(price_out_per_token)


def assemble_and_validate_tensor(
    scored_labels: List[Dict[str, Any]]
) -> Tuple[pd.DataFrame, List[Dict[str, Any]]]:
    """
    Assemble the canonical 17-column outcome tensor from scored labels, computing a
    real cost axis and enforcing strict dtypes, key uniqueness, and the error
    exclusivity invariant.

    Rationale (Eqs. 3, 36; detailed tasklist Task 11, Step 1). The cost term C(S)
    (Eq. 3), QAC, and the cost budget all require a real per-row cost_usd =
    prompt_tokens*price_in + completion_tokens*price_out (Eq. 36). The prior
    implementation blindly zero-filled cost_usd, nullifying the cost axis. This
    version computes cost via the kernel when records carry resolved per-token
    prices, otherwise uses the upstream-resolved cost_usd, ZERO-FILLS ONLY genuinely
    missing telemetry (counted and logged), and FAILS CLOSED if the entire cost
    column is zero — refusing to ship a dead cost axis. The canonical schema, strict
    dtypes, composite-key uniqueness, and the exclusivity invariant Σ e_* = 1 -
    correct are preserved exactly.

    Parameters
    ----------
    scored_labels : List[Dict[str, Any]]
        The unified scored generation records from Task 10 (carrying correctness,
        the eight e_* flags, provenance, token counts, and possibly resolved
        per-token prices 'price_in'/'price_out').

    Returns
    -------
    Tuple[pd.DataFrame, List[Dict[str, Any]]]
        - primary_tensor : the strictly typed 17-column DataFrame (judge_rep == 0).
        - judge_replicates : records with judge_rep > 0.

    Raises
    ------
    RuntimeError
        If the composite key is not unique, the exclusivity invariant fails, or the
        cost column is identically zero (degenerate cost axis — pin a price table).
    """
    # Separate the primary (judge_rep == 0) records from replicated judge records.
    primary_records: List[Dict[str, Any]] = [r for r in scored_labels if r.get("judge_rep", 0) == 0]
    # Collect the adjudication replicates (judge_rep > 0) for the variance decomposition.
    judge_replicates: List[Dict[str, Any]] = [r for r in scored_labels if r.get("judge_rep", 0) > 0]

    # Materialize the primary records into a DataFrame.
    df = pd.DataFrame(primary_records)

    # Guard against an empty assembly (no primary records is a fatal upstream failure).
    if df.empty:
        raise RuntimeError("Tensor Assembly Failed: no primary (judge_rep == 0) records to assemble.")

    # -------------------------------------------------------------------------
    # Cost axis (Eq. 36): compute from tokens x pinned price when available.
    # -------------------------------------------------------------------------
    # Determine whether the records carry resolved per-token prices for a kernel computation.
    has_price_fields = {"prompt_tokens", "completion_tokens", "price_in", "price_out"}.issubset(df.columns)
    # Track how many rows had genuinely missing cost telemetry (for honest logging).
    missing_cost_rows: int = 0

    if has_price_fields:
        # Vectorized cost: prompt_tokens*price_in + completion_tokens*price_out (Eq. 36).
        computed_cost = (
            df["prompt_tokens"].astype("float64").fillna(0.0) * df["price_in"].astype("float64").fillna(0.0)
            + df["completion_tokens"].astype("float64").fillna(0.0) * df["price_out"].astype("float64").fillna(0.0)
        )
        # Where a positive upstream cost_usd already exists, prefer it; else use the kernel result.
        if "cost_usd" in df.columns:
            # Use the upstream cost where present and positive, otherwise the kernel-computed value.
            upstream = df["cost_usd"].astype("float64")
            df["cost_usd"] = upstream.where(upstream.fillna(0.0) > 0.0, computed_cost)
        else:
            # No upstream cost column: adopt the kernel-computed cost.
            df["cost_usd"] = computed_cost
    else:
        # No price fields: rely on the upstream-resolved cost_usd (computed at the generation stage).
        if "cost_usd" not in df.columns:
            # Create the column as missing telemetry to be flagged below.
            df["cost_usd"] = np.nan
        # Count rows lacking a usable cost so the zero-fill is auditable, not silent.
        missing_cost_rows = int(df["cost_usd"].isna().sum() + (df["cost_usd"].fillna(0.0) <= 0.0).sum())

    # Ensure latency exists (member-max latency feeds L(S), Eq. 3); missing telemetry -> 0.0.
    if "latency_s" not in df.columns:
        df["latency_s"] = 0.0
    # Ensure provenance string columns exist.
    for col in ("provider_version", "system_fingerprint"):
        if col not in df.columns:
            df[col] = ""

    # Define the canonical 17-column byte-level release schema (§9.2d); token columns are excluded.
    canonical_columns: List[str] = [
        "task_id", "model_id", "rep", "correct",
        "e_reason", "e_comp", "e_halluc", "e_ctx",
        "e_plan", "e_tool", "e_safety", "e_refusal",
        "cost_usd", "latency_s", "provider_version",
        "system_fingerprint", "seed",
    ]
    # Project to the canonical columns in canonical order (tokens/prices used only to derive cost).
    df = df[canonical_columns].copy()

    # -------------------------------------------------------------------------
    # Strict dtype coercion for memory efficiency and mathematical correctness.
    # -------------------------------------------------------------------------
    # Categorical keys.
    df["task_id"] = df["task_id"].astype("category")
    df["model_id"] = df["model_id"].astype("category")
    # Replicate index and correctness as small integers.
    df["rep"] = df["rep"].astype("int16")
    df["correct"] = df["correct"].astype("int8")
    # The eight binding error columns as int8.
    error_cols: List[str] = ["e_reason", "e_comp", "e_halluc", "e_ctx", "e_plan", "e_tool", "e_safety", "e_refusal"]
    for col in error_cols:
        df[col] = df[col].astype("int8")
    # Cost and latency as float64; zero-fill ONLY genuinely missing telemetry (already counted).
    df["cost_usd"] = df["cost_usd"].fillna(0.0).astype("float64")
    df["latency_s"] = df["latency_s"].fillna(0.0).astype("float64")
    # Provenance columns as category, fingerprint nulls as empty strings.
    df["provider_version"] = df["provider_version"].fillna("").astype("category")
    df["system_fingerprint"] = df["system_fingerprint"].fillna("").astype("category")
    # Seed as a 64-bit integer.
    df["seed"] = df["seed"].astype("int64")

    # Emit an honest log if any rows had missing cost telemetry that was zero-filled.
    if missing_cost_rows > 0:
        logger.warning("Tensor assembly: %d row(s) had missing/zero cost telemetry and were zero-filled.", missing_cost_rows)

    # -------------------------------------------------------------------------
    # D2 integrity guard: refuse to ship a dead (identically zero) cost axis.
    # -------------------------------------------------------------------------
    # If every cost is non-positive, C(S), QAC, and the cost budget are all inert -> fail closed.
    if not (df["cost_usd"] > 0.0).any():
        raise RuntimeError(
            "Tensor Assembly Failed: cost_usd is identically zero across all rows. "
            "The cost axis (C(S), Eq. 3; QAC; cost budget) is nullified. Pin a numeric per-model "
            "price table (cost_per_token) and compute cost = tokens x price (Eq. 36) at the generation "
            "stage or supply 'price_in'/'price_out' on the records."
        )

    # -------------------------------------------------------------------------
    # Composite-key uniqueness.
    # -------------------------------------------------------------------------
    # Each (task_id, model_id, rep) must be unique.
    if df.duplicated(subset=["task_id", "model_id", "rep"]).any():
        raise RuntimeError("Tensor Assembly Failed: composite key (task_id, model_id, rep) is not unique.")

    # -------------------------------------------------------------------------
    # Exclusivity invariant (vectorized): Σ e_* == 1 - correct.
    # -------------------------------------------------------------------------
    # Row-wise sum of the eight error flags.
    error_sums = df[error_cols].sum(axis=1)
    # Expected per-row error count: 0 if correct, exactly 1 if incorrect.
    expected_sums = 1 - df["correct"]
    # Boolean mask of rows satisfying the invariant.
    invariant_mask = (error_sums == expected_sums)
    # Any violation halts assembly with a diagnostic sample.
    if not invariant_mask.all():
        violators = df[~invariant_mask][["task_id", "model_id", "correct"] + error_cols]
        raise RuntimeError(
            f"Tensor Assembly Failed: exclusivity invariant violated on {len(violators)} row(s).\n"
            f"Sample violators:\n{violators.head()}"
        )

    # Return the validated primary tensor and the judge replicates.
    return df, judge_replicates

# ------------------------------------------------------------------------------
# Task 11, Step 2: Canonicalize the tensor and compute the SHA-256 hash
# ------------------------------------------------------------------------------

def canonicalize_and_hash_tensor(df: pd.DataFrame) -> Tuple[pd.DataFrame, str]:
    """
    Stable-sorts the tensor, fixes float precision, and computes a strict
    byte-level SHA-256 hash to guarantee Tier-1 reproducibility.

    Parameters
    ----------
    df : pd.DataFrame
        The validated primary outcome tensor.

    Returns
    -------
    Tuple[pd.DataFrame, str]
        - canonical_df: The sorted and re-indexed DataFrame.
        - tensor_sha256: The cryptographic hash of the canonical bytes.
    """
    # 1. Fill NaN/None in string provenance columns with empty strings
    # This prevents serialization differences between null types across platforms
    canonical_df = df.copy()
    for col in ["provider_version", "system_fingerprint"]:
        # Convert to object temporarily to fill, then back to category
        canonical_df[col] = canonical_df[col].astype(object).fillna("").astype("category")

    # 2. Stable sort using mergesort to guarantee deterministic ordering
    sort_keys = ["task_id", "model_id", "rep"]
    canonical_df = canonical_df.sort_values(by=sort_keys, kind="mergesort")

    # 3. Reset the index so the pandas internal index does not leak into the hash
    canonical_df = canonical_df.reset_index(drop=True)

    # 4. Serialize to JSON with fixed double precision
    # orient='records' creates a list of dictionaries
    json_str = canonical_df.to_json(orient="records", double_precision=15)

    # 5. Parse and re-serialize with strict key sorting
    # pandas to_json does not guarantee key order within the dictionaries;
    # json.dumps with sort_keys=True enforces strict byte-level canonicalization.
    parsed_records = json.loads(json_str)
    canonical_json_bytes = json.dumps(
        parsed_records,
        sort_keys=True,
        separators=(',', ':')
    ).encode('utf-8')

    # 6. Compute the SHA-256 hash
    tensor_sha256 = hashlib.sha256(canonical_json_bytes).hexdigest()

    return canonical_df, tensor_sha256

# ------------------------------------------------------------------------------
# Task 11, Step 3: Freeze the tensor and define the estimator projection
# ------------------------------------------------------------------------------

def create_estimator_projection(
    canonical_df: pd.DataFrame,
    judge_replicates: List[Dict[str, Any]]
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Creates the 13-column estimator projection and formats the judge replicates.

    Parameters
    ----------
    canonical_df : pd.DataFrame
        The canonicalized primary outcome tensor.
    judge_replicates : List[Dict[str, Any]]
        The list of records where judge_rep > 0.

    Returns
    -------
    Tuple[pd.DataFrame, pd.DataFrame]
        - estimator_projection: The 13-column view for downstream math functions.
        - judge_replicates_df: The formatted DataFrame of judge replicates.
    """
    # 1. Define the 13-column estimator projection
    # This omits 'rep' and the provenance triple, providing only the analytic core
    projection_columns = [
        "task_id", "model_id", "correct",
        "e_reason", "e_comp", "e_halluc", "e_ctx",
        "e_plan", "e_tool", "e_safety", "e_refusal",
        "cost_usd", "latency_s"
    ]

    # Create a deep copy to ensure downstream estimators cannot mutate the primary artifact
    estimator_projection = canonical_df[projection_columns].copy(deep=True)

    # 2. Format the judge replicates into a DataFrame
    if judge_replicates:
        judge_replicates_df = pd.DataFrame(judge_replicates)
        # Apply the same basic dtype coercions for consistency
        judge_replicates_df["task_id"] = judge_replicates_df["task_id"].astype("category")
        judge_replicates_df["model_id"] = judge_replicates_df["model_id"].astype("category")
        judge_replicates_df["correct"] = judge_replicates_df["correct"].astype("int8")
    else:
        # Return an empty DataFrame with the correct columns if no replicates exist
        judge_replicates_df = pd.DataFrame(columns=canonical_df.columns.tolist() + ["judge_rep"])

    return estimator_projection, judge_replicates_df

# ------------------------------------------------------------------------------
# Task 11, Orchestrator Function
# ------------------------------------------------------------------------------

def build_outcome_tensor(
    scored_labels: List[Dict[str, Any]],
    config: Mapping[str, Any]
) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame, str]:
    """
    Orchestrates Task 11: Assembles the outcome tensor, enforces mathematical
    invariants, computes the canonical SHA-256 hash, and enforces the Tier-1 gate.

    Parameters
    ----------
    scored_labels : List[Dict[str, Any]]
        The unified list of scored generation records.
    config : Mapping[str, Any]
        The immutable study configuration.

    Returns
    -------
    Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame, str]
        - primary_tensor: The 17-column canonical released artifact.
        - estimator_projection: The 13-column analytic core.
        - judge_replicates_df: The DataFrame of adjudication replicates.
        - tensor_sha256: The cryptographic hash of the primary tensor.

    Raises
    ------
    RuntimeError
        If the computed hash does not match the configured hash (Tier-1 violation).
    """
    # Step 1: Assemble the tensor and enforce the exclusivity invariant
    raw_tensor, judge_replicates = assemble_and_validate_tensor(scored_labels)

    # Step 2: Canonicalize the tensor and compute the SHA-256 hash
    primary_tensor, computed_hash = canonicalize_and_hash_tensor(raw_tensor)

    # Step 3: Create the estimator projection and format judge replicates
    estimator_projection, judge_replicates_df = create_estimator_projection(
        canonical_df=primary_tensor,
        judge_replicates=judge_replicates
    )

    # --- Enforce the Tier-1 Reproducibility Gate ---
    expected_hash: Optional[str] = config["reproducibility"].get("tensor_sha256")

    if expected_hash is not None:
        # Verify Mode: The run must exactly reproduce the published hash
        if computed_hash != expected_hash:
            raise RuntimeError(
                f"Tier-1 Reproducibility Violation: Computed tensor hash does not match the configuration.\n"
                f"Expected: {expected_hash}\n"
                f"Computed: {computed_hash}\n"
                f"This indicates silent drift in the data-generating substrate."
            )
    else:
        pass

    return primary_tensor, estimator_projection, judge_replicates_df, computed_hash
````

````python
# Task 12 — Coalition Utility Function (utility)

# ==============================================================================
# Task 12: Coalition Utility Function
# ==============================================================================

# ------------------------------------------------------------------------------
# Task 12, Helper: Pre-computation Factory
# ------------------------------------------------------------------------------

def build_utility_function(
    estimator_projection: pd.DataFrame,
    sigma: np.ndarray,
    config: Mapping[str, Any],
    qualified_pool: List[str],
    judge_model_id: str
) -> Callable[[List[int]], float]:
    """
    Factory function that pre-computes the dense numerical matrices required for
    high-performance utility evaluation and returns a closure v(S) optimized for
    the Monte Carlo hot loop.

    Parameters
    ----------
    estimator_projection : pd.DataFrame
        The 13-column analytic core tensor from Task 11.
    sigma : np.ndarray
        The n x n error covariance matrix from Task 15.
    config : Mapping[str, Any]
        The immutable study configuration containing utility weights.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    judge_model_id : str
        The identifier for the judge model to extract its cost and latency.

    Returns
    -------
    Callable[[List[int]], float]
        A pure function v(S) that accepts a list of integer model indices and
        returns the scalar utility of that coalition.
    """
    # 1. Extract utility preference weights (Eq. 3)
    weights = config["estimation"]["utility_weights"]
    alpha: float = weights["alpha"]
    beta: float = weights["beta"]
    gamma: float = weights["gamma"]
    delta: float = weights["delta"]

    n_models = len(qualified_pool)

    # 2. Pre-compute the dense correctness matrix A (tasks x models)
    # The protocol specifies that the binary collapse (majority vote across replicates)
    # happens BEFORE the coalition utility is evaluated.

    # Group by task and model, compute the mean correctness across replicates
    mean_correctness = estimator_projection.groupby(["task_id", "model_id"], observed=True)["correct"].mean()

    # Unstack to create a dense matrix: rows are tasks, columns are models
    # Reindex columns to strictly enforce the canonical model ordering
    A_df = mean_correctness.unstack(level="model_id").reindex(columns=qualified_pool)

    # Threshold at > 0.5 to resolve the majority vote per (task, model) pair
    # Convert to a contiguous C-ordered NumPy array for maximum cache efficiency
    A_matrix: np.ndarray = (A_df.values > 0.5).astype(np.int8)

    # 3. Pre-compute the reliability vector r (mean correctness across all tasks)
    # Used for deterministic tie-breaking in even-sized coalitions
    r_vec: np.ndarray = A_matrix.mean(axis=0)

    # 4. Pre-compute cost and latency vectors
    # Extract the mean cost and latency per model across all tasks
    cost_series = estimator_projection.groupby("model_id", observed=True)["cost_usd"].mean()
    latency_series = estimator_projection.groupby("model_id", observed=True)["latency_s"].mean()

    c_vec: np.ndarray = cost_series.reindex(qualified_pool).fillna(0.0).values.astype(np.float64)
    l_vec: np.ndarray = latency_series.reindex(qualified_pool).fillna(0.0).values.astype(np.float64)

    # 5. Extract Judge cost and latency
    # If the judge is in the pool, use its values; otherwise, extract from the tensor if present, else default to 0
    if judge_model_id in cost_series.index:
        c_j: float = float(cost_series[judge_model_id])
        l_j: float = float(latency_series[judge_model_id])
    else:
        c_j = 0.0
        l_j = 0.0

    # 6. Pre-compute normalization constants to ensure commensurability [0,1]
    # Maximum possible cost: sum of all models + judge
    c_max_possible: float = np.sum(c_vec) + c_j
    if c_max_possible == 0.0: c_max_possible = 1.0 # Prevent division by zero

    # Maximum possible latency: max of any single model + judge
    l_max_possible: float = np.max(l_vec) + l_j if len(l_vec) > 0 else 1.0
    if l_max_possible == 0.0: l_max_possible = 1.0

    # Maximum possible risk: the maximum variance (diagonal element) of any single model
    # Since w^T Sigma w for a single model is just its variance
    r_max_possible: float = np.max(np.diag(sigma)) if sigma.size > 0 else 1.0
    if r_max_possible == 0.0: r_max_possible = 1.0

    # --------------------------------------------------------------------------
    # Task 12, Steps 1-3: The Closure v(S)
    # --------------------------------------------------------------------------
    def v(S: List[int]) -> float:
        """
        Evaluates the utility of coalition S.

        Parameters
        ----------
        S : List[int]
            List of integer indices representing the models in the coalition.

        Returns
        -------
        float
            The scalar utility v(S).
        """
        s_size = len(S)
        if s_size == 0:
            return 0.0

        # --- Step 1: Simulated-Fusion Quality (Eq. 3a) ---
        # Slice the dense matrix for the models in S
        A_S = A_matrix[:, S]

        # Compute the row-wise sum (number of correct models per task)
        row_sums = np.sum(A_S, axis=1)

        # Strict majority threshold
        threshold = s_size / 2.0
        majority_mask = row_sums > threshold

        # Tie-breaking for even-sized coalitions
        if s_size % 2 == 0:
            # Identify rows where the vote is exactly tied
            tie_mask = (row_sums == threshold)
            if np.any(tie_mask):
                # Find the index (within S) of the most reliable model
                best_model_idx_in_S = np.argmax(r_vec[S])
                # For tied rows, the coalition is correct iff the most reliable model is correct
                majority_mask[tie_mask] = A_S[tie_mask, best_model_idx_in_S].astype(bool)

        # Expected quality is the mean of the resolved correctness mask
        q_val = np.mean(majority_mask)

        # --- Step 2: Cost, Latency, and Risk ---
        # Cost is additive over members plus the judge
        c_val = np.sum(c_vec[S]) + c_j

        # Latency is the member-max plus the judge (parallel fan-out assumption)
        l_val = np.max(l_vec[S]) + l_j

        # Risk is the quadratic form w^T Sigma w
        # Construct the uniform weight vector for the set S
        w = np.zeros(n_models, dtype=np.float64)
        w[S] = 1.0 / s_size

        # Compute w^T Sigma w efficiently
        r_val = np.dot(w, np.dot(sigma, w))

        # --- Step 3: Normalization and Composition (Eq. 3) ---
        # Normalize penalties to [0,1] to ensure commensurability with Quality
        c_norm = c_val / c_max_possible
        l_norm = l_val / l_max_possible
        r_norm = r_val / r_max_possible

        # Compose the final utility
        utility = (alpha * q_val) - (beta * r_norm) - (gamma * c_norm) - (delta * l_norm)

        return float(utility)

    return v

# ------------------------------------------------------------------------------
# Task 12, Orchestrator Function
# ------------------------------------------------------------------------------

def utility(
    S: List[int],
    estimator_projection: pd.DataFrame,
    sigma: np.ndarray,
    config: Mapping[str, Any],
    qualified_pool: List[str]
) -> float:
    """
    Orchestrates Task 12: Evaluates the utility of a single coalition S.

    Note: For high-performance Monte Carlo loops, it is strongly recommended to
    use `build_utility_function` directly to avoid redundant matrix pre-computation
    on every call. This function is provided for API completeness and single-shot
    evaluations.

    Parameters
    ----------
    S : List[int]
        List of integer indices representing the models in the coalition.
    estimator_projection : pd.DataFrame
        The 13-column analytic core tensor.
    sigma : np.ndarray
        The n x n error covariance matrix.
    config : Mapping[str, Any]
        The immutable study configuration.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.

    Returns
    -------
    float
        The scalar utility v(S).
    """
    # Extract the judge model ID from the configuration
    judge_model_id: str = config["models"]["judge"]

    # Build the optimized closure (incurs pre-computation overhead)
    v_func = build_utility_function(
        estimator_projection=estimator_projection,
        sigma=sigma,
        config=config,
        qualified_pool=qualified_pool,
        judge_model_id=judge_model_id
    )

    # Evaluate and return the utility for the specific coalition
    return v_func(S)
````

````python
# Task 13 — Marginal Ensemble Contribution via Permutation Shapley (Algorithm 1, Stage A) (estimate_mec)

# =======================================================================================
# Task 13: Marginal Ensemble Contribution via Permutation Shapley (Algorithm 1, Stage A)
# =======================================================================================

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 1: Determine the Monte Carlo budget K and generate permutations
# -------------------------------------------------------------------------------------------------------------------------------

def determine_k_and_generate_permutations(
    n_models: int,
    config: Mapping[str, Any],
    rng: np.random.Generator,
    epsilon: Optional[float] = None,
    delta: Optional[float] = None
) -> Tuple[int, np.ndarray]:
    """
    Determines the Monte Carlo budget K (using the Hoeffding bound if precision
    parameters are provided) and generates K uniform permutations of the model indices.

    Equation 21: K >= ((b-a)^2 / 2*epsilon^2) * ln(2/delta)
    Equation 6a: P^{(k)} ~ Unif(S_n)

    Parameters
    ----------
    n_models : int
        The number of qualified models in the pool (n).
    config : Mapping[str, Any]
        The immutable study configuration containing Monte Carlo settings.
    rng : np.random.Generator
        The dedicated random number generator for Monte Carlo sampling.
    epsilon : Optional[float], optional
        The desired precision tolerance for the Hoeffding bound.
    delta : Optional[float], optional
        The failure probability for the Hoeffding bound.

    Returns
    -------
    Tuple[int, np.ndarray]
        - K: The final integer number of permutations to evaluate.
        - permutations: A 2D NumPy array of shape (K, n_models) containing the integer indices.

    Raises
    ------
    ValueError
        If the configured coalition sampler is not 'permutation'.
    """
    # Extract Monte Carlo configuration
    mc_config = config["estimation"]["monte_carlo"]
    sampler_type: str = mc_config["coalition_sampler"]
    k_config: int = mc_config["K"]

    # Validate that the sampler is the unbiased permutation Shapley estimator
    if sampler_type != "permutation":
        raise ValueError(
            f"Unsupported coalition_sampler: '{sampler_type}'. "
            f"Task 13 strictly implements the 'permutation' sampler (Eq. 6a)."
        )

    # Calculate K based on the Hoeffding bound if precision parameters are provided
    k_final: int = k_config
    if epsilon is not None and delta is not None:
        # The utility function v(S) is normalized to [0, 1], so the range (b-a) = 1.0
        range_sq = 1.0
        # Compute the Hoeffding sample complexity bound (Eq. 21)
        k_bound = math.ceil((range_sq / (2.0 * (epsilon ** 2))) * math.log(2.0 / delta))
        # Use the maximum of the configured K and the theoretically required K
        k_final = max(k_config, k_bound)

    # Initialize an empty array to hold the permutations
    # Shape: (K, n), dtype: int32 (sufficient for n <= 20 or even much larger)
    permutations = np.empty((k_final, n_models), dtype=np.int32)

    # Generate K independent, uniformly distributed permutations
    # We use the dedicated RNG to ensure Tier-1 bit-exact reproducibility
    base_indices = np.arange(n_models, dtype=np.int32)
    for k in range(k_final):
        permutations[k, :] = rng.permutation(base_indices)

    return k_final, permutations

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 2: Compute marginal contributions incrementally
# -------------------------------------------------------------------------------------------------------------------------------

def compute_incremental_marginals(
    permutations: np.ndarray,
    v_func: Callable[[List[int]], float],
    qualified_pool: List[str]
) -> Tuple[np.ndarray, List[Dict[str, Any]]]:
    """
    Evaluates the marginal utility contribution of each model across all permutations
    using an optimized incremental evaluation strategy.

    Equation 6a: v(P^{(k)}_{< j} U {M_j}) - v(P^{(k)}_{< j})

    Parameters
    ----------
    permutations : np.ndarray
        The 2D array of shape (K, n) containing the permutation indices.
    v_func : Callable[[List[int]], float]
        The optimized utility closure from Task 12.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.

    Returns
    -------
    Tuple[np.ndarray, List[Dict[str, Any]]]
        - mec_sums: A 1D array of shape (n,) containing the sum of marginals per model.
        - coalition_records: A list of dictionaries recording the exact provenance of each evaluation.
    """
    k_total, n_models = permutations.shape

    # Initialize the accumulation array for the marginal sums
    # dtype float64 is used to prevent precision loss during accumulation
    mec_sums = np.zeros(n_models, dtype=np.float64)

    # Initialize the list to store the provenance records for Tier-1 reproducibility
    coalition_records: List[Dict[str, Any]] = []

    # Iterate over each of the K permutations
    for k in range(k_total):
        # Extract the current permutation sequence
        current_perm = permutations[k, :]

        # Initialize the prefix (coalition S) as empty
        prefix_indices: List[int] = []
        # The utility of the empty set is defined as 0.0
        v_without: float = 0.0

        # Iterate incrementally down the permutation sequence
        for step in range(n_models):
            # The target model to be added to the coalition
            target_idx: int = current_perm[step]

            # Form the new coalition by appending the target model
            prefix_with: List[int] = prefix_indices + [target_idx]

            # Evaluate the utility of the new coalition
            v_with: float = v_func(prefix_with)

            # Compute the instantaneous marginal contribution (Eq. 4)
            delta: float = v_with - v_without

            # Accumulate the marginal contribution for the target model
            mec_sums[target_idx] += delta

            # Map the integer indices in the prefix back to their string model_ids
            coalition_ids: List[str] = [qualified_pool[idx] for idx in prefix_indices]
            target_model_id: str = qualified_pool[target_idx]

            # Record the exact provenance of this evaluation
            coalition_records.append({
                "target_model": target_model_id,
                "k": k,
                "coalition_ids": coalition_ids,
                "v_with": v_with,
                "v_without": v_without,
                "delta": delta
            })

            # The utility 'with' the target becomes the utility 'without' for the next step
            # This halves the total number of v_func evaluations required
            v_without = v_with
            prefix_indices = prefix_with

    return mec_sums, coalition_records

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 3: Average marginals and format outputs
# -------------------------------------------------------------------------------------------------------------------------------

def format_mec_outputs(
    mec_sums: np.ndarray,
    k_total: int,
    coalition_records: List[Dict[str, Any]],
    qualified_pool: List[str]
) -> Tuple[Dict[str, float], pd.DataFrame]:
    """
    Averages the accumulated marginals to compute MEC_hat and formats the
    provenance records into a strictly typed pandas DataFrame.

    Equation 6: MEC_hat_j = (1 / K) * sum(...)

    Parameters
    ----------
    mec_sums : np.ndarray
        The 1D array containing the sum of marginals per model.
    k_total : int
        The total number of permutations evaluated (K).
    coalition_records : List[Dict[str, Any]]
        The list of provenance dictionaries.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.

    Returns
    -------
    Tuple[Dict[str, float], pd.DataFrame]
        - mec_hat: A dictionary mapping model_id to its estimated MEC.
        - coalition_samples: The strictly typed DataFrame of all evaluations.
    """
    # Compute the expected marginal ensemble contribution (Eq. 6)
    mec_averages = mec_sums / float(k_total)

    # Map the averages back to the string model_ids to prevent downstream indexing bugs
    mec_hat: Dict[str, float] = {
        model_id: float(mec_averages[idx])
        for idx, model_id in enumerate(qualified_pool)
    }

    # Convert the provenance records into a pandas DataFrame
    coalition_samples = pd.DataFrame(coalition_records)

    # Enforce the exact schema and dtypes specified in the protocol (§9.3(f))
    expected_columns = ["target_model", "k", "coalition_ids", "v_with", "v_without", "delta"]
    coalition_samples = coalition_samples[expected_columns]

    # Coerce dtypes for memory efficiency and strictness
    coalition_samples["target_model"] = coalition_samples["target_model"].astype("category")
    coalition_samples["k"] = coalition_samples["k"].astype("int32")
    coalition_samples["v_with"] = coalition_samples["v_with"].astype("float64")
    coalition_samples["v_without"] = coalition_samples["v_without"].astype("float64")
    coalition_samples["delta"] = coalition_samples["delta"].astype("float64")
    # coalition_ids remains as 'object' because it contains lists of strings

    return mec_hat, coalition_samples

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def estimate_mec(
    qualified_pool: List[str],
    v_func: Callable[[List[int]], float],
    config: Mapping[str, Any],
    rng_bundle: Dict[str, np.random.Generator],
    epsilon: Optional[float] = None,
    delta: Optional[float] = None
) -> Tuple[Dict[str, float], pd.DataFrame]:
    """
    Orchestrates Task 13: Estimates the Marginal Ensemble Contribution (MEC) for
    each model using the Permutation Shapley estimator (Algorithm 1, Stage A).

    Parameters
    ----------
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    v_func : Callable[[List[int]], float]
        The optimized utility closure from Task 12.
    config : Mapping[str, Any]
        The immutable study configuration.
    rng_bundle : Dict[str, np.random.Generator]
        The bundle of independent random number generators.
    epsilon : Optional[float], optional
        The desired precision tolerance for the Hoeffding bound.
    delta : Optional[float], optional
        The failure probability for the Hoeffding bound.

    Returns
    -------
    Tuple[Dict[str, float], pd.DataFrame]
        - mec_hat: A dictionary mapping model_id to its estimated MEC.
        - coalition_samples: The strictly typed DataFrame of all evaluations.

    Raises
    ------
    ValueError
        If the configured coalition sampler is not 'permutation'.
    """
    # Extract the dedicated RNG for Monte Carlo sampling to ensure Tier-1 reproducibility
    mc_rng = rng_bundle["monte_carlo_sampling_seed"]
    n_models = len(qualified_pool)

    # Step 1: Determine K and generate the uniform permutations
    k_total, permutations = determine_k_and_generate_permutations(
        n_models=n_models,
        config=config,
        rng=mc_rng,
        epsilon=epsilon,
        delta=delta
    )

    # Step 2: Compute the marginal contributions using incremental evaluation
    mec_sums, coalition_records = compute_incremental_marginals(
        permutations=permutations,
        v_func=v_func,
        qualified_pool=qualified_pool
    )

    # Step 3: Average the marginals and format the output structures
    mec_hat, coalition_samples = format_mec_outputs(
        mec_sums=mec_sums,
        k_total=k_total,
        coalition_records=coalition_records,
        qualified_pool=qualified_pool
    )

    return mec_hat, coalition_samples
````

````python
# Task 14 — Conditional Information-Gain Estimation (Algorithm 1, Stage C) (estimate_ig)

# ==============================================================================
# Task 14: Conditional Information-Gain Estimation (Algorithm 1, Stage C)
# ==============================================================================

# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 1: Extract and format the correctness matrix and ground truth
# -------------------------------------------------------------------------------------------------------------------------------

def prepare_correctness_data(
    estimator_projection: pd.DataFrame,
    qualified_pool: List[str]
) -> Tuple[np.ndarray, np.ndarray]:
    """
    Prepares the dense binary correctness matrix A and the ground-truth signal Z.

    The target Z is defined explicitly as the task's ground-truth-correctness
    signal (solvability), distinct from the candidate's own indicator A_j, to
    prevent the estimand from collapsing to zero. Z_t = 1 if any model in the
    pool solves task t, else 0.

    Parameters
    ----------
    estimator_projection : pd.DataFrame
        The 13-column analytic core tensor containing the 'correct' column.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.

    Returns
    -------
    Tuple[np.ndarray, np.ndarray]
        - A_matrix: A 2D NumPy array of shape (m_tasks, n_models) with binary indicators.
        - Z_vec: A 1D NumPy array of shape (m_tasks,) representing task solvability.
    """
    # Group by task and model, compute the mean correctness across replicates
    # observed=True ensures we only group by existing combinations
    mean_correctness = estimator_projection.groupby(["task_id", "model_id"], observed=True)["correct"].mean()

    # Unstack to create a dense matrix: rows are tasks, columns are models
    # Reindex columns to strictly enforce the canonical model ordering
    A_df = mean_correctness.unstack(level="model_id").reindex(columns=qualified_pool)

    # Threshold at > 0.5 to resolve the majority vote per (task, model) pair
    # Convert to a contiguous C-ordered NumPy array of 8-bit integers
    A_matrix: np.ndarray = (A_df.values > 0.5).astype(np.int8)

    # Define the target Z explicitly as the task's ground-truth-correctness signal.
    # To ensure Z is distinct from A_j and H(Z) > 0, we define Z_t as the indicator
    # of whether the oracle ensemble (the union of all models) solves the task.
    # Z_t = max_{i \in N} A_{it}
    Z_vec: np.ndarray = np.max(A_matrix, axis=1).astype(np.int8)

    return A_matrix, Z_vec

# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 2: Compute Conditional Mutual Information with Miller-Madow Correction
# -------------------------------------------------------------------------------------------------------------------------------

def _compute_corrected_conditional_entropy(
    Z: np.ndarray,
    C: np.ndarray,
    alpha: float = 0.5
) -> float:
    """
    Computes the conditional entropy H(Z | C) with Laplace smoothing and the
    Miller-Madow finite-sample bias correction.

    Equation: H_MM = H_plug + (m_hat - 1) / (2 * N_stratum)

    Parameters
    ----------
    Z : np.ndarray
        The 1D array of target variables (shape: m_tasks).
    C : np.ndarray
        The 1D or 2D array of conditioning variables. If 2D, rows are treated as
        joint conditioning tuples.
    alpha : float, optional
        The Laplace smoothing parameter, by default 0.5.

    Returns
    -------
    float
        The bias-corrected conditional entropy in bits.
    """
    n_samples = len(Z)
    if n_samples == 0:
        return 0.0

    # If C is 2D (e.g., conditioning on A_j and c_S), we need to group by unique rows.
    # We achieve this by finding unique rows and their inverse mapping.
    if C.ndim > 1:
        # axis=0 finds unique rows. return_inverse gives the stratum index for each sample.
        unique_c, inverse_indices = np.unique(C, axis=0, return_inverse=True)
    else:
        unique_c, inverse_indices = np.unique(C, return_inverse=True)

    n_strata = len(unique_c)

    # The support of Z is binary {0, 1}, so the maximum number of categories is 2.
    # We define the support size explicitly to ensure consistent smoothing.
    z_support_size = 2

    total_entropy = 0.0

    # Iterate over each unique stratum defined by the conditioning variable(s) C
    for stratum_idx in range(n_strata):
        # Find all samples belonging to the current stratum
        stratum_mask = (inverse_indices == stratum_idx)
        Z_stratum = Z[stratum_mask]

        # N_stratum is the number of observations BEFORE smoothing
        N_stratum = len(Z_stratum)

        if N_stratum == 0:
            continue

        # Count the occurrences of Z=0 and Z=1 in this stratum
        # minlength=2 ensures the bincount array always has size 2, even if only 0s or 1s exist
        counts = np.bincount(Z_stratum, minlength=z_support_size)

        # m_hat is the number of OBSERVED support categories in this stratum
        m_hat = np.sum(counts > 0)

        # Apply Laplace smoothing: add alpha to all joint cell counts
        smoothed_counts = counts + alpha
        smoothed_total = N_stratum + (z_support_size * alpha)

        # Compute the plug-in probability distribution
        probs = smoothed_counts / smoothed_total

        # Compute the plug-in entropy (base 2 for bits)
        # Add a small epsilon to probs to prevent log2(0), though smoothing should prevent this
        h_plug = -np.sum(probs * np.log2(probs + 1e-12))

        # Apply the Miller-Madow finite-sample bias correction
        # The correction term is (m_hat - 1) / (2 * N_stratum)
        # We use the natural log conversion factor (1 / ln(2)) because the theoretical
        # Miller-Madow derivation assumes natural logarithms, but we report in bits.
        # Correction in bits: ((m_hat - 1) / (2 * N_stratum)) * (1 / ln(2))
        if N_stratum > 0:
            mm_correction = ((m_hat - 1) / (2.0 * N_stratum)) * (1.0 / np.log(2))
        else:
            mm_correction = 0.0

        h_mm = h_plug + mm_correction

        # Accumulate the count-weighted average of per-stratum entropies
        # The weight is the empirical probability of the stratum: N_stratum / n_samples
        stratum_weight = N_stratum / float(n_samples)
        total_entropy += stratum_weight * h_mm

    return total_entropy

def compute_conditional_information_gain(
    j_idx: int,
    S_indices: List[int],
    A_matrix: np.ndarray,
    Z_vec: np.ndarray,
    alpha: float = 0.5
) -> float:
    """
    Computes the operational information gain IG(M_j | S) = I(A_j; Z | A_S)
    using correct-count conditioning and Miller-Madow correction.

    Equation 9a: IG_hat = H_hat(Z | c_S) - H_hat(Z | A_j, c_S)

    Parameters
    ----------
    j_idx : int
        The integer index of the target model M_j.
    S_indices : List[int]
        The list of integer indices representing the coalition S.
    A_matrix : np.ndarray
        The dense binary correctness matrix.
    Z_vec : np.ndarray
        The ground-truth solvability vector.
    alpha : float, optional
        The Laplace smoothing parameter, by default 0.5.

    Returns
    -------
    float
        The estimated conditional information gain in bits.
    """
    # Extract the correctness indicator vector for the target model M_j
    A_j = A_matrix[:, j_idx]

    if not S_indices:
        # If the coalition is empty, IG(M_j | empty) = I(A_j; Z) = H(Z) - H(Z | A_j)
        # We use a dummy conditioning variable (all zeros) to reuse the entropy function
        dummy_c = np.zeros_like(Z_vec)
        h_z = _compute_corrected_conditional_entropy(Z_vec, dummy_c, alpha)
        h_z_given_aj = _compute_corrected_conditional_entropy(Z_vec, A_j, alpha)
        return max(0.0, h_z - h_z_given_aj)

    # Extract the correctness matrix for the coalition S
    A_S = A_matrix[:, S_indices]

    # Compute the correct-count sufficient statistic c_S
    # c_S = sum_{i \in S} A_{S,i}
    # This collapses the conditioning support from 2^|S| to |S|+1 cells
    c_S = np.sum(A_S, axis=1)

    # Compute the first term: H_hat(Z | c_S)
    h_z_given_cs = _compute_corrected_conditional_entropy(Z_vec, c_S, alpha)

    # Compute the second term: H_hat(Z | A_j, c_S)
    # We stack A_j and c_S column-wise to form a 2D conditioning array
    joint_cond = np.column_stack((A_j, c_S))
    h_z_given_aj_cs = _compute_corrected_conditional_entropy(Z_vec, joint_cond, alpha)

    # Compute the final Information Gain
    # Mutual information must be non-negative; we clip at 0.0 to handle minor
    # numerical noise or slight over-corrections from the Miller-Madow term.
    ig_val = max(0.0, h_z_given_cs - h_z_given_aj_cs)

    return ig_val

# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 3: Average IG over Permutation Coalitions
# -------------------------------------------------------------------------------------------------------------------------------

def average_ig_over_coalitions(
    coalition_samples: pd.DataFrame,
    A_matrix: np.ndarray,
    Z_vec: np.ndarray,
    qualified_pool: List[str],
    alpha: float = 0.5
) -> Dict[str, float]:
    """
    Averages the conditional information gain over the exact permutation-sampled
    coalitions used for the MEC term, enforcing the unified-conditioning decision.

    Parameters
    ----------
    coalition_samples : pd.DataFrame
        The strictly typed DataFrame containing the exact coalitions evaluated in Task 13.
    A_matrix : np.ndarray
        The dense binary correctness matrix.
    Z_vec : np.ndarray
        The ground-truth solvability vector.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    alpha : float, optional
        The Laplace smoothing parameter, by default 0.5.

    Returns
    -------
    Dict[str, float]
        A dictionary mapping model_id to its average conditional information gain.
    """
    # Create a reverse lookup dictionary for fast string-to-index mapping
    model_to_idx: Dict[str, int] = {m_id: idx for idx, m_id in enumerate(qualified_pool)}

    # Initialize the output dictionary
    ig_dict: Dict[str, float] = {m_id: 0.0 for m_id in qualified_pool}

    # Group the coalition samples by the target model
    # observed=True ensures we only iterate over models that actually appear
    grouped_samples = coalition_samples.groupby("target_model", observed=True)

    for target_model_id, group_df in grouped_samples:
        # Resolve the integer index of the target model
        j_idx = model_to_idx[target_model_id]

        ig_sum = 0.0
        k_count = len(group_df)

        if k_count == 0:
            continue

        # Iterate over the exact coalitions evaluated for this target model
        for _, row in group_df.iterrows():
            # Extract the coalition IDs and map them to integer indices
            coalition_ids: List[str] = row["coalition_ids"]
            S_indices: List[int] = [model_to_idx[m_id] for m_id in coalition_ids]

            # Compute the operational information gain for this specific coalition
            ig_val = compute_conditional_information_gain(
                j_idx=j_idx,
                S_indices=S_indices,
                A_matrix=A_matrix,
                Z_vec=Z_vec,
                alpha=alpha
            )

            ig_sum += ig_val

        # Compute the expected information gain over the coalition distribution
        ig_dict[target_model_id] = ig_sum / float(k_count)

    return ig_dict

# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def estimate_ig(
    estimator_projection: pd.DataFrame,
    coalition_samples: pd.DataFrame,
    qualified_pool: List[str],
    config: Mapping[str, Any]
) -> Dict[str, float]:
    """
    Orchestrates Task 14: Estimates the Conditional Information Gain (IG) for
    each model using outcome-based CMI with Miller-Madow correction.

    Parameters
    ----------
    estimator_projection : pd.DataFrame
        The 13-column analytic core tensor.
    coalition_samples : pd.DataFrame
        The DataFrame containing the exact coalitions evaluated in Task 13.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    config : Mapping[str, Any]
        The immutable study configuration.

    Returns
    -------
    Dict[str, float]
        A dictionary mapping model_id to its estimated Information Gain.

    Raises
    ------
    ValueError
        If the configured IG estimator is not 'outcome_based_cmi'.
    """
    # Extract configuration parameters
    ig_config = config["estimation"]["information_gain"]
    estimator_type: str = ig_config["estimator"]
    alpha: float = ig_config["smoothing_alpha"]

    # Validate that the estimator is the outcome-based CMI specified in the protocol
    if estimator_type != "outcome_based_cmi":
        raise ValueError(
            f"Unsupported information_gain estimator: '{estimator_type}'. "
            f"Task 14 strictly implements the 'outcome_based_cmi' estimator (Eq. 9)."
        )

    # Step 1: Extract and format the correctness matrix and ground truth
    A_matrix, Z_vec = prepare_correctness_data(
        estimator_projection=estimator_projection,
        qualified_pool=qualified_pool
    )

    # Step 2 & 3: Compute CMI with Miller-Madow correction and average over coalitions
    ig_dict = average_ig_over_coalitions(
        coalition_samples=coalition_samples,
        A_matrix=A_matrix,
        Z_vec=Z_vec,
        qualified_pool=qualified_pool,
        alpha=alpha
    )

    return ig_dict
````

````python
# Task 15 — Error Covariance & Marginal Risk (Algorithm 1, Stage B) (estimate_risk)

# ==============================================================================
# Task 15: Error Covariance & Marginal Risk (Algorithm 1, Stage B)
# ==============================================================================

# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 1: Extract and format the error vectors for covariance estimation
# -------------------------------------------------------------------------------------------------------------------------------

def extract_error_category_matrices(
    estimator_projection: pd.DataFrame,
    qualified_pool: List[str],
    error_taxonomy: List[str]
) -> Dict[str, np.ndarray]:
    """
    Extracts the error vectors from the outcome tensor and formats them into
    dense 2D matrices (tasks x models) for each error category.

    This step collapses the replicate axis by computing the mean error rate
    per (task, model) pair, providing a stable signal for covariance estimation.

    Parameters
    ----------
    estimator_projection : pd.DataFrame
        The 13-column analytic core tensor containing the 'e_*' columns.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    error_taxonomy : List[str]
        The list of the 8 error category column names.

    Returns
    -------
    Dict[str, np.ndarray]
        A dictionary mapping each error category name to a 2D NumPy array
        of shape (m_tasks, n_models).
    """
    # Initialize the dictionary to hold the dense matrices
    e_dict: Dict[str, np.ndarray] = {}

    # Iterate over each of the 8 error categories defined in the taxonomy
    for error_col in error_taxonomy:
        # Group by task and model, compute the mean error rate across replicates
        # observed=True ensures we only group by existing combinations
        mean_error = estimator_projection.groupby(["task_id", "model_id"], observed=True)[error_col].mean()

        # Unstack to create a dense matrix: rows are tasks, columns are models
        # Reindex columns to strictly enforce the canonical model ordering
        e_df = mean_error.unstack(level="model_id").reindex(columns=qualified_pool)

        # Fill any missing values with 0.0 (assuming no error if the record is missing,
        # though the tensor assembly should guarantee a complete grid)
        e_df = e_df.fillna(0.0)

        # Convert to a contiguous C-ordered NumPy array of 64-bit floats
        # Float64 is required for numerical stability in covariance estimation
        e_matrix: np.ndarray = e_df.values.astype(np.float64)

        # Store the matrix in the dictionary
        e_dict[error_col] = e_matrix

    return e_dict

# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 2: Estimate the covariance matrix using Ledoit-Wolf shrinkage
# -------------------------------------------------------------------------------------------------------------------------------

def _ledoit_wolf_diagonal_target(X: np.ndarray) -> np.ndarray:
    """
    Computes the Ledoit-Wolf shrinkage estimator of the covariance matrix,
    strictly shrinking towards a diagonal target matrix (Target D in Schäfer
    and Strimmer 2005).

    This analytical implementation guarantees fidelity to the configuration
    `shrinkage_target: "diagonal"`, avoiding the scaled-identity target
    used by default in some standard libraries.

    Parameters
    ----------
    X : np.ndarray
        A 2D array of shape (m_tasks, n_models) representing the centered data.

    Returns
    -------
    np.ndarray
        The shrunk n x n covariance matrix.
    """
    m, n = X.shape
    if m <= 1:
        # Fallback for degenerate cases (should not occur in practice)
        return np.zeros((n, n), dtype=np.float64)

    # 1. Compute the sample covariance matrix S
    # S = (1 / (m - 1)) * X^T * X
    S = np.dot(X.T, X) / (m - 1)

    # 2. Define the shrinkage target T (Diagonal matrix of sample variances)
    # T_ij = S_ii if i == j else 0
    T = np.diag(np.diag(S))

    # 3. Compute the optimal shrinkage intensity lambda*
    # We use the analytical formula for Target D (diagonal, unequal variances)

    # Compute the variance of the sample covariance elements
    # var(s_ij) = (m / (m-1)^3) * sum_{k=1}^m (x_ki * x_kj - s_ij)^2
    # To compute this efficiently, we compute the squared deviations

    # X_ij_prod has shape (m, n, n) where X_ij_prod[k, i, j] = X[k, i] * X[k, j]
    # To avoid massive memory allocation for large n, we compute the sum of variances
    # iteratively or using broadcasting carefully.

    sum_var_s_ij = 0.0
    sum_sq_diff = 0.0

    # We only need to compute the variance for the off-diagonal elements
    # because the target T matches S exactly on the diagonal (so T_ii - S_ii = 0)
    for i in range(n):
        for j in range(i + 1, n):
            # x_ki * x_kj for all k
            x_prod = X[:, i] * X[:, j]
            # Sample variance of the product
            var_s_ij = np.var(x_prod, ddof=1) / m
            sum_var_s_ij += 2.0 * var_s_ij # Multiply by 2 for symmetry

            # Squared difference between S and T
            # Since T_ij = 0 for i != j, (S_ij - T_ij)^2 = S_ij^2
            sum_sq_diff += 2.0 * (S[i, j] ** 2)

    # If the off-diagonal elements are all exactly zero, no shrinkage is needed
    if sum_sq_diff == 0.0:
        return S

    # Compute the optimal shrinkage intensity lambda*
    # lambda* = sum(var(s_ij)) / sum((s_ij - t_ij)^2)
    lambda_star = sum_var_s_ij / sum_sq_diff

    # Constrain lambda* to the valid range [0, 1]
    lambda_star = max(0.0, min(1.0, lambda_star))

    # 4. Compute the shrunk covariance matrix
    # Sigma* = (1 - lambda*) * S + lambda* * T
    Sigma_shrunk = (1.0 - lambda_star) * S + lambda_star * T

    return Sigma_shrunk

def compute_block_covariance(
    e_dict: Dict[str, np.ndarray],
    n_models: int
) -> np.ndarray:
    """
    Computes the total error covariance matrix by estimating the Ledoit-Wolf
    shrunk covariance separately for each error category and summing them.

    Equation 10: Sigma = Cov(E)

    Parameters
    ----------
    e_dict : Dict[str, np.ndarray]
        The dictionary of 2D error matrices per category.
    n_models : int
        The number of qualified models (n).

    Returns
    -------
    np.ndarray
        The total n x n covariance matrix Sigma.
    """
    # Initialize the total covariance matrix with zeros
    sigma_total = np.zeros((n_models, n_models), dtype=np.float64)

    # Iterate over each error category matrix
    for cat_name, E_matrix in e_dict.items():
        # Center the data: X_c = X - mean(X)
        # Covariance requires centered data
        E_mean = np.mean(E_matrix, axis=0)
        E_centered = E_matrix - E_mean

        # Compute the Ledoit-Wolf shrunk covariance for this specific category
        # shrinking towards the diagonal target
        sigma_cat = _ledoit_wolf_diagonal_target(E_centered)

        # Sum the category covariances to form the total covariance matrix
        # This is mathematically valid because the error categories are mutually
        # exclusive per row (enforced in Task 11), meaning the cross-category
        # covariances are structurally zero or negligible.
        sigma_total += sigma_cat

    return sigma_total

# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 3: Compute Marginal Risk Contribution (MRC)
# -------------------------------------------------------------------------------------------------------------------------------

def compute_marginal_risk(
    sigma: np.ndarray,
    qualified_pool: List[str],
    config: Mapping[str, Any]
) -> Tuple[float, Dict[str, float]]:
    """
    Computes the portfolio risk R(w) and the Marginal Risk Contribution (MRC)
    for each model using the reference weight vector.

    Equation 11: R(w) = w^T * Sigma * w
    Equation 12: MRC_j = (Sigma * w)_j

    Parameters
    ----------
    sigma : np.ndarray
        The n x n total error covariance matrix.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    config : Mapping[str, Any]
        The immutable study configuration.

    Returns
    -------
    Tuple[float, Dict[str, float]]
        - r_val: The scalar portfolio risk R(w).
        - mrc_dict: A dictionary mapping model_id to its MRC.

    Raises
    ------
    ValueError
        If the reference weight vector is invalid or missing.
    """
    n_models = len(qualified_pool)

    # Extract the reference weight vector from the configuration
    ref_w_config = config["estimation"].get("reference_weight_w")

    # Resolve the weight vector
    if ref_w_config == "simplex_uniform" or ref_w_config is None:
        # If None, we fallback to uniform to prevent pipeline collapse,
        # but this should ideally be caught by Task 2 validation.
        w_vec = np.ones(n_models, dtype=np.float64) / float(n_models)
    elif isinstance(ref_w_config, list):
        w_vec = np.array(ref_w_config, dtype=np.float64)
        # Validate length
        if len(w_vec) != n_models:
            raise ValueError(f"reference_weight_w length ({len(w_vec)}) does not match pool size ({n_models}).")
        # Validate simplex property (sum to 1, non-negative)
        if not np.isclose(np.sum(w_vec), 1.0) or np.any(w_vec < 0.0):
            raise ValueError("reference_weight_w must be a valid probability simplex (sum to 1, non-negative).")
    else:
        raise ValueError(f"Invalid reference_weight_w configuration: {ref_w_config}")

    # Compute the Marginal Risk Contribution vector (Eq. 12)
    # MRC = Sigma * w (matrix-vector multiplication)
    mrc_vec = np.dot(sigma, w_vec)

    # Compute the scalar portfolio risk (Eq. 11)
    # R(w) = w^T * (Sigma * w) = w^T * MRC
    r_val = float(np.dot(w_vec, mrc_vec))

    # Format the MRC vector as a dictionary keyed by model_id
    mrc_dict: Dict[str, float] = {
        model_id: float(mrc_vec[idx])
        for idx, model_id in enumerate(qualified_pool)
    }

    return r_val, mrc_dict

# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def estimate_risk(
    estimator_projection: pd.DataFrame,
    qualified_pool: List[str],
    config: Mapping[str, Any]
) -> Tuple[pd.DataFrame, Dict[str, float], float]:
    """
    Orchestrates Task 15: Estimates the error covariance matrix using Ledoit-Wolf
    shrinkage (blocked by category) and computes the Marginal Risk Contribution.

    Parameters
    ----------
    estimator_projection : pd.DataFrame
        The 13-column analytic core tensor.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    config : Mapping[str, Any]
        The immutable study configuration.

    Returns
    -------
    Tuple[pd.DataFrame, Dict[str, float], float]
        - sigma_df: The n x n covariance matrix as a DataFrame with model_id indices.
        - mrc_dict: A dictionary mapping model_id to its MRC.
        - r_val: The scalar portfolio risk R(w).

    Raises
    ------
    ValueError
        If the configured covariance estimator is not 'ledoit_wolf' or the
        shrinkage target is not 'diagonal'.
    """
    # Extract configuration parameters
    cov_config = config["estimation"]["covariance"]
    estimator_type: str = cov_config["estimator"]
    shrinkage_target: str = cov_config["shrinkage_target"]
    block_by_category: bool = cov_config["block_by_error_category"]
    error_taxonomy: List[str] = config["schemas"]["error_taxonomy"]

    # Validate that the estimator matches the protocol specifications
    if estimator_type != "ledoit_wolf" or shrinkage_target != "diagonal" or not block_by_category:
        raise ValueError(
            f"Unsupported covariance configuration. Task 15 strictly implements "
            f"estimator='ledoit_wolf', shrinkage_target='diagonal', and "
            f"block_by_error_category=True."
        )

    n_models = len(qualified_pool)

    # Step 1: Extract and format the error vectors for each category
    e_dict = extract_error_category_matrices(
        estimator_projection=estimator_projection,
        qualified_pool=qualified_pool,
        error_taxonomy=error_taxonomy
    )

    # Step 2: Estimate the total covariance matrix using Ledoit-Wolf shrinkage
    sigma_matrix = compute_block_covariance(
        e_dict=e_dict,
        n_models=n_models
    )

    # Step 3: Compute the Marginal Risk Contribution (MRC) and portfolio risk
    r_val, mrc_dict = compute_marginal_risk(
        sigma=sigma_matrix,
        qualified_pool=qualified_pool,
        config=config
    )

    # Wrap the NumPy covariance matrix in a pandas DataFrame for self-describing output
    # Both the index and columns are set to the canonical qualified_pool
    sigma_df = pd.DataFrame(
        sigma_matrix,
        index=qualified_pool,
        columns=qualified_pool
    )

    return sigma_df, mrc_dict, r_val
````

````python
# Task 16 — FMEC Composition (compose_fmec)

# ==============================================================================
# Task 16: FMEC Composition
# ==============================================================================

# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 1: Combine the three primitives into the FMEC score
# -------------------------------------------------------------------------------------------------------------------------------

def compute_composite_fmec_scores(
    mec_hat: Dict[str, float],
    ig: Dict[str, float],
    mrc: Dict[str, float],
    qualified_pool: List[str],
    config: Mapping[str, Any]
) -> List[Dict[str, Any]]:
    """
    Combines the three estimation primitives into the Final Marginal Ensemble
    Contribution (FMEC) score for each model.

    Equation 13: FMEC_j = MEC_hat_j + eta * IG_j - lambda * MRC_j

    Note on Scale Dependence: The three terms are in different native units
    (utility, bits, error-variance). As per the protocol, they are combined raw,
    making the (eta, lambda) hyperparameters scale-dependent conversion factors.
    Interpretability relies on the downstream robustness grid analysis.

    Parameters
    ----------
    mec_hat : Dict[str, float]
        Dictionary mapping model_id to its expected marginal utility contribution.
    ig : Dict[str, float]
        Dictionary mapping model_id to its conditional information gain.
    mrc : Dict[str, float]
        Dictionary mapping model_id to its marginal risk contribution.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    config : Mapping[str, Any]
        The immutable study configuration containing the FMEC hyperparameters.

    Returns
    -------
    List[Dict[str, Any]]
        A list of dictionaries, each representing a row for the FMEC table.

    Raises
    ------
    KeyError
        If any model in the qualified pool is missing from any of the input dictionaries.
    """
    # Extract the combination hyperparameters from the configuration
    fmec_params = config["estimation"]["fmec_hyperparams"]
    eta: float = fmec_params["eta"]
    lam: float = fmec_params["lambda"]

    fmec_rows: List[Dict[str, Any]] = []

    # Iterate strictly over the canonical pool to guarantee ordering and alignment
    for model_id in qualified_pool:
        try:
            # Explicitly extract values to fail fast on upstream dictionary corruption
            mec_val = mec_hat[model_id]
            ig_val = ig[model_id]
            mrc_val = mrc[model_id]
        except KeyError as e:
            raise KeyError(
                f"FMEC Composition Failed: Model '{model_id}' is missing from "
                f"one of the estimator outputs. Missing key: {str(e)}"
            )

        # Compute the composite FMEC score (Eq. 13)
        # Rewards utility and novelty, penalizes correlated risk
        fmec_val = mec_val + (eta * ig_val) - (lam * mrc_val)

        # Construct the row record
        fmec_rows.append({
            "model_id": model_id,
            "mec_hat": mec_val,
            "ig": ig_val,
            "mrc": mrc_val,
            "fmec": fmec_val
        })

    return fmec_rows

# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 2: Assemble the FMEC table
# -------------------------------------------------------------------------------------------------------------------------------

def assemble_fmec_dataframe(fmec_rows: List[Dict[str, Any]]) -> pd.DataFrame:
    """
    Converts the list of FMEC records into a strictly typed pandas DataFrame
    matching the schema defined in Section 9.3(h).

    Since bootstrap validation (Algorithm 3) occurs after this stage, the
    uncertainty and stability columns are initialized with placeholders to
    satisfy the structural contract.

    Parameters
    ----------
    fmec_rows : List[Dict[str, Any]]
        The list of computed FMEC records from Step 1.

    Returns
    -------
    pd.DataFrame
        The strictly typed FMEC DataFrame.
    """
    # Convert the list of dictionaries to a DataFrame
    df = pd.DataFrame(fmec_rows)

    # Initialize the bootstrap-dependent columns with mathematical placeholders
    # np.nan for confidence intervals, 0.0 for selection frequency
    df["ci_low"] = np.nan
    df["ci_high"] = np.nan
    df["sel_freq"] = 0.0

    # Define the exact canonical column sequence required by the schema
    canonical_columns = [
        "model_id", "mec_hat", "ig", "mrc", "fmec", "ci_low", "ci_high", "sel_freq"
    ]

    # Reorder columns to match the schema
    df = df[canonical_columns]

    # Enforce strict dtypes for memory efficiency and downstream compatibility
    df["model_id"] = df["model_id"].astype("category")

    # All numerical columns are cast to float64
    float_cols = ["mec_hat", "ig", "mrc", "fmec", "ci_low", "ci_high", "sel_freq"]
    for col in float_cols:
        df[col] = df[col].astype("float64")

    return df

# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 3: Persist the FMEC table
# -------------------------------------------------------------------------------------------------------------------------------

def persist_fmec_table(
    fmec_table: pd.DataFrame,
    config: Mapping[str, Any],
    run_dir: Path
) -> None:
    """
    Serializes the FMEC table to disk as a self-describing JSON artifact,
    including the hyperparameters and unit metadata required for interpretation.

    Parameters
    ----------
    fmec_table : pd.DataFrame
        The strictly typed FMEC DataFrame.
    config : Mapping[str, Any]
        The immutable study configuration.
    run_dir : Path
        The path to the immutable run directory.
    """
    # Extract hyperparameters for metadata documentation
    fmec_params = config["estimation"]["fmec_hyperparams"]

    # Construct the self-describing artifact dictionary
    artifact: Dict[str, Any] = {
        "metadata": {
            "description": "Final Marginal Ensemble Contribution (FMEC) estimates.",
            "equation": "FMEC_j = MEC_hat_j + eta * IG_j - lambda * MRC_j",
            "hyperparameters": {
                "eta": fmec_params["eta"],
                "lambda": fmec_params["lambda"]
            },
            "units": {
                "mec_hat": "Utility (normalized [0,1])",
                "ig": "Bits (base 2)",
                "mrc": "Variance/Covariance units",
                "fmec": "Composite score (scale-dependent)"
            },
            "note": "ci_low, ci_high, and sel_freq are placeholders until Algorithm 3 (Bootstrap) is executed."
        },
        # Convert the DataFrame to a list of dictionaries for JSON serialization
        "data": fmec_table.to_dict(orient="records")
    }

    # Define the output path
    output_path: Path = run_dir / "fmec_estimates.json"

    # Serialize and write to disk
    with open(output_path, "w", encoding="utf-8") as f:
        json.dump(artifact, f, indent=2)

# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def compose_fmec(
    mec_hat: Dict[str, float],
    ig: Dict[str, float],
    mrc: Dict[str, float],
    qualified_pool: List[str],
    config: Mapping[str, Any],
    run_dir: Path
) -> pd.DataFrame:
    """
    Orchestrates Task 16: Combines the three estimation primitives into the
    FMEC score, assembles the strictly typed DataFrame, and persists the
    self-describing artifact to disk.

    Parameters
    ----------
    mec_hat : Dict[str, float]
        Dictionary mapping model_id to its expected marginal utility contribution.
    ig : Dict[str, float]
        Dictionary mapping model_id to its conditional information gain.
    mrc : Dict[str, float]
        Dictionary mapping model_id to its marginal risk contribution.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    config : Mapping[str, Any]
        The immutable study configuration.
    run_dir : Path
        The path to the immutable run directory.

    Returns
    -------
    pd.DataFrame
        The strictly typed FMEC DataFrame, serving as the central ranking primitive.
    """
    # Step 1: Compute the composite scores using Eq. 13
    fmec_rows = compute_composite_fmec_scores(
        mec_hat=mec_hat,
        ig=ig,
        mrc=mrc,
        qualified_pool=qualified_pool,
        config=config
    )

    # Step 2: Assemble the strictly typed pandas DataFrame
    fmec_table = assemble_fmec_dataframe(fmec_rows=fmec_rows)

    # Step 3: Persist the table and its metadata to disk
    persist_fmec_table(
        fmec_table=fmec_table,
        config=config,
        run_dir=run_dir
    )

    return fmec_table
````

````python
# Task 17 — Constrained Ensemble Selection (Algorithm 2) (select_ensemble)

# ==============================================================================
# Task 17: Constrained Ensemble Selection (Algorithm 2)
# ==============================================================================

# -------------------------------------------------------------------------------------------------------------------------------
# Task 17, Step 1: Formulate the constrained optimization program
# -------------------------------------------------------------------------------------------------------------------------------

def prepare_optimization_vectors(
    fmec_table: pd.DataFrame,
    estimator_projection: pd.DataFrame,
    qualified_pool: List[str],
    config: Mapping[str, Any]
) -> Tuple[np.ndarray, np.ndarray, np.ndarray, float, float, Dict[str, float]]:
    """
    Prepares the dense 1D NumPy arrays required for fast, vectorized feasibility
    checks and objective evaluation during the greedy search.

    Parameters
    ----------
    fmec_table : pd.DataFrame
        The strictly typed FMEC DataFrame containing the objective coefficients.
    estimator_projection : pd.DataFrame
        The 13-column analytic core tensor for computing cost and latency.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    config : Mapping[str, Any]
        The immutable study configuration containing budgets and judge info.

    Returns
    -------
    Tuple[np.ndarray, np.ndarray, np.ndarray, float, float, Dict[str, float]]
        - fmec_vec: 1D array of FMEC scores.
        - c_vec: 1D array of mean costs per model.
        - l_vec: 1D array of mean latencies per model.
        - c_j: Scalar cost of the judge model.
        - l_j: Scalar latency of the judge model.
        - budgets: Dictionary containing C_max, L_max, R_max.
    """
    # 1. Extract the FMEC objective coefficients
    # Ensure the table is strictly aligned with the canonical qualified_pool
    fmec_aligned = fmec_table.set_index("model_id").reindex(qualified_pool)
    fmec_vec: np.ndarray = fmec_aligned["fmec"].values.astype(np.float64)

    # 2. Extract mean cost and latency per model across all tasks
    cost_series = estimator_projection.groupby("model_id", observed=True)["cost_usd"].mean()
    latency_series = estimator_projection.groupby("model_id", observed=True)["latency_s"].mean()

    c_vec: np.ndarray = cost_series.reindex(qualified_pool).fillna(0.0).values.astype(np.float64)
    l_vec: np.ndarray = latency_series.reindex(qualified_pool).fillna(0.0).values.astype(np.float64)

    # 3. Extract the Judge's cost and latency overhead
    # The judge's overhead must be included in the ensemble's total cost/latency
    # even if the judge is not explicitly selected as a panel member.
    judge_model_id: str = config["models"]["judge"]
    if judge_model_id in cost_series.index:
        c_j: float = float(cost_series[judge_model_id])
        l_j: float = float(latency_series[judge_model_id])
    else:
        c_j = 0.0
        l_j = 0.0

    # 4. Extract the optimization budgets
    opt_config = config["optimization"]
    budgets: Dict[str, float] = {
        "C_max": float(opt_config["C_max"]),
        "L_max": float(opt_config["L_max"]),
        "R_max": float(opt_config["R_max"])
    }

    return fmec_vec, c_vec, l_vec, c_j, l_j, budgets

# -------------------------------------------------------------------------------------------------------------------------------
# Task 17, Step 2: Execute the budget-feasible greedy algorithm
# -------------------------------------------------------------------------------------------------------------------------------

def execute_greedy_selection(
    fmec_vec: np.ndarray,
    c_vec: np.ndarray,
    l_vec: np.ndarray,
    sigma: np.ndarray,
    c_j: float,
    l_j: float,
    budgets: Dict[str, float]
) -> List[int]:
    """
    Executes Algorithm 2: Greedy Constrained FMEC Ensemble Selection.

    Iteratively adds the feasible candidate that maximizes the marginal utility
    gain. Halts when no feasible candidate remains or the maximum gain is <= 0.

    Parameters
    ----------
    fmec_vec : np.ndarray
        1D array of FMEC scores.
    c_vec : np.ndarray
        1D array of mean costs per model.
    l_vec : np.ndarray
        1D array of mean latencies per model.
    sigma : np.ndarray
        2D array (n x n) error covariance matrix.
    c_j : float
        Scalar cost of the judge model.
    l_j : float
        Scalar latency of the judge model.
    budgets : Dict[str, float]
        Dictionary containing C_max, L_max, R_max.

    Returns
    -------
    List[int]
        A list of integer indices representing the selected ensemble S*.
    """
    n_models = len(fmec_vec)
    S: List[int] = []
    current_u: float = 0.0

    # Epsilon tolerance for floating-point constraint checks
    eps = 1e-9

    while True:
        best_gain = -float('inf')
        best_candidate = -1

        # Evaluate all candidates not currently in the ensemble
        for i in range(n_models):
            if i in S:
                continue

            # Form the hypothetical new ensemble S'
            S_prime = S + [i]
            s_size = len(S_prime)

            # 1. Check Cost Feasibility
            # Cost is additive over members plus the judge
            c_val = np.sum(c_vec[S_prime]) + c_j
            if c_val > budgets["C_max"] + eps:
                continue

            # 2. Check Latency Feasibility
            # Latency is the member-max plus the judge (parallel fan-out)
            l_val = np.max(l_vec[S_prime]) + l_j
            if l_val > budgets["L_max"] + eps:
                continue

            # 3. Check Risk Feasibility
            # Risk is the quadratic form w^T Sigma w
            # We use the uniform weight rule: w_i = 1/|S'| for i in S'
            w = np.zeros(n_models, dtype=np.float64)
            w[S_prime] = 1.0 / s_size
            r_val = np.dot(w, np.dot(sigma, w))

            if r_val > budgets["R_max"] + eps:
                continue

            # 4. Compute Objective Utility U(S')
            # U(w) = sum_i w_i * FMEC_i
            u_val = np.sum(w * fmec_vec)

            # 5. Compute Marginal Gain
            delta_u = u_val - current_u

            # Update the best candidate if this one provides a strictly better gain
            if delta_u > best_gain:
                best_gain = delta_u
                best_candidate = i

        # Halting Condition: Break if no feasible candidate was found OR
        # the best marginal gain is not strictly positive.
        if best_candidate == -1 or best_gain <= eps:
            break

        # Add the best candidate to the ensemble and update the current utility
        S.append(best_candidate)

        # Recompute the exact utility of the new set to avoid accumulation errors
        w_new = np.zeros(n_models, dtype=np.float64)
        w_new[S] = 1.0 / len(S)
        current_u = np.sum(w_new * fmec_vec)

    return S

# -------------------------------------------------------------------------------------------------------------------------------
# Task 17, Step 3: Emit the selected ensemble, weights, and active constraints
# -------------------------------------------------------------------------------------------------------------------------------

def format_selection_outputs(
    S_indices: List[int],
    qualified_pool: List[str],
    c_vec: np.ndarray,
    l_vec: np.ndarray,
    sigma: np.ndarray,
    c_j: float,
    l_j: float,
    budgets: Dict[str, float]
) -> Tuple[List[str], Dict[str, float], Dict[str, Any]]:
    """
    Formats the integer indices into string IDs, computes the final simplex weights,
    and determines which constraints are active (binding) at the optimum.

    Parameters
    ----------
    S_indices : List[int]
        The list of integer indices representing the selected ensemble.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    c_vec, l_vec, sigma : np.ndarray
        The cost, latency, and covariance arrays.
    c_j, l_j : float
        The judge cost and latency.
    budgets : Dict[str, float]
        The optimization budgets.

    Returns
    -------
    Tuple[List[str], Dict[str, float], Dict[str, Any]]
        - S_FMEC: List of selected model_ids.
        - w_dict: Dictionary mapping model_id to its simplex weight.
        - active_constraints: Dictionary detailing constraint utilization and binding status.
    """
    n_models = len(qualified_pool)
    s_size = len(S_indices)

    # 1. Map integer indices to string model_ids
    S_FMEC: List[str] = [qualified_pool[idx] for idx in S_indices]

    # 2. Construct the weight dictionary
    # Weights are uniform on the selected set, 0 otherwise
    w_dict: Dict[str, float] = {m_id: 0.0 for m_id in qualified_pool}
    if s_size > 0:
        weight_val = 1.0 / float(s_size)
        for m_id in S_FMEC:
            w_dict[m_id] = weight_val

    # 3. Compute final constraint values
    if s_size > 0:
        c_final = float(np.sum(c_vec[S_indices]) + c_j)
        l_final = float(np.max(l_vec[S_indices]) + l_j)

        w_vec = np.zeros(n_models, dtype=np.float64)
        w_vec[S_indices] = 1.0 / s_size
        r_final = float(np.dot(w_vec, np.dot(sigma, w_vec)))
    else:
        c_final = c_j
        l_final = l_j
        r_final = 0.0

    # 4. Determine active (binding) constraints
    # A constraint is considered active if it is within 1% of the budget limit.
    # This provides explainability for why the greedy search halted.
    tolerance = 0.01

    def _is_active(val: float, budget: float) -> bool:
        if budget == 0.0: return False
        return (val / budget) >= (1.0 - tolerance)

    active_constraints: Dict[str, Any] = {
        "cost": {
            "value": c_final,
            "budget": budgets["C_max"],
            "utilization_pct": (c_final / budgets["C_max"] * 100) if budgets["C_max"] > 0 else 0.0,
            "is_binding": _is_active(c_final, budgets["C_max"])
        },
        "latency": {
            "value": l_final,
            "budget": budgets["L_max"],
            "utilization_pct": (l_final / budgets["L_max"] * 100) if budgets["L_max"] > 0 else 0.0,
            "is_binding": _is_active(l_final, budgets["L_max"])
        },
        "risk": {
            "value": r_final,
            "budget": budgets["R_max"],
            "utilization_pct": (r_final / budgets["R_max"] * 100) if budgets["R_max"] > 0 else 0.0,
            "is_binding": _is_active(r_final, budgets["R_max"])
        }
    }

    return S_FMEC, w_dict, active_constraints

# -------------------------------------------------------------------------------------------------------------------------------
# Task 17, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def select_ensemble(
    fmec_table: pd.DataFrame,
    estimator_projection: pd.DataFrame,
    sigma_df: pd.DataFrame,
    qualified_pool: List[str],
    config: Mapping[str, Any]
) -> Tuple[List[str], Dict[str, float], Dict[str, Any]]:
    """
    Orchestrates Task 17: Prepares optimization vectors, executes the greedy
    budget-feasible selection algorithm, and formats the outputs including
    active constraint diagnostics.

    Parameters
    ----------
    fmec_table : pd.DataFrame
        The strictly typed FMEC DataFrame.
    estimator_projection : pd.DataFrame
        The 13-column analytic core tensor.
    sigma_df : pd.DataFrame
        The n x n covariance matrix as a DataFrame.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    config : Mapping[str, Any]
        The immutable study configuration.

    Returns
    -------
    Tuple[List[str], Dict[str, float], Dict[str, Any]]
        - S_FMEC: List of selected model_ids.
        - w_dict: Dictionary mapping model_id to its simplex weight.
        - active_constraints: Dictionary detailing constraint utilization.
    """
    # Extract the underlying NumPy array from the covariance DataFrame
    # Ensure it is strictly aligned with the canonical pool
    sigma_aligned = sigma_df.reindex(index=qualified_pool, columns=qualified_pool)
    sigma_matrix: np.ndarray = sigma_aligned.values.astype(np.float64)

    # Step 1: Prepare the dense 1D arrays for fast feasibility checks
    fmec_vec, c_vec, l_vec, c_j, l_j, budgets = prepare_optimization_vectors(
        fmec_table=fmec_table,
        estimator_projection=estimator_projection,
        qualified_pool=qualified_pool,
        config=config
    )

    # Step 2: Execute Algorithm 2 (Greedy Constrained Selection)
    S_indices = execute_greedy_selection(
        fmec_vec=fmec_vec,
        c_vec=c_vec,
        l_vec=l_vec,
        sigma=sigma_matrix,
        c_j=c_j,
        l_j=l_j,
        budgets=budgets
    )

    # Step 3: Format the outputs and compute active constraint diagnostics
    S_FMEC, w_dict, active_constraints = format_selection_outputs(
        S_indices=S_indices,
        qualified_pool=qualified_pool,
        c_vec=c_vec,
        l_vec=l_vec,
        sigma=sigma_matrix,
        c_j=c_j,
        l_j=l_j,
        budgets=budgets
    )

    return S_FMEC, w_dict, active_constraints
````

````python
# Task 18 — Fusion-Simulation-Gap Audit (fusion_sim_audit)

# ==============================================================================
# Task 18: Fusion-Simulation-Gap Audit
# ==============================================================================

# -------------------------------------------------------------------------------------------------------------------------------
# Task 18, Step 1: Sample stratified coalitions and compute simulated quality
# -------------------------------------------------------------------------------------------------------------------------------

def sample_coalitions_and_simulate_quality(
    estimator_projection: pd.DataFrame,
    qualified_pool: List[str],
    config: Mapping[str, Any],
    rng: np.random.Generator
) -> pd.DataFrame:
    """
    Samples coalitions stratified by size and computes their simulated quality
    using the deterministic majority-vote surrogate.

    Equation 3a: Q_hat_maj(S) = (1 / |T|) * sum( I[ sum(A_it) > |S|/2 ] )

    Parameters
    ----------
    estimator_projection : pd.DataFrame
        The 13-column analytic core tensor.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    config : Mapping[str, Any]
        The immutable study configuration.
    rng : np.random.Generator
        The dedicated random number generator for sampling.

    Returns
    -------
    pd.DataFrame
        A strictly typed DataFrame containing the sampled coalitions, their sizes,
        and their simulated qualities (q_hat).
    """
    n_models = len(qualified_pool)
    audit_config = config["fusion_sim_audit"]
    total_samples: int = audit_config["audit_n_coalitions"]

    # 1. Pre-compute the dense correctness matrix A and reliability vector r
    # Group by task and model, compute the mean correctness across replicates
    mean_correctness = estimator_projection.groupby(["task_id", "model_id"], observed=True)["correct"].mean()
    A_df = mean_correctness.unstack(level="model_id").reindex(columns=qualified_pool)

    # Convert to a contiguous C-ordered NumPy array of 8-bit integers
    A_matrix: np.ndarray = (A_df.values > 0.5).astype(np.int8)

    # Compute the reliability vector r (mean correctness across all tasks)
    r_vec: np.ndarray = A_matrix.mean(axis=0)

    # 2. Determine the number of samples per coalition size (stratification)
    # We stratify across sizes k in {1, ..., n_models}
    samples_per_size = total_samples // n_models
    remainder = total_samples % n_models

    size_allocations = {k: samples_per_size for k in range(1, n_models + 1)}
    # Distribute the remainder to the first few sizes to ensure exactly total_samples
    for k in range(1, remainder + 1):
        size_allocations[k] += 1

    audit_records: List[Dict[str, Any]] = []

    # 3. Sample coalitions and compute Q_hat
    base_indices = np.arange(n_models, dtype=np.int32)

    for k, n_samples in size_allocations.items():
        if n_samples == 0:
            continue

        # Calculate the maximum possible combinations for size k
        max_combinations = math.comb(n_models, k)
        # Cap the actual samples to prevent infinite loops during unique sampling
        actual_samples = min(n_samples, max_combinations)

        # Use a set to guarantee uniqueness of sampled combinations without replacement
        sampled_sets: Set[Tuple[int, ...]] = set()
        attempts = 0
        max_attempts = actual_samples * 100 # Generous limit for random sampling

        while len(sampled_sets) < actual_samples and attempts < max_attempts:
            # Generate a random permutation and take the first k elements
            perm = rng.permutation(base_indices)
            combo = tuple(sorted(perm[:k]))
            sampled_sets.add(combo)
            attempts += 1

        # Compute Q_hat for each uniquely sampled coalition
        for combo in sampled_sets:
            S = list(combo)
            s_size = len(S)

            # Slice the dense matrix for the models in S
            A_S = A_matrix[:, S]

            # Compute the row-wise sum (number of correct models per task)
            row_sums = np.sum(A_S, axis=1)

            # Strict majority threshold
            threshold = s_size / 2.0
            majority_mask = row_sums > threshold

            # Tie-breaking for even-sized coalitions
            if s_size % 2 == 0:
                tie_mask = (row_sums == threshold)
                if np.any(tie_mask):
                    # Find the index (within S) of the most reliable model
                    best_model_idx_in_S = np.argmax(r_vec[S])
                    # For tied rows, the coalition is correct iff the most reliable model is correct
                    majority_mask[tie_mask] = A_S[tie_mask, best_model_idx_in_S].astype(bool)

            # Expected quality is the mean of the resolved correctness mask
            q_hat = float(np.mean(majority_mask))

            audit_records.append({
                "coalition_indices": S,
                "coalition_size": s_size,
                "q_hat": q_hat
            })

    # Convert to a strictly typed DataFrame
    audit_df = pd.DataFrame(audit_records)
    audit_df["coalition_size"] = audit_df["coalition_size"].astype("int32")
    audit_df["q_hat"] = audit_df["q_hat"].astype("float64")

    return audit_df

# -------------------------------------------------------------------------------------------------------------------------------
# Task 18, Step 2: Execute realized fusion and fit the audit regression
# -------------------------------------------------------------------------------------------------------------------------------

def execute_realized_fusion_and_regress(
    audit_df: pd.DataFrame,
    model_answer: List[Dict[str, Any]],
    candidate_solution: List[Dict[str, Any]],
    task_record: pd.DataFrame,
    fuse_callable: Callable[[List[Dict[str, Any]], str, Optional[str]], Dict[str, Any]],
    qualified_pool: List[str],
    config: Mapping[str, Any],
    api_key: str
) -> Tuple[pd.DataFrame, float, float, float, float]:
    """
    Executes realized fusion on raw outputs and fits the simulation-gap regression (Section 4.2.1).

    This function implements the OLS regression model: Q_f(S) = beta_0 + beta_1 * Q_hat(S) + u(S).
    It calls the remediated `fuse_callable` over raw text completions, scores the
    resulting fused outputs, and computes the OLS parameters along with the
    residual standard deviation (sigma_u) using N-2 degrees of freedom.

    Parameters
    ----------
    audit_df : pd.DataFrame
        DataFrame containing sampled coalitions and their simulated quality (q_hat).
    model_answer : List[Dict[str, Any]]
        Cleansed raw free-form generation records.
    candidate_solution : List[Dict[str, Any]]
        Cleansed raw structured solution records.
    task_record : pd.DataFrame
        Pruned task DataFrame containing prompt and scoring_rule.
    fuse_callable : Callable
        The remediated fusion closure from specify_fusion_operator (Task 8).
    qualified_pool : List[str]
        The canonical ordered list of qualified model IDs.
    config : Mapping[str, Any]
        The immutable study configuration.
    api_key : str
        The resolved API key required for live judge calls during the audit.

    Returns
    -------
    Tuple[pd.DataFrame, float, float, float, float]
        - audit_df: Updated DataFrame with the 'q_f' column populated.
        - slope (beta_1): The regression slope coefficient.
        - intercept (beta_0): The regression intercept coefficient.
        - r_squared (R^2): The coefficient of determination.
        - residual_sd (sigma_u): The standard deviation of the residuals (Eq. 20a input).
    """
    # Combine the cleansed raw records into a single list for rapid lookup
    all_records = model_answer + candidate_solution

    # Create a dense dictionary mapping task_id to its scoring rule and prompt
    task_lookup: Dict[str, Dict[str, Any]] = task_record.set_index("task_id")[["scoring_rule", "prompt"]].to_dict(orient="index")

    # Group the raw records by task_id and model_id to optimize search time to O(1)
    # This prevents the nested loop from executing O(N) searches per task evaluation
    records_map: Dict[str, Dict[str, List[Dict[str, Any]]]] = {}
    for rec in all_records:
        t_id: str = rec["task_id"]
        m_id: str = rec["model_id"]
        if t_id not in records_map:
            records_map[t_id] = {}
        if m_id not in records_map[t_id]:
            records_map[t_id][m_id] = []
        records_map[t_id][m_id].append(rec)

    # Initialize the list to store the realized quality values (Q_f)
    q_f_values: List[float] = []

    # Iterate over each coalition in the audit DataFrame
    for _, row in audit_df.iterrows():
        # Retrieve the model indices included in this specific coalition
        coalition_indices: List[int] = row["coalition_indices"]
        # Map indices back to their canonical string model IDs
        coalition_ids: Set[str] = set([qualified_pool[idx] for idx in coalition_indices])

        # Initialize the list of synthetic records for this coalition
        synthetic_records: List[Dict[str, Any]] = []

        # Evaluate the fusion operator on a task-by-task basis
        for t_id, task_meta in task_lookup.items():
            rule = task_meta["scoring_rule"]
            prompt = task_meta["prompt"]

            # Extract the candidate solutions for the models present in the coalition
            candidates: List[Dict[str, Any]] = []
            if t_id in records_map:
                for m_id in coalition_ids:
                    if m_id in records_map[t_id]:
                        # Append all replicates for this model on this task
                        candidates.extend(records_map[t_id][m_id])

            # If no candidates exist for this task, skip it to avoid biasing the average
            if not candidates:
                continue

            # Execute the fully functional fusion operator
            # This generates a real fused text, making actual judge calls if required
            fusion_result = fuse_callable(candidates, rule, prompt)
            fused_answer = fusion_result.get("fused_answer", "")

            # Construct the synthetic record expected by Task 10 scorers
            # We map the fused answer to both 'answer' and 'thesis' keys
            synthetic_record = {
                "task_id": t_id,
                "model_id": "fused_ensemble", # Dummy ID representing the ensemble
                "rep": 0,
                "answer": fused_answer,
                "thesis": fused_answer,
                "cost_usd": 0.0,
                "latency_s": 0.0,
                "provider_version": "fusion_audit",
                "system_fingerprint": "fusion_audit",
                "seed": 0
            }
            synthetic_records.append(synthetic_record)

        # Score the synthetic records using deterministic comparators (Task 10)
        scored_labels, pending_judge = execute_deterministic_scoring(
            records=synthetic_records,
            task_record=task_record
        )

        # Score the remaining open-ended records using the LLM judge (Task 10)
        # This executes actual network calls, strictly avoiding mocks
        judge_labels, _ = execute_judge_adjudication(
            pending_records=pending_judge,
            task_record=task_record,
            config=config,
            api_key=api_key,
            human_labels=None
        )

        # Combine all scored outputs
        all_scored_labels = scored_labels + judge_labels

        # Compute the realized quality Q_f(S) as the mean correctness over the tasks
        if all_scored_labels:
            correct_count = sum(1 for lbl in all_scored_labels if lbl["correct"] == 1)
            q_f = float(correct_count) / float(len(all_scored_labels))
        else:
            q_f = 0.0

        q_f_values.append(q_f)

    # Append the realized quality values to the audit DataFrame
    audit_df["q_f"] = q_f_values
    audit_df["q_f"] = audit_df["q_f"].astype("float64")

    # Extract independent (Q_hat) and dependent (Q_f) variables
    x = audit_df["q_hat"].values
    y = audit_df["q_f"].values

    # Handle degenerate zero-variance cases to prevent division by zero
    if np.var(x) == 0.0 or np.var(y) == 0.0:
        slope, intercept, r_value, p_value, std_err = 0.0, float(np.mean(y)), 0.0, 1.0, 0.0
        residuals = y - intercept
    else:
        # Solve the OLS regression parameters analytically (Eq. 18-OLS)
        slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
        residuals = y - (intercept + slope * x)

    # Compute the coefficient of determination R^2
    r_squared = float(r_value ** 2)

    # Compute the residual standard deviation (sigma_u) using N-2 degrees of freedom (Eq. 18-SE)
    n_samples = len(residuals)
    if n_samples > 2:
        sigma_u = float(np.sqrt(np.sum(residuals ** 2) / (n_samples - 2)))
    else:
        sigma_u = 0.0

    # Return the mutated DataFrame and OLS parameters
    return audit_df, float(slope), float(intercept), r_squared, sigma_u

# -------------------------------------------------------------------------------------------------------------------------------
# Task 18, Step 3: Flag strata and format outputs
# -------------------------------------------------------------------------------------------------------------------------------

def flag_strata_and_format(
    audit_df: pd.DataFrame,
    slope: float,
    r_squared: float,
    sigma_u: float,
    config: Mapping[str, Any]
) -> Tuple[Dict[str, float], List[int]]:
    """
    Identifies coalition-size strata that exceed the pre-registered tolerance
    and formats the final diagnostic dictionary.

    Parameters
    ----------
    audit_df : pd.DataFrame
        The DataFrame containing Q_hat and Q_f.
    slope, r_squared, sigma_u : float
        The regression diagnostics.
    config : Mapping[str, Any]
        The immutable study configuration.

    Returns
    -------
    Tuple[Dict[str, float], List[int]]
        - diagnostics: Dictionary containing slope, r_squared, and sigma_u.
        - flagged_strata: List of coalition sizes that exceeded the tolerance.
    """
    # Compute absolute error for each coalition
    audit_df["abs_error"] = np.abs(audit_df["q_f"] - audit_df["q_hat"])

    # Group by coalition size and find the maximum absolute error per stratum
    stratum_errors = audit_df.groupby("coalition_size")["abs_error"].max()

    flagged_strata: List[int] = []

    # Extract the pre-registered tolerance
    tolerance = config["fusion_sim_audit"].get("per_stratum_tolerance")

    # The protocol notes that the tolerance is "pre-registered but NO numeric value
    # is given in the paper". We enforce strict type checking to prevent runtime errors.
    if tolerance is not None and isinstance(tolerance, (int, float)):
        # Identify strata where the max error exceeds the tolerance
        violators = stratum_errors[stratum_errors > float(tolerance)]
        flagged_strata = violators.index.tolist()

    diagnostics: Dict[str, float] = {
        "slope": slope,
        "r_squared": r_squared,
        "residual_sd_sigma_u": sigma_u
    }

    return diagnostics, flagged_strata

# -------------------------------------------------------------------------------------------------------------------------------
# Task 18, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def fusion_sim_audit(
    estimator_projection: pd.DataFrame,
    model_answer: List[Dict[str, Any]],
    candidate_solution: List[Dict[str, Any]],
    task_record: pd.DataFrame,
    fuse_callable: Callable[[List[Dict[str, Any]], str, Optional[str]], Dict[str, Any]],
    qualified_pool: List[str],
    config: Mapping[str, Any],
    rng_bundle: Dict[str, np.random.Generator]
) -> Tuple[Dict[str, float], List[int]]:
    """
    Orchestrates Task 18: Samples stratified coalitions, executes the active
    fusion operator on raw outputs, and fits the OLS simulation-gap regression (Eq. 18-Reg).

    This remediated orchestrator securely resolves the API key at runtime and passes
    it along with the estimator projection to support the remediated, active
    execute_realized_fusion_and_regress function.

    Parameters
    ----------
    estimator_projection : pd.DataFrame
        The 13-column analytic core outcomes tensor.
    model_answer : List[Dict[str, Any]]
        Cleansed raw free-form generation records.
    candidate_solution : List[Dict[str, Any]]
        Cleansed raw structured solution records.
    task_record : pd.DataFrame
        Pruned task DataFrame containing prompt and scoring rules.
    fuse_callable : Callable
        The active, weight-aware fusion closure from Task 8.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    config : Mapping[str, Any]
        The immutable study configuration.
    rng_bundle : Dict[str, np.random.Generator]
        The bundle of independent, seeded random number generators.

    Returns
    -------
    Tuple[Dict[str, float], List[int]]
        - diagnostics: Dictionary containing slope, r_squared, and sigma_u.
        - flagged_strata: List of coalition sizes that exceeded the tolerance threshold.

    Raises
    ------
    RuntimeError
        If the API key environment variable is missing or empty.
    """
    # Securely resolve the API key at runtime for the live judge calls
    api_key_env_var: str = config["gateway"]["api_key_env_var"]
    api_key: Optional[str] = os.environ.get(api_key_env_var)
    if not api_key:
        raise RuntimeError(
            f"Fusion Simulation Audit Failed: Environment variable '{api_key_env_var}' "
            f"is missing. Live judge calls are required for the audit."
        )

    # Extract the dedicated global RNG from the bundle for stratified sampling
    global_rng = rng_bundle["global_seed"]

    # Step 1: Sample coalitions stratified by size and compute simulated quality Q_hat (Eq. 3a)
    # sample_coalitions_and_simulate_quality is assumed available
    audit_df = sample_coalitions_and_simulate_quality(
        estimator_projection=estimator_projection,
        qualified_pool=qualified_pool,
        config=config,
        rng=global_rng
    )

    # Step 2: Execute realized fusion on raw outputs and fit the OLS regression
    # This calls the active fuse_callable, executing actual LLM completions and OLS math
    # execute_realized_fusion_and_regress is assumed available
    audit_df, slope, intercept, r_squared, sigma_u = execute_realized_fusion_and_regress(
        audit_df=audit_df,
        model_answer=model_answer,
        candidate_solution=candidate_solution,
        task_record=task_record,
        fuse_callable=fuse_callable,
        qualified_pool=qualified_pool,
        config=config,
        api_key=api_key
    )

    # Step 3: Flag strata and format output diagnostics
    # flag_strata_and_format is assumed available
    diagnostics, flagged_strata = flag_strata_and_format(
        audit_df=audit_df,
        slope=slope,
        r_squared=r_squared,
        sigma_u=sigma_u,
        config=config
    )

    return diagnostics, flagged_strata
````

````python
# Task 19 — Nested Bootstrap, Variance Components & Stability (Algorithm 3) (bootstrap_validate)

# ==============================================================================
# Task 19: Nested Bootstrap, Variance Components & Stability (Algorithm 3)
# ==============================================================================

# -------------------------------------------------------------------------------------------------------------------------------
# Task 19, Step 1: Implement the Nested Cluster Bootstrap Resampling Logic
# -------------------------------------------------------------------------------------------------------------------------------

def generate_nested_bootstrap_samples(
    primary_tensor: pd.DataFrame,
    judge_replicates_df: pd.DataFrame,
    config: Mapping[str, Any],
    rng: np.random.Generator,
    b_iterations: int,
    resample_mode: str = "nested_cluster"
) -> Iterator[pd.DataFrame]:
    """
    Generates resampled outcome tensors using a nested cluster bootstrap.

    Levels of resampling (Eq. 20a):
    1. Outer: Resample tasks with replacement (sigma^2_task).
    2. Inner: Resample generation replicates within tasks (sigma^2_gen).
    3. Adjudication: Resample judge replicates for the subsample (sigma^2_adj).

    Parameters
    ----------
    primary_tensor : pd.DataFrame
        The 17-column canonical released artifact (judge_rep == 0).
    judge_replicates_df : pd.DataFrame
        The DataFrame of adjudication replicates (judge_rep > 0).
    config : Mapping[str, Any]
        The immutable study configuration.
    rng : np.random.Generator
        The dedicated random number generator for the bootstrap.
    b_iterations : int
        The number of bootstrap iterations to generate.
    resample_mode : str, optional
        The specific axis to resample. Options: "nested_cluster" (all),
        "task_only", "gen_only", "adj_only". Used for variance decomposition.

    Yields
    ------
    Iterator[pd.DataFrame]
        A generator yielding the resampled primary tensor for each iteration.
    """
    # Extract replication parameters
    r_count: int = config["uncertainty"]["replication"]["R"]

    # Identify unique tasks and models
    unique_tasks = primary_tensor["task_id"].unique()
    n_tasks = len(unique_tasks)

    # Create a fast lookup for judge replicates
    # Key: (task_id, model_id), Value: list of available judge_reps (including 0)
    has_judge_reps = not judge_replicates_df.empty
    if has_judge_reps:
        combined_df = pd.concat([primary_tensor, judge_replicates_df], ignore_index=True)
        # We only need the label columns for overwriting
        label_cols = ["correct", "e_reason", "e_comp", "e_halluc", "e_ctx", "e_plan", "e_tool", "e_safety", "e_refusal"]
        judge_lookup = combined_df.set_index(["task_id", "model_id", "judge_rep"])[label_cols].to_dict(orient="index")

        # Identify which tasks actually have judge replicates > 0
        tasks_with_adj = judge_replicates_df["task_id"].unique()
    else:
        judge_lookup = {}
        tasks_with_adj = []

    # Pre-sort the primary tensor to ensure deterministic .iloc behavior
    sorted_tensor = primary_tensor.sort_values(by=["task_id", "model_id", "rep"]).reset_index(drop=True)

    # Create a mapping from task_id to its row indices in the sorted tensor
    # This allows extremely fast vectorized extraction of task blocks
    task_to_indices = sorted_tensor.groupby("task_id", observed=True).groups

    for b in range(b_iterations):
        # --- Level 1: Task Resampling (Outer Cluster) ---
        if resample_mode in {"nested_cluster", "task_only"}:
            # Draw N tasks with replacement
            sampled_tasks = rng.choice(unique_tasks, size=n_tasks, replace=True)
        else:
            # Keep tasks fixed
            sampled_tasks = unique_tasks

        resampled_indices = []
        synthetic_task_ids = []

        # --- Level 2: Generation Resampling (Inner) ---
        for i, task in enumerate(sampled_tasks):
            # Create a synthetic ID to preserve multiplicity if a task is drawn multiple times
            syn_task_id = f"{task}_b{b}_i{i}"

            # Get the row indices for this task in the original tensor
            base_idx = task_to_indices[task]

            if resample_mode in {"nested_cluster", "gen_only"}:
                # We need to resample the 'rep' axis.
                # Since the tensor is sorted by (task, model, rep), we can reshape the indices
                # to (n_models, R), sample along the R axis, and flatten.
                n_models_in_task = len(base_idx) // r_count
                idx_matrix = np.array(base_idx).reshape((n_models_in_task, r_count))

                # Draw R indices with replacement from 0..R-1
                rep_choices = rng.choice(r_count, size=r_count, replace=True)

                # Select the corresponding columns from the index matrix
                sampled_idx_matrix = idx_matrix[:, rep_choices]
                task_resampled_idx = sampled_idx_matrix.flatten().tolist()
            else:
                # Keep generations fixed
                task_resampled_idx = list(base_idx)

            resampled_indices.extend(task_resampled_idx)
            # Extend the synthetic IDs to match the number of rows extracted for this task
            synthetic_task_ids.extend([syn_task_id] * len(task_resampled_idx))

        # Construct the new DataFrame using fast integer indexing
        resampled_df = sorted_tensor.iloc[resampled_indices].copy()

        # Overwrite the task_id with the synthetic IDs to preserve statistical weight
        resampled_df["task_id"] = synthetic_task_ids
        # We must also keep track of the original task_id for the adjudication step
        resampled_df["original_task_id"] = [task for task in sampled_tasks for _ in range(len(task_to_indices[task]))]

        # --- Level 3: Adjudication Resampling ---
        if has_judge_reps and resample_mode in {"nested_cluster", "adj_only"}:
            # Identify rows belonging to tasks that have adjudication replicates
            adj_mask = resampled_df["original_task_id"].isin(tasks_with_adj)

            if adj_mask.any():
                # Extract the relevant rows
                adj_rows = resampled_df[adj_mask]

                # For each unique (original_task_id, model_id) pair, draw a single judge_rep
                # This ensures the same judge draw is applied across all generation replicates of that pair
                unique_pairs = adj_rows[["original_task_id", "model_id"]].drop_duplicates()

                # Assuming R_judge = 5 (indices 0 to 4)
                r_judge = config["uncertainty"]["replication"]["R_judge"]

                # Draw judge reps
                drawn_judge_reps = rng.choice(r_judge, size=len(unique_pairs), replace=True)

                # Create a mapping from (task, model) to the drawn judge_rep
                pair_to_jrep = {
                    (row["original_task_id"], row["model_id"]): jrep
                    for (_, row), jrep in zip(unique_pairs.iterrows(), drawn_judge_reps)
                }

                # Apply the overwrites
                for idx, row in adj_rows.iterrows():
                    orig_t = row["original_task_id"]
                    m_id = row["model_id"]
                    jrep = pair_to_jrep[(orig_t, m_id)]

                    if jrep > 0:
                        # Fetch the labels from the lookup
                        lookup_key = (orig_t, m_id, jrep)
                        if lookup_key in judge_lookup:
                            new_labels = judge_lookup[lookup_key]
                            # Overwrite the columns in the resampled DataFrame
                            for col, val in new_labels.items():
                                resampled_df.at[idx, col] = val

        # Drop the temporary original_task_id column
        resampled_df = resampled_df.drop(columns=["original_task_id"])

        # Ensure task_id is categorical for downstream performance
        resampled_df["task_id"] = resampled_df["task_id"].astype("category")

        yield resampled_df

# -------------------------------------------------------------------------------------------------------------------------------
# Task 19, Step 2: Execute the Estimation-Selection Chain per Bootstrap Replicate
# -------------------------------------------------------------------------------------------------------------------------------

def _execute_single_bootstrap_chain(
    resampled_tensor: pd.DataFrame,
    qualified_pool: List[str],
    config: Mapping[str, Any],
    b_index: int
) -> Tuple[Dict[str, float], List[str]]:
    """
    Executes the full estimation and selection pipeline for a single resampled tensor.
    This function is designed to be stateless and picklable for multiprocessing.

    Parameters
    ----------
    resampled_tensor : pd.DataFrame
        The resampled primary tensor.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    config : Mapping[str, Any]
        The immutable study configuration.
    b_index : int
        The bootstrap iteration index, used to seed the local Monte Carlo RNG.

    Returns
    -------
    Tuple[Dict[str, float], List[str]]
        - fmec_dict: Dictionary mapping model_id to its FMEC score.
        - S_b: The selected ensemble list.
    """
    # 1. Create the estimator projection
    # We pass an empty list for judge_replicates as they have already been applied
    estimator_projection, _ = create_estimator_projection(resampled_tensor, [])

    # 2. Spawn a local, independent RNG for Monte Carlo sampling
    # Seed is deterministically derived from the base MC seed and the iteration index
    base_mc_seed = config["reproducibility"]["seed_ledger"]["monte_carlo_sampling_seed"]
    local_mc_rng = np.random.default_rng(base_mc_seed + b_index)
    local_rng_bundle = {"monte_carlo_sampling_seed": local_mc_rng}

    # 3. Estimate Risk (Sigma is needed for utility evaluation)
    sigma_df, mrc_dict, _ = estimate_risk(
        estimator_projection=estimator_projection,
        qualified_pool=qualified_pool,
        config=config
    )
    sigma_matrix = sigma_df.reindex(index=qualified_pool, columns=qualified_pool).values.astype(np.float64)

    # 4. Build the optimized utility closure
    judge_model_id = config["models"]["judge"]
    v_func = build_utility_function(
        estimator_projection=estimator_projection,
        sigma=sigma_matrix,
        config=config,
        qualified_pool=qualified_pool,
        judge_model_id=judge_model_id
    )

    # 5. Estimate MEC
    mec_hat, coalition_samples = estimate_mec(
        qualified_pool=qualified_pool,
        v_func=v_func,
        config=config,
        rng_bundle=local_rng_bundle
    )

    # 6. Estimate Information Gain
    ig_dict = estimate_ig(
        estimator_projection=estimator_projection,
        coalition_samples=coalition_samples,
        qualified_pool=qualified_pool,
        config=config
    )

    # 7. Compose FMEC
    fmec_table = compose_fmec(
        mec_hat=mec_hat,
        ig=ig_dict,
        mrc=mrc_dict,
        qualified_pool=qualified_pool,
        config=config,
        run_dir=Path(".") # Dummy path, we only need the DataFrame returned
    )

    # Extract the FMEC dictionary for easy return
    fmec_dict = fmec_table.set_index("model_id")["fmec"].to_dict()

    # 8. Select Ensemble
    S_b, _, _ = select_ensemble(
        fmec_table=fmec_table,
        estimator_projection=estimator_projection,
        sigma_df=sigma_df,
        qualified_pool=qualified_pool,
        config=config
    )

    return fmec_dict, S_b

# -------------------------------------------------------------------------------------------------------------------------------
# Task 19, Step 3: Compute BCa Intervals, Variance Components, and Stability Metrics
# -------------------------------------------------------------------------------------------------------------------------------

def _compute_bca_intervals(
    bootstrap_fmecs: Dict[str, np.ndarray],
    point_fmecs: Dict[str, float],
    jackknife_fmecs: Dict[str, np.ndarray],
    ci_level: float = 0.95
) -> Dict[str, Tuple[float, float]]:
    """
    Computes Bias-Corrected and Accelerated (BCa) confidence intervals.

    Parameters
    ----------
    bootstrap_fmecs : Dict[str, np.ndarray]
        Dictionary mapping model_id to an array of B bootstrap FMEC scores.
    point_fmecs : Dict[str, float]
        Dictionary mapping model_id to its point estimate FMEC score.
    jackknife_fmecs : Dict[str, np.ndarray]
        Dictionary mapping model_id to an array of jackknife FMEC scores.
    ci_level : float, optional
        The confidence level, by default 0.95.

    Returns
    -------
    Dict[str, Tuple[float, float]]
        Dictionary mapping model_id to (ci_low, ci_high).
    """
    alpha = 1.0 - ci_level
    z_alpha = norm.ppf(alpha / 2.0)
    z_1_minus_alpha = norm.ppf(1.0 - alpha / 2.0)

    bca_intervals = {}

    for m_id, boot_dist in bootstrap_fmecs.items():
        theta_hat = point_fmecs[m_id]
        B = len(boot_dist)

        # 1. Compute Bias-Correction factor (z0)
        # z0 = Phi^-1( proportion of boot_dist < theta_hat )
        prop_less = np.sum(boot_dist < theta_hat) / float(B)
        # Handle edge cases to prevent inf
        prop_less = max(min(prop_less, 0.9999), 0.0001)
        z0 = norm.ppf(prop_less)

        # 2. Compute Acceleration factor (a) using the jackknife distribution
        jk_dist = jackknife_fmecs[m_id]
        jk_mean = np.mean(jk_dist)
        diffs = jk_mean - jk_dist

        num = np.sum(diffs ** 3)
        den = 6.0 * (np.sum(diffs ** 2) ** 1.5)

        # Handle zero denominator
        a = num / den if den > 1e-12 else 0.0

        # 3. Compute adjusted percentiles
        # alpha_1 = Phi( z0 + (z0 + z_alpha) / (1 - a*(z0 + z_alpha)) )
        z_low = z0 + (z0 + z_alpha) / (1.0 - a * (z0 + z_alpha))
        z_high = z0 + (z0 + z_1_minus_alpha) / (1.0 - a * (z0 + z_1_minus_alpha))

        pct_low = norm.cdf(z_low) * 100.0
        pct_high = norm.cdf(z_high) * 100.0

        # Ensure percentiles are within [0, 100]
        pct_low = max(min(pct_low, 100.0), 0.0)
        pct_high = max(min(pct_high, 100.0), 0.0)

        # 4. Extract the bounds from the bootstrap distribution
        ci_low = float(np.percentile(boot_dist, pct_low))
        ci_high = float(np.percentile(boot_dist, pct_high))

        bca_intervals[m_id] = (ci_low, ci_high)

    return bca_intervals

def compute_validation_metrics(
    bootstrap_results: List[Tuple[Dict[str, float], List[str]]],
    point_fmec_table: pd.DataFrame,
    S_star: List[str],
    qualified_pool: List[str],
    config: Mapping[str, Any],
    primary_tensor: pd.DataFrame,
    judge_replicates_df: pd.DataFrame,
    rng_bundle: Dict[str, np.random.Generator]
) -> Tuple[pd.DataFrame, Dict[str, float], float, Dict[str, float]]:
    """
    Compute BCa intervals, the single-axis variance decomposition WITH the
    fusion-simulation residual added in quadrature, and stability metrics.

    Rationale (Eq. 20a; detailed tasklist Task 19, Step 2). Total uncertainty is
    Var(theta_hat) = sigma2_task + sigma2_gen + sigma2_adj, AND — where the
    evaluation uses the fusion surrogate quality — the fusion-simulation residual
    sigma_u must be added in quadrature (Eq. 20a). The prior implementation omitted
    sigma_u entirely. This version reads sigma_u from the config mapping (injected by
    the orchestrator from the Task 18 fusion diagnostics; default 0.0 with a logged
    warning), reports sigma2_task/gen/adj separately, and forms
    sigma2_total = sigma2_task + sigma2_gen + sigma2_adj + sigma_u^2. The decomposed
    functional is the pool-mean FMEC; BCa uses a delete-one-task jackknife; RSP is
    the exact-set-match frequency (Eq. 51). The variance-component bootstrap size is
    config-sourced (a deliberate approximation relative to the headline B).

    Parameters
    ----------
    bootstrap_results : List[Tuple[Dict[str, float], List[str]]]
        Accumulated (fmec_by_model, selected_set) from the B nested bootstraps.
    point_fmec_table : pd.DataFrame
        The FMEC table on the full data (carrying 'model_id', 'fmec').
    S_star : List[str]
        The ensemble selected on the full data.
    qualified_pool : List[str]
        The canonical ordered model IDs.
    config : Mapping[str, Any]
        The study configuration; supplies ci_level, the variance-component B, and
        (runtime-injected) the fusion residual sigma_u.
    primary_tensor, judge_replicates_df : pd.DataFrame
        Raw data for the targeted variance decomposition and jackknife.
    rng_bundle : Dict[str, np.random.Generator]
        The RNG bundle (uses the dedicated 'bootstrap_seed' stream).

    Returns
    -------
    Tuple[pd.DataFrame, Dict[str, float], float, Dict[str, float]]
        - updated_fmec_table : FMEC table with ci_low/ci_high/sel_freq populated.
        - variance_components : {sigma2_task, sigma2_gen, sigma2_adj, sigma2_fusion,
          sigma2_total} (Eq. 20a).
        - RSP : the Recommendation-Stability Probability (Eq. 51).
        - selection_freq : model_id -> bootstrap selection frequency.
    """
    # Number of nested bootstrap replicates supplied.
    B = len(bootstrap_results)
    # Guard against an empty bootstrap (division by zero downstream).
    if B == 0:
        raise ValueError("compute_validation_metrics requires a non-empty bootstrap_results list.")

    # Extract the point FMEC per model for BCa centering.
    point_fmecs: Dict[str, float] = point_fmec_table.set_index("model_id")["fmec"].to_dict()

    # Pre-allocate the per-model bootstrap FMEC arrays and selection counters.
    boot_fmecs_arrays: Dict[str, np.ndarray] = {m_id: np.zeros(B) for m_id in qualified_pool}
    selection_counts: Dict[str, int] = {m_id: 0 for m_id in qualified_pool}
    # Counter for exact-set matches (RSP numerator, Eq. 51).
    exact_match_count = 0
    # The selected set as a set for order-invariant comparison.
    S_star_set = set(S_star)

    # Reorganize the bootstrap results into per-model arrays and selection tallies.
    for b, (fmec_dict, S_b) in enumerate(bootstrap_results):
        # Record each model's resampled FMEC and accumulate its selection frequency.
        for m_id in qualified_pool:
            boot_fmecs_arrays[m_id][b] = fmec_dict[m_id]
            if m_id in S_b:
                selection_counts[m_id] += 1
        # Increment the exact-match counter when the resampled selection equals S_star.
        if set(S_b) == S_star_set:
            exact_match_count += 1

    # Recommendation-Stability Probability: fraction of resamples reproducing S_star exactly (Eq. 51).
    RSP = float(exact_match_count) / float(B)
    # Per-model bootstrap selection frequency.
    selection_freq: Dict[str, float] = {m_id: float(c) / float(B) for m_id, c in selection_counts.items()}

    # ---- Delete-one-task jackknife for the BCa acceleration factor ----
    # Unique tasks define the clusters for the delete-one-cluster jackknife.
    unique_tasks = primary_tensor["task_id"].unique()
    # Number of task clusters.
    m_tasks = len(unique_tasks)
    # Pre-allocate per-model jackknife FMEC arrays.
    jackknife_fmecs: Dict[str, np.ndarray] = {m_id: np.zeros(m_tasks) for m_id in qualified_pool}
    # Recompute the estimation chain on each delete-one-task subset.
    for i, task_to_drop in enumerate(unique_tasks):
        # Drop one task cluster (consistent with the nested cluster resampling).
        jk_tensor = primary_tensor[primary_tensor["task_id"] != task_to_drop].copy()
        # Re-run the deterministic estimation chain with a unique seed offset.
        jk_fmec_dict, _ = _execute_single_bootstrap_chain(
            resampled_tensor=jk_tensor, qualified_pool=qualified_pool, config=config, b_index=999999 + i
        )
        # Store the jackknife FMEC per model.
        for m_id in qualified_pool:
            jackknife_fmecs[m_id][i] = jk_fmec_dict[m_id]

    # ---- BCa intervals (z0 bias correction + jackknife acceleration) ----
    # Confidence level from config (e.g., 0.95).
    ci_level = config["uncertainty"]["bootstrap"]["ci_level"]
    # Compute the BCa lower/upper bounds per model.
    bca_intervals = _compute_bca_intervals(
        bootstrap_fmecs=boot_fmecs_arrays, point_fmecs=point_fmecs,
        jackknife_fmecs=jackknife_fmecs, ci_level=ci_level
    )

    # Update the FMEC table with the BCa intervals and selection frequencies.
    updated_fmec_table = point_fmec_table.copy()
    updated_fmec_table["ci_low"] = updated_fmec_table["model_id"].map(lambda x: bca_intervals[x][0])
    updated_fmec_table["ci_high"] = updated_fmec_table["model_id"].map(lambda x: bca_intervals[x][1])
    updated_fmec_table["sel_freq"] = updated_fmec_table["model_id"].map(selection_freq)

    # ---- Single-axis variance components (Eq. 20a) ----
    # Variance-component bootstrap size, sourced from config (deliberate approximation; documented).
    v_comp_B = int(config["uncertainty"]["bootstrap"].get("variance_component_B", 100))
    # The dedicated bootstrap RNG stream (seed 13) ensures independence from the MC stream.
    v_rng = rng_bundle["bootstrap_seed"]

    # Helper: estimate one variance component by resampling a single axis and taking the variance of
    # the decomposed functional (the pool-mean FMEC) across resamples.
    def _get_variance_for_mode(mode: str) -> float:
        # Generate v_comp_B single-axis resamples for the requested axis.
        generator = generate_nested_bootstrap_samples(
            primary_tensor, judge_replicates_df, config, v_rng, v_comp_B, resample_mode=mode
        )
        # Accumulate the pool-mean FMEC across resamples (the functional theta being decomposed).
        mean_fmecs: List[float] = []
        for b, tensor in enumerate(generator):
            # Re-run the deterministic estimation chain on the single-axis resample.
            f_dict, _ = _execute_single_bootstrap_chain(tensor, qualified_pool, config, b)
            # Record the pool-mean FMEC for this resample.
            mean_fmecs.append(np.mean(list(f_dict.values())))
        # Return the unbiased sample variance (ddof=1) of the functional across resamples.
        return float(np.var(mean_fmecs, ddof=1))

    # Task-axis variance component.
    sigma2_task = _get_variance_for_mode("task_only")
    # Generation-axis variance component.
    sigma2_gen = _get_variance_for_mode("gen_only")
    # Adjudication-axis variance component (only if adjudication replicates exist).
    sigma2_adj = _get_variance_for_mode("adj_only") if not judge_replicates_df.empty else 0.0

    # ---- Fusion-simulation residual sigma_u in quadrature (Eq. 20a) ----
    # Read sigma_u from the config mapping at the documented runtime-injected path (default 0.0).
    sigma_u: float = 0.0
    try:
        # The orchestrator injects the Task 18 fusion residual here from fusion_diagnostics['sigma_u'].
        sigma_u = float(_get_nested_value(config, "uncertainty.fusion_residual_sigma_u"))
    except (KeyError, TypeError, ValueError):
        # If unavailable, default to 0.0 and warn so the omission is visible (not silent).
        logger.warning(
            "compute_validation_metrics: fusion residual sigma_u not found at "
            "'uncertainty.fusion_residual_sigma_u'; defaulting sigma2_fusion to 0.0 (Eq. 20a quadrature incomplete)."
        )
        sigma_u = 0.0
    # The fusion variance contribution is sigma_u squared.
    sigma2_fusion = float(sigma_u) ** 2

    # Total uncertainty: the three single-axis components plus the fusion residual in quadrature (Eq. 20a).
    sigma2_total = sigma2_task + sigma2_gen + sigma2_adj + sigma2_fusion

    # Assemble the variance-component report (each component separate, plus the fusion term and total).
    variance_components: Dict[str, float] = {
        "sigma2_task": sigma2_task,       # task-axis component
        "sigma2_gen": sigma2_gen,         # generation-axis component
        "sigma2_adj": sigma2_adj,         # adjudication-axis component
        "sigma2_fusion": sigma2_fusion,   # fusion-simulation residual contribution (Eq. 20a)
        "sigma2_total": sigma2_total,     # total uncertainty including sigma_u in quadrature
    }

    # Return the updated FMEC table, the variance decomposition, RSP, and selection frequencies.
    return updated_fmec_table, variance_components, RSP, selection_freq

# -------------------------------------------------------------------------------------------------------------------------------
# Task 19, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def bootstrap_validate(
    primary_tensor: pd.DataFrame,
    judge_replicates_df: pd.DataFrame,
    point_fmec_table: pd.DataFrame,
    S_star: List[str],
    qualified_pool: List[str],
    config: Mapping[str, Any],
    rng_bundle: Dict[str, np.random.Generator]
) -> Tuple[pd.DataFrame, Dict[str, float], float, Dict[str, float]]:
    """
    Orchestrates Task 19: Executes the nested cluster bootstrap, computes BCa
    confidence intervals, isolates variance components, and calculates stability.

    Parameters
    ----------
    primary_tensor : pd.DataFrame
        The 17-column canonical released artifact.
    judge_replicates_df : pd.DataFrame
        The DataFrame of adjudication replicates.
    point_fmec_table : pd.DataFrame
        The FMEC table computed on the full dataset.
    S_star : List[str]
        The ensemble selected on the full dataset.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    config : Mapping[str, Any]
        The immutable study configuration.
    rng_bundle : Dict[str, np.random.Generator]
        The bundle of independent random number generators.

    Returns
    -------
    Tuple[pd.DataFrame, Dict[str, float], float, Dict[str, float]]
        - updated_fmec_table: The FMEC table with CI and sel_freq populated.
        - variance_components: Dict containing sigma^2_task, gen, adj.
        - RSP: The Recommendation-Stability Probability.
        - selection_freq: Dict mapping model_id to selection frequency.
    """
    B = config["uncertainty"]["bootstrap"]["B"]
    boot_rng = rng_bundle["bootstrap_seed"]

    # Step 1: Initialize the generator for nested bootstrap samples
    tensor_generator = generate_nested_bootstrap_samples(
        primary_tensor=primary_tensor,
        judge_replicates_df=judge_replicates_df,
        config=config,
        rng=boot_rng,
        b_iterations=B,
        resample_mode="nested_cluster"
    )

    bootstrap_results: List[Tuple[Dict[str, float], List[str]]] = []

    # Step 2: Execute the estimation-selection chain in parallel
    # We use ProcessPoolExecutor to bypass the GIL and fully utilize the H100s/A100s
    # max_workers is set to None to let Python optimize based on available cores
    with concurrent.futures.ProcessPoolExecutor() as executor:
        futures = []
        for b, resampled_tensor in enumerate(tensor_generator):
            future = executor.submit(
                _execute_single_bootstrap_chain,
                resampled_tensor,
                qualified_pool,
                config,
                b
            )
            futures.append(future)

        # Collect results as they complete
        for future in concurrent.futures.as_completed(futures):
            bootstrap_results.append(future.result())

    # Step 3: Compute BCa intervals, variance components, and stability metrics
    updated_fmec_table, variance_components, RSP, selection_freq = compute_validation_metrics(
        bootstrap_results=bootstrap_results,
        point_fmec_table=point_fmec_table,
        S_star=S_star,
        qualified_pool=qualified_pool,
        config=config,
        primary_tensor=primary_tensor,
        judge_replicates_df=judge_replicates_df,
        rng_bundle=rng_bundle
    )

    return updated_fmec_table, variance_components, RSP, selection_freq
````

````python
# Task 20 — Confirmatory Hypothesis Testing & Falsification (confirmatory_tests)

# ==============================================================================
# Task 20: Confirmatory Hypothesis Testing & Falsification
# ==============================================================================

# -------------------------------------------------------------------------------------------------------------------------------
# Task 20, Helper Function: Safe Feasible Top-K Selection
# -------------------------------------------------------------------------------------------------------------------------------

def _get_feasible_top_k(
    ranked_models: List[str],
    target_k: int,
    c_vec: np.ndarray,
    l_vec: np.ndarray,
    sigma: np.ndarray,
    c_j: float,
    l_j: float,
    budgets: Dict[str, float],
    qualified_pool: List[str]
) -> List[str]:
    """
    Selects the top K models from a ranked list while strictly enforcing cost,
    latency, and risk budgets. Ensures all additions are budget-feasible.

    Parameters
    ----------
    ranked_models : List[str]
        The list of model IDs ordered by their respective ranking metric.
    target_k : int
        The target cardinality constraint (K) for the selected ensemble.
    c_vec : np.ndarray
        1D array of shape (n_models,) representing mean model inference costs.
    l_vec : np.ndarray
        1D array of shape (n_models,) representing mean model latencies.
    sigma : np.ndarray
        2D array of shape (n_models, n_models) representing the error covariance matrix.
    c_j : float
        The constant inference cost of the judge model.
    l_j : float
        The constant latency of the judge model.
    budgets : Dict[str, float]
        Dictionary containing the budget boundaries: 'C_max', 'L_max', and 'R_max'.
    qualified_pool : List[str]
        The canonical ordered list of qualified model IDs defining the indexing.

    Returns
    -------
    List[str]
        The list of selected model IDs of length <= target_k that satisfies all budgets.

    Raises
    ------
    TypeError
        If input types do not match the expected signatures.
    ValueError
        If vectors or matrix dimensions are mismatched, or if numeric inputs are invalid.
    """
    # Validate that target_k is a positive integer
    if not isinstance(target_k, int) or target_k < 0:
        raise TypeError("The parameter 'target_k' must be a non-negative integer.")

    # Validate that c_j is a non-negative float
    if not isinstance(c_j, (int, float)) or c_j < 0.0:
        raise ValueError("The parameter 'c_j' must be a non-negative float.")

    # Validate that l_j is a non-negative float
    if not isinstance(l_j, (int, float)) or l_j < 0.0:
        raise ValueError("The parameter 'l_j' must be a non-negative float.")

    # Validate the structure and presence of required keys in the budgets dictionary
    required_budgets = {"C_max", "L_max", "R_max"}
    if not isinstance(budgets, dict) or not required_budgets.issubset(budgets.keys()):
        raise ValueError(f"The 'budgets' dictionary must contain the keys: {required_budgets}")

    # Validate that c_vec is a 1D NumPy array matching the length of the qualified pool
    n_models = len(qualified_pool)
    if not isinstance(c_vec, np.ndarray) or c_vec.ndim != 1 or len(c_vec) != n_models:
        raise ValueError(f"The 'c_vec' parameter must be a 1D NumPy array of length {n_models}.")

    # Validate that l_vec is a 1D NumPy array matching the length of the qualified pool
    if not isinstance(l_vec, np.ndarray) or l_vec.ndim != 1 or len(l_vec) != n_models:
        raise ValueError(f"The 'l_vec' parameter must be a 1D NumPy array of length {n_models}.")

    # Validate that sigma is a 2D square NumPy array matching the dimensions of the pool
    if not isinstance(sigma, np.ndarray) or sigma.shape != (n_models, n_models):
        raise ValueError(f"The 'sigma' parameter must be a square 2D NumPy array of shape ({n_models}, {n_models}).")

    # Initialize the list of selected integer indices
    S_indices: List[int] = []
    # Define a small epsilon to prevent floating-point boundary violations
    eps = 1e-9

    # Iterate over the ranked model IDs to select feasible members
    for model_id in ranked_models:
        # Halt the search if we have successfully met the target cardinality K
        if len(S_indices) == target_k:
            break

        # Validate that the model ID exists within our canonical qualified pool
        if model_id not in qualified_pool:
            raise ValueError(f"Model ID '{model_id}' in ranked list is not present in the qualified pool.")

        # Resolve the canonical integer index of the candidate model
        idx = qualified_pool.index(model_id)
        # Form the hypothetical new index set S'
        S_prime = S_indices + [idx]
        s_size = len(S_prime)

        # Compute the hypothetical total cost (additive cost of members + judge)
        c_val = np.sum(c_vec[S_prime]) + c_j
        # Enforce the cost constraint; skip this model if it violates the budget
        if c_val > budgets["C_max"] + eps:
            continue

        # Compute the hypothetical total latency (max latency among members + judge)
        l_val = np.max(l_vec[S_prime]) + l_j
        # Enforce the latency constraint; skip this model if it violates the budget
        if l_val > budgets["L_max"] + eps:
            continue

        # Construct the uniform weight vector w for the hypothetical ensemble S'
        w = np.zeros(n_models, dtype=np.float64)
        w[S_prime] = 1.0 / s_size
        # Compute the hypothetical quadratic portfolio risk: R(w) = w^T * Sigma * w
        r_val = np.dot(w, np.dot(sigma, w))
        # Enforce the risk constraint; skip this model if it violates the budget
        if r_val > budgets["R_max"] + eps:
            continue

        # If the candidate satisfies all constraints, commit its index to the selected set
        S_indices.append(idx)

    # Map the selected integer indices back to their canonical string model IDs
    return [qualified_pool[i] for i in S_indices]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 20, Step 1: Construct the Baseline Ensembles
# -------------------------------------------------------------------------------------------------------------------------------

def construct_baselines(
    S_FMEC: List[str],
    point_fmec_table: pd.DataFrame,
    qualification: pd.DataFrame,
    estimator_projection: pd.DataFrame,
    sigma_df: pd.DataFrame,
    qualified_pool: List[str],
    config: Mapping[str, Any],
    rng: np.random.Generator
) -> Dict[str, List[str]]:
    """
    Constructs the six required baseline ensembles (Bench, Shapley, Info, Random,
    CorrMin, and GreedyUtility), ensuring each strictly adheres to the target
    cardinality K and the budget envelope.

    Parameters
    ----------
    S_FMEC : List[str]
        The FMEC-selected ensemble defining the target cardinality K.
    point_fmec_table : pd.DataFrame
        The FMEC table containing MEC (mec_hat) and IG (ig) estimates.
    qualification : pd.DataFrame
        The qualification table containing the composite benchmark ranks (pct_rank).
    estimator_projection : pd.DataFrame
        The 13-column analytic core tensor containing cost and latency telemetry.
    sigma_df : pd.DataFrame
        The n x n error covariance DataFrame.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    config : Mapping[str, Any]
        The immutable study configuration.
    rng : np.random.Generator
        The dedicated random number generator for random baseline construction.

    Returns
    -------
    Dict[str, List[str]]
        A dictionary mapping the six baseline names to their selected model ID lists.

    Raises
    ------
    TypeError
        If input DataFrames or parameters do not match their expected types.
    ValueError
        If input matrices or lists are structurally inconsistent or mismatched.
    """
    # Validate that S_FMEC is a list of strings
    if not isinstance(S_FMEC, list) or not all(isinstance(m, str) for m in S_FMEC):
        raise TypeError("Parameter 'S_FMEC' must be a list of string model IDs.")

    # Validate that point_fmec_table is a valid pandas DataFrame
    if not isinstance(point_fmec_table, pd.DataFrame):
        raise TypeError("Parameter 'point_fmec_table' must be a pandas DataFrame.")

    # Validate that qualification is a valid pandas DataFrame
    if not isinstance(qualification, pd.DataFrame):
        raise TypeError("Parameter 'qualification' must be a pandas DataFrame.")

    # Validate that estimator_projection is a valid pandas DataFrame
    if not isinstance(estimator_projection, pd.DataFrame):
        raise TypeError("Parameter 'estimator_projection' must be a pandas DataFrame.")

    # Validate that sigma_df is a valid pandas DataFrame
    if not isinstance(sigma_df, pd.DataFrame):
        raise TypeError("Parameter 'sigma_df' must be a pandas DataFrame.")

    # Validate that qualified_pool is a non-empty list of strings
    if not isinstance(qualified_pool, list) or len(qualified_pool) == 0:
        raise ValueError("Parameter 'qualified_pool' must be a non-empty list of strings.")

    # Define the target cardinality K from the FMEC selection
    target_k = len(S_FMEC)
    # Initialize the output dictionary for baselines
    baselines: Dict[str, List[str]] = {}

    # Handle the boundary case of an empty selection to prevent divisions by zero
    if target_k == 0:
        # Return empty lists for all six baseline configurations
        return {name: [] for name in ["Bench", "Random", "CorrMin", "Shapley", "Info", "GreedyUtility"]}

    # Extract optimization parameters, costs, latencies, and budgets from the config
    fmec_vec, c_vec, l_vec, c_j, l_j, budgets = prepare_optimization_vectors(
        point_fmec_table, estimator_projection, qualified_pool, config
    )
    # Extract and align the covariance matrix to the canonical pool ordering
    sigma_matrix = sigma_df.reindex(index=qualified_pool, columns=qualified_pool).values.astype(np.float64)

    # --- 1. Benchmark top-K (S_Bench) ---
    # Reindex qualification to match the canonical pool
    qual_aligned = qualification.set_index("model_id").reindex(qualified_pool)
    # Sort the models by their composite benchmark percentile rank descending
    ranked_bench = qual_aligned.sort_values("pct_rank", ascending=False).index.tolist()
    # Select the top K feasible models under budgets
    baselines["Bench"] = _get_feasible_top_k(ranked_bench, target_k, c_vec, l_vec, sigma_matrix, c_j, l_j, budgets, qualified_pool)

    # --- 2. Shapley-only (S_Shapley) ---
    # Sort the models by their expected marginal utility (mec_hat) descending
    ranked_shapley = point_fmec_table.sort_values("mec_hat", ascending=False)["model_id"].tolist()
    # Select the top K feasible models under budgets
    baselines["Shapley"] = _get_feasible_top_k(ranked_shapley, target_k, c_vec, l_vec, sigma_matrix, c_j, l_j, budgets, qualified_pool)

    # --- 3. Information-only (S_Info) ---
    # Sort the models by their conditional information gain (ig) descending
    ranked_info = point_fmec_table.sort_values("ig", ascending=False)["model_id"].tolist()
    # Select the top K feasible models under budgets
    baselines["Info"] = _get_feasible_top_k(ranked_info, target_k, c_vec, l_vec, sigma_matrix, c_j, l_j, budgets, qualified_pool)

    # --- 4. Random (S_Random) ---
    # Create a local copy of the qualified pool to shuffle
    shuffled_pool = list(qualified_pool)
    # Use the dedicated random number generator to shuffle the pool deterministically
    rng.shuffle(shuffled_pool)
    # Select the first K feasible models under budgets from the shuffled pool
    baselines["Random"] = _get_feasible_top_k(shuffled_pool, target_k, c_vec, l_vec, sigma_matrix, c_j, l_j, budgets, qualified_pool)

    # --- 5. Correlation-minimized (S_CorrMin) ---
    # Exhaustive search over all subsets of size K to find the budget-feasible set minimizing off-diagonal covariance
    best_corr = float('inf')
    best_corr_S: List[int] = []
    n_models = len(qualified_pool)
    eps = 1e-9

    # Generate all combinations of size K
    for combo in itertools.combinations(range(n_models), target_k):
        S_prime = list(combo)

        # Enforce the cost constraint
        if np.sum(c_vec[S_prime]) + c_j > budgets["C_max"] + eps:
            continue
        # Enforce the latency constraint
        if np.max(l_vec[S_prime]) + l_j > budgets["L_max"] + eps:
            continue

        # Construct the uniform weight vector w for this candidate combination
        w = np.zeros(n_models, dtype=np.float64)
        w[S_prime] = 1.0 / target_k
        # Enforce the portfolio risk constraint
        if np.dot(w, np.dot(sigma_matrix, w)) > budgets["R_max"] + eps:
            continue

        # Slice the covariance matrix for the subset S'
        sub_sigma = sigma_matrix[np.ix_(S_prime, S_prime)]
        # Compute the sum of off-diagonal covariance elements (correlations)
        off_diag_sum = np.sum(sub_sigma) - np.trace(sub_sigma)

        # Update the best candidate set if this one has lower correlation
        if off_diag_sum < best_corr:
            best_corr = off_diag_sum
            best_corr_S = S_prime

    # Map the best integer indices back to string model IDs
    baselines["CorrMin"] = [qualified_pool[i] for i in best_corr_S]

    # --- 6. Greedy-utility (S_GreedyUtility) ---
    # Build the optimized utility function closure mapping sets to raw expected utility v(S)
    v_func = build_utility_function(estimator_projection, sigma_matrix, config, qualified_pool, config["models"]["judge"])

    S_greedy: List[int] = []
    current_u = 0.0

    # Iteratively build the greedy-utility ensemble up to target_k
    while len(S_greedy) < target_k:
        best_gain = -float('inf')
        best_candidate = -1

        # Scan all available candidates
        for i in range(n_models):
            if i in S_greedy:
                continue
            S_prime = S_greedy + [i]

            # Enforce the cost constraint
            if np.sum(c_vec[S_prime]) + c_j > budgets["C_max"] + eps:
                continue
            # Enforce the latency constraint
            if np.max(l_vec[S_prime]) + l_j > budgets["L_max"] + eps:
                continue
            # Enforce the risk constraint
            w = np.zeros(n_models, dtype=np.float64)
            w[S_prime] = 1.0 / len(S_prime)
            if np.dot(w, np.dot(sigma_matrix, w)) > budgets["R_max"] + eps:
                continue

            # Compute raw expected utility
            u_val = v_func(S_prime)
            # Compute the marginal utility gain
            delta_u = u_val - current_u

            # Select the candidate maximizing raw utility gain
            if delta_u > best_gain:
                best_gain = delta_u
                best_candidate = i

        # Halt if no feasible candidate remains
        if best_candidate == -1:
            break
        S_greedy.append(best_candidate)
        # Update the baseline utility for the next round
        current_u = v_func(S_greedy)

    # Map the final greedy indices back to model IDs
    baselines["GreedyUtility"] = [qualified_pool[i] for i in S_greedy]

    return baselines

# -------------------------------------------------------------------------------------------------------------------------------
# Task 20, Helper Function: Compute Task-by-Task Utility
# -------------------------------------------------------------------------------------------------------------------------------

def _compute_per_task_utility(
    S: List[str],
    estimator_projection: pd.DataFrame,
    sigma_df: pd.DataFrame,
    qualified_pool: List[str],
    config: Mapping[str, Any]
) -> np.ndarray:
    """
    Computes the utility v(S) for a specific ensemble on a task-by-task basis.
    This is required for paired statistical testing (H1, H5).

    Parameters
    ----------
    S : List[str]
        The list of model IDs included in the ensemble.
    estimator_projection : pd.DataFrame
        The 13-column analytic core tensor containing correct, cost, and latency columns.
    sigma_df : pd.DataFrame
        The n x n covariance matrix.
    qualified_pool : List[str]
        The canonical ordered list of model IDs defining the indexing.
    config : Mapping[str, Any]
        The immutable study configuration containing utility weights and judge info.

    Returns
    -------
    np.ndarray
        A 1D NumPy array of shape (n_tasks,) representing the utility of the ensemble per task.

    Raises
    ------
    TypeError
        If input types do not match the expected signatures.
    ValueError
        If vectors or matrix dimensions are mismatched, or if model IDs are invalid.
    """
    # Validate that S is a list of strings
    if not isinstance(S, list) or not all(isinstance(m, str) for m in S):
        raise TypeError("The parameter 'S' must be a list of string model IDs.")

    # Validate that estimator_projection is a valid pandas DataFrame
    if not isinstance(estimator_projection, pd.DataFrame):
        raise TypeError("The parameter 'estimator_projection' must be a pandas DataFrame.")

    # Validate that sigma_df is a valid pandas DataFrame
    if not isinstance(sigma_df, pd.DataFrame):
        raise TypeError("The parameter 'sigma_df' must be a pandas DataFrame.")

    # Validate that qualified_pool is a non-empty list of strings
    if not isinstance(qualified_pool, list) or len(qualified_pool) == 0:
        raise ValueError("The parameter 'qualified_pool' must be a non-empty list of strings.")

    # Calculate the total number of unique tasks in the evaluation set
    unique_tasks = estimator_projection["task_id"].unique()
    n_tasks = len(unique_tasks)

    # Boundary case: if the ensemble is empty, return a task-length vector of zeros
    if not S:
        return np.zeros(n_tasks, dtype=np.float64)

    # Validate that all models in S exist within our canonical qualified pool
    for m in S:
        if m not in qualified_pool:
            raise ValueError(f"Model ID '{m}' in ensemble S is not present in the qualified pool.")

    # Extract utility preference weights from the configuration [Eq. 3]
    weights = config["estimation"]["utility_weights"]
    alpha: float = weights["alpha"]
    beta: float = weights["beta"]
    gamma: float = weights["gamma"]
    delta: float = weights["delta"]

    # --- 1. Compute Per-Task Quality Q_t(S) ---
    # Group by task and model, compute the mean correctness across replicates
    mean_correctness = estimator_projection.groupby(["task_id", "model_id"], observed=True)["correct"].mean()
    # Unstack to form the dense (tasks x models) DataFrame
    A_df = mean_correctness.unstack(level="model_id").reindex(columns=qualified_pool)
    # Threshold at > 0.5 to resolve the majority vote per (task, model) pair
    A_matrix = (A_df.values > 0.5).astype(np.int8)

    # Resolve the integer indices of the ensemble S
    S_indices = [qualified_pool.index(m) for m in S]
    s_size = len(S_indices)

    # Extract the sub-matrix for the models in S
    A_S = A_matrix[:, S_indices]
    # Compute the row-wise sum (number of correct models per task)
    row_sums = np.sum(A_S, axis=1)
    # Apply the majority threshold
    threshold = s_size / 2.0
    majority_mask = row_sums > threshold

    # Execute reliability-based tie-breaking for even-sized ensembles [Eq. 3a]
    if s_size % 2 == 0:
        r_vec = A_matrix.mean(axis=0)
        tie_mask = (row_sums == threshold)
        if np.any(tie_mask):
            best_model_idx_in_S = np.argmax(r_vec[S_indices])
            majority_mask[tie_mask] = A_S[tie_mask, best_model_idx_in_S].astype(bool)

    # Convert the boolean correctness mask to a task-varying quality vector
    q_vec = majority_mask.astype(np.float64)

    # --- 2. Compute Constant Cost, Latency, and Risk ---
    # Cost, latency, and risk are constant across tasks for a fixed ensemble S.
    cost_series = estimator_projection.groupby("model_id", observed=True)["cost_usd"].mean()
    latency_series = estimator_projection.groupby("model_id", observed=True)["latency_s"].mean()

    # Extract cost and latency vectors aligned with the qualified pool
    c_vec = cost_series.reindex(qualified_pool).fillna(0.0).values
    l_vec = latency_series.reindex(qualified_pool).fillna(0.0).values

    # Extract the judge's overhead
    judge_id = config["models"]["judge"]
    c_j = float(cost_series.get(judge_id, 0.0))
    l_j = float(latency_series.get(judge_id, 0.0))

    # Compute the raw cost of the ensemble (additive cost of members + judge)
    c_val = np.sum(c_vec[S_indices]) + c_j
    # Compute the raw latency of the ensemble (max latency of members + judge)
    l_val = np.max(l_vec[S_indices]) + l_j

    # Construct the uniform weight vector w for the ensemble S
    w = np.zeros(len(qualified_pool), dtype=np.float64)
    w[S_indices] = 1.0 / s_size
    # Align the covariance matrix
    sigma_matrix = sigma_df.reindex(index=qualified_pool, columns=qualified_pool).values
    # Compute the raw portfolio risk: R(w) = w^T * Sigma * w [Eq. 11]
    r_val = np.dot(w, np.dot(sigma_matrix, w))

    # --- 3. Normalization Constants ---
    # Normalization prevents cost, latency, or risk from numerically dominating the objective.
    c_max = np.sum(c_vec) + c_j or 1.0
    l_max = (np.max(l_vec) + l_j) if len(l_vec) > 0 else 1.0
    r_max = np.max(np.diag(sigma_matrix)) if sigma_matrix.size > 0 else 1.0

    # Normalize each penalty to [0, 1]
    c_norm = c_val / c_max
    l_norm = l_val / l_max
    r_norm = r_val / r_max

    # --- 4. Compose Per-Task Utility ---
    # U_t = alpha * Q_t - beta * R_norm - gamma * C_norm - delta * L_norm [Eq. 3]
    u_vec = (alpha * q_vec) - (beta * r_norm) - (gamma * c_norm) - (delta * l_norm)

    return u_vec

# -------------------------------------------------------------------------------------------------------------------------------
# Task 20, Step 2: Execute the five pre-registered hypothesis tests
# -------------------------------------------------------------------------------------------------------------------------------

def execute_raw_hypothesis_tests(
    S_FMEC: List[str],
    baselines: Dict[str, List[str]],
    point_fmec_table: pd.DataFrame,
    qualification: pd.DataFrame,
    estimator_projection: pd.DataFrame,
    sigma_df: pd.DataFrame,
    qualified_pool: List[str],
    config: Mapping[str, Any],
    rng: np.random.Generator,
    primary_tensor: pd.DataFrame,
    judge_replicates_df: pd.DataFrame
) -> Tuple[Dict[str, float], Dict[str, np.ndarray]]:
    """
    Execute the raw (unadjusted) tests for H1-H5, returning p-values and the
    per-task utility arrays for downstream effect sizes.

    Rationale (Eqs. 44-48a; detailed tasklist Task 20, Step 1). H1 utility
    dominance (Shapiro-gated paired t / Wilcoxon, one-sided); H2 IG-MEC Spearman
    with a permutation null; H3 one-sided risk reduction of FMEC vs the lambda=0
    ablation, with the quadratic risk w_S^T Σ w_S evaluated under the CORRECT
    per-set uniform weights for each ensemble independently (fixing the stale-w_vec
    bug that forced p=1.0); H4 a genuine cluster-bootstrap test of Kendall's tau
    against the 0.80 boundary (replacing hardcoded sentinel p-values), retaining the
    n<10 underpowered gate; H5 open-weight non-inferiority with the EXACT cost-tied
    margin Delta_U = max(gamma*(C_front - C_open), 0.02*|U_front|) (Eq. 48a; the
    stray constant 0.02 removed).

    Parameters
    ----------
    S_FMEC : List[str]
        The FMEC-selected ensemble.
    baselines : Dict[str, List[str]]
        Baseline ensembles (must include 'Bench').
    point_fmec_table : pd.DataFrame
        FMEC table carrying 'mec_hat', 'ig', 'fmec', 'model_id'.
    qualification : pd.DataFrame
        Qualification table with 'pct_rank' and 'params_b'.
    estimator_projection : pd.DataFrame
        The 13-column analytic core tensor.
    sigma_df : pd.DataFrame
        The n x n covariance matrix.
    qualified_pool : List[str]
        Canonical ordered model IDs.
    config : Mapping[str, Any]
        Study configuration.
    rng : np.random.Generator
        Dedicated RNG for permutation/bootstrap nulls.
    primary_tensor, judge_replicates_df : pd.DataFrame
        Raw data for the H3 resampling.

    Returns
    -------
    Tuple[Dict[str, float], Dict[str, np.ndarray]]
        - raw_p_values : {H1..H5} unadjusted p-values (H4 may be NaN if underpowered).
        - utility_arrays : per-task utility vectors for effect sizes.

    Raises
    ------
    TypeError
        If core inputs are not of the expected types.
    """
    # Validate the principal input types (fail fast on structural inconsistency).
    if not isinstance(S_FMEC, list) or not all(isinstance(m, str) for m in S_FMEC):
        raise TypeError("Parameter 'S_FMEC' must be a list of string model IDs.")
    if not isinstance(baselines, dict):
        raise TypeError("Parameter 'baselines' must be a dictionary.")
    for name, obj in (("point_fmec_table", point_fmec_table), ("qualification", qualification),
                      ("estimator_projection", estimator_projection), ("sigma_df", sigma_df)):
        if not isinstance(obj, pd.DataFrame):
            raise TypeError(f"Parameter '{name}' must be a pandas DataFrame.")

    # Output containers for p-values and the per-task utility arrays.
    raw_p_values: Dict[str, float] = {}
    utility_arrays: Dict[str, np.ndarray] = {}

    # ---- H1: Utility Dominance (Eq. 44) ----
    # The benchmark-top-K baseline ensemble.
    S_Bench = baselines.get("Bench", [])
    # Per-task utility vectors for FMEC and the benchmark baseline.
    u_fmec = _compute_per_task_utility(S_FMEC, estimator_projection, sigma_df, qualified_pool, config)
    u_bench = _compute_per_task_utility(S_Bench, estimator_projection, sigma_df, qualified_pool, config)
    # Persist the arrays for the effect-size step.
    utility_arrays["u_fmec"] = u_fmec
    utility_arrays["u_bench"] = u_bench
    # Per-task differences.
    diffs = u_fmec - u_bench
    # Shapiro-Wilk normality gate selects the paired test.
    _, p_shapiro = stats.shapiro(diffs)
    if p_shapiro > 0.05:
        # Normal differences: one-sided paired t-test (FMEC > Bench).
        res = stats.ttest_rel(u_fmec, u_bench, alternative="greater")
        raw_p_values["H1"] = float(res.pvalue)
    else:
        # Non-normal: one-sided Wilcoxon signed-rank (guard the all-zero-difference degeneracy).
        if np.all(diffs == 0):
            raw_p_values["H1"] = 1.0
        else:
            res = stats.wilcoxon(u_fmec, u_bench, alternative="greater")
            raw_p_values["H1"] = float(res.pvalue)

    # ---- H2: IG-MEC Coupling (Eq. 45) ----
    # IG and MEC vectors across the pool.
    ig_vals = point_fmec_table["ig"].values
    mec_vals = point_fmec_table["mec_hat"].values
    # Observed Spearman rank correlation.
    rho_obs, _ = stats.spearmanr(ig_vals, mec_vals)
    # Permutation null of fixed size: permute MEC to break the coupling.
    n_perms = 10000
    perm_rhos = np.zeros(n_perms)
    for i in range(n_perms):
        # One permutation of MEC against fixed IG.
        perm_mec = rng.permutation(mec_vals)
        perm_rhos[i], _ = stats.spearmanr(ig_vals, perm_mec)
    # One-sided empirical p-value: proportion of permuted rho at or above the observed.
    raw_p_values["H2"] = float(np.sum(perm_rhos >= rho_obs) / n_perms)

    # ---- H3: Risk Reduction (Eq. 46) ----
    # Build the risk-agnostic config (lambda = 0) preserving eta.
    config_norisk = dict(config)
    config_norisk["estimation"] = dict(config["estimation"])
    config_norisk["estimation"]["fmec_hyperparams"] = {
        "eta": config["estimation"]["fmec_hyperparams"]["eta"], "lambda": 0.0,
    }
    # Recompute the FMEC objective with lambda = 0 (no risk penalty).
    fmec_norisk = point_fmec_table.copy()
    fmec_norisk["fmec"] = fmec_norisk["mec_hat"] + (config_norisk["estimation"]["fmec_hyperparams"]["eta"] * fmec_norisk["ig"])
    # Prepare vectors and run the greedy selector to obtain S_NoRisk.
    fmec_vec, c_vec, l_vec, c_j, l_j, budgets = prepare_optimization_vectors(
        fmec_norisk, estimator_projection, qualified_pool, config_norisk
    )
    sigma_matrix = sigma_df.reindex(index=qualified_pool, columns=qualified_pool).values.astype(np.float64)
    S_NoRisk_indices = execute_greedy_selection(fmec_vec, c_vec, l_vec, sigma_matrix, c_j, l_j, budgets)
    S_NoRisk = [qualified_pool[i] for i in S_NoRisk_indices]

    # Module-level (loop-invariant) helper: portfolio risk under per-set uniform weights (Eq. 11).
    # CORRECTED: builds and uses the per-set weight vector w_S (the stale-w_vec bug is fixed).
    def _portfolio_risk(S_list: List[str], sig: np.ndarray) -> float:
        # An empty set carries unit (maximal) normalized risk by convention.
        if not S_list:
            return 1.0
        # Construct the uniform simplex weight vector on the members of S (Eq. 11 weights).
        w_S = np.zeros(len(qualified_pool))
        # Indices of the set members in the canonical ordering.
        idx = [qualified_pool.index(m) for m in S_list]
        # Equal weight 1/|S| on each member.
        w_S[idx] = 1.0 / len(idx)
        # Quadratic portfolio risk R(w) = w^T Σ w (Eq. 11), evaluated with the CORRECT per-set weights.
        return float(np.dot(w_S, np.dot(sig, w_S)))

    # Targeted bootstrap to form the FMEC vs NoRisk risk distributions.
    b_h3 = 1000
    r_fmec_dist = np.zeros(b_h3)
    r_norisk_dist = np.zeros(b_h3)
    # Generate nested-cluster resamples for the risk distributions.
    tensor_generator = generate_nested_bootstrap_samples(
        primary_tensor, judge_replicates_df, config, rng, b_h3, resample_mode="nested_cluster"
    )
    # For each resample, re-estimate Σ and evaluate both portfolios' risks.
    for b, resampled_tensor in enumerate(tensor_generator):
        # Build the analytic-core projection on the resample.
        est_proj, _ = create_estimator_projection(resampled_tensor, [])
        # Re-estimate the covariance on the resampled data.
        sig_df, _, _ = estimate_risk(est_proj, qualified_pool, config)
        # Align the covariance to the canonical ordering.
        sig_mat = sig_df.reindex(index=qualified_pool, columns=qualified_pool).values
        # Evaluate FMEC and NoRisk risk under their OWN per-set weights.
        r_fmec_dist[b] = _portfolio_risk(S_FMEC, sig_mat)
        r_norisk_dist[b] = _portfolio_risk(S_NoRisk, sig_mat)
    # One-sided test (R_FMEC < R_NoRisk): p = proportion of resamples where FMEC risk is NOT lower.
    raw_p_values["H3"] = float(np.sum(r_fmec_dist >= r_norisk_dist) / b_h3)

    # ---- H4: Insufficiency of Benchmark Rank (Eq. 47) — genuine bootstrap test ----
    # Align qualification to the canonical pool ordering.
    qual_aligned = qualification.set_index("model_id").reindex(qualified_pool)
    # Benchmark composite ranks and point FMEC scores (the two paired orderings).
    bench_scores = qual_aligned["pct_rank"].values.astype(np.float64)
    fmec_scores = point_fmec_table.set_index("model_id").reindex(qualified_pool)["fmec"].values.astype(np.float64)
    # Observed Kendall's tau between the benchmark and FMEC orderings.
    tau_obs, _ = stats.kendalltau(bench_scores, fmec_scores)
    # The discordance boundary tau_threshold = 1 - delta_tau (e.g., 0.80).
    delta_tau = config["stat_tests"]["delta_tau"]
    tau_threshold = 1.0 - delta_tau
    # Number of models (the bootstrap resampling unit).
    n_models = len(qualified_pool)
    # Power gate: Kendall's tau is uninformative for very small pools.
    if n_models < 10:
        # Report underpowered as NaN (handled downstream; NOT falsified).
        raw_p_values["H4"] = np.nan
    else:
        # Cluster-bootstrap the paired (bench, fmec) vectors to form the sampling distribution of tau.
        b_tau = 10000
        tau_boot = np.empty(b_tau)
        # Integer index base for resampling model pairs with replacement.
        base_idx = np.arange(n_models)
        for i in range(b_tau):
            # Resample model indices with replacement (a cluster bootstrap over the paired ranks).
            samp = rng.choice(base_idx, size=n_models, replace=True)
            # Recompute tau on the resampled pairs (tau-b handles ties from duplicate draws).
            tau_i, _ = stats.kendalltau(bench_scores[samp], fmec_scores[samp])
            # Store the resampled tau (NaN-guard for degenerate resamples).
            tau_boot[i] = tau_i if np.isfinite(tau_i) else 0.0
        # Smoothed one-sided bootstrap p-value for H0: tau >= tau_threshold (benchmark sufficient).
        # Small p => the boundary is implausible given the data => support H4 (discordance).
        raw_p_values["H4"] = float((1.0 + np.sum(tau_boot >= tau_threshold)) / (b_tau + 1.0))

    # ---- H5: Open-Weight Non-Inferiority (Eqs. 48, 48a) ----
    # Open-weight models in S_FMEC (params_b present).
    open_models = qual_aligned[qual_aligned["params_b"].notna()].index.tolist()
    S_Open = [m for m in S_FMEC if m in open_models]
    # Identify the best frontier model by mean per-task utility.
    frontier_candidates = config["models"]["frontier_baselines"]
    best_frontier = None
    best_u = -float("inf")
    for m in frontier_candidates:
        # Only consider frontier candidates that qualified.
        if m in qualified_pool:
            # Mean per-task utility of the singleton frontier model.
            u_m = np.mean(_compute_per_task_utility([m], estimator_projection, sigma_df, qualified_pool, config))
            # Track the best frontier model.
            if u_m > best_u:
                best_u = u_m
                best_frontier = m
    # The frontier comparator set (possibly empty).
    M_Frontier = [best_frontier] if best_frontier else []
    # Per-task utility vectors for the open subset and the frontier comparator.
    u_open = _compute_per_task_utility(S_Open, estimator_projection, sigma_df, qualified_pool, config)
    u_front = _compute_per_task_utility(M_Frontier, estimator_projection, sigma_df, qualified_pool, config)
    # Mean per-model costs for the cost-tied margin.
    cost_series = estimator_projection.groupby("model_id", observed=True)["cost_usd"].mean()
    c_open = sum(cost_series.get(m, 0.0) for m in S_Open)
    c_front = sum(cost_series.get(m, 0.0) for m in M_Frontier)
    # Cost-penalty weight gamma and the frontier mean utility.
    gamma = config["estimation"]["utility_weights"]["gamma"]
    u_front_mean = float(np.mean(u_front))
    # EXACT cost-tied non-inferiority margin (Eq. 48a): max(gamma*(C_front - C_open), 0.02*|U_front|).
    # The stray bare 0.02 constant present in the prior implementation is removed.
    delta_u = max(gamma * (c_front - c_open), 0.02 * abs(u_front_mean))
    # Non-inferiority shift: (U_open - U_front + Delta_U); H0: shift <= 0.
    ni_diffs = u_open - u_front + delta_u
    # Shapiro gate selects the one-sample test.
    _, p_shapiro_ni = stats.shapiro(ni_diffs)
    if p_shapiro_ni > 0.05:
        # Normal: one-sample t-test (mean shift > 0 => non-inferior).
        res_ni = stats.ttest_1samp(ni_diffs, 0.0, alternative="greater")
        raw_p_values["H5"] = float(res_ni.pvalue)
    else:
        # Non-normal: one-sided Wilcoxon (guard the all-zero degeneracy).
        if np.all(ni_diffs == 0):
            raw_p_values["H5"] = 1.0
        else:
            res_ni = stats.wilcoxon(ni_diffs, alternative="greater")
            raw_p_values["H5"] = float(res_ni.pvalue)

    # Return the unadjusted p-values and the per-task utility arrays.
    return raw_p_values, utility_arrays

# -------------------------------------------------------------------------------------------------------------------------------
# Task 20, Step 3: Apply Holm-Bonferroni FWER control and compute effect sizes
# -------------------------------------------------------------------------------------------------------------------------------

def apply_fwer_and_effect_sizes(
    raw_p_values: Dict[str, float],
    utility_arrays: Dict[str, np.ndarray],
    config: Mapping[str, Any]
) -> Tuple[Dict[str, Dict[str, Any]], Dict[str, float]]:
    """
    Applies the Holm-Bonferroni step-down procedure to control the Family-Wise
    Error Rate (FWER) and computes standardized effect sizes (Cohen's d, Cliff's
    delta, and Relative Utility Gain).

    Parameters
    ----------
    raw_p_values : Dict[str, float]
        Dictionary mapping hypothesis names (H1-H5) to their unadjusted p-values.
    utility_arrays : Dict[str, np.ndarray]
        Dictionary containing the task-level utility vectors (u_fmec, u_bench).
    config : Mapping[str, Any]
        The immutable study configuration containing 'alpha_level'.

    Returns
    -------
    Tuple[Dict[str, Dict[str, Any]], Dict[str, float]]
        - test_results: Dict mapping hypothesis to its raw, adjusted p-value,
          and rejection decision.
        - effect_sizes: Dict containing Cohen's d, Cliff's delta, and RUG.

    Raises
    ------
    TypeError
        If input types do not match the expected signatures.
    ValueError
        If inputs are structurally inconsistent.
    """
    # Validate that raw_p_values is a dictionary
    if not isinstance(raw_p_values, dict):
        raise TypeError("The parameter 'raw_p_values' must be a dictionary.")

    # Validate that utility_arrays is a dictionary containing the required keys
    if not isinstance(utility_arrays, dict) or "u_fmec" not in utility_arrays or "u_bench" not in utility_arrays:
        raise ValueError("The 'utility_arrays' must contain 'u_fmec' and 'u_bench' vectors.")

    # Extract the pre-registered significance level alpha
    alpha: float = config["stat_tests"]["alpha_level"]

    # Filter out NaN p-values to prevent them from corrupting the step-down sequence (e.g., underpowered H4)
    valid_p = {k: v for k, v in raw_p_values.items() if not np.isnan(v)}

    # Sort the valid p-values in ascending order
    sorted_tests = sorted(valid_p.items(), key=lambda item: item[1])

    # Initialize the test results dictionary with default values
    test_results: Dict[str, Dict[str, Any]] = {
        k: {"p_raw": v, "p_adj": np.nan, "reject": False}
        for k, v in raw_p_values.items()
    }

    m = len(sorted_tests)
    # The rejection flag remains True until a hypothesis fails to clear its adjusted threshold
    reject_flag = True

    # Iterate over the sorted hypotheses to apply the step-down Holm-Bonferroni logic
    for j, (h_name, p_val) in enumerate(sorted_tests, start=1):
        # Compute the adjusted significance threshold for step j
        threshold = alpha / (m - j + 1)

        # Compute the adjusted p-value: p_val * (m - j + 1), capped at 1.0
        p_adj = min(1.0, p_val * (m - j + 1))
        # Enforce the mathematical monotonicity constraint: p_adj(j) >= p_adj(j-1)
        if j > 1:
            prev_h_name = sorted_tests[j-2][0]
            p_adj = max(p_adj, test_results[prev_h_name]["p_adj"])

        # Update the results dictionary
        test_results[h_name]["p_adj"] = p_adj

        # A hypothesis is rejected iff all preceding hypotheses were rejected AND its p-value is <= its threshold
        if reject_flag and p_val <= threshold:
            test_results[h_name]["reject"] = True
        else:
            # Once a single hypothesis fails, the rejection chain is broken
            reject_flag = False
            test_results[h_name]["reject"] = False

    # Compute standardized effect sizes (Treatment vs Benchmark)
    u_f = utility_arrays["u_fmec"]
    u_b = utility_arrays["u_bench"]

    effect_sizes: Dict[str, float] = {}

    if len(u_f) > 0 and len(u_b) > 0:
        # Compute sample means
        mean_f, mean_b = np.mean(u_f), np.mean(u_b)
        # Compute sample variances with Bessel's correction (ddof=1)
        var_f, var_b = np.var(u_f, ddof=1), np.var(u_b, ddof=1)

        # Compute the pooled standard deviation [Eq. 49]
        s_pooled = np.sqrt((var_f + var_b) / 2.0)
        # Compute Cohen's d (Eq. 49), preventing division-by-zero
        effect_sizes["cohens_d"] = float((mean_f - mean_b) / s_pooled) if s_pooled > 1e-12 else 0.0

        # Compute Cliff's delta (Eq. 50) using highly optimized broadcasting
        # diff_matrix[i, j] = sign( u_f[i] - u_b[j] )
        diff_matrix = np.sign(u_f[:, None] - u_b)
        effect_sizes["cliffs_delta"] = float(np.mean(diff_matrix))

        # Compute the Relative Utility Gain (Eq. 18)
        effect_sizes["relative_utility_gain"] = float((mean_f - mean_b) / abs(mean_b)) if mean_b != 0.0 else 0.0

    return test_results, effect_sizes


# -------------------------------------------------------------------------------------------------------------------------------
# Task 20, Step 4: Apply falsification rules and format final verdicts
# -------------------------------------------------------------------------------------------------------------------------------

def apply_falsification_rules(
    test_results: Dict[str, Dict[str, Any]],
    effect_sizes: Dict[str, float],
    rsp: float,
    config: Mapping[str, Any]
) -> Dict[str, Any]:
    """
    Evaluate the Section 11.13 falsification rules and assign the overall status as
    the MOST SEVERE triggered verdict (severity-ranked, not last-rule-wins).

    Rationale (Section 11.13; detailed tasklist Task 20, Step 3). Each hypothesis is
    marked supported/rejected/inconclusive and the recommendation
    stable/unstable/inconclusive. The prior implementation overwrote `status`
    sequentially, so a falsified-AND-unstable result was reported merely as
    "unstable", losing the more severe verdict. This version evaluates every rule
    independently, accumulates all reasons, and sets overall_status to the most
    severe triggered verdict via the explicit ranking
    falsified > operationally_void > unstable > inconclusive > stable. The
    underpowered-H4 carve-out and the negligibility conjunction (d<0.20 AND
    delta<0.147) are preserved.

    Parameters
    ----------
    test_results : Dict[str, Dict[str, Any]]
        Per-hypothesis results, each with at least 'reject' (bool) and 'p_raw'.
    effect_sizes : Dict[str, float]
        Point effect sizes ('cohens_d', 'cliffs_delta').
    rsp : float
        The Recommendation-Stability Probability (Eq. 51).
    config : Mapping[str, Any]
        Study configuration with 'stat_tests' thresholds.

    Returns
    -------
    Dict[str, Any]
        {'overall_status', 'falsification_reasons', 'hypotheses', 'effect_sizes',
        'stability'} with the severity-ranked overall status.

    Raises
    ------
    TypeError
        If inputs are not of the expected types.
    """
    # Validate input types.
    if not isinstance(test_results, dict):
        raise TypeError("The parameter 'test_results' must be a dictionary.")
    if not isinstance(effect_sizes, dict):
        raise TypeError("The parameter 'effect_sizes' must be a dictionary.")
    if not isinstance(rsp, (int, float)):
        raise TypeError("The parameter 'rsp' must be a float or integer.")

    # Pre-registered decision thresholds.
    thresholds = config["stat_tests"]

    # Explicit severity ranking: larger rank == more severe (drives the final status).
    severity_rank: Dict[str, int] = {
        "stable": 0, "inconclusive": 1, "unstable": 2, "operationally_void": 3, "falsified": 4,
    }
    # Accumulate every triggered verdict and its reason independently.
    triggered: List[str] = []
    reasons: List[str] = []

    # Rule 1: H1 (Utility Dominance) failed to reject the null -> falsified.
    if not test_results.get("H1", {}).get("reject", False):
        triggered.append("falsified")
        reasons.append("H1 (Utility Dominance) failed to reject the null.")

    # Rule 2: H4 (Insufficiency of Rank) failed AND was adequately powered -> falsified.
    p_h4_raw = test_results.get("H4", {}).get("p_raw", 0.0)
    if (not test_results.get("H4", {}).get("reject", False)) and (not (isinstance(p_h4_raw, float) and np.isnan(p_h4_raw))):
        triggered.append("falsified")
        reasons.append("H4 (Insufficiency of Rank) failed; ranks are too concordant (tau > 0.80).")

    # Rule 3: negligible effect sizes (BOTH floors) -> operationally void.
    d_val = effect_sizes.get("cohens_d", 0.0)
    delta_val = effect_sizes.get("cliffs_delta", 0.0)
    if d_val < thresholds["negligible_d"] and delta_val < thresholds["negligible_cliffs_delta"]:
        triggered.append("operationally_void")
        reasons.append(
            f"Effect sizes are negligible (Cohen's d={d_val:.3f} < {thresholds['negligible_d']} "
            f"AND Cliff's delta={delta_val:.3f} < {thresholds['negligible_cliffs_delta']})."
        )

    # Rule 4: stability below the RSP floor -> unstable.
    if rsp < thresholds["rsp_floor"]:
        triggered.append("unstable")
        reasons.append(
            f"Recommendation is unstable (RSP={rsp:.3f} is below the required floor of {thresholds['rsp_floor']})."
        )

    # The overall status is the MOST SEVERE triggered verdict; default to 'stable' if none triggered.
    if triggered:
        # Select the triggered verdict with the maximum severity rank.
        overall_status = max(triggered, key=lambda v: severity_rank[v])
    else:
        # No rule triggered: the recommendation is stable.
        overall_status = "stable"

    # Assemble the structured verdict (reasons preserve every triggered rule, not just the most severe).
    verdicts: Dict[str, Any] = {
        "overall_status": overall_status,                                   # severity-ranked status
        "falsification_reasons": reasons if reasons else ["None. All thresholds met."],
        "hypotheses": test_results,                                         # per-hypothesis detail
        "effect_sizes": effect_sizes,                                       # effect-size detail
        "stability": {"RSP": rsp},                                          # stability scalar
    }

    # Return the verdict dictionary.
    return verdicts


# -------------------------------------------------------------------------------------------------------------------------------
# Task 20, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def confirmatory_tests(
    S_FMEC: List[str],
    point_fmec_table: pd.DataFrame,
    qualification: pd.DataFrame,
    estimator_projection: pd.DataFrame,
    sigma_df: pd.DataFrame,
    qualified_pool: List[str],
    rsp: float,
    config: Mapping[str, Any],
    rng_bundle: Dict[str, np.random.Generator],
    primary_tensor: pd.DataFrame,
    judge_replicates_df: pd.DataFrame
) -> Dict[str, Any]:
    """
    Orchestrates Task 20: Constructs the six budget-feasible baseline ensembles,
    executes the five pre-registered hypothesis tests, applies Holm-Bonferroni
    FWER control, computes effect sizes, and evaluates the strict falsification rules.

    Parameters
    ----------
    S_FMEC : List[str]
        The FMEC-selected ensemble.
    point_fmec_table : pd.DataFrame
        The FMEC table containing point estimates.
    qualification : pd.DataFrame
        The qualification table containing benchmark ranks.
    estimator_projection : pd.DataFrame
        The 13-column analytic core tensor.
    sigma_df : pd.DataFrame
        The n x n covariance matrix.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    rsp : float
        The Recommendation-Stability Probability from Task 19.
    config : Mapping[str, Any]
        The immutable study configuration.
    rng_bundle : Dict[str, np.random.Generator]
        The bundle of independent random number generators.
    primary_tensor : pd.DataFrame
        The primary 17-column outcomes tensor.
    judge_replicates_df : pd.DataFrame
        The DataFrame of adjudication replicates.

    Returns
    -------
    Dict[str, Any]
        A structured dictionary containing final verdicts, overall status, and
        the detailed reasons for any failures.
    """
    # Extract the dedicated global RNG from the bundle
    global_rng = rng_bundle["global_seed"]

    # Step 1: Construct the six budget-feasible baseline ensembles
    baselines = construct_baselines(
        S_FMEC=S_FMEC,
        point_fmec_table=point_fmec_table,
        qualification=qualification,
        estimator_projection=estimator_projection,
        sigma_df=sigma_df,
        qualified_pool=qualified_pool,
        config=config,
        rng=global_rng
    )

    # Step 2: Execute the raw hypothesis tests for H1-H5 (including H3 risk bootstrap)
    raw_p_values, utility_arrays = execute_raw_hypothesis_tests(
        S_FMEC=S_FMEC,
        baselines=baselines,
        point_fmec_table=point_fmec_table,
        qualification=qualification,
        estimator_projection=estimator_projection,
        sigma_df=sigma_df,
        qualified_pool=qualified_pool,
        config=config,
        rng=global_rng,
        primary_tensor=primary_tensor,
        judge_replicates_df=judge_replicates_df
    )

    # Step 3: Apply the Holm-Bonferroni FWER control and compute effect sizes
    test_results, effect_sizes = apply_fwer_and_effect_sizes(
        raw_p_values=raw_p_values,
        utility_arrays=utility_arrays,
        config=config
    )

    # Step 4: Apply the strict pre-registered falsification rules
    verdicts = apply_falsification_rules(
        test_results=test_results,
        effect_sizes=effect_sizes,
        rsp=rsp,
        config=config
    )

    return verdicts
````

````python
# Task 21 — Oracle-Regret Evaluation for Small Pools (Algorithm 4) (oracle_regret)

# ==============================================================================
# Task 21: Oracle-Regret Evaluation for Small Pools (Algorithm 4)
# ==============================================================================

# -------------------------------------------------------------------------------------------------------------------------------
# Task 21, Step 1: Enumerate all feasible subsets and compute their utilities
# -------------------------------------------------------------------------------------------------------------------------------

def enumerate_feasible_subsets(
    qualified_pool: List[str],
    estimator_projection: pd.DataFrame,
    sigma_df: pd.DataFrame,
    config: Mapping[str, Any]
) -> Tuple[List[Tuple[List[int], float]], Callable[[List[int]], float]]:
    """
    Exhaustively enumerates all 2^n - 1 non-empty subsets of the candidate pool,
    filters them by the strict cost, latency, and risk budgets, and computes
    the utility v(S) for the feasible subsets.

    Parameters
    ----------
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    estimator_projection : pd.DataFrame
        The 13-column analytic core tensor.
    sigma_df : pd.DataFrame
        The n x n covariance matrix.
    config : Mapping[str, Any]
        The immutable study configuration.

    Returns
    -------
    Tuple[List[Tuple[List[int], float]], Callable[[List[int]], float]]
        - feasible_evaluations: A list of tuples, each containing a feasible
          subset (as integer indices) and its scalar utility.
        - v_func: The optimized utility closure used for evaluation.
    """
    n_models = len(qualified_pool)

    # 1. Prepare the dense 1D arrays for fast feasibility checks
    # We pass a dummy fmec_table as it is not needed for feasibility or raw utility
    dummy_fmec = pd.DataFrame({"model_id": qualified_pool, "fmec": np.zeros(n_models)})
    _, c_vec, l_vec, c_j, l_j, budgets = prepare_optimization_vectors(
        fmec_table=dummy_fmec,
        estimator_projection=estimator_projection,
        qualified_pool=qualified_pool,
        config=config
    )

    # Extract the underlying NumPy array from the covariance DataFrame
    sigma_matrix = sigma_df.reindex(index=qualified_pool, columns=qualified_pool).values.astype(np.float64)

    # 2. Instantiate the optimized utility closure
    judge_model_id = config["models"]["judge"]
    v_func = build_utility_function(
        estimator_projection=estimator_projection,
        sigma=sigma_matrix,
        config=config,
        qualified_pool=qualified_pool,
        judge_model_id=judge_model_id
    )

    feasible_evaluations: List[Tuple[List[int], float]] = []
    eps = 1e-9

    # 3. Exhaustive Enumeration
    # Iterate over all possible subset sizes k from 1 to n
    for k in range(1, n_models + 1):
        # Generate all combinations of size k
        for combo in itertools.combinations(range(n_models), k):
            S_prime = list(combo)

            # --- Fast Feasibility Gates ---
            # Short-circuit evaluation by checking the cheapest constraints first

            # Gate 1: Cost
            c_val = np.sum(c_vec[S_prime]) + c_j
            if c_val > budgets["C_max"] + eps:
                continue

            # Gate 2: Latency
            l_val = np.max(l_vec[S_prime]) + l_j
            if l_val > budgets["L_max"] + eps:
                continue

            # Gate 3: Risk
            w = np.zeros(n_models, dtype=np.float64)
            w[S_prime] = 1.0 / k
            r_val = np.dot(w, np.dot(sigma_matrix, w))
            if r_val > budgets["R_max"] + eps:
                continue

            # --- Full Utility Evaluation ---
            # If the subset passes all budget gates, compute its full utility
            u_val = v_func(S_prime)
            feasible_evaluations.append((S_prime, u_val))

    return feasible_evaluations, v_func

# -------------------------------------------------------------------------------------------------------------------------------
# Task 21, Step 2: Identify the oracle ensemble and compute regret
# -------------------------------------------------------------------------------------------------------------------------------

def compute_oracle_regret_metrics(
    feasible_evaluations: List[Tuple[List[int], float]],
    S_FMEC: List[str],
    qualified_pool: List[str],
    v_func: Callable[[List[int]], float]
) -> Tuple[float, float, List[str]]:
    """
    Identifies the oracle-optimal ensemble from the feasible set and computes
    the absolute and relative regret of the FMEC-selected ensemble.

    Equation 28: S* = argmax_S v(S)
    Equation 29: Regret(S_hat) = v(S*) - v(S_hat)

    Parameters
    ----------
    feasible_evaluations : List[Tuple[List[int], float]]
        The list of feasible subsets and their utilities.
    S_FMEC : List[str]
        The FMEC-selected ensemble.
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    v_func : Callable[[List[int]], float]
        The optimized utility closure.

    Returns
    -------
    Tuple[float, float, List[str]]
        - absolute_regret: v(S*) - v(S_FMEC)
        - relative_regret: (v(S*) - v(S_FMEC)) / v(S*)
        - S_star: The oracle-optimal ensemble as a list of string IDs.

    Raises
    ------
    RuntimeError
        If no subsets are feasible under the given budgets.
    """
    if not feasible_evaluations:
        raise RuntimeError(
            "Oracle Evaluation Failed: No subsets are feasible under the current "
            "cost, latency, and risk budgets. The budget envelope is too restrictive."
        )

    # 1. Identify the Oracle Ensemble S*
    # Find the tuple with the maximum utility
    best_tuple = max(feasible_evaluations, key=lambda item: item[1])
    S_star_indices = best_tuple[0]
    v_S_star = best_tuple[1]

    # Map the integer indices back to string model IDs
    S_star = [qualified_pool[i] for i in S_star_indices]

    # 2. Evaluate the FMEC-selected ensemble
    # Map the string IDs to integer indices
    S_FMEC_indices = [qualified_pool.index(m_id) for m_id in S_FMEC]

    # Re-evaluate using the exact same closure to guarantee perfect comparability
    v_S_FMEC = v_func(S_FMEC_indices)

    # 3. Compute Regret Metrics
    absolute_regret = v_S_star - v_S_FMEC

    # Handle potential division by zero for relative regret
    if abs(v_S_star) > 1e-12:
        relative_regret = absolute_regret / v_S_star
    else:
        relative_regret = 0.0 if absolute_regret == 0.0 else float('inf')

    return float(absolute_regret), float(relative_regret), S_star

# -------------------------------------------------------------------------------------------------------------------------------
# Task 21, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def oracle_regret(
    qualified_pool: List[str],
    estimator_projection: pd.DataFrame,
    sigma_df: pd.DataFrame,
    config: Mapping[str, Any],
    S_FMEC: List[str]
) -> Tuple[float, float, List[str]]:
    """
    Orchestrates Task 21: Checks the enumeration cap, exhaustively evaluates all
    feasible subsets, identifies the oracle optimum, and computes regret.

    Parameters
    ----------
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    estimator_projection : pd.DataFrame
        The 13-column analytic core tensor.
    sigma_df : pd.DataFrame
        The n x n covariance matrix.
    config : Mapping[str, Any]
        The immutable study configuration.
    S_FMEC : List[str]
        The FMEC-selected ensemble.

    Returns
    -------
    Tuple[float, float, List[str]]
        - absolute_regret: v(S*) - v(S_FMEC)
        - relative_regret: (v(S*) - v(S_FMEC)) / v(S*)
        - S_star: The oracle-optimal ensemble.
        Returns (NaN, NaN, []) if the pool size exceeds the enumeration cap.
    """
    n_models = len(qualified_pool)
    max_pool_size = config["oracle_eval"]["max_pool_size_for_enumeration"]

    # Enforce the computational bound defined in the protocol
    if n_models > max_pool_size:
        # Log a warning in a real system; here we return NaNs to indicate intractability
        return float('nan'), float('nan'), []

    # Step 1: Enumerate and evaluate all feasible subsets
    feasible_evaluations, v_func = enumerate_feasible_subsets(
        qualified_pool=qualified_pool,
        estimator_projection=estimator_projection,
        sigma_df=sigma_df,
        config=config
    )

    # Step 2: Identify the oracle and compute regret
    absolute_regret, relative_regret, S_star = compute_oracle_regret_metrics(
        feasible_evaluations=feasible_evaluations,
        S_FMEC=S_FMEC,
        qualified_pool=qualified_pool,
        v_func=v_func
    )

    return absolute_regret, relative_regret, S_star
````

````python
# Task 22 — Assemble the End-to-End Orchestrator (run_pipeline)

# ==============================================================================
# Task 22: Assemble the End-to-End Orchestrator (run_pipeline)
# ==============================================================================

# -------------------------------------------------------------------------------------------------------------------------------
# Task 22, Helper Structures: PipelineCache, EstimationOutputs, and ResultsBundle
# -------------------------------------------------------------------------------------------------------------------------------

@dataclass(frozen=True)
class PipelineCache:
    """
    An immutable, frozen dataclass caching the deterministic substrate (Tasks 1-11).

    This class is the single source of truth for the data-generating substrate.
    By enforcing `frozen=True`, we guarantee that parallel process sweeps
    during robustness and sensitivity testing (Task 23) cannot silently mutate
    the state. This re-implementation explicitly includes the `qualification_df`
    field to resolve the downstream data fabrication deficiency.

    Attributes
    ----------
    primary_tensor : pd.DataFrame
        The 17-column canonical released artifact (judge_rep == 0).
    estimator_projection : pd.DataFrame
        The 13-column analytic core tensor used for downstream estimators.
    judge_replicates_df : pd.DataFrame
        The DataFrame of adjudication replicates (judge_rep > 0).
    qualified_pool : List[str]
        The canonical ordered list of model IDs.
    fuse_callable : Callable[[List[Dict[str, Any]], str, Optional[str]], Dict[str, Any]]
        The specified, functional fusion operator closure.
    rng_bundle : Dict[str, np.random.Generator]
        The bundle of independent, seeded random number generators.
    run_dir : Path
        The path to the timestamped run directory.
    base_config : Mapping[str, Any]
        The immutable study configuration.
    task_record_pruned : pd.DataFrame
        The task DataFrame with contaminated items removed.
    model_answer_clean : List[Dict[str, Any]]
        The cleansed raw free-form generation records.
    candidate_solution_clean : List[Dict[str, Any]]
        The cleansed raw structured solution records.
    qualification_df : pd.DataFrame
        The authentic, mutated qualification DataFrame (output of Task 6).
    """
    primary_tensor: pd.DataFrame
    estimator_projection: pd.DataFrame
    judge_replicates_df: pd.DataFrame
    qualified_pool: List[str]
    fuse_callable: Callable[[List[Dict[str, Any]], str, Optional[str]], Dict[str, Any]]
    rng_bundle: Dict[str, np.random.Generator]
    run_dir: Path
    base_config: Mapping[str, Any]
    task_record_pruned: pd.DataFrame
    model_answer_clean: List[Dict[str, Any]]
    candidate_solution_clean: List[Dict[str, Any]]
    qualification_df: pd.DataFrame  # The critical, remediated benchmark DataFrame

    def __post_init__(self) -> None:
        """
        Executes strict, post-initialization type and state validation.
        """
        # Validate that the primary outcome tensor is a pandas DataFrame
        if not isinstance(self.primary_tensor, pd.DataFrame):
            raise TypeError("primary_tensor must be an instance of pandas.DataFrame")
        # Validate that the 13-column analytic core is a pandas DataFrame
        if not isinstance(self.estimator_projection, pd.DataFrame):
            raise TypeError("estimator_projection must be an instance of pandas.DataFrame")
        # Validate that the judge replicates table is a pandas DataFrame
        if not isinstance(self.judge_replicates_df, pd.DataFrame):
            raise TypeError("judge_replicates_df must be an instance of pandas.DataFrame")
        # Validate that the pruned evaluation task set is a pandas DataFrame
        if not isinstance(self.task_record_pruned, pd.DataFrame):
            raise TypeError("task_record_pruned must be an instance of pandas.DataFrame")
        # Validate that the authentic qualification table is a pandas DataFrame
        if not isinstance(self.qualification_df, pd.DataFrame):
            raise TypeError("qualification_df must be an instance of pandas.DataFrame")
        # Validate that the qualified pool is a non-empty list of strings
        if not isinstance(self.qualified_pool, list) or not all(isinstance(x, str) for x in self.qualified_pool):
            raise TypeError("qualified_pool must be a list of strings")
        if len(self.qualified_pool) == 0:
            raise ValueError("qualified_pool cannot be empty")
        # Validate that the run directory is a pathlib Path
        if not isinstance(self.run_dir, Path):
            raise TypeError("run_dir must be an instance of pathlib.Path")

@dataclass
class EstimationOutputs:
    """
    A strictly typed dataclass encapsulating the outputs of the pure
    estimation-to-selection segment (Tasks 12-18).

    Attributes
    ----------
    active_config : Mapping[str, Any]
        The configuration used for this estimation run (potentially perturbed).
    sigma_df : pd.DataFrame
        The n x n error covariance matrix.
    fmec_table : pd.DataFrame
        The strictly typed FMEC DataFrame containing point estimates.
    S_FMEC : List[str]
        The list of selected model IDs.
    w_dict : Dict[str, float]
        Dictionary mapping model_id to its simplex weight.
    active_constraints : Dict[str, Any]
        Dictionary detailing constraint utilization and binding status.
    fusion_diagnostics : Dict[str, float]
        Dictionary containing slope, r_squared, and sigma_u from the audit.
    """
    active_config: Mapping[str, Any]
    sigma_df: pd.DataFrame
    fmec_table: pd.DataFrame
    S_FMEC: List[str]
    w_dict: Dict[str, float]
    active_constraints: Dict[str, Any]
    fusion_diagnostics: Dict[str, float]


@dataclass
class ResultsBundle:
    """
    A strictly typed dataclass encapsulating the final outputs of the entire
    FMEC pipeline, ready for reporting and serialization (Task 24).

    Attributes
    ----------
    fmec_table : pd.DataFrame
        The FMEC table updated with confidence intervals and selection frequencies.
    S_FMEC : List[str]
        The list of selected model IDs.
    w : Dict[str, float]
        Dictionary mapping model_id to its simplex weight.
    active_constraints : Dict[str, Any]
        Dictionary detailing constraint utilization and binding status.
    CIs : Dict[str, Tuple[float, float]]
        Dictionary mapping model_id to its (ci_low, ci_high) BCa interval.
    variance_components : Dict[str, float]
        Dictionary containing sigma2_task, sigma2_gen, and sigma2_adj.
    RSP : float
        The Recommendation-Stability Probability (Eq. 51).
    selection_freq : Dict[str, float]
        Dictionary mapping model_id to its bootstrap selection frequency.
    verdicts : Dict[str, Any]
        Structured dictionary containing final hypothesis verdicts and status.
    regret : Dict[str, Any]
        Dictionary containing absolute regret, relative regret, and the oracle ensemble.
    """
    fmec_table: pd.DataFrame
    S_FMEC: List[str]
    w: Dict[str, float]
    active_constraints: Dict[str, Any]
    CIs: Dict[str, Tuple[float, float]]
    variance_components: Dict[str, float]
    RSP: float
    selection_freq: Dict[str, float]
    verdicts: Dict[str, Any]
    regret: Dict[str, Any]

# -------------------------------------------------------------------------------------------------------------------------------
# Task 22, Step 1: Deterministic ingestion-to-tensor segment
# -------------------------------------------------------------------------------------------------------------------------------

def execute_ingestion_to_tensor_segment(
    raw_config: Dict[str, Any],
    model_answer_raw: List[Dict[str, Any]],
    candidate_solution_raw: List[Dict[str, Any]],
    task_record_raw: List[Dict[str, Any]],
    qualification_raw: pd.DataFrame,
    raw_proxy_texts: List[str]
) -> PipelineCache:
    """
    Executes the deterministic data-generating substrate (Tasks 1-11).

    This segment runs exactly once per study. It ingests raw data, audits for
    contamination, collects LLM responses, adjudicates correctness, and builds
    the canonical outcome tensor, enforcing the Tier-1 reproducibility gate.
    This re-implementation explicitly packages the authentic, mutated
    qualification DataFrame (Task 6 output) into the cached state.

    Parameters
    ----------
    raw_config : Dict[str, Any]
        The raw, unvalidated study configuration dictionary.
    model_answer_raw : List[Dict[str, Any]]
        The raw free-form generation records.
    candidate_solution_raw : List[Dict[str, Any]]
        The raw structured solution records.
    task_record_raw : List[Dict[str, Any]]
        The raw task and ground truth records.
    qualification_raw : pd.DataFrame
        The raw benchmark eligibility DataFrame.
    raw_proxy_texts : List[str]
        The raw text corpus used for the contamination audit.

    Returns
    -------
    PipelineCache
        The frozen cache containing the canonical tensor and intermediate states.

    Raises
    ------
    RuntimeError
        If the computed tensor SHA-256 hash does not match the configuration.
    """
    # Task 1: Load configuration, redact gateway credentials, and initialize seed ledger
    frozen_config, run_dir, rng_bundle = load_config_and_seeds(raw_config)

    # Task 2: Validate configuration parameter appropriateness and cross-field invariants
    _ = validate_config(frozen_config, run_dir)

    # Task 3: Ingest raw data structures and enforce strict schema compliance
    ma_df, cs_df, tr_df, qual_df = ingest_and_validate_inputs(
        model_answer_raw, candidate_solution_raw, task_record_raw, qualification_raw, frozen_config, run_dir
    )

    # Task 4: Cleanse raw records and canonicalize the input DataFrames
    cleansed_data = cleanse_raw(ma_df, cs_df, tr_df, qual_df)

    # Extract the cleansed DataFrames from the canonicalized packaging
    ma_clean_df = cleansed_data["model_answer"]["data"]
    cs_clean_df = cleansed_data["candidate_solution"]["data"]
    tr_clean_df = cleansed_data["task_record"]["data"]
    qual_clean_df = cleansed_data["qualification"]["data"]

    # Task 5: Execute the pre-registered benchmark contamination audit (lexical + behavioral)
    tr_pruned_df, _ = contamination_audit(
        task_record=tr_clean_df,
        qualification=qual_clean_df,
        config=frozen_config,
        run_dir=run_dir,
        raw_proxy_texts=raw_proxy_texts
    )

    # Task 6: Apply the Layer-1 candidate qualification eligibility gate
    qualified_pool, qual_mutated_df = qualify(qual_clean_df, frozen_config, run_dir)

    # Task 7: Stratify the qualified candidate pool and persist metadata
    _ = stratify(qualified_pool, qual_mutated_df, frozen_config, run_dir)

    # Task 8: Build the stateless fusion operator specification closure
    # We pass dummy quality, cost, and latency vectors as they are bypassed for verifier_augmented
    dummy_vec = np.zeros(len(qualified_pool))
    fuse_callable, _, _, _ = build_fusion_operator(
        frozen_config, rng_bundle, qualified_pool, dummy_vec, dummy_vec, dummy_vec, dummy_vec
    )

    # Task 9: Collect responses from the qualified pool using parallel fan-out and sandboxing
    ma_records, cs_records = collect_responses(qualified_pool, tr_pruned_df, frozen_config)

    # Task 10: Score and adjudicate responses (deterministic grounding + LLM judge)
    scored_labels = score_and_adjudicate(
        model_answer=ma_records,
        candidate_solution=cs_records,
        task_record=tr_pruned_df,
        config=frozen_config,
        rng_bundle=rng_bundle,
        human_labels=None # Human audit is assumed to be an offline process for this pipeline
    )

    # Task 11: Assemble the canonical outcome-and-error tensor
    primary_tensor, estimator_projection, judge_replicates_df, computed_hash = build_outcome_tensor(
        scored_labels=scored_labels,
        config=frozen_config
    )

    # Enforce the Tier-1 Reproducibility Gate
    expected_hash = frozen_config["reproducibility"].get("tensor_sha256")
    # If a hash is pre-registered in the config, assert that our computed hash is identical
    if expected_hash is not None and expected_hash != "<populated upon execution of the reference instantiation>":
        if computed_hash != expected_hash:
            # Raise an explicit error to prevent running inference on corrupted or modified data
            raise RuntimeError(
                f"Tier-1 Reproducibility Violation: Computed tensor hash does not match the configuration.\n"
                f"Expected: {expected_hash}\n"
                f"Computed: {computed_hash}\n"
                f"Pipeline halted to prevent inference on drifted data."
            )

    # Package all deterministic intermediates into the immutable PipelineCache dataclass
    # We explicitly pass qual_mutated_df (authentic benchmark ranks) to the qualification_df field
    return PipelineCache(
        primary_tensor=primary_tensor,
        estimator_projection=estimator_projection,
        judge_replicates_df=judge_replicates_df,
        qualified_pool=qualified_pool,
        fuse_callable=fuse_callable,
        rng_bundle=rng_bundle,
        run_dir=run_dir,
        base_config=frozen_config,
        task_record_pruned=tr_pruned_df,
        model_answer_clean=ma_records,
        candidate_solution_clean=cs_records,
        qualification_df=qual_mutated_df  # Persist authentic qualification table
    )

# -------------------------------------------------------------------------------------------------------------------------------
# Task 22, Step 2: Pure estimation-to-selection segment
# -------------------------------------------------------------------------------------------------------------------------------

def _apply_perturbation_hook(
    base_config: Mapping[str, Any],
    perturbation: Optional[Dict[str, Any]],
    run_dir: Path
) -> Mapping[str, Any]:
    """
    Safely applies allowed perturbations to the configuration for robustness sweeps.

    Parameters
    ----------
    base_config : Mapping[str, Any]
        The immutable base configuration.
    perturbation : Optional[Dict[str, Any]]
        A dictionary of allowed configuration overrides.
    run_dir : Path
        The path to the immutable run directory (used for re-validation).

    Returns
    -------
    Mapping[str, Any]
        The perturbed configuration dictionary.

    Raises
    ------
    ValueError
        If the perturbation dictionary contains unauthorized keys.
    """
    # If no perturbation is provided, return the base configuration unmodified
    if not perturbation:
        return base_config

    # Define the strict allow-list of perturbable keys to prevent data-generating substrate mutation
    allowed_keys = {
        "utility_weights",
        "fmec_hyperparams",
        "K",
        "covariance_estimator",
        "information_gain_estimator"
    }

    # Validate that the perturbation only contains allowed keys
    invalid_keys = set(perturbation.keys()) - allowed_keys
    if invalid_keys:
        # Raise an error to prevent unauthorized experimental manipulation
        raise ValueError(f"Perturbation hook rejected unauthorized keys: {invalid_keys}")

    # Deep copy the base configuration to prevent mutating the global state
    # We must convert the FrozenDict back to a standard dict for mutation
    mutable_config = copy.deepcopy(dict(base_config))

    # Apply the specific perturbations to their respective nested locations
    if "utility_weights" in perturbation:
        mutable_config["estimation"]["utility_weights"] = perturbation["utility_weights"]
    if "fmec_hyperparams" in perturbation:
        mutable_config["estimation"]["fmec_hyperparams"] = perturbation["fmec_hyperparams"]
    if "K" in perturbation:
        mutable_config["estimation"]["monte_carlo"]["K"] = perturbation["K"]
    if "covariance_estimator" in perturbation:
        mutable_config["estimation"]["covariance"]["estimator"] = perturbation["covariance_estimator"]
    if "information_gain_estimator" in perturbation:
        mutable_config["estimation"]["information_gain"]["estimator"] = perturbation["information_gain_estimator"]

    # Re-validate the perturbed configuration to ensure structural soundness (Task 2)
    _ = validate_config(mutable_config, run_dir)

    # Return the perturbed configuration
    return mutable_config

def execute_estimation_selection_segment(
    cache: PipelineCache,
    perturbation: Optional[Dict[str, Any]] = None
) -> EstimationOutputs:
    """
    Execute the deterministic estimation-and-selection segment (Tasks 12-17), running
    the live-judge fusion audit ONLY on the baseline (un-perturbed) run.

    Rationale (detailed tasklist Task 22, Step 2). This segment must be a
    deterministic function of the cached tensor with no LLM generation. The prior
    implementation included `fusion_sim_audit` (which issues live judge calls) on
    every perturbation, breaking determinism and making robustness (Task 23)
    intractable. This version runs the fusion audit only when `perturbation is None`
    (the baseline run that supplies sigma_u to the Eq. 20a quadrature); perturbed runs
    return NaN/zero fusion diagnostics and are pure functions of the cached tensor.

    Parameters
    ----------
    cache : PipelineCache
        The frozen cache containing the canonical tensor and substrate.
    perturbation : Optional[Dict[str, Any]], optional
        Allowed configuration overrides; when None, this is the baseline run.

    Returns
    -------
    EstimationOutputs
        Estimation primitives, FMEC table, selected ensemble, and fusion diagnostics
        (real on the baseline run; NaN/zero on perturbed runs).
    """
    # Apply the allow-listed perturbation to obtain the active configuration.
    active_config = _apply_perturbation_hook(cache.base_config, perturbation, cache.run_dir)

    # Task 15: covariance (Eq. 10, Ledoit-Wolf) and marginal risk (Eq. 12).
    sigma_df, mrc_dict, _ = estimate_risk(
        estimator_projection=cache.estimator_projection,
        qualified_pool=cache.qualified_pool, config=active_config,
    )
    # Aligned covariance array for O(1) utility evaluation.
    sigma_matrix = sigma_df.reindex(index=cache.qualified_pool, columns=cache.qualified_pool).values.astype(np.float64)

    # Task 12: build the dense utility closure (Eq. 3/3a).
    judge_model_id = active_config["models"]["judge"]
    v_func = build_utility_function(
        estimator_projection=cache.estimator_projection, sigma=sigma_matrix,
        config=active_config, qualified_pool=cache.qualified_pool, judge_model_id=judge_model_id,
    )

    # Task 13: permutation-Shapley MEC (Eq. 6a).
    mec_hat, coalition_samples = estimate_mec(
        qualified_pool=cache.qualified_pool, v_func=v_func, config=active_config, rng_bundle=cache.rng_bundle,
    )

    # Task 14: outcome-based CMI information gain (Eq. 9a).
    ig_dict = estimate_ig(
        estimator_projection=cache.estimator_projection, coalition_samples=coalition_samples,
        qualified_pool=cache.qualified_pool, config=active_config,
    )

    # Task 16: compose FMEC (Eq. 13).
    fmec_table = compose_fmec(
        mec_hat=mec_hat, ig=ig_dict, mrc=mrc_dict,
        qualified_pool=cache.qualified_pool, config=active_config, run_dir=cache.run_dir,
    )

    # Task 17: budget-feasible greedy selection (Eq. 14).
    S_FMEC, w_dict, active_constraints = select_ensemble(
        fmec_table=fmec_table, estimator_projection=cache.estimator_projection,
        sigma_df=sigma_df, qualified_pool=cache.qualified_pool, config=active_config,
    )

    # Task 18: fusion-simulation audit — ONLY on the baseline run (no live judge calls under perturbation).
    if perturbation is None:
        # Baseline run: compute the realized fusion gap and sigma_u (feeds the Eq. 20a quadrature).
        fusion_diagnostics, _ = fusion_sim_audit(
            estimator_projection=cache.estimator_projection, model_answer=cache.model_answer_clean,
            candidate_solution=cache.candidate_solution_clean, task_record=cache.task_record_pruned,
            fuse_callable=cache.fuse_callable, qualified_pool=cache.qualified_pool,
            config=active_config, rng_bundle=cache.rng_bundle,
        )
    else:
        # Perturbed run: skip the live-judge audit; this segment stays deterministic and network-free.
        # sigma_u is 0.0 here because perturbed runs (robustness) do not enter the inference segment.
        fusion_diagnostics = {"slope": float("nan"), "r_squared": float("nan"), "sigma_u": 0.0}

    # Package the estimation outputs.
    return EstimationOutputs(
        active_config=active_config, sigma_df=sigma_df, fmec_table=fmec_table,
        S_FMEC=S_FMEC, w_dict=w_dict, active_constraints=active_constraints,
        fusion_diagnostics=fusion_diagnostics,
    )

# -------------------------------------------------------------------------------------------------------------------------------
# Task 22, Step 3: Inference segment and bundle assembly
# -------------------------------------------------------------------------------------------------------------------------------

def execute_inference_and_bundle_segment(
    cache: PipelineCache,
    estimation_outputs: Any # Expecting EstimationOutputs dataclass
) -> Any: # Returns ResultsBundle dataclass
    """
    Executes the statistical inference segment (Tasks 19-21) and assembles the final bundle.

    This segment is a deterministic function of the cached tensor. This re-implementation
    completely removes the previous dummy_qual data fabrication hack. It extracts the
    authentic qualification_df from the cache and passes it directly to the
    confirmatory tests, ensuring the econometric integrity of H4 (Eq. 47) and H5 (Eq. 48).

    Parameters
    ----------
    cache : PipelineCache
        The frozen, remediated cache containing the canonical tensor and authentic qualification data.
    estimation_outputs : Any
        The outputs from the estimation-to-selection segment (Task 22, Step 2).

    Returns
    -------
    ResultsBundle
        The strictly typed final results bundle containing all 10 required fields.
    """
    # Extract point estimates and configurations from the estimation segment
    active_config = estimation_outputs.active_config
    fmec_table = estimation_outputs.fmec_table
    S_FMEC = estimation_outputs.S_FMEC
    sigma_df = estimation_outputs.sigma_df

    # Task 19: Execute the nested cluster bootstrap validation (Algorithm 3)
    # This computes BCa confidence intervals, variance components, and the RSP
    updated_fmec_table, variance_components, rsp, selection_freq = bootstrap_validate(
        primary_tensor=cache.primary_tensor,
        judge_replicates_df=cache.judge_replicates_df,
        point_fmec_table=fmec_table,
        S_star=S_FMEC,
        qualified_pool=cache.qualified_pool,
        config=active_config,
        rng_bundle=cache.rng_bundle
    )

    # Task 20: Execute Confirmatory Hypothesis Testing & Falsification
    # We extract the AUTHENTIC qualification_df from the cache to preserve econometric validity
    # This allows H4 to calculate Kendall's tau and H5 to subset open-weight models accurately
    verdicts = confirmatory_tests(
        S_FMEC=S_FMEC,
        point_fmec_table=updated_fmec_table,
        qualification=cache.qualification_df,  # Pass authentic qualification DataFrame
        estimator_projection=cache.estimator_projection,
        sigma_df=sigma_df,
        qualified_pool=cache.qualified_pool,
        rsp=rsp,
        config=active_config,
        rng_bundle=cache.rng_bundle,
        primary_tensor=cache.primary_tensor,
        judge_replicates_df=cache.judge_replicates_df
    )

    # Task 21: Execute Oracle-Regret Evaluation for Small Pools (Algorithm 4)
    # Exhaustively enumerates all subsets if pool size <= 20 to compute regret
    abs_regret, rel_regret, S_star = oracle_regret(
        qualified_pool=cache.qualified_pool,
        estimator_projection=cache.estimator_projection,
        sigma_df=sigma_df,
        config=active_config,
        S_FMEC=S_FMEC
    )

    # Package the regret metrics into a structured dictionary
    regret_dict = {
        "absolute_regret": abs_regret,
        "relative_regret": rel_regret,
        "oracle_ensemble": S_star
    }

    # Extract Confidence Intervals from the updated FMEC table for the results bundle
    cis = {
        row["model_id"]: (row["ci_low"], row["ci_high"])
        for _, row in updated_fmec_table.iterrows()
    }

    # Assemble and return the final ResultsBundle dataclass (Task 22 contract)
    return ResultsBundle(
        fmec_table=updated_fmec_table,
        S_FMEC=S_FMEC,
        w=estimation_outputs.w_dict,
        active_constraints=estimation_outputs.active_constraints,
        CIs=cis,
        variance_components=variance_components,
        RSP=rsp,
        selection_freq=selection_freq,
        verdicts=verdicts,
        regret=regret_dict
    )

# -------------------------------------------------------------------------------------------------------------------------------
# Task 22, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def run_pipeline(
    raw_config: Dict[str, Any],
    model_answer_raw: List[Dict[str, Any]],
    candidate_solution_raw: List[Dict[str, Any]],
    task_record_raw: List[Dict[str, Any]],
    qualification_raw: pd.DataFrame,
    raw_proxy_texts: List[str],
    cached_state: Optional[Any] = None, # Expecting PipelineCache
    perturbation: Optional[Dict[str, Any]] = None
) -> Tuple[Any, Any]: # Returns Tuple[ResultsBundle, PipelineCache]
    """
    Orchestrates Task 22: Assembles the End-to-End Orchestrator.

    This remediated orchestrator glues the deterministic data-generating substrate,
    the pure estimation segment, and the statistical inference segment.
    It ensures that the authentic qualification table is passed through
    the cache to the confirmatory tests, preventing any data fabrication.

    Parameters
    ----------
    raw_config : Dict[str, Any]
        The raw study configuration dictionary.
    model_answer_raw : List[Dict[str, Any]]
        The raw free-form generation records.
    candidate_solution_raw : List[Dict[str, Any]]
        The raw structured solution records.
    task_record_raw : List[Dict[str, Any]]
        The raw task and ground truth records.
    qualification_raw : pd.DataFrame
        The raw benchmark eligibility DataFrame.
    raw_proxy_texts : List[str]
        The raw text corpus used for the contamination audit.
    cached_state : Optional[PipelineCache], optional
        A previously computed pipeline cache to bypass Stage 1 and 2.
    perturbation : Optional[Dict[str, Any]], optional
        A dictionary of allowed configuration overrides.

    Returns
    -------
    Tuple[ResultsBundle, PipelineCache]
        - results_bundle: The final dataclass containing all required artifacts.
        - cache: The frozen pipeline cache containing the authentic qualification DataFrame.
    """
    # Step 1: Execute or bypass the deterministic data-generating substrate
    if cached_state is None:
        # If no cache is provided, execute the full deterministic segment (Tasks 1-11)
        # execute_ingestion_to_tensor_segment is assumed available (Task 22, Step 1)
        cache = execute_ingestion_to_tensor_segment(
            raw_config=raw_config,
            model_answer_raw=model_answer_raw,
            candidate_solution_raw=candidate_solution_raw,
            task_record_raw=task_record_raw,
            qualification_raw=qualification_raw,
            raw_proxy_texts=raw_proxy_texts
        )
    else:
        # Inject the cached state to bypass expensive I/O and gateway calls
        cache = cached_state

    # Step 2: Execute the pure estimation-to-selection segment (Tasks 12-18)
    # This applies the perturbation hook and computes point-estimate FMEC
    # execute_estimation_selection_segment is assumed available (Task 22, Step 2)
    estimation_outputs = execute_estimation_selection_segment(
        cache=cache,
        perturbation=perturbation
    )

    # Step 3: Execute the statistical inference segment (Tasks 19-21)
    # This extracts the authentic qualification_df from the cache to run the tests
    # execute_inference_and_bundle_segment is assumed available (Task 22, Step 3)
    results_bundle = execute_inference_and_bundle_segment(
        cache=cache,
        estimation_outputs=estimation_outputs
    )

    return results_bundle, cache
````

````python
# Task 23 — Robustness & Sensitivity Analysis (robustness)

# ==============================================================================
# Task 23: Robustness & Sensitivity Analysis (robustness)
# ==============================================================================

@dataclass
class RobustnessReport:
    """
    A strictly typed dataclass encapsulating the results of the robustness and
    sensitivity analysis (Task 23).

    Attributes
    ----------
    stability_surface : pd.DataFrame
        DataFrame detailing the selection probability mass for each unique ensemble
        across the Dirichlet preference draws.
    max_stability_mass : float
        The maximum probability mass accumulated by any single ensemble.
    is_fragile : bool
        True if max_stability_mass < 0.90, indicating the recommendation is
        highly sensitive to arbitrary utility weight choices.
    ablation_results : Dict[str, Dict[str, Any]]
        Dictionary mapping ablation identifiers to their results (selected ensemble,
        whether it flipped compared to baseline, and utility delta).
    """
    stability_surface: pd.DataFrame
    max_stability_mass: float
    is_fragile: bool
    ablation_results: Dict[str, Dict[str, Any]]

# -------------------------------------------------------------------------------------------------------------------------------
# Task 23, Step 1: Utility-weight robustness analysis (Dirichlet Sweeps)
# -------------------------------------------------------------------------------------------------------------------------------

def _compute_loo_mec(v_func: Callable[[List[int]], float], qualified_pool: List[str]) -> Dict[str, float]:
    """
    Compute leave-one-out marginal contributions directly from the utility closure
    (Ablation A3): MEC_j = v(N) - v(N \\ {M_j}).

    Rationale (Section 11.7, Ablation A3; Eq. 30-LOO). The leave-one-out estimand
    credits each model by the loss in grand-coalition value when it is removed. This
    is computed directly via the O(1) utility closure, avoiding any modification to
    the Task 13 permutation estimator and any sampler-override perturbation.

    Parameters
    ----------
    v_func : Callable[[List[int]], float]
        The set-utility closure mapping member indices to v(S).
    qualified_pool : List[str]
        The canonical ordered model IDs (defining the grand coalition N).

    Returns
    -------
    Dict[str, float]
        Mapping model_id -> leave-one-out MEC.
    """
    # Index set of the grand coalition N = {0, ..., n-1}.
    n = len(qualified_pool)
    all_idx = list(range(n))
    # Value of the grand coalition v(N).
    v_full = v_func(all_idx)
    # Compute MEC_j = v(N) - v(N \ {j}) for each model j.
    loo: Dict[str, float] = {}
    for j, m_id in enumerate(qualified_pool):
        # Grand coalition with model j removed.
        without_j = [i for i in all_idx if i != j]
        # Marginal value of j relative to the rest of the pool (Eq. 30-LOO).
        loo[m_id] = float(v_full - v_func(without_j))
    # Return the leave-one-out contributions keyed by model_id.
    return loo


def _estimate_select_minimal(
    cache: PipelineCache,
    perturbation: Optional[Dict[str, Any]],
    draw_index: int,
    *,
    cached_sigma_df: pd.DataFrame,
    cached_sigma_matrix: np.ndarray,
    cached_mrc_dict: Dict[str, float],
    cached_ig_dict: Dict[str, float]
) -> Dict[str, Any]:
    """
    Minimal estimation entry point: recompute ONLY MEC and selection for a utility-
    weight perturbation, reusing the cached weight-independent Sigma/MRC/IG.

    Rationale (Section 11.7; detailed tasklist Task 23, Step 1). Information gain and
    the covariance do NOT depend on the utility weights; recomputing them per draw is
    the intractability the blueprint forbids. This path rebuilds the utility closure
    with the perturbed weights, recomputes MEC (the only weight-dependent term),
    composes FMEC = MEC + eta*IG - lambda*MRC INLINE (no compose_fmec disk write), and
    selects the ensemble — with no inference and no fusion. Determinism is guaranteed
    by re-seeding the Monte Carlo generator from (mc_root, draw_index).

    Parameters
    ----------
    cache : PipelineCache
        The frozen substrate (tensor, pool, rng_bundle, run_dir, base_config).
    perturbation : Optional[Dict[str, Any]]
        The utility-weight override for this draw.
    draw_index : int
        The draw index used to deterministically derive the MC stream.
    cached_sigma_df, cached_sigma_matrix : pd.DataFrame, np.ndarray
        The precomputed (weight-independent) covariance, as a DataFrame and array.
    cached_mrc_dict, cached_ig_dict : Dict[str, float]
        The precomputed (weight-independent) MRC and IG vectors.

    Returns
    -------
    Dict[str, Any]
        {'S_FMEC': List[str], 'fmec_table': pd.DataFrame, 'w_dict': Dict[str, float]}.
    """
    # Build the active config by applying ONLY the utility-weight override (a valid simplex by
    # construction), avoiding re-validation and its per-draw report write.
    active_config = copy.deepcopy(dict(cache.base_config))
    # Apply the perturbed utility weights if provided.
    if perturbation and "utility_weights" in perturbation:
        active_config["estimation"]["utility_weights"] = perturbation["utility_weights"]

    # Derive a deterministic per-draw Monte Carlo generator from the MC root and the draw index.
    mc_root = cache.base_config["reproducibility"]["seed_ledger"]["monte_carlo_sampling_seed"]
    # SeedSequence with a draw-specific spawn key yields an independent, reproducible MC stream.
    draw_mc_seq = np.random.SeedSequence(entropy=mc_root, spawn_key=(int(draw_index),))
    # Copy the RNG bundle and replace ONLY the MC stream with the per-draw generator.
    draw_rng_bundle = dict(cache.rng_bundle)
    draw_rng_bundle["monte_carlo_sampling_seed"] = np.random.Generator(np.random.PCG64(draw_mc_seq))

    # Rebuild the utility closure with the perturbed weights and the CACHED covariance (weight-independent).
    judge_model_id = active_config["models"]["judge"]
    v_func = build_utility_function(
        estimator_projection=cache.estimator_projection, sigma=cached_sigma_matrix,
        config=active_config, qualified_pool=cache.qualified_pool, judge_model_id=judge_model_id,
    )
    # Recompute MEC (the only weight-dependent term) via the permutation Shapley estimator (Eq. 6a).
    mec_hat, _ = estimate_mec(
        qualified_pool=cache.qualified_pool, v_func=v_func, config=active_config, rng_bundle=draw_rng_bundle,
    )

    # Compose FMEC inline (Eq. 13): FMEC_j = MEC_j + eta*IG_j - lambda*MRC_j (no disk write).
    eta = active_config["estimation"]["fmec_hyperparams"]["eta"]
    lam = active_config["estimation"]["fmec_hyperparams"]["lambda"]
    # Assemble the minimal FMEC rows reusing cached IG and MRC.
    rows: List[Dict[str, Any]] = []
    for m in cache.qualified_pool:
        # Eq. 13 evaluated with the recomputed MEC and the cached IG/MRC.
        fmec_val = mec_hat[m] + eta * cached_ig_dict[m] - lam * cached_mrc_dict[m]
        rows.append({"model_id": m, "mec_hat": mec_hat[m], "ig": cached_ig_dict[m], "mrc": cached_mrc_dict[m], "fmec": fmec_val})
    # Materialize the FMEC table required by the selector.
    fmec_table = pd.DataFrame(rows)

    # Run the budget-feasible greedy selection (Eq. 14) reusing the cached covariance.
    S_FMEC, w_dict, _ = select_ensemble(
        fmec_table=fmec_table, estimator_projection=cache.estimator_projection,
        sigma_df=cached_sigma_df, qualified_pool=cache.qualified_pool, config=active_config,
    )
    # Return only the quantities needed by the robustness aggregator.
    return {"S_FMEC": S_FMEC, "fmec_table": fmec_table, "w_dict": w_dict}


def _execute_single_robustness_draw(
    draw_index: int,
    weights: np.ndarray,
    run_pipeline_callable: Callable,
    cache: PipelineCache
) -> Dict[str, Any]:
    """
    Execute one utility-weight robustness draw via the ESTIMATION-ONLY path (no
    inference, no fusion), recording the selected ensemble and its utility.

    Rationale (Section 11.7; detailed tasklist Task 23, Step 1). The prior
    implementation invoked the full run_pipeline (including the B=10,000 nested
    bootstrap and confirmatory tests) per draw — intractable and contrary to the
    minimal-recomputation mandate. Here, `run_pipeline_callable` is the dedicated
    estimation-only callable (constructed by `execute_dirichlet_sweeps` with the
    cached Sigma/MRC/IG bound), invoked as callable(cache, perturbation, draw_index);
    it recomputes only MEC + selection. The function remains stateless and picklable
    for multiprocessing.

    Parameters
    ----------
    draw_index : int
        Unique identifier for this Monte Carlo draw.
    weights : np.ndarray
        Length-4 Dirichlet-drawn weights (alpha, beta, gamma, delta).
    run_pipeline_callable : Callable
        The dedicated estimation-only callable (cache, perturbation, draw_index) ->
        {'S_FMEC', 'fmec_table', 'w_dict'}. Retained name for interface compatibility.
    cache : PipelineCache
        The frozen data-generating substrate.

    Returns
    -------
    Dict[str, Any]
        The draw index, the weights used, the selected ensemble, the utility, and a
        status flag.
    """
    # Construct the perturbation targeting ONLY the utility weights.
    perturbation = {
        "utility_weights": {
            "alpha": float(weights[0]), "beta": float(weights[1]),
            "gamma": float(weights[2]), "delta": float(weights[3]),
        }
    }
    try:
        # Invoke the estimation-only path (NOT the full pipeline): recomputes MEC + selection only.
        result = run_pipeline_callable(cache, perturbation, draw_index)
        # Extract the selected ensemble and sort it for order-invariant comparison.
        selected_ensemble = tuple(sorted(result["S_FMEC"]))
        # Retrieve the FMEC table and the selection weights for the utility computation.
        fmec_table = result["fmec_table"]
        w_dict = result["w_dict"]
        # Compute U(w) = sum_i w_i * FMEC_i over the selected members.
        fmec_by_model = fmec_table.set_index("model_id")["fmec"].to_dict()
        utility_val = float(sum(w_dict[m] * fmec_by_model[m] for m in w_dict if m in fmec_by_model))
        # Return the structured draw record.
        return {
            "draw_index": draw_index,
            "alpha": float(weights[0]), "beta": float(weights[1]),
            "gamma": float(weights[2]), "delta": float(weights[3]),
            "selected_ensemble": selected_ensemble, "utility": utility_val, "status": "SUCCESS",
        }
    except Exception as e:
        # Surface a failed draw without crashing the entire pool.
        return {"draw_index": draw_index, "status": f"FAILED: {str(e)}"}


@dataclass(frozen=True)
class _EstimationDispatchContext:
    """
    A lightweight, picklable substrate for the estimation-only robustness workers.

    Rationale (Section 11.7; detailed tasklist Task 23, Step 1). The full PipelineCache
    carries `fuse_callable`, which is a closure returned by `specify_fusion_operator`
    and therefore NOT picklable; submitting the full cache to a ProcessPoolExecutor
    raises `AttributeError: Can't pickle local object 'specify_fusion_operator.<locals>.fuse'`.
    The estimation-only path (`_estimate_select_minimal`) never reads `fuse_callable`
    (fusion is skipped under perturbation, per the Task 22 correction); it reads only the
    four fields below. This frozen dataclass carries exactly those fields — all picklable
    and, for the RNG bundle, stream-faithful across a pickle round-trip — so the process
    pool can serialize it without touching the fusion closure.

    Attributes
    ----------
    estimator_projection : pd.DataFrame
        The 13-column analytic-core tensor (read by the utility closure and the selector).
    qualified_pool : List[str]
        The canonical ordered model IDs.
    base_config : Mapping[str, Any]
        The active-config source and the Monte Carlo root seed.
    rng_bundle : Dict[str, np.random.Generator]
        The RNG bundle; only the 'monte_carlo_sampling_seed' stream is consumed (and it is
        re-seeded per draw), but the full bundle is carried for behavioural parity.
    """
    # The analytic-core projection consumed by build_utility_function and select_ensemble.
    estimator_projection: pd.DataFrame
    # The canonical ordered model identifiers.
    qualified_pool: List[str]
    # The configuration mapping (active-config source + MC root).
    base_config: Mapping[str, Any]
    # The RNG bundle (picklable, stream-faithful); the MC stream is re-seeded per draw.
    rng_bundle: Dict[str, np.random.Generator]


def execute_dirichlet_sweeps(
    run_pipeline_callable: Callable,
    cache: PipelineCache,
    n_draws: int,
    rng: np.random.Generator
) -> pd.DataFrame:
    """
    Draw utility-weight vectors from a symmetric Dirichlet and evaluate each via the
    estimation-only path, precomputing the weight-independent primitives ONCE and
    dispatching a PICKLABLE context (never the fusion closure) to the worker pool.

    Rationale (Section 11.7; detailed tasklist Task 23, Step 1). (alpha, beta, gamma,
    delta) ~ Dirichlet(1,1,1,1). Sigma and IG are weight-independent and are
    precomputed once; each draw then recomputes only MEC and selection (via
    `_estimate_select_minimal`), with no inference and no fusion. The correction here
    is the dispatch substrate: the prior version submitted the full `cache` to the
    ProcessPoolExecutor, but `cache.fuse_callable` is a closure and is NOT picklable,
    so `executor.submit(...)` raised an AttributeError and the sweep could never run in
    parallel. Because the estimation-only worker path reads only
    {estimator_projection, qualified_pool, base_config, rng_bundle}, this version
    constructs a frozen, picklable `_EstimationDispatchContext` carrying exactly those
    fields (excluding the fusion closure) and submits THAT to the workers.

    Parameters
    ----------
    run_pipeline_callable : Callable
        Retained for interface compatibility; the function constructs and uses the
        dedicated estimation-only callable mandated by Section 11.7.
    cache : PipelineCache
        The frozen data-generating substrate (its picklable subset is forwarded to the
        workers via `_EstimationDispatchContext`; `fuse_callable` is deliberately excluded).
    n_draws : int
        Number of Monte Carlo draws (e.g., 10,000).
    rng : np.random.Generator
        Dedicated RNG for the Dirichlet draws.

    Returns
    -------
    pd.DataFrame
        Results of all successful draws.
    """
    # Announce the sweep.
    logger.info(f"Initiating {n_draws} Dirichlet robustness sweeps (estimation-only, cached IG/covariance)...")

    # ---- Precompute the weight-independent primitives ONCE (Sigma, MRC, IG) ----
    # Estimate the covariance and marginal-risk vector (independent of the utility weights).
    sigma_df, mrc_dict, _ = estimate_risk(
        estimator_projection=cache.estimator_projection, qualified_pool=cache.qualified_pool, config=cache.base_config,
    )
    # Materialize the aligned covariance array for the utility closure.
    sigma_matrix = sigma_df.reindex(index=cache.qualified_pool, columns=cache.qualified_pool).values.astype(np.float64)
    # Build a baseline utility closure solely to obtain the coalition samples for IG.
    judge_model_id = cache.base_config["models"]["judge"]
    v_func_base = build_utility_function(
        estimator_projection=cache.estimator_projection, sigma=sigma_matrix,
        config=cache.base_config, qualified_pool=cache.qualified_pool, judge_model_id=judge_model_id,
    )
    # Obtain the coalition samples (weight-independent permutation set) for the IG estimator.
    _, coalition_samples = estimate_mec(
        qualified_pool=cache.qualified_pool, v_func=v_func_base, config=cache.base_config, rng_bundle=cache.rng_bundle,
    )
    # Estimate IG once (independent of the utility weights), reused across all draws.
    ig_dict = estimate_ig(
        estimator_projection=cache.estimator_projection, coalition_samples=coalition_samples,
        qualified_pool=cache.qualified_pool, config=cache.base_config,
    )

    # Bind the cached primitives into the picklable estimation-only callable.
    est_callable = functools.partial(
        _estimate_select_minimal,
        cached_sigma_df=sigma_df, cached_sigma_matrix=sigma_matrix,
        cached_mrc_dict=mrc_dict, cached_ig_dict=ig_dict,
    )

    # ---- CORRECTION: build a PICKLABLE dispatch context (excludes the fuse_callable closure) ----
    # The estimation-only worker reads only these four fields; carrying the full cache would
    # drag in `fuse_callable` (a closure from specify_fusion_operator) and break pickling.
    dispatch_context = _EstimationDispatchContext(
        estimator_projection=cache.estimator_projection,
        qualified_pool=cache.qualified_pool,
        base_config=cache.base_config,
        rng_bundle=cache.rng_bundle,
    )

    # Draw N weight vectors from the symmetric Dirichlet(1,1,1,1).
    weight_draws = rng.dirichlet(np.ones(4), size=n_draws)
    # Accumulate successful draw records.
    results: List[Dict[str, Any]] = []

    # Dispatch the draws in parallel; each worker runs the estimation-only path against the
    # picklable context (NOT the full cache), so submission no longer fails on the fusion closure.
    with concurrent.futures.ProcessPoolExecutor() as executor:
        # Submit one estimation-only draw per Dirichlet sample, forwarding the picklable context.
        futures = [
            executor.submit(_execute_single_robustness_draw, i, weight_draws[i], est_callable, dispatch_context)
            for i in range(n_draws)
        ]
        # Collect results as they complete (order-invariant aggregation).
        for future in concurrent.futures.as_completed(futures):
            res = future.result()
            # Keep successes; log failures.
            if res.get("status") == "SUCCESS":
                results.append(res)
            else:
                logger.warning(f"Robustness draw failed: {res.get('status')}")

    # Materialize the successful draws into a DataFrame.
    df_results = pd.DataFrame(results)
    # Report completion.
    logger.info(f"Completed {len(df_results)} successful robustness sweeps.")
    # Return the sweep results.
    return df_results

# -------------------------------------------------------------------------------------------------------------------------------
# Task 23, Step 2: Hyperparameter and estimator ablations
# -------------------------------------------------------------------------------------------------------------------------------

def execute_ablations(
    run_pipeline_callable: Callable,
    cache: PipelineCache,
    baseline_bundle: Any
) -> Dict[str, Dict[str, Any]]:
    """
    Execute the six pre-registered ablations via MINIMAL, dependency-edge-mapped
    recomputation on the cached substrate (no inference, no fusion).

    Rationale (Section 11.7, Ablations A1-A6; detailed tasklist Task 23, Step 2).
    Each ablation recomputes ONLY what its perturbation invalidates:
      A1/A2 ((eta,lambda)) -> recompose FMEC + select from cached MEC/IG/MRC;
      A4 (K)               -> recompute MEC + IG (coalition set changes), reuse Sigma/MRC;
      A5 (covariance)      -> recompute Sigma + MRC + MEC, reuse IG;
      A6 (IG estimator)    -> recompute IG, reuse MEC/Sigma/MRC;
      A3 (LOO MEC)         -> compute MEC_j = v(N) - v(N\\{j}) directly via the closure.
    A3 is functional WITHOUT any sampler-override perturbation or Task 13 edit. A5/A6
    attempt the alternative estimator and report an explicit 'NOT_IMPLEMENTED' status
    if the current estimator rejects it (the blueprint-sanctioned skip), rather than a
    silent generic failure.

    Parameters
    ----------
    run_pipeline_callable : Callable
        Retained for interface compatibility; minimal recomputation is used instead.
    cache : PipelineCache
        The frozen, remediated data-generating substrate.
    baseline_bundle : Any
        The ResultsBundle (or EstimationOutputs) from the unperturbed baseline run.

    Returns
    -------
    Dict[str, Dict[str, Any]]
        Mapping ablation name -> {'status', 'selected_ensemble', 'flipped'}.
    """
    # Announce the ablation suite.
    logger.info("Initiating dependency-mapped hyperparameter and estimator ablations...")
    # The baseline ensemble as a sorted tuple for order-invariant flip detection.
    baseline_ensemble = tuple(sorted(baseline_bundle.S_FMEC))

    # ---- Precompute baseline primitives ONCE on the cached substrate ----
    base_config = cache.base_config
    # Baseline covariance + marginal risk.
    sigma_df_base, mrc_base, _ = estimate_risk(cache.estimator_projection, cache.qualified_pool, base_config)
    # Aligned baseline covariance array.
    sigma_matrix_base = sigma_df_base.reindex(index=cache.qualified_pool, columns=cache.qualified_pool).values.astype(np.float64)
    # Baseline utility closure.
    judge_model_id = base_config["models"]["judge"]
    v_func_base = build_utility_function(cache.estimator_projection, sigma_matrix_base, base_config, cache.qualified_pool, judge_model_id)
    # Baseline MEC + coalition samples.
    mec_base, coalition_samples_base = estimate_mec(cache.qualified_pool, v_func_base, base_config, cache.rng_bundle)
    # Baseline IG.
    ig_base = estimate_ig(cache.estimator_projection, coalition_samples_base, cache.qualified_pool, base_config)

    # Helper: compose FMEC inline (Eq. 13) and select, reusing the supplied components.
    def _compose_and_select(mec: Dict[str, float], ig: Dict[str, float], mrc: Dict[str, float],
                            eta: float, lam: float, sigma_df: pd.DataFrame, cfg: Mapping[str, Any]) -> tuple:
        # Build the FMEC rows from the supplied MEC/IG/MRC and (eta, lambda).
        rows = [
            {"model_id": m, "mec_hat": mec[m], "ig": ig[m], "mrc": mrc[m],
             "fmec": mec[m] + eta * ig[m] - lam * mrc[m]}
            for m in cache.qualified_pool
        ]
        # Materialize the FMEC table.
        tbl = pd.DataFrame(rows)
        # Select the ensemble under the supplied covariance and config.
        S, _, _ = select_ensemble(tbl, cache.estimator_projection, sigma_df, cache.qualified_pool, cfg)
        # Return the selected set.
        return S

    # Baseline (eta, lambda) for the recomposition ablations.
    eta0 = base_config["estimation"]["fmec_hyperparams"]["eta"]
    lam0 = base_config["estimation"]["fmec_hyperparams"]["lambda"]

    # Results accumulator.
    ablation_results: Dict[str, Dict[str, Any]] = {}

    # Helper to record an ablation outcome (selected set + flip status).
    def _record(name: str, S: List[str]) -> None:
        # Sort for order-invariant comparison.
        sel = tuple(sorted(S))
        # Store the result with the flip indicator.
        ablation_results[name] = {"status": "SUCCESS", "selected_ensemble": sel, "flipped": (sel != baseline_ensemble)}

    # ---- A1: no IG (eta = 0) -> recompose + select from cached MEC/IG/MRC ----
    try:
        S = _compose_and_select(mec_base, ig_base, mrc_base, 0.0, lam0, sigma_df_base, base_config)
        _record("A1_no_ig", S)
    except Exception as e:
        ablation_results["A1_no_ig"] = {"status": f"FAILED: {e}", "selected_ensemble": tuple(), "flipped": False}

    # ---- A2: no risk (lambda = 0) -> recompose + select ----
    try:
        S = _compose_and_select(mec_base, ig_base, mrc_base, eta0, 0.0, sigma_df_base, base_config)
        _record("A2_no_risk", S)
    except Exception as e:
        ablation_results["A2_no_risk"] = {"status": f"FAILED: {e}", "selected_ensemble": tuple(), "flipped": False}

    # ---- A3: leave-one-out MEC -> compute directly via the closure, reuse IG/MRC ----
    try:
        # Direct LOO contributions MEC_j = v(N) - v(N\{j}) (Eq. 30-LOO).
        mec_loo = _compute_loo_mec(v_func_base, cache.qualified_pool)
        # Recompose with baseline (eta, lambda) and select.
        S = _compose_and_select(mec_loo, ig_base, mrc_base, eta0, lam0, sigma_df_base, base_config)
        _record("A3_loo_mec", S)
    except Exception as e:
        ablation_results["A3_loo_mec"] = {"status": f"FAILED: {e}", "selected_ensemble": tuple(), "flipped": False}

    # ---- A4: K sweep -> recompute MEC + IG (coalition set changes), reuse Sigma/MRC ----
    for k_val in (64, 128, 256, 1024):
        name = f"A4_K_{k_val}"
        try:
            # Active config with the perturbed K.
            cfg_k = copy.deepcopy(dict(base_config))
            cfg_k["estimation"]["monte_carlo"]["K"] = k_val
            # Recompute MEC at the new K (utility closure unchanged: K does not affect v).
            mec_k, coalitions_k = estimate_mec(cache.qualified_pool, v_func_base, cfg_k, cache.rng_bundle)
            # Recompute IG on the new coalition set.
            ig_k = estimate_ig(cache.estimator_projection, coalitions_k, cache.qualified_pool, cfg_k)
            # Recompose with baseline (eta, lambda) and select.
            S = _compose_and_select(mec_k, ig_k, mrc_base, eta0, lam0, sigma_df_base, cfg_k)
            _record(name, S)
        except Exception as e:
            ablation_results[name] = {"status": f"FAILED: {e}", "selected_ensemble": tuple(), "flipped": False}

    # ---- A5: empirical covariance -> recompute Sigma + MRC + MEC (v depends on Sigma), reuse IG ----
    try:
        # Active config selecting the empirical covariance estimator.
        cfg_cov = copy.deepcopy(dict(base_config))
        cfg_cov["estimation"]["covariance"]["estimator"] = "empirical"
        # Attempt to re-estimate the covariance with the alternative estimator.
        sigma_df_emp, mrc_emp, _ = estimate_risk(cache.estimator_projection, cache.qualified_pool, cfg_cov)
        # Aligned empirical covariance array.
        sigma_matrix_emp = sigma_df_emp.reindex(index=cache.qualified_pool, columns=cache.qualified_pool).values.astype(np.float64)
        # Rebuild the utility closure under the empirical covariance (v depends on Sigma).
        v_func_emp = build_utility_function(cache.estimator_projection, sigma_matrix_emp, cfg_cov, cache.qualified_pool, judge_model_id)
        # Recompute MEC under the empirical covariance; reuse cached IG (IG is covariance-independent).
        mec_emp, _ = estimate_mec(cache.qualified_pool, v_func_emp, cfg_cov, cache.rng_bundle)
        # Recompose and select under the empirical covariance.
        S = _compose_and_select(mec_emp, ig_base, mrc_emp, eta0, lam0, sigma_df_emp, cfg_cov)
        _record("A5_cov_empirical", S)
    except (ValueError, NotImplementedError) as e:
        # The current estimator does not implement 'empirical': report explicitly (blueprint-sanctioned skip).
        logger.warning(f"Ablation A5 (empirical covariance) not available: {e}")
        ablation_results["A5_cov_empirical"] = {"status": "NOT_IMPLEMENTED", "selected_ensemble": tuple(), "flipped": False}
    except Exception as e:
        ablation_results["A5_cov_empirical"] = {"status": f"FAILED: {e}", "selected_ensemble": tuple(), "flipped": False}

    # ---- A6: entropy-reduction IG -> recompute IG only, reuse MEC/Sigma/MRC ----
    try:
        # Active config selecting the entropy-reduction IG estimator.
        cfg_ig = copy.deepcopy(dict(base_config))
        cfg_ig["estimation"]["information_gain"]["estimator"] = "entropy_reduction"
        # Attempt to recompute IG with the alternative estimator on the baseline coalition set.
        ig_alt = estimate_ig(cache.estimator_projection, coalition_samples_base, cache.qualified_pool, cfg_ig)
        # Recompose with cached MEC/MRC and select.
        S = _compose_and_select(mec_base, ig_alt, mrc_base, eta0, lam0, sigma_df_base, cfg_ig)
        _record("A6_ig_entropy_reduction", S)
    except (ValueError, NotImplementedError) as e:
        # The current estimator does not implement 'entropy_reduction': report explicitly.
        logger.warning(f"Ablation A6 (entropy-reduction IG) not available: {e}")
        ablation_results["A6_ig_entropy_reduction"] = {"status": "NOT_IMPLEMENTED", "selected_ensemble": tuple(), "flipped": False}
    except Exception as e:
        ablation_results["A6_ig_entropy_reduction"] = {"status": f"FAILED: {e}", "selected_ensemble": tuple(), "flipped": False}

    # Report completion.
    logger.info("Completed dependency-mapped ablations.")
    # Return the ablation comparison table.
    return ablation_results

# -------------------------------------------------------------------------------------------------------------------------------
# Task 23, Step 3: Aggregate the robustness surface and flag flips
# -------------------------------------------------------------------------------------------------------------------------------

def compute_stability_surface(
    sweep_results: pd.DataFrame,
    rsp_floor: float
) -> Tuple[pd.DataFrame, float, bool]:
    """
    Aggregates the Dirichlet sweep results to compute the selection probability
    mass for each unique ensemble and evaluates the fragility of the recommendation.

    Parameters
    ----------
    sweep_results : pd.DataFrame
        The DataFrame containing the results of the Dirichlet sweeps.
    rsp_floor : float
        The pre-registered stability floor (e.g., 0.90).

    Returns
    -------
    Tuple[pd.DataFrame, float, bool]
        - stability_surface: DataFrame detailing the probability mass per ensemble.
        - max_mass: The maximum probability mass achieved by any ensemble.
        - is_fragile: True if max_mass < rsp_floor.
    """
    if sweep_results.empty:
        return pd.DataFrame(), 0.0, True

    n_total = len(sweep_results)

    # Convert the tuple representations back to strings for grouping
    sweep_results["ensemble_str"] = sweep_results["selected_ensemble"].apply(lambda x: ",".join(x))

    # Group by the unique ensemble string and count occurrences
    surface_df = sweep_results.groupby("ensemble_str").size().reset_index(name="count")

    # Compute the probability mass (frequency)
    surface_df["probability_mass"] = surface_df["count"] / float(n_total)

    # Sort descending by probability mass
    surface_df = surface_df.sort_values(by="probability_mass", ascending=False).reset_index(drop=True)

    # Extract the maximum stability mass
    max_mass = float(surface_df["probability_mass"].max()) if not surface_df.empty else 0.0

    # Evaluate fragility against the pre-registered floor
    is_fragile = max_mass < rsp_floor

    return surface_df, max_mass, is_fragile

# -------------------------------------------------------------------------------------------------------------------------------
# Task 23, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def robustness(
    run_pipeline_callable: Callable,
    cache: Any, # Expecting PipelineCache
    baseline_bundle: Any # Expecting ResultsBundle
) -> Any: # Returns RobustnessReport
    """
    Orchestrates Task 23: Coordinates Dirichlet sweeps and ablations.

    This remediated orchestrator integrates the remediated execute_ablations
    function to accurately include Ablation A3 (Leave-one-out MEC, Eq. 30).
    It evaluates recommendation stability and reports the fragility status.

    Parameters
    ----------
    run_pipeline_callable : Callable
        The Task 22 orchestrator function (run_pipeline).
    cache : PipelineCache
        The frozen, remediated data-generating substrate containing the authentic qualification data.
    baseline_bundle : ResultsBundle
        The ResultsBundle from the unperturbed baseline run.

    Returns
    -------
    RobustnessReport
        A strictly typed dataclass containing the stability surface and ablation outcomes.
    """
    # Extract the configuration parameters
    n_draws = cache.base_config["stat_tests"].get("robustness_draws", 10000)
    rsp_floor = cache.base_config["stat_tests"].get("rsp_floor", 0.90)

    # Extract the global RNG stream for the Dirichlet draws to ensure reproducibility
    global_rng = cache.rng_bundle["global_seed"]

    # Step 1: Execute the 10,000 Dirichlet sweeps in parallel (ProcessPoolExecutor)
    # execute_dirichlet_sweeps is assumed available
    sweep_results_df = execute_dirichlet_sweeps(
        run_pipeline_callable=run_pipeline_callable,
        cache=cache,
        n_draws=n_draws,
        rng=global_rng
    )

    # Step 2: Execute the pre-registered ablations (including the remediated A3)
    # execute_ablations is assumed available
    ablation_results = execute_ablations(
        run_pipeline_callable=run_pipeline_callable,
        cache=cache,
        baseline_bundle=baseline_bundle
    )

    # Step 3: Compute the stability surface and evaluate fragility
    # compute_stability_surface is assumed available
    stability_surface, max_mass, is_fragile = compute_stability_surface(
        sweep_results=sweep_results_df,
        rsp_floor=rsp_floor
    )

    if is_fragile:
        logger.warning(
            f"ROBUSTNESS WARNING: The recommendation is fragile. No ensemble achieved "
            f"a stability mass >= {rsp_floor}. Max mass observed: {max_mass:.3f}."
        )
    else:
        logger.info(
            f"ROBUSTNESS PASSED: The recommendation is stable across preference perturbations. "
            f"Max mass observed: {max_mass:.3f} >= {rsp_floor}."
        )

    # Package the results into the strictly typed RobustnessReport dataclass (Task 23 contract)
    return RobustnessReport(
        stability_surface=stability_surface,
        max_stability_mass=max_mass,
        is_fragile=is_fragile,
        ablation_results=ablation_results
    )
````

````python
# Task 24 — Reporting, Visualization & Reproducibility Packaging (report_and_package)

# ===================================================================================
# Task 24: Reporting, Visualization & Reproducibility Packaging (report_and_package)
# ===================================================================================

# -------------------------------------------------------------------------------------------------------------------------------
# Task 24, Step 1: Generate the Figure Suite
# -------------------------------------------------------------------------------------------------------------------------------

def generate_figure_suite(
    bundle: ResultsBundle,
    cache: PipelineCache,
    sigma_df: pd.DataFrame,
    run_dir: Path
) -> None:
    """
    Generates and saves the five mandated high-resolution plots: Efficient Frontier,
    FMEC CI Forest, Covariance Heatmap, RSP Bar, and Variance Components.

    Parameters
    ----------
    bundle : ResultsBundle
        The final results dataclass containing FMEC tables and stability metrics.
    cache : PipelineCache
        The frozen data-generating substrate containing the estimator projection.
    sigma_df : pd.DataFrame
        The n x n error covariance matrix.
    run_dir : Path
        The path to the immutable run directory where figures will be saved.
    """
    # Extract plotting configuration parameters from the base config
    plot_config = cache.base_config.get("plots", {})
    # Default to 300 DPI for publication-grade resolution
    dpi = plot_config.get("dpi", 300)
    # Default to PDF format for vector graphics
    fmt = plot_config.get("figure_format", "pdf")
    # Extract font scaling for readability
    font_scale = plot_config.get("font_scale", 1.1)

    # Apply professional seaborn theme for consistent aesthetics
    sns.set_theme(style="whitegrid", font_scale=font_scale)

    # --- Plot 1: Efficient Frontier (Cost vs. Quality) ---
    # Initialize the figure and axes explicitly to manage memory
    fig1, ax1 = plt.subplots(figsize=(10, 6))

    # Compute individual model quality (mean correctness) and cost
    # observed=True ensures we only group by existing categorical combinations
    mean_correctness = cache.estimator_projection.groupby("model_id", observed=True)["correct"].mean()
    mean_cost = cache.estimator_projection.groupby("model_id", observed=True)["cost_usd"].mean()

    # Plot each individual model in the qualified pool
    for m_id in cache.qualified_pool:
        # Extract quality and cost, defaulting to 0.0 if missing
        q = mean_correctness.get(m_id, 0.0)
        c = mean_cost.get(m_id, 0.0)
        # Scatter plot the individual model
        ax1.scatter(c, q, label=m_id, alpha=0.7, s=100)
        # Annotate the point with the model ID
        ax1.annotate(m_id, (c, q), xytext=(5, 5), textcoords='offset points', fontsize=9)

    # Compute S_FMEC quality and cost
    # Map the string IDs in S_FMEC to their integer indices in the qualified pool
    S_indices = [cache.qualified_pool.index(m) for m in bundle.S_FMEC]

    # Reconstruct the dense correctness matrix A_it to compute majority vote quality
    A_df = cache.estimator_projection.groupby(["task_id", "model_id"], observed=True)["correct"].mean().unstack().reindex(columns=cache.qualified_pool)
    A_matrix = (A_df.values > 0.5).astype(np.int8)

    if S_indices:
        # Slice the matrix for the selected ensemble
        A_S = A_matrix[:, S_indices]
        # Compute row-wise sums (number of correct models per task)
        row_sums = np.sum(A_S, axis=1)
        # Apply the strict majority threshold (Eq. 3a)
        q_fmec = float(np.mean(row_sums > (len(S_indices) / 2.0)))

        # Compute the total cost: sum of members plus the judge (Eq. 36)
        judge_id = cache.base_config["models"]["judge"]
        c_j = mean_cost.get(judge_id, 0.0)
        c_fmec = float(np.sum([mean_cost.get(m, 0.0) for m in bundle.S_FMEC]) + c_j)

        # Plot the selected ensemble as a distinct red star
        ax1.scatter(c_fmec, q_fmec, color='red', marker='*', s=300, label='S_FMEC', edgecolor='black')
        ax1.annotate('S_FMEC', (c_fmec, q_fmec), xytext=(10, -10), textcoords='offset points', color='red', weight='bold')

    # Set titles and labels
    ax1.set_title("Efficient Frontier: Quality vs. Cost")
    ax1.set_xlabel("Expected Cost per Task (USD)")
    ax1.set_ylabel("Expected Quality (Majority Correctness)")
    # Place legend outside the plot area
    ax1.legend(loc='lower right', bbox_to_anchor=(1.3, 0))
    plt.tight_layout()
    # Save the figure to the run directory
    fig1.savefig(run_dir / f"efficient_frontier.{fmt}", dpi=dpi, format=fmt)
    # Explicitly close the figure to prevent memory leaks
    plt.close(fig1)

    # --- Plot 2: FMEC CI Forest Plot ---
    fig2, ax2 = plt.subplots(figsize=(10, 6))

    # Sort the FMEC table by the FMEC score ascending so the highest is at the top of the y-axis
    sorted_fmec = bundle.fmec_table.sort_values(by="fmec", ascending=True).reset_index(drop=True)

    # Extract data arrays
    models = sorted_fmec["model_id"]
    fmec_vals = sorted_fmec["fmec"]

    # Calculate asymmetric errors for matplotlib errorbar: shape (2, N) -> [lower_errors, upper_errors]
    # Error is the distance from the point estimate, not the absolute bound
    lower_err = fmec_vals - sorted_fmec["ci_low"]
    upper_err = sorted_fmec["ci_high"] - fmec_vals
    errors = np.vstack([lower_err, upper_err])

    # Plot the point estimates with horizontal error bars
    ax2.errorbar(fmec_vals, models, xerr=errors, fmt='o', color='black', ecolor='gray', capsize=5, markersize=8)

    # Annotate each point with its bootstrap selection frequency
    for i, (_, row) in enumerate(sorted_fmec.iterrows()):
        ax2.annotate(f"Sel Freq: {row['sel_freq']:.2f}",
                     (row['ci_high'], i),
                     xytext=(10, -3), textcoords='offset points', fontsize=9, color='blue')

    ax2.set_title("FMEC Point Estimates with 95% BCa Confidence Intervals")
    ax2.set_xlabel("Final Marginal Ensemble Contribution (FMEC)")
    ax2.set_ylabel("Candidate Models")
    plt.tight_layout()
    fig2.savefig(run_dir / f"fmec_ci_forest.{fmt}", dpi=dpi, format=fmt)
    plt.close(fig2)

    # --- Plot 3: Covariance Heatmap ---
    fig3, ax3 = plt.subplots(figsize=(10, 8))
    # Extract heatmap specific configurations
    cmap = plot_config.get("covariance_heatmap", {}).get("cmap", "viridis")
    annot = plot_config.get("covariance_heatmap", {}).get("annot", True)

    # Render the heatmap using seaborn (Eq. 10 visualization)
    sns.heatmap(sigma_df, cmap=cmap, annot=annot, fmt=".4f", ax=ax3, cbar_kws={'label': 'Error Covariance'})
    ax3.set_title("Error-Mode Covariance Matrix ($\widehat{\Sigma}$)")
    plt.tight_layout()
    fig3.savefig(run_dir / f"covariance_heatmap.{fmt}", dpi=dpi, format=fmt)
    plt.close(fig3)

    # --- Plot 4: RSP Bar Chart ---
    fig4, ax4 = plt.subplots(figsize=(6, 6))

    # Plot the single scalar RSP value (Eq. 51)
    ax4.bar(["Recommendation Stability\nProbability (RSP)"], [bundle.RSP], color='steelblue', edgecolor='black')
    # Set y-axis limits to [0, 1] for probability
    ax4.set_ylim(0, 1.05)

    # Add pre-registered reference lines
    ref_lines = plot_config.get("rsp_bar", {}).get("threshold_lines", [0.5, 0.9])
    for line in ref_lines:
        # Color the 0.9 floor red, others orange
        color = 'red' if line == 0.9 else 'orange'
        ax4.axhline(y=line, color=color, linestyle='--', label=f'Threshold {line}')

    ax4.set_title("Ensemble Selection Stability")
    ax4.set_ylabel("Probability")
    ax4.legend()
    plt.tight_layout()
    fig4.savefig(run_dir / f"rsp_bar.{fmt}", dpi=dpi, format=fmt)
    plt.close(fig4)

    # --- Plot 5: Variance Components ---
    fig5, ax5 = plt.subplots(figsize=(8, 6))

    # Extract the variance components (Eq. 20a)
    components = bundle.variance_components
    labels = ["Task Sampling\n($\sigma^2_{\mathrm{task}}$)",
              "Generation Noise\n($\sigma^2_{\mathrm{gen}}$)",
              "Adjudication Noise\n($\sigma^2_{\mathrm{adj}}$)"]
    values = [components.get("sigma2_task", 0.0),
              components.get("sigma2_gen", 0.0),
              components.get("sigma2_adj", 0.0)]

    # Plot the components as a bar chart
    ax5.bar(labels, values, color=['#4c72b0', '#55a868', '#c44e52'], edgecolor='black')
    ax5.set_title("Variance Components Decomposition (Eq. 20a)")
    ax5.set_ylabel("Variance")
    plt.tight_layout()
    fig5.savefig(run_dir / f"variance_components_bar.{fmt}", dpi=dpi, format=fmt)
    plt.close(fig5)

# -------------------------------------------------------------------------------------------------------------------------------
# Task 24, Step 2: Generate Scientific Report (Falsification & Disclosures)
# -------------------------------------------------------------------------------------------------------------------------------

def generate_scientific_report(
    bundle: ResultsBundle,
    robustness_report: RobustnessReport,
    cache: PipelineCache,
    run_dir: Path
) -> None:
    """
    Render the clinical Markdown report with a built-in honesty guard that refuses
    to render populated-looking numbers when the underlying artifacts are unfilled.

    Rationale (Section 11.13; detailed tasklist Task 24, Step 2). The report must
    state the recommendation as stable/unstable/inconclusive (severity-ranked),
    report failed hypotheses plainly, disclose missing artifacts/unavailable
    versions/unresolved contamination, and — critically — NEVER render fabricated
    numbers from unfilled artifacts. The prior implementation rendered FMEC values,
    p-values, and effect sizes unconditionally. This version inspects finiteness and
    provenance before rendering each number, emitting an explicit
    'NOT EXECUTED / artifact unfilled' marker otherwise, and softens the Tier-1
    bit-exact claim unless the tensor hash is real and verified.

    Parameters
    ----------
    bundle : ResultsBundle
        The final results dataclass (point estimates, verdicts, stability).
    robustness_report : RobustnessReport
        The robustness surface and fragility flag.
    cache : PipelineCache
        The frozen data-generating substrate (config, pruned tasks, qualification).
    run_dir : Path
        The immutable run directory.

    Returns
    -------
    None
        The report is written to `<run_dir>/scientific_report.md`.
    """
    # Sentinel string rendered wherever an underlying value is unfilled/non-finite.
    UNFILLED = "NOT EXECUTED / artifact unfilled"

    # Local guard: True iff x is a finite real number (gates every numeric render).
    def _finite(x: Any) -> bool:
        # Reject None and non-numerics, then test finiteness.
        try:
            return x is not None and np.isfinite(float(x))
        except (TypeError, ValueError):
            return False

    # Local renderer: format a float to `fmt` if finite, else the UNFILLED marker.
    def _num(x: Any, fmt: str = "{:.3f}") -> str:
        # Render the number only when it is a finite real value.
        return fmt.format(float(x)) if _finite(x) else UNFILLED

    # Accumulator for the Markdown lines.
    report_lines: List[str] = []

    # --- Header ---
    report_lines.append("# Scientific Report: FMEC Ensemble Optimization")
    report_lines.append(f"**Run Directory:** `{run_dir.name}`\n")

    # --- Section 1: Overall Recommendation Status (severity-ranked) ---
    report_lines.append("## 1. Overall Recommendation Status")
    # The severity-ranked overall status (from apply_falsification_rules).
    status = str(bundle.verdicts.get("overall_status", "UNKNOWN")).upper()
    report_lines.append(f"**Status:** {status}")
    # List the explicit falsification reasons whenever the status is not STABLE.
    if status != "STABLE":
        report_lines.append("\n**Falsification Reasons:**")
        for reason in bundle.verdicts.get("falsification_reasons", []):
            report_lines.append(f"- {reason}")
    # Render the selected ensemble; an empty selection is an unexecuted/empty artifact.
    if bundle.S_FMEC:
        report_lines.append(f"\n**Selected Ensemble ($S_{{\\mathrm{{FMEC}}}}$):** `{bundle.S_FMEC}`")
    else:
        report_lines.append(f"\n**Selected Ensemble ($S_{{\\mathrm{{FMEC}}}}$):** {UNFILLED}")
    # Render RSP and the fragility flag, gating RSP on finiteness.
    report_lines.append(f"**Recommendation-Stability Probability (RSP):** {_num(bundle.RSP)}")
    report_lines.append(f"**Robustness Fragility Flag:** {robustness_report.is_fragile}\n")

    # --- Section 2: Confirmatory Hypothesis Tests ---
    report_lines.append("## 2. Confirmatory Hypothesis Tests (Holm-Bonferroni $\\alpha=0.05$)")
    # The per-hypothesis verdict dictionary.
    hypotheses = bundle.verdicts.get("hypotheses", {})
    # If no hypotheses were evaluated, render the unexecuted marker for the whole section.
    if not hypotheses:
        report_lines.append(f"- {UNFILLED} (confirmatory tests did not run).")
    else:
        # Render each hypothesis, distinguishing underpowered (NaN by design) from unexecuted.
        for h_name, h_data in hypotheses.items():
            # Map the boolean reject flag to clinical text.
            verdict = "SUPPORTED" if h_data.get("reject", False) else "FALSIFIED / FAILED TO REJECT NULL"
            # The adjusted p-value (may be NaN for an underpowered test).
            p_adj = h_data.get("p_adj", float("nan"))
            # The raw p-value provenance disambiguates 'underpowered' from 'unevaluated'.
            p_raw = h_data.get("p_raw", float("nan"))
            if _finite(p_adj):
                # A finite adjusted p-value renders normally.
                report_lines.append(f"- **{h_name}:** {verdict} (Adjusted $p$-value: {p_adj:.4e})")
            elif isinstance(p_raw, float) and np.isnan(p_raw):
                # A NaN raw p-value indicates a deliberately underpowered test.
                report_lines.append(f"- **{h_name}:** UNDERPOWERED (insufficient pool size; not falsified).")
            else:
                # Otherwise the test result is genuinely unfilled.
                report_lines.append(f"- **{h_name}:** {UNFILLED}")

    # --- Section 3: Effect Sizes ---
    report_lines.append("\n## 3. Effect Sizes")
    # The effect-size dictionary.
    effects = bundle.verdicts.get("effect_sizes", {})
    # Render each effect size gated on finiteness (no zero-fabrication).
    report_lines.append(f"- **Cohen's $d$:** {_num(effects.get('cohens_d'))}")
    report_lines.append(f"- **Cliff's $\\delta$:** {_num(effects.get('cliffs_delta'))}")
    report_lines.append(f"- **Relative Utility Gain (RUG):** {_num(effects.get('relative_utility_gain'))}\n")

    # --- Section 4: Methodological Disclosures ---
    report_lines.append("## 4. Methodological Disclosures")
    # Corpus fixity for the two non-canonical domains.
    non_canonical = cache.task_record_pruned[cache.task_record_pruned["domain"].isin(["finance", "long_context"])]
    if not non_canonical.empty:
        # Report demotion when any non-canonical task is non-confirmatory.
        if not non_canonical["is_confirmatory"].all():
            report_lines.append("- **Corpus Fixity:** Non-canonical domains lacked content hashes and were demoted to EXPLORATORY (excluded from H1-H5).")
        else:
            report_lines.append("- **Corpus Fixity:** Non-canonical domains provided content hashes and were admitted to confirmatory analysis.")
    # Tier-1 hash status (establish vs verified).
    expected_hash = cache.base_config["reproducibility"].get("tensor_sha256")
    is_establish = expected_hash is None or expected_hash == "<populated upon execution of the reference instantiation>"
    if is_establish:
        report_lines.append("- **Tier-1 Status:** ESTABLISH run. The tensor hash was computed and recorded but not verified against a prior baseline.")
    else:
        report_lines.append("- **Tier-1 Status:** VERIFIED. The computed tensor hash matches the configuration.")

    # --- Section 4b: Missing artifacts, unavailable versions, unresolved contamination ---
    report_lines.append("\n### 4b. Missing Artifacts & Open Disclosures")
    # Disclose the three deferred/None artifacts honestly (never fabricated).
    repro = cache.base_config.get("reproducibility", {})
    estim = cache.base_config.get("estimation", {})
    fusion_cfg = cache.base_config.get("fusion_sim_audit", {})
    # tensor_sha256 disclosure.
    if is_establish:
        report_lines.append("- **tensor_sha256:** established on this run (no prior baseline to verify against).")
    # reference_weight_w disclosure.
    if estim.get("reference_weight_w") is None:
        report_lines.append("- **reference_weight_w:** UNFILLED — MRC reference weights were not specified.")
    # per_stratum_tolerance disclosure.
    if fusion_cfg.get("per_stratum_tolerance") is None:
        report_lines.append("- **fusion per_stratum_tolerance:** UNFILLED — fusion-gap stratum flagging operated as a no-op.")
    # Unavailable model versions: any empty provider_version on the released tensor.
    missing_versions = (
        cache.primary_tensor.assign(_pv=cache.primary_tensor["provider_version"].astype(str))
        .loc[lambda d: d["_pv"].str.strip() == "", "model_id"].astype(str).unique().tolist()
    )
    if missing_versions:
        report_lines.append(f"- **Unavailable model versions:** {sorted(missing_versions)} lacked a recorded provider_version (Tier-2 drift bound widened).")
    # Unresolved contamination: surface from the persisted contamination report if present.
    contam_path = run_dir / "contamination_report.json"
    if contam_path.exists():
        try:
            # Load the contamination report to disclose any flagged/pruned items.
            with open(contam_path, "r", encoding="utf-8") as f:
                contam = json.load(f)
            # Report a pruned-item count if available.
            pruned = contam.get("n_pruned_items", contam.get("pruned_count"))
            if pruned:
                report_lines.append(f"- **Contamination:** {pruned} item(s) were flagged and pruned (drop-for-all rule, threshold 0.10).")
        except (json.JSONDecodeError, OSError):
            # If the report is unreadable, disclose that explicitly rather than omitting it.
            report_lines.append("- **Contamination:** report present but unreadable during packaging.")

    # --- Section 5: Two-Tier Reproducibility Claims (bit-exact gated on a verified hash) ---
    report_lines.append("\n## 5. Two-Tier Reproducibility Claims")
    if is_establish:
        # In establish mode, the bit-exact guarantee is pending verification.
        report_lines.append("- **Tier 1 (Computational):** A canonical tensor hash was recorded this run; bit-exact reproducibility downstream of the frozen tensor and seeds is claimable only once a subsequent run verifies against it.")
    else:
        # In verify mode, the bit-exact guarantee holds against the matched hash.
        report_lines.append("- **Tier 1 (Computational):** Bit-exact reproducibility downstream of the frozen tensor and seeds is verified against the matched tensor hash.")
    # Tier-2 wording references the reported variance components, gated on their finiteness.
    vc = bundle.variance_components or {}
    if all(_finite(vc.get(k)) for k in ("sigma2_task", "sigma2_gen", "sigma2_adj")):
        report_lines.append("- **Tier 2 (Empirical):** Re-collection is bounded and instrumented via the reported variance components ($\\sigma^2_{\\mathrm{task}}, \\sigma^2_{\\mathrm{gen}}, \\sigma^2_{\\mathrm{adj}}$) and the fusion residual in quadrature.")
    else:
        report_lines.append(f"- **Tier 2 (Empirical):** {UNFILLED} (variance components not fully computed).")

    # Serialize the report to disk.
    report_path = run_dir / "scientific_report.md"
    with open(report_path, "w", encoding="utf-8") as f:
        f.write("\n".join(report_lines))

# -------------------------------------------------------------------------------------------------------------------------------
# Task 24, Step 3: Persist Reproducibility Package
# -------------------------------------------------------------------------------------------------------------------------------

def persist_reproducibility_package(
    bundle: Any,
    cache: Any,
    robustness_report: Any,
    sigma_df: pd.DataFrame,
    coalition_samples: pd.DataFrame,
    run_dir: Path
) -> Path:
    """
    Serialize all artifacts with a mechanically verifiable tensor hash, release the
    exact hashed byte representation, mint a content-addressed persistent identifier,
    and archive the package — with a packaging self-test.

    Rationale (Section 13.3; detailed tasklist Task 24, Step 3; Task 11, Steps 2-3).
    Tier-1 verifiability requires that (a) the establish run records the REAL computed
    tensor hash (not a sentinel) and (b) the released artifact be byte-identical to the
    representation that was hashed. The prior implementation wrote
    'ESTABLISHED_ON_THIS_RUN' and released a CSV while hashing a JSON canonicalization,
    so the published hash was unverifiable. This version recomputes the canonical hash
    from the cached tensor, releases the exact hashed bytes
    (outcome_tensor.canonical.json), keeps the CSV with its own .sha256, records a
    content-addressed persistent identifier, and self-tests that the released canonical
    bytes re-hash to the published hash.

    Parameters
    ----------
    bundle : ResultsBundle
        Final results (FMEC table, CIs, selection, variance components, RSP).
    cache : PipelineCache
        Frozen substrate (primary tensor, config, qualification, manifest source).
    robustness_report : RobustnessReport
        Robustness surface and ablation results.
    sigma_df : pd.DataFrame
        The error covariance matrix.
    coalition_samples : pd.DataFrame
        The evaluated coalition samples.
    run_dir : Path
        The immutable run directory.

    Returns
    -------
    Path
        Path to the generated .tar.gz archive.

    Raises
    ------
    RuntimeError
        If the packaging self-test fails (released bytes do not re-hash to the hash).
    IOError
        If writing artifacts fails.
    """
    # 1. Prompt templates (exact strings used for generation).
    with open(run_dir / "prompts.json", "w", encoding="utf-8") as f:
        json.dump(cache.base_config.get("prompts", {}), f, indent=2)

    # 2. Seed ledger (entropy provenance).
    with open(run_dir / "seeds.json", "w", encoding="utf-8") as f:
        json.dump(cache.base_config.get("reproducibility", {}).get("seed_ledger", {}), f, indent=2)

    # 3a. Recompute the canonical tensor and its REAL SHA-256 from the cached tensor (Task 11, Step 2).
    canonical_df, computed_tensor_hash = canonicalize_and_hash_tensor(cache.primary_tensor)

    # 3b. Re-serialize the EXACT bytes that were hashed (mirroring canonicalize_and_hash_tensor).
    # to_json with fixed precision, parsed, then dumped with sorted keys and tight separators.
    json_str = canonical_df.to_json(orient="records", double_precision=15)
    parsed_records = json.loads(json_str)
    canonical_bytes = json.dumps(parsed_records, sort_keys=True, separators=(",", ":")).encode("utf-8")
    # Release the exact hashed representation so a reader can mechanically re-derive the hash.
    with open(run_dir / "outcome_tensor.canonical.json", "wb") as f:
        f.write(canonical_bytes)

    # 3c. Release the human-friendly CSV for convenience, with its own content hash.
    cache.primary_tensor.to_csv(run_dir / "outcome_tensor.csv", index=False)
    # Compute and publish the CSV's SHA-256 so the CSV is independently verifiable.
    with open(run_dir / "outcome_tensor.csv", "rb") as f:
        csv_hash = hashlib.sha256(f.read()).hexdigest()
    with open(run_dir / "outcome_tensor.csv.sha256", "w", encoding="utf-8") as f:
        f.write(csv_hash)

    # 4. Write the REAL tensor hash (never a sentinel). In verify mode, prefer the pinned hash.
    expected_hash = cache.base_config["reproducibility"].get("tensor_sha256")
    is_establish = expected_hash is None or expected_hash == "<populated upon execution of the reference instantiation>"
    # In establish mode write the computed hash; in verify mode write the pinned (matched) hash.
    hash_to_write = computed_tensor_hash if is_establish else str(expected_hash)
    with open(run_dir / "tensor_sha256.txt", "w", encoding="utf-8") as f:
        f.write(hash_to_write)

    # 4b. Packaging self-test: the released canonical bytes MUST re-hash to the published hash.
    recomputed = hashlib.sha256(canonical_bytes).hexdigest()
    # The released canonical JSON must reproduce the computed hash by construction.
    if recomputed != computed_tensor_hash:
        raise RuntimeError(
            "Packaging self-test failed: released canonical bytes do not re-hash to the computed tensor hash "
            f"({recomputed} != {computed_tensor_hash})."
        )
    # In verify mode, additionally assert the computed hash matches the pinned hash (Tier-1 gate echo).
    if (not is_establish) and (computed_tensor_hash != str(expected_hash)):
        raise RuntimeError(
            "Packaging self-test failed: computed tensor hash does not match the pinned tensor_sha256 "
            f"({computed_tensor_hash} != {expected_hash})."
        )

    # 5. Coalition samples and covariance.
    coalition_samples.to_json(run_dir / "coalition_samples.json", orient="records", indent=2)
    sigma_df.to_csv(run_dir / "covariance.csv", index=True)

    # 6. Authentic qualification table.
    cache.qualification_df.to_csv(run_dir / "qualification.csv", index=False)

    # 7. FMEC estimates, variance components, bootstrap outputs.
    bundle.fmec_table.to_json(run_dir / "fmec_estimates_final.json", orient="records", indent=2)
    with open(run_dir / "variance_components.json", "w", encoding="utf-8") as f:
        json.dump(bundle.variance_components, f, indent=2)
    with open(run_dir / "bootstrap_outputs.json", "w", encoding="utf-8") as f:
        json.dump({"CIs": bundle.CIs, "selection_freq": bundle.selection_freq, "RSP": bundle.RSP}, f, indent=2)

    # 8. Selected ensemble and weights.
    with open(run_dir / "selected_ensemble.json", "w", encoding="utf-8") as f:
        json.dump({"S_FMEC": bundle.S_FMEC, "weights": bundle.w}, f, indent=2)

    # 9. Model-version manifest (unique model/version/fingerprint triples).
    manifest_df = cache.primary_tensor[["model_id", "provider_version", "system_fingerprint"]].drop_duplicates()
    manifest_df.to_json(run_dir / "model_version_manifest.json", orient="records", indent=2)

    # 10. Robustness report and stability surface.
    with open(run_dir / "robustness_report.json", "w", encoding="utf-8") as f:
        json.dump({
            "max_stability_mass": robustness_report.max_stability_mass,
            "is_fragile": robustness_report.is_fragile,
            "ablation_results": robustness_report.ablation_results,
        }, f, indent=2)
    robustness_report.stability_surface.to_csv(run_dir / "stability_surface.csv", index=False)

    # 11. Ensure the contamination report exists in the package.
    contamination_path = run_dir / "contamination_report.json"
    if not contamination_path.exists():
        with open(contamination_path, "w", encoding="utf-8") as f:
            json.dump({"note": "Contamination report was missing during packaging."}, f, indent=2)

    # 12. Create the immutable compressed archive of the entire run directory.
    archive_base_name = str(run_dir.resolve())
    archive_path_str = shutil.make_archive(
        base_name=archive_base_name, format="gztar",
        root_dir=str(run_dir.parent), base_dir=run_dir.name,
    )
    archive_path = Path(archive_path_str)

    # 13. Compute the archive's SHA-256 (streamed) for ultimate provenance.
    sha256_hash = hashlib.sha256()
    with open(archive_path, "rb") as f:
        for byte_block in iter(lambda: f.read(4096), b""):
            sha256_hash.update(byte_block)
    archive_digest = sha256_hash.hexdigest()
    # Write the archive hash as a sibling file.
    with open(f"{archive_path_str}.sha256", "w", encoding="utf-8") as f:
        f.write(archive_digest)

    # 14. Record a content-addressed persistent identifier (DOI left null and documented).
    persistent_identifier = {
        "content_id": f"sha256:{archive_digest}",                 # immutable content-addressed identifier
        "archive_sha256": archive_digest,                         # archive content hash
        "tensor_sha256": hash_to_write,                           # the published, verifiable tensor hash
        "tensor_canonical_artifact": "outcome_tensor.canonical.json",  # exact hashed representation
        "doi": None,                                              # to be minted (e.g., Zenodo); never fabricated
        "note": "Content-addressed identifier; mint a DOI for the archive to obtain a citable persistent identifier.",
        "tier1_self_test": "passed",                              # released bytes re-hash to the published hash
    }
    # Write the persistent-identifier record as a sibling of the archive.
    with open(f"{archive_path_str}.persistent_identifier.json", "w", encoding="utf-8") as f:
        json.dump(persistent_identifier, f, indent=2)

    # Return the archive path.
    return archive_path

# -------------------------------------------------------------------------------------------------------------------------------
# Task 24, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def report_and_package(
    bundle: ResultsBundle,
    cache: PipelineCache,
    robustness_report: RobustnessReport,
    sigma_df: pd.DataFrame,
    coalition_samples: pd.DataFrame
) -> Path:
    """
    Orchestrates Task 24: Generates the mandated figure suite, writes the
    clinical scientific report, serializes all artifacts, and packages the
    run into an immutable, cryptographically hashed archive.

    Parameters
    ----------
    bundle : ResultsBundle
        The final results dataclass.
    cache : PipelineCache
        The frozen data-generating substrate.
    robustness_report : RobustnessReport
        The robustness analysis results.
    sigma_df : pd.DataFrame
        The error covariance matrix.
    coalition_samples : pd.DataFrame
        The DataFrame of evaluated coalitions.

    Returns
    -------
    Path
        The file path to the generated .tar.gz reproducibility archive.

    Raises
    ------
    RuntimeError
        If the run directory does not exist or is not writable.
    """
    run_dir = cache.run_dir

    # Validate that the run directory exists before attempting I/O
    if not run_dir.exists() or not run_dir.is_dir():
        raise RuntimeError(f"Packaging Failed: Run directory {run_dir} does not exist.")

    # Step 1: Generate the Figure Suite
    generate_figure_suite(
        bundle=bundle,
        cache=cache,
        sigma_df=sigma_df,
        run_dir=run_dir
    )

    # Step 2: Generate Scientific Report (Falsification & Disclosures)
    generate_scientific_report(
        bundle=bundle,
        robustness_report=robustness_report,
        cache=cache,
        run_dir=run_dir
    )

    # Step 3: Persist Reproducibility Package and Create Archive
    archive_path = persist_reproducibility_package(
        bundle=bundle,
        cache=cache,
        robustness_report=robustness_report,
        sigma_df=sigma_df,
        coalition_samples=coalition_samples,
        run_dir=run_dir
    )

    return archive_path
````

````python
# Top-Level Orchestrator

def run_experimental_suite(
    raw_config: Dict[str, Any],
    model_answer_raw: List[Dict[str, Any]],
    candidate_solution_raw: List[Dict[str, Any]],
    task_record_raw: List[Dict[str, Any]],
    qualification_raw: pd.DataFrame,
    raw_proxy_texts: List[str]
) -> Path:
    """
    Orchestrates the entire end-to-end FMEC research and optimization workflow.

    This ultimate coordinator sequentially executes:
    1. Phase 1: Data collection, scoring, and outcome tensor canonicalization.
    2. Phase 2: Expected MEC, CMI Information Gain, and Ledoit-Wolf Covariance estimation.
    3. Phase 3: Robustness sweeps (10,000 Dirichlet draws) and pre-registered ablations.
    4. Phase 4: Figure generation, scientific reporting, and reproducibility packaging.
    It guarantees complete state isolation, prevents look-ahead bias, and maintains
    the strict analytical boundaries required for top-tier funds.

    Parameters
    ----------
    raw_config : Dict[str, Any]
        The raw study configuration dictionary.
    model_answer_raw : List[Dict[str, Any]]
        The raw free-form generation records.
    candidate_solution_raw : List[Dict[str, Any]]
        The raw structured solution records.
    task_record_raw : List[Dict[str, Any]]
        The raw task and ground truth records.
    qualification_raw : pd.DataFrame
        The raw benchmark eligibility DataFrame.
    raw_proxy_texts : List[str]
        The raw text corpus used for the contamination audit.

    Returns
    -------
    Path
        The file path to the final, cryptographically hashed reproducibility archive (.tar.gz).
    """
    logger.info("==========================================================")
    logger.info("Starting End-to-End FMEC Experimental Suite Optimization...")
    logger.info("==========================================================")

    # Input Validation: Assert that the raw datasets are non-empty to prevent pipeline collapse
    assert len(model_answer_raw) > 0, "Input Error: model_answer_raw cannot be empty."
    assert len(candidate_solution_raw) > 0, "Input Error: candidate_solution_raw cannot be empty."
    assert len(task_record_raw) > 0, "Input Error: task_record_raw cannot be empty."
    assert len(qualification_raw) > 0, "Input Error: qualification_raw cannot be empty."
    assert len(raw_proxy_texts) > 0, "Input Error: raw_proxy_texts cannot be empty."

    # --------------------------------------------------------------------------
    # PHASE 1: Data-Generating Substrate & Baseline Point Estimation
    # --------------------------------------------------------------------------
    logger.info("Executing Phase 1: Ingestion, Collection, and Tensor Assembly...")
    # Execute run_pipeline with no cached state and no perturbations
    # This runs the deterministic substrate (Tasks 1-11) and baseline estimation (Tasks 12-21)
    baseline_bundle, cache = run_pipeline(
        raw_config=raw_config,
        model_answer_raw=model_answer_raw,
        candidate_solution_raw=candidate_solution_raw,
        task_record_raw=task_record_raw,
        qualification_raw=qualification_raw,
        raw_proxy_texts=raw_proxy_texts,
        cached_state=None, # Execute Step 1 (generate cache)
        perturbation=None  # Baseline run
    )

    logger.info(f"Baseline Outcome Tensor Hash: {cache.base_config['reproducibility'].get('tensor_sha256')}")
    logger.info(f"Baseline Selected Ensemble: {baseline_bundle.S_FMEC}")

    # --------------------------------------------------------------------------
    # PHASE 2: Parallel Robustness & Sensitivity Analysis
    # --------------------------------------------------------------------------
    logger.info("Executing Phase 2: Dirichlet Sweeps and Ablations...")
    # Execute robustness sweeps over the cached state (Tasks 22-23)
    # Reuses the cached outcome tensor to bypass expensive network collections
    robustness_report = robustness(
        run_pipeline_callable=run_pipeline,
        cache=cache,
        baseline_bundle=baseline_bundle
    )

    # --------------------------------------------------------------------------
    # PHASE 3: Extraction of Point-Estimate Primitives
    # --------------------------------------------------------------------------
    logger.info("Executing Phase 3: Extracting Point-Estimate Primitives...")
    # To package the exact point-estimate primitives (covariance matrix, coalition samples)
    # without introducing look-ahead bias, we execute a lightweight, cached estimation run
    # execute_estimation_selection_segment is assumed available
    estimation_outputs = execute_estimation_selection_segment(
        cache=cache,
        perturbation=None
    )

    # Extract the covariance DataFrame
    sigma_df = estimation_outputs.sigma_df

    # Extract the exact permutation coalition samples generated during the baseline run
    # We re-run MEC estimation on the cached state to deterministically extract the samples
    # estimate_mec is assumed available (Task 13)
    _, coalition_samples = estimate_mec(
        qualified_pool=cache.qualified_pool,
        v_func=build_utility_function(
            estimator_projection=cache.estimator_projection,
            sigma=sigma_df.values,
            config=cache.base_config,
            qualified_pool=cache.qualified_pool,
            judge_model_id=cache.base_config["models"]["judge"]
        ),
        config=cache.base_config,
        rng_bundle=cache.rng_bundle
    )

    # --------------------------------------------------------------------------
    # PHASE 4: Reporting, Visualization & Packaging
    # --------------------------------------------------------------------------
    logger.info("Executing Phase 4: Visualization, Reporting, and Archiving...")

    # Generate the figures (Step 1), Markdown report (Step 2), and compress (Step 3)
    # We pass our newly refactored persist_reproducibility_package logic inside this step
    # generate_figure_suite, generate_scientific_report are assumed available (Task 24)
    generate_figure_suite(
        bundle=baseline_bundle,
        cache=cache,
        sigma_df=sigma_df,
        run_dir=cache.run_dir
    )

    generate_scientific_report(
        bundle=baseline_bundle,
        robustness_report=robustness_report,
        cache=cache,
        run_dir=cache.run_dir
    )

    archive_path = persist_reproducibility_package(
        bundle=baseline_bundle,
        cache=cache,
        robustness_report=robustness_report,
        sigma_df=sigma_df,
        coalition_samples=coalition_samples,
        run_dir=cache.run_dir
    )

    logger.info("==========================================================")
    logger.info("FMEC Experimental Suite Optimization Complete!")
    logger.info(f"RSP Stability Mass: {baseline_bundle.RSP:.3f}")
    logger.info(f"Fragility Verdict: {robustness_report.is_fragile}")
    logger.info(f"Provenance Archive: {archive_path}")
    logger.info("==========================================================")

    # Return the file path to the final immutable tarball
    return archive_path
````
